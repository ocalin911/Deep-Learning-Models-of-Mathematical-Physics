# **Inverse Sturm--Liouville Problem**

The frequencies of vibration are determined by the eigenvalues $\{\omega_n\}$ satisfying

$$
X_n''(x) = -\omega_n^2 \rho(x) X_n(x),
\qquad X_n(0)=X_n(L)=0,
$$

with associated eigenfunctions $X_n(x)$.
Sturm--Liouville theory ensures that the eigenvalues are real, positive, simple, and increase to infinity.

The inverse problem may now be stated as follows:
Given the frequency spectrum $\{\omega_n\}$, determine the weight density $\rho(x)$.


```python
import numpy as np
import scipy.sparse as sp
import scipy.sparse.linalg as spla

# --- problem setup ---
L = 10.0
N = 15000                  # interior grid points (increase for higher accuracy)
x = np.linspace(0, L, N+2)
h = x[1] - x[0]
xi = x[1:-1]              # interior nodes

# density: rho(x) = ((x-5)^2 + 1) / (10 + 10^3/12), normalized to ∫_0^10 rho = 1
rho = ((xi - 5.0)**2 + 1.0) / (L + L**3/12.0)

# (optional) check normalization
mass_check = np.trapz(((x - 5.0)**2 + 1.0) / (L + L**3/12.0), x)
print(f"∫ rho dx ≈ {mass_check:.8f}")

# --- build matrices ---
# Stiffness (FD for -d^2/dx^2 with Dirichlet): tri-diagonal [1, -2, 1]/h^2
main = 2.0 * np.ones(N) / h**2
off  = -1.0 * np.ones(N-1) / h**2
A = sp.diags([off, main, off], offsets=[-1, 0, 1], format='csc')

# Mass matrix: diagonal with rho(x_i)
B = sp.diags(rho, offsets=0, format='csc')

# --- solve generalized eigenproblem: A v = λ B v, with λ = ω^2 ---
k = 20  # number of smallest eigenpairs wanted
# shift-invert around sigma=0 to target the smallest eigenvalues
evals, evecs = spla.eigsh(A, k=k, M=B, sigma=0.0, which='LM')

# frequencies
omegas = np.sqrt(np.maximum(evals, 0.0))

print("First few eigenfrequencies ω_n:")
for n, w in enumerate(sorted(omegas), 1):
    print(f"  n={n:2d}: ω_{n} ≈ {w:.8f}")

# (optional) normalize eigenmodes in ρ-weighted inner product:
def rho_weighted_norm(v):
    return np.sqrt(np.trapz(rho * v**2, xi))

modes = []
for j in range(k):
    v = evecs[:, j]
    v = v / rho_weighted_norm(v)
    modes.append(v)
# modes[j] holds the j-th normalized mode on interior nodes xi
```

```python
# pinn_sturm_liouville_q_N20.py
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt

# -----------------------------
# Problem setup
# -----------------------------
L = 10.0

# Provided eigenfrequencies (20 values)
omegas = np.array([
    1.39823803, 2.03687761, 3.51833599, 4.26718726, 5.64515014,
    6.48404027, 7.78151322, 8.69059828, 9.92854681, 10.88995860,
    12.08489683, 13.08425539, 14.24851690, 15.27496910, 16.41750934,
    17.46314468, 18.59036801, 19.64952673, 20.76597542, 21.83464725
], dtype=np.float64)

# Ground-truth q(x) = rho(x) for reference plotting
def q_true(x):  # rho(x)
    return ((x - 5.0)**2 + 1.0) / (L + L**3/12.0)

# Collocation grid and trapezoidal weights
m = 2560
x_np = np.linspace(0.0, L, m).reshape(-1, 1).astype(np.float64)
w_np = np.ones((m, 1), dtype=np.float64) * (L/(m-1))
w_np[0, 0]  *= 0.5
w_np[-1, 0] *= 0.5

# Normalize x to [-1,1] for NN input; add squared feature (a symmetry cue)
x_center  = L / 2.0
x_scale   = L / 2.0
x_norm_np = ((x_np - x_center) / x_scale).astype(np.float64)
feats_np  = np.concatenate([x_norm_np, x_norm_np**2], axis=1)

# -----------------------------
# Model builder (N=20 fixed)
# -----------------------------
N = 20

def build_model(N):
    inp = tf.keras.Input(shape=(2,), dtype=tf.float64)  # [x_norm, x_norm^2]

    # Shared trunk
    h = tf.keras.layers.Dense(64, activation='swish', dtype=tf.float64)(inp)
    h = tf.keras.layers.Dense(64, activation='swish', dtype=tf.float64)(h)

    # q(x) head (pre-softplus)
    q_raw = tf.keras.layers.Dense(1, activation=None, dtype=tf.float64, name='q_raw')(h)

    # N heads for eigenfunctions (pre-BC embedding)
    X_heads = []
    for k in range(N):
        hk = tf.keras.layers.Dense(64, activation='swish', dtype=tf.float64)(h)
        hk = tf.keras.layers.Dense(64, activation='swish', dtype=tf.float64)(hk)
        outk = tf.keras.layers.Dense(1, activation=None, dtype=tf.float64, name=f'nn_k_{k+1}')(hk)
        X_heads.append(outk)

    return tf.keras.Model(inputs=inp, outputs=[q_raw] + X_heads)

# -----------------------------
# Training step (persistent tape + symmetry penalty)
# -----------------------------
@tf.function(jit_compile=False)
def train_step(model, x_norm_tf, x_phys_tf, w_tf, omegas_sel, optimizer, lambda_sym):
    omegas_list = tf.unstack(omegas_sel)  # avoid iterating a symbolic Tensor

    with tf.GradientTape(persistent=True) as tape:
        tape.watch(x_phys_tf)  # derivatives w.r.t. physical x

        # Features
        feats = tf.concat([x_norm_tf, tf.square(x_norm_tf)], axis=1)

        outs    = model(feats, training=True)
        q_raw   = outs[0]      # (m,1)
        nn_ks   = outs[1:]     # list of (m,1)

        # Positivity + normalization: q(x) >= 0 and ∫ q = 1 (string total mass)
        q_pos   = tf.nn.softplus(q_raw)
        q       = q_pos / tf.reduce_sum(q_pos * w_tf)

        # PDE residual losses across modes
        losses = []
        for j, omega in enumerate(omegas_list):
            nnk = nn_ks[j]
            # Dirichlet hard-embedding: X_k(0)=X_k(L)=0
            Xk = x_phys_tf * (L - x_phys_tf) * nnk

            # First and second derivative w.r.t. x
            Xk_x  = tape.gradient(Xk, x_phys_tf)
            Xk_xx = tape.gradient(Xk_x, x_phys_tf)

            # Residual of: X'' + ω^2 q X = 0  (equivalently -X'' = ω^2 q X)
            res    = Xk_xx + (omega**2) * q * Xk
            loss_k = tf.reduce_mean(tf.square(res))
            losses.append(loss_k)

        loss_pde = tf.add_n(losses)

        # Symmetry penalty q(x) ≈ q(L-x) (uniform grid => reverse)
        q_mirror = tf.reverse(q, axis=[0])
        L_sym    = tf.reduce_mean(tf.square(q - q_mirror))

        # Total loss
        loss_total = loss_pde + lambda_sym * L_sym

    grads = tape.gradient(loss_total, model.trainable_variables)
    del tape
    optimizer.apply_gradients(zip(grads, model.trainable_variables))
    return loss_total, loss_pde, L_sym

# -----------------------------
# Training loop (N = 20 only)
# -----------------------------
def train_N20(epochs=8000, lr=8e-4, lambda_sym=1, seed=2024):
    tf.keras.backend.clear_session()
    tf.random.set_seed(seed)
    np.random.seed(seed)

    model = build_model(N)
    optimizer = tf.keras.optimizers.Adam(learning_rate=lr)

    # Tensors
    x_norm_tf  = tf.convert_to_tensor(x_norm_np)
    x_phys_tf  = tf.convert_to_tensor(x_np)
    w_tf       = tf.convert_to_tensor(w_np)
    omegas_sel = tf.convert_to_tensor(omegas[:N], dtype=tf.float64)

    Ltot_hist, Lpde_hist, Lsym_hist = [], [], []
    for ep in range(epochs):
        Ltot, Lpde, Lsym = train_step(model, x_norm_tf, x_phys_tf, w_tf,
                                      omegas_sel, optimizer, lambda_sym)
        Ltot_hist.append(float(Ltot.numpy()))
        Lpde_hist.append(float(Lpde.numpy()))
        Lsym_hist.append(float(Lsym.numpy()))
        # Simple LR schedule
        if (ep + 1) in (400, 650):
            optimizer.learning_rate.assign(optimizer.learning_rate * 0.3)

    # Extract learned q(x) on the grid
    outs    = model.predict(feats_np, verbose=0)
    q_raw   = outs[0]
    q_pos   = tf.nn.softplus(tf.convert_to_tensor(q_raw)).numpy()
    q_rec   = (q_pos / np.sum(q_pos * w_np)).reshape(-1)
    return q_rec, np.array(Ltot_hist), np.array(Lpde_hist), np.array(Lsym_hist)

# -----------------------------
# Run (N = 20) and plots
# -----------------------------
q_rec, Ltot, Lpde, Lsym = train_N20(epochs=800, lr=2.2e-2, lambda_sym=0.85, seed=2024)

# L2 error for reference
def trapz_L2_error(f, g, x, w):
    return np.sqrt(np.sum(((f - g)**2) * w))

q_true_vals = q_true(x_np)
err_L2 = trapz_L2_error(q_rec.reshape(-1,1), q_true_vals, x_np, w_np)
print(f"L2 error ||q_rec - q_true|| ≈ {err_L2:.4e}")

# Plot q(x): true vs recovered
plt.figure(figsize=(9,5))
plt.plot(x_np, q_true_vals, 'k--', label='True $q(x)$')
plt.plot(x_np, q_rec, label='Recovered $q_\\theta(x)$')
plt.xlabel('$x$')
plt.ylabel('$q(x)$')
plt.title('True vs. recovered $q(x)$ with N=20 frequencies')
plt.legend()
plt.tight_layout()
plt.show()

# Plot losses
plt.figure(figsize=(9,5))
plt.plot(Ltot, label='Total loss')
plt.plot(Lpde, label='PDE loss')
plt.plot(Lsym, label='Symmetry loss')
plt.yscale('log')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.title('Losses across epochs (N=20)')
plt.legend()
plt.tight_layout()
plt.show()
```

# **Deep Learning for GPS--INS Sensor Fusion**

We consider a vehicle moving in a plane with unit speed along the unit circle


$$
x(t) = (\cos t, \sin t), \qquad t \in [0, 2\pi].
$$


The initial conditions are given as $x_0 = (1,0)$ and $v_0 = (0,1)$.
The available measurements consist of noisy position data


$$
x_j = \big(\cos \tau_j + \epsilon_j^1, \; \sin \tau_j + \epsilon_j^2\big),
$$


and noisy acceleration data


$$
\ddot{x}_j = \big(-\cos t_j + \eta_j^1, \; -\sin t_j + \eta_j^2\big),
$$


where $\epsilon_j^k$ and $\eta_j^k$ are Gaussian noise terms with standard deviation equal to $5\%$ of the signal amplitude.
We used $N_T = 300$ INS acceleration measurements and $M_T = 5$ GPS position measurements.


The trajectory was modeled with a neural network having three hidden layers of 64 neurons each, with $\tanh$ activation
and an $L^2$ regularization penalty of $10^{-5}$ per layer.
Training was performed for 9,000 epochs using the Adam optimizer with a learning rate of $10^{-4}$.

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, Model, regularizers
import matplotlib.pyplot as plt

# -----------------------------
# 1. True trajectory (circle)
# -----------------------------
def true_traj(t):
    return np.stack([np.cos(t), np.sin(t)], axis=1)

def true_accel(t):
    return np.stack([-np.cos(t), -np.sin(t)], axis=1)

# Initial conditions
x0 = np.array([1.0, 0.0], dtype=np.float32)
v0 = np.array([0.0, 1.0], dtype=np.float32)

# -----------------------------
# 2. Generate noisy INS + GPS data
# -----------------------------
N_T = 300              # INS samples (dense)
M_T = 5             # GPS samples (sparser)
noise_pos = 0.05
noise_acc = 0.05

# INS times
t_ins = np.linspace(0, 2*np.pi, N_T)
x_true_ins = true_traj(t_ins)
a_true_ins = true_accel(t_ins)
x_ins = x_true_ins + noise_pos * np.random.randn(N_T, 2)
a_ins = a_true_ins + noise_acc * np.random.randn(N_T, 2)

# GPS times (subset)
tau_gps = np.linspace(0, 2*np.pi, M_T)
x_true_gps = true_traj(tau_gps)
x_gps = x_true_gps + noise_pos * np.random.randn(M_T, 2)

# Tensor format
t_ins_tf = tf.convert_to_tensor(t_ins.reshape(-1,1), dtype=tf.float32)
a_ins_tf = tf.convert_to_tensor(a_ins, dtype=tf.float32)
tau_gps_tf = tf.convert_to_tensor(tau_gps.reshape(-1,1), dtype=tf.float32)
x_gps_tf = tf.convert_to_tensor(x_gps, dtype=tf.float32)

# -----------------------------
# 3. Free neural network nn(theta;t)
# -----------------------------
def build_free_nn(hidden_layers=3, hidden_units=64, reg=1e-5):
    inputs = layers.Input(shape=(1,))
    x = inputs
    for _ in range(hidden_layers):
        x = layers.Dense(hidden_units, activation="tanh",
                         kernel_regularizer=regularizers.l2(reg))(x)
    outputs = layers.Dense(2)(x)
    return Model(inputs, outputs)

free_nn = build_free_nn()

# Embedded trajectory model with position + velocity constraints
def trajectory_model(t):
    nn_out = free_nn(t)  # shape (N,2)
    return x0 + t * v0 + t**2 * nn_out


# -----------------------------
# 4. Loss function: INS + GPS
# -----------------------------
lambda_gps = N_T / M_T  # balance factor

def loss_fn():
    # INS loss (acceleration)
    with tf.GradientTape(persistent=True) as tape2:
        tape2.watch(t_ins_tf)
        with tf.GradientTape(persistent=True) as tape:
            tape.watch(t_ins_tf)
            x_pred_ins = trajectory_model(t_ins_tf)
        dxdt = tape.gradient(x_pred_ins, t_ins_tf)
    d2xdt2 = tape2.gradient(dxdt, t_ins_tf)
    del tape, tape2
    loss_ins = tf.reduce_mean(tf.square(d2xdt2 - a_ins_tf))

    # GPS loss (positions)
    x_pred_gps = trajectory_model(tau_gps_tf)
    loss_gps = tf.reduce_mean(tf.square(x_pred_gps - x_gps_tf))

    # Total loss
    return loss_ins + 1*lambda_gps * loss_gps

optimizer = tf.keras.optimizers.Adam(1e-4)

@tf.function
def train_step():
    with tf.GradientTape() as tape:
        loss = loss_fn()
    grads = tape.gradient(loss, free_nn.trainable_variables)
    optimizer.apply_gradients(zip(grads, free_nn.trainable_variables))
    return loss

# -----------------------------
# 5. Training loop
# -----------------------------
loss_history = []
epochs = 9000
for epoch in range(epochs):
    loss = train_step()
    loss_history.append(loss.numpy())
    if epoch % 500 == 0:
        print(f"Epoch {epoch}, Loss: {loss.numpy():.6f}")

# -----------------------------
# 6. Evaluate + plots
# -----------------------------
t_test = np.linspace(0, 2*np.pi, 300).reshape(-1,1).astype(np.float32)
x_pred = trajectory_model(tf.convert_to_tensor(t_test)).numpy()
x_true = true_traj(t_test.flatten())

# Plot recovered vs true trajectory
plt.figure(figsize=(5,5))
plt.plot(x_true[:,0], x_true[:,1], 'k--', label="True trajectory (circle)")
plt.plot(x_pred[:,0], x_pred[:,1], 'r', label="Recovered trajectory (NN)")
plt.scatter(x_gps[:,0], x_gps[:,1], s=30, c='b', alpha=0.6, label="GPS noisy")
plt.scatter(x_ins[:,0], x_ins[:,1], s=10, c='g', alpha=0.3, label="INS noisy")
plt.axis("equal")
plt.legend()
#plt.title("Trajectory Recovery with INS only")
#plt.title("Trajectory Recovery with GPS only")
plt.title("Trajectory Recovery with INS + GPS fusion")
plt.show()

# Plot loss curve
plt.figure(figsize=(5,5))
plt.plot(loss_history)
plt.yscale("log")
plt.xlabel("Epoch")
plt.ylabel("Loss")
#plt.title("Training Loss (INS)")
#plt.title("Training Loss (GPS)")
plt.title("Training Loss (INS + GPS Fusion)")
plt.show()
```

**Brachistocrone Curve**

The number of solutions of the nonlinear equation $$\frac{\cos u - 1}{u-\sin u} = m$$
is equal to the number of intersections of the graphs of the functions $y=m$ and $  y =\frac{\cos u - 1}{u-\sin u}.$

 For $m<0$ there is at least one solution. For $m=0$ there are infinitely many solutions.

 We multiplied by 20 for visibility reasons.


```python
import numpy as np
import matplotlib.pyplot as plt

# Define the function f(u)
def f(u):
    # To avoid division by zero at u = 0, use np.where
    denominator = u - np.sin(u)
    return np.where(denominator != 0, 20 * (np.cos(u) - 1) / denominator, np.nan)

# Generate u values in the interval (0, 40]
u_vals = np.linspace(0.01, 40, 1000)  # start at 0.01 to avoid singularity at 0

# Compute f(u)
f_vals = f(u_vals)

# Plotting
plt.figure(figsize=(8, 5))
plt.plot(u_vals, f_vals, label=r"$f(u) = \frac{20(\cos u - 1)}{u - \sin u}$")
plt.xlim(0, 40)
plt.ylim(-20, 5)
plt.xlabel("u")
plt.ylabel("f(u)")
plt.title("Plot of $f(u) = 20(\cos u - 1)/(u - \sin u)$")
plt.grid(True)
plt.axhline(0, color='black', linewidth=0.5)
plt.axhline(-3.5, color='red', linestyle='--', linewidth=1.2, label=r"$y = -3.5$")
plt.axvline(0, color='black', linewidth=0.5)
plt.legend()
plt.tight_layout()
plt.show()
```

The following code produces a learned curve (in blue) for the cycloid (in dashed orange).

```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from scipy.optimize import fsolve

# 1) Boundary points
A = np.array([0.0, 1.0])   # (x0, y0)
B = np.array([1.0, 0.0])   # (x1, y1)

# 2) Discretize parameter t ∈ [0,1]
n_points = 1200
t_values = np.linspace(0, 1, n_points).reshape(-1,1).astype(np.float32)

# 3) Build the neural network model
inp = keras.Input(shape=(1,))
h = layers.Dense(32, activation='tanh')(inp)
h = layers.Dense(32, activation='tanh')(h)
h = layers.Dense(32, activation='tanh')(h)
h = layers.Dense(32, activation='tanh')(h)
h = layers.Dense(32, activation='tanh')(h)
h = layers.Dense(32, activation='tanh')(h)
h = layers.Dense(32, activation='tanh')(h)
x_out = layers.Dense(1)(h)
y_out = layers.Dense(1, activation='softplus')(h)  # ensures y>0
model = keras.Model(inputs=inp, outputs=[x_out, y_out])

# 4) Loss components
def boundary_loss(x_pred, y_pred):
    x0, y0 = x_pred[0],   y_pred[0]
    x1, y1 = x_pred[-1],  y_pred[-1]
    return (
        tf.square(x0 - A[0]) +
        tf.square(y0 - A[1]) +
        30.0*tf.square(x1 - B[0]) +
        30.0*tf.square(y1 - B[1])
    )

@tf.function
def lagrangian_loss(t_tensor):
    with tf.GradientTape(persistent=True) as tape:
        tape.watch(t_tensor)
        x_pred, y_pred = model(t_tensor)
    dx_dt = tape.gradient(x_pred, t_tensor)
    dy_dt = tape.gradient(y_pred, t_tensor)
    y_safe = tf.maximum(y_pred, 1e-12)
    integrand = tf.sqrt(dx_dt**2 + dy_dt**2) / tf.sqrt(y_safe)
    dt = t_tensor[1] - t_tensor[0]
    return tf.reduce_sum(integrand * dt)


@tf.function
def total_loss(y_true, y_pred):
    # ignore y_true / y_pred from compile; compute from t_values
    t_tensor = tf.constant(t_values)
    x_pred, y_pred = model(t_tensor)
    return 100.0 * boundary_loss(x_pred, y_pred) + lagrangian_loss(t_tensor)

# 5) Compile & train
model.compile(optimizer=keras.optimizers.Adam(1e-2), loss=total_loss)
history = model.fit(
    t_values,
    [np.zeros_like(t_values), np.zeros_like(t_values)],
    epochs=10,
    verbose=0
)

# 6) Predict the learned curve
x_pred, y_pred = model.predict(t_values)

# 7) Compute the exact cycloid passing through (0,1)→(1,0)
# Solve θ - sinθ = 1 - cosθ for θ_max
f = lambda th: th - np.sin(th) - (1 - np.cos(th))
theta_max = fsolve(f, 2.0)[0]
R = 1.0 / (theta_max - np.sin(theta_max))

theta = np.linspace(0, theta_max, 500)
x_c = R * (theta - np.sin(theta))
y_c = 1.0 - R * (1 - np.cos(theta))

# 8) Plot: Learned curve + Exact cycloid + Endpoints, and Loss
plt.figure(figsize=(12,4))

# Left: curves
plt.subplot(1,2,1)
plt.plot(x_pred, y_pred,  '-', lw=2, label='Learned curve')
plt.plot(x_c,     y_c,    '--', lw=2, label='Exact cycloid')
plt.scatter([A[0], B[0]], [A[1], B[1]], color='red', zorder=5, label='Endpoints')
plt.title("Brachistochrone: NN vs Exact Cycloid")
plt.xlabel("x"); plt.ylabel("y")
plt.legend(); plt.grid(True)
plt.gca().set_aspect('equal')

# Right: loss over epochs
plt.subplot(1,2,2)
plt.plot(history.history['loss'], '-g', lw=2)
plt.title("Training Loss over Epochs")
plt.xlabel("Epoch"); plt.ylabel("Loss")
plt.grid(True)

plt.tight_layout()
plt.show()
```

The next code computes the brachistocrone using the Snell's conservation law. As it can be seen, the learning is perfect. The graph on the risht side provides the conservation law.

```python
import numpy as np
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import matplotlib.pyplot as plt
from scipy.optimize import fsolve

# 1) Discretize x ∈ [0,1]
N = 5000
x_vals = np.linspace(0, 1, N + 1).reshape(-1, 1).astype(np.float32)
x_tf   = tf.constant(x_vals)

# 2) Build the neural network for perturbation g(x)
def build_model():
    inp = keras.Input(shape=(1,))
    h = layers.Dense(32, activation='tanh')(inp)
    h = layers.Dense(32, activation='tanh')(h)
    h = layers.Dense(32, activation='tanh')(h)
    h = layers.Dense(32, activation='tanh')(h)
    out = layers.Dense(1)(h)
    return keras.Model(inputs=inp, outputs=out)

model = build_model()
optimizer = keras.optimizers.Adam(learning_rate=1e-3)

# 3) Parametrize f(x) so that f(0)=1, f(1)=0
def f_and_df(x):
    with tf.GradientTape() as tape:
        tape.watch(x)
        g = model(x)
        f = 1.0 - x + x*(1.0 - x)*g   # boundary conditions
    df = tape.gradient(f, x)
    return f, df

# 4) Loss function
def variance_loss():
    f, df = f_and_df(x_tf)
    q = f * (1.0 + df**2)
    q_mean = tf.reduce_mean(q)
    return tf.reduce_mean((q - q_mean)**2)

@tf.function
def compute_loss():
    return  variance_loss()

@tf.function
def train_step():
    with tf.GradientTape() as tape:
        loss = compute_loss()
    grads = tape.gradient(loss, model.trainable_variables)
    optimizer.apply_gradients(zip(grads, model.trainable_variables))
    return loss

# 5) Train
epochs = 3000
for epoch in range(epochs):
    L = train_step()
    if epoch % 200 == 0:
        print(f"Epoch {epoch}: loss = {L.numpy().item():.6f}")

# 6) Evaluate f(x) and q(x)
f_pred_tf, df_pred_tf = f_and_df(x_tf)
f_np  = f_pred_tf.numpy().flatten()
df_np = df_pred_tf.numpy().flatten()
q_np  = f_np * (1 + df_np**2)
K2_est = q_np.mean()

# 7) Compute exact cycloid passing through (0,1)→(1,0)
theta_eq = lambda θ: θ - np.sin(θ) - (1 - np.cos(θ))
theta_max = fsolve(theta_eq, 2.0)[0]
R = 1.0 / (theta_max - np.sin(theta_max))
θ = np.linspace(0, theta_max, 500)
x_c = R * (θ - np.sin(θ))
y_c = 1.0 - R * (1.0 - np.cos(θ))

# 8) Plot both the learned curve vs cycloid and the variance q(x)
plt.figure(figsize=(12,5))

# Left: Learned brachistochrone vs cycloid
plt.subplot(1,2,1)
plt.plot(1 - x_vals, 1 - f_np, '-b', linewidth=2, label='Learned curve')
plt.plot(x_c, y_c, '--', color='orange', linewidth=2, label='Exact cycloid')
plt.scatter([0,1], [1,0], color='red', zorder=5, label='Endpoints')
plt.gca().set_aspect('equal')
plt.title('Learned vs. Exact Cycloid')
plt.xlabel('x'); plt.ylabel('y')
plt.legend(); plt.grid(True)

# Right: Variance q(x)
plt.subplot(1,2,2)
plt.plot(1-x_vals, q_np, '-m', label=r'$q(x)=f(x)\,(1+f\'(x)^2)$')
plt.hlines(K2_est, 0, 1, 'k', '--', label=r'$\overline{q}=K^2$')
plt.title('Variance-Minimized $q(x)$')
plt.xlabel('x'); plt.ylabel('q(x)')
plt.legend(); plt.grid(True)

plt.tight_layout()
plt.show()
```

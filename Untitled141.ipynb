# **Manifold learning: the upper hemisphere**


**Positions only**
We generate $N=1000$ data points
$
q_i = (q_i^1,q_i^2,q_i^3)
$
on the upper hemisphere of the unit sphere (i.e.\ with $q_i^3>0$), to which we
add $5\%$ Gaussian noise.
We then train an autoencoder
$
A_{\phi,\theta} = (E_{\phi}, \Phi_{\theta})
$
on this dataset in order to recover the parametrization of the surface.
Training is performed on $80\%$ of the data, while the remaining $20\%$ is used
for testing.

We plot and analyze both the training and testing reconstruction errors, and we
compare the true spherical surface with the surface recovered by the trained
autoencoder.



```python
import numpy as np
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import matplotlib.pyplot as plt
from tensorflow.keras import regularizers

reg = regularizers.l2(1e-4)   # try 1e-4, 1e-5, etc.

# -----------------------------
# 1. Generate synthetic data
# -----------------------------

np.random.seed(42)
N = 1000

# Sample latent points uniformly in the unit disk U in R^2
r = np.sqrt(np.random.rand(N))           # radius ~ sqrt(U) for uniform in disk
theta = 2 * np.pi * np.random.rand(N)    # angle

xi1 = r * np.cos(theta)
xi2 = r * np.sin(theta)
Xi = np.stack([xi1, xi2], axis=1)        # shape (N, 2)

# True (noise-free) embedding on upper hemisphere: q = (xi1, xi2, sqrt(1 - r^2))
z = np.sqrt(1.0 - (xi1**2 + xi2**2))
Q_true = np.stack([xi1, xi2, z], axis=1)  # shape (N, 3)

# Add 5% Gaussian noise
noise_level = 0.05
Q_noisy = Q_true + noise_level * np.random.randn(*Q_true.shape)

# -----------------------------
# 2. Train / test split (80/20)
# -----------------------------
N_train = int(0.8 * N)
indices = np.random.permutation(N)

train_idx = indices[:N_train]
test_idx  = indices[N_train:]

Q_train = Q_noisy[train_idx]
Q_test  = Q_noisy[test_idx]

# -----------------------------
# 3. Define encoder and decoder
# -----------------------------

# Encoder: R^3 -> R^2
input_q = keras.Input(shape=(3,), name="input_q")
x = layers.Dense(32, activation="swish", kernel_regularizer=reg)(input_q)
x = layers.Dense(16, activation="swish", kernel_regularizer=reg)(x)
latent = layers.Dense(2, activation="linear", kernel_regularizer=reg,
                      name="latent")(x)
encoder = keras.Model(input_q, latent, name="encoder")

# Decoder: R^2 -> R^3
input_xi = keras.Input(shape=(2,), name="input_xi")
y = layers.Dense(16, activation="swish", kernel_regularizer=reg)(input_xi)
y = layers.Dense(32, activation="swish", kernel_regularizer=reg)(y)
output_q = layers.Dense(3, activation="linear", kernel_regularizer=reg,
                        name="recon_q")(y)
decoder = keras.Model(input_xi, output_q, name="decoder")

# Autoencoder: q -> xi -> q_hat
ae_input = keras.Input(shape=(3,), name="ae_input")
xi_hat = encoder(ae_input)
q_hat = decoder(xi_hat)
autoencoder = keras.Model(inputs=ae_input, outputs=q_hat, name="autoencoder")

autoencoder.compile(optimizer=keras.optimizers.Adam(1e-3),
                    loss="mse")

autoencoder.summary()

# -----------------------------
# 4. Train the autoencoder
# -----------------------------

history = autoencoder.fit(
    Q_train, Q_train,
    validation_data=(Q_test, Q_test),
    epochs=1000,
    batch_size=16,
    verbose=0
)

# -----------------------------
# 5. Plot training / test errors
# -----------------------------

plt.figure(figsize=(6,4))
plt.plot(history.history["loss"], label="train")
plt.plot(history.history["val_loss"], label="test")
plt.yscale("log")
plt.xlabel("Epoch")
plt.ylabel("MSE loss")
plt.title("Reconstruction error on noisy upper hemisphere")
plt.legend()
plt.tight_layout()
plt.show()

# -----------------------------
# 6. Compare true vs reconstructed surface
# -----------------------------

# Use the *true* clean points to see how well the learned parametrization fits
Q_recon = autoencoder.predict(Q_true)

# Simple 3D scatter plot: true vs reconstructed
from mpl_toolkits.mplot3d import Axes3D  # noqa: F401

fig = plt.figure(figsize=(8,4))

ax1 = fig.add_subplot(121, projection="3d")
ax1.scatter(Q_true[:,0], Q_true[:,1], Q_true[:,2], c='b', label='true', s=20)
ax1.set_title("True upper hemisphere")
ax1.set_xlabel("x")
ax1.set_ylabel("y")
ax1.set_zlabel("z")

ax2 = fig.add_subplot(122, projection="3d")
ax2.scatter(Q_recon[:,0], Q_recon[:,1], Q_recon[:,2], c='r', label='recon', s=20)
ax2.set_title("Reconstructed surface")
ax2.set_xlabel("x")
ax2.set_ylabel("y")
ax2.set_zlabel("z")

plt.tight_layout()
plt.show()

# -----------------------------
# 7. Inspect the learned parametrization
# -----------------------------

# Sample a fine grid in the latent disk and push it through the decoder
n_grid = 40
r_grid = np.linspace(0, 0.98, n_grid)
theta_grid = np.linspace(0, 2*np.pi, n_grid)

grid_points = []
for rg in r_grid:
    for th in theta_grid:
        grid_points.append([rg*np.cos(th), rg*np.sin(th)])
grid_points = np.array(grid_points)

surface_learned = decoder.predict(grid_points)

fig = plt.figure(figsize=(5,4))
ax = fig.add_subplot(111, projection="3d")
ax.scatter(surface_learned[:,0], surface_learned[:,1], surface_learned[:,2],
           s=5)
ax.set_title("Learned parametrization via decoder")
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_zlabel("z")
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt

# -----------------------------
# 1. Build a latent grid in U
# -----------------------------
n_grid = 40
xi1_lin = np.linspace(-0.9, 0.9, n_grid)
xi2_lin = np.linspace(-0.9, 0.9, n_grid)

XI1, XI2 = np.meshgrid(xi1_lin, xi2_lin)  # shape (n_grid, n_grid)

# Arrays to store learned metric coefficients
g11_learned = np.full_like(XI1, np.nan, dtype=np.float64)
g12_learned = np.full_like(XI1, np.nan, dtype=np.float64)
g22_learned = np.full_like(XI1, np.nan, dtype=np.float64)

# -----------------------------
# 2. Compute metric via decoder
# -----------------------------
for i in range(n_grid):
    for j in range(n_grid):
        xi1 = XI1[i, j]
        xi2 = XI2[i, j]
        # Skip points outside the unit disk
        if xi1**2 + xi2**2 >= 1.0:
            continue

        # Single point as Tensor
        xi = tf.constant([xi1, xi2], dtype=tf.float32)

        with tf.GradientTape() as tape:
            tape.watch(xi)
            q = decoder(tf.reshape(xi, (1, 2)))  # shape (1,3)
            q = tf.reshape(q, (3,))             # shape (3,)

        # Jacobian dq/dxi: shape (3,2)
        J = tape.jacobian(q, xi)  # d q^a / d xi^i
        # Columns: X1 = ∂Φ/∂xi1, X2 = ∂Φ/∂xi2
        X1 = J[:, 0]  # shape (3,)
        X2 = J[:, 1]  # shape (3,)

        # Inner products in R^3
        g11 = tf.tensordot(X1, X1, axes=1).numpy()
        g12 = tf.tensordot(X1, X2, axes=1).numpy()
        g22 = tf.tensordot(X2, X2, axes=1).numpy()

        g11_learned[i, j] = g11
        g12_learned[i, j] = g12
        g22_learned[i, j] = g22

# -----------------------------
# 3. True spherical metric
# -----------------------------
XI1_np = XI1
XI2_np = XI2
denom = 1.0 - XI1_np**2 - XI2_np**2

# Mask outside the disk to avoid division by zero
mask = denom > 0

g11_true = np.full_like(denom, np.nan, dtype=np.float64)
g12_true = np.full_like(denom, np.nan, dtype=np.float64)
g22_true = np.full_like(denom, np.nan, dtype=np.float64)

g11_true[mask] = (1.0 - XI2_np[mask]**2) / denom[mask]
g12_true[mask] = (XI1_np[mask] * XI2_np[mask]) / denom[mask]
g22_true[mask] = (1.0 - XI1_np[mask]**2) / denom[mask]

# -----------------------------
# 4. Contour plots: learned vs true
# -----------------------------
def plot_metric_comparison(Z_learned, Z_true, title):
    plt.figure(figsize=(8, 3))

    # Learned
    plt.subplot(1, 2, 1)
    cs1 = plt.contourf(XI1, XI2, Z_learned, levels=20, cmap="viridis")
    plt.colorbar(cs1)
    plt.title(title + " (learned)")
    plt.xlabel(r"$\xi^1$")
    plt.ylabel(r"$\xi^2$")
    plt.gca().set_aspect("equal")

    # True
    plt.subplot(1, 2, 2)
    cs2 = plt.contourf(XI1, XI2, Z_true, levels=20, cmap="viridis")
    plt.colorbar(cs2)
    plt.title(title + " (true)")
    plt.xlabel(r"$\xi^1$")
    plt.ylabel(r"$\xi^2$")
    plt.gca().set_aspect("equal")

    plt.tight_layout()
    plt.show()

plot_metric_comparison(g11_learned, g11_true, r"$g_{11}$")
plot_metric_comparison(g12_learned, g12_true, r"$g_{12}$")
plot_metric_comparison(g22_learned, g22_true, r"$g_{22}$")
```

**Learning latent coordinates** We plot the first two components of the decoder in terms of $\xi^1$ and $\xi^2$. We obtain almost linear graphs, which suggests that the autoencoder has learned an affine transformation of the true latent coordinates
$\hat{\xi} = A\xi +b$, where $A$ is appoximately a scaled rotation composed by a reflection.

This agrees with the fact that autoencoders are identifiable only up to a diffeomorphism (here essentially an affine map), unless additional geometric constraints are added.

Next is  a Keras/TensorFlow snippet that does the following things:

1. samples a few  $\xi$  points in the unit disk,

2. computes the learned normal from your trained decoder,

3. computes the true normal analytically,

4. evaluates the angle and provides a table.

```python
import numpy as np
import tensorflow as tf

# Assume: trained decoder model exists: decoder : R^2 -> R^3

def learned_normal_and_point(xi):
    """
    Given xi in R^2, compute:
    - q_learned = Phi_theta(xi)
    - n_learned = dPhi/dxi1 x dPhi/dxi2 at xi
    """
    xi = tf.convert_to_tensor(xi, dtype=tf.float32)  # shape (2,)
    with tf.GradientTape() as tape:
        tape.watch(xi)
        q = decoder(tf.reshape(xi, (1, 2)))  # (1,3)
        q = tf.reshape(q, (3,))             # (3,)
    J = tape.jacobian(q, xi)                # (3,2)
    X1 = J[:, 0]
    X2 = J[:, 1]
    n_learned = tf.linalg.cross(X1, X2)     # (3,)
    return q, n_learned

def angle_between_normals_at_learned_point(xi):
    xi = tf.convert_to_tensor(xi, dtype=tf.float32)
    q_learned, n_learn = learned_normal_and_point(xi)

    # true radial normal at the learned point
    n_true = q_learned / (tf.norm(q_learned) + 1e-8)

    dot = tf.tensordot(n_true, n_learn, axes=1)
    norm_true = tf.norm(n_true)
    norm_learn = tf.norm(n_learn)

    # use absolute value to ignore orientation
    cos_theta = tf.abs(dot) / (norm_true * norm_learn + 1e-8)
    cos_theta = tf.clip_by_value(cos_theta, -1.0, 1.0)
    theta = tf.acos(cos_theta)
    return theta


# ---------------------------------------------------
# Sample a few xi in the unit disk and compute angles
# ---------------------------------------------------

num_samples = 10
angles_rad = []
angles_deg = []
xis = []

np.random.seed(0)
for _ in range(num_samples):
    # sample uniformly in the unit disk
    r = np.sqrt(np.random.rand())
    th = 2 * np.pi * np.random.rand()
    xi1 = r * np.cos(th)
    xi2 = r * np.sin(th)
    xi = np.array([xi1, xi2], dtype=np.float32)

    theta = angle_between_normals_at_learned_point(xi).numpy()
    angles_rad.append(theta)
    angles_deg.append(theta * 180.0 / np.pi)
    xis.append(xi)

print("ξ^1      ξ^2      angle (rad)   angle (deg)")
for xi, th_r, th_d in zip(xis, angles_rad, angles_deg):
    print(f"{xi[0]: .4f}  {xi[1]: .4f}   {th_r: .6f}    {th_d: .3f}")
```

# **Elastica Curves**

An Euler elastica curve is a smooth curve in the plane (or space) that minimizes the bending energy functional
$$E[c] =  \int_0^T \kappa(s)^2 \, ds$$
where $c(s)$ is the curve parametrized by arc-length
$s$, $T$ is the total length,
$
κ(s)$ is the curvature at
$
s$.
The boundary constraints are: $c(0)=A$, $c(T) = B$, $\dot c(0) = v_A$ and $\dot c(T)= v_B$.

**Hard Constraints Approach**

We consider a curve $c_0(s)$ which satisfies the above boundary contraints and consider

$$c_{\theta}(s) = c_0(s) + \phi(s) nn_{\theta} (s) , $$
where $\phi(s) = s^2 (T-s)^2$ and $nn_{\theta}(s)$ a free neural network.

We also consider the curve to be unit-speed, $\|\dot c(t)\|=1$, a condition added as a soft constraint.

```python
# ============================================================
# Stable bending-energy version (no curvature division)
# Hard endpoint pos/vel via Hermite; soft unit-speed; ∫||c''||^2
# ============================================================

import numpy as np
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import matplotlib.pyplot as plt

tf.keras.backend.set_floatx("float64")
np.random.seed(0); tf.random.set_seed(0)

# --- Problem data ---
A   = np.array([0.0, 0.0], dtype=np.float64)
B   = np.array([3.0, 0.0], dtype=np.float64)
v_A = np.array([1.0, 0.0], dtype=np.float64)
v_B = np.array([-1.0, 0.0], dtype=np.float64)

r = np.linalg.norm(B - A)
T = float(r + 2.0)

A_tf  = tf.constant(A[None,:],  tf.float64)
B_tf  = tf.constant(B[None,:],  tf.float64)
vA_tf = tf.constant(v_A[None,:], tf.float64)
vB_tf = tf.constant(v_B[None,:], tf.float64)

# --- Hermite baseline (hard endpoints) ---
def hermite_baseline(s):
    u = s / T
    u2, u3 = u*u, u*u*u
    h00 =  2*u3 - 3*u2 + 1
    h10 =      u3 - 2*u2 + u
    h01 = -2*u3 + 3*u2
    h11 =      u3 -   u2

    c = (h00*A_tf + h10*(T*vA_tf) + h01*B_tf + h11*(T*vB_tf))

    dh00 = 6*u2 - 6*u
    dh10 = 3*u2 - 4*u + 1
    dh01 = -6*u2 + 6*u
    dh11 = 3*u2 - 2*u

    dc_du = (dh00*A_tf + dh10*(T*vA_tf) + dh01*B_tf + dh11*(T*vB_tf))
    dc_ds = dc_du / T
    return c, dc_ds

# --- Small MLP for interior bump ---
def build_bump(width=64, depth=3, act="tanh", last_scale=0.1):
    s_in = keras.Input(shape=(1,))
    x = s_in
    for _ in range(depth):
        x = layers.Dense(width, activation=act,
                         kernel_initializer=keras.initializers.GlorotUniform())(x)
    # small output init to prevent big second derivatives at start
    out = layers.Dense(2, activation=None,
                       kernel_initializer=keras.initializers.RandomUniform(-last_scale, last_scale),
                       bias_initializer=keras.initializers.Zeros())(x)
    return keras.Model(s_in, out)

bump = build_bump(width=64, depth=3, act="tanh", last_scale=0.05)

# --- Hard embedding: c = c0 + phi * n,  phi = s^2 (T-s)^2 ---
def embedded_curve_and_derivs(s):
    c0, dc0 = hermite_baseline(s)
    phi  = tf.square(s) * tf.square(T - s)
    n    = bump(s)
    c    = c0 + phi * n

    with tf.GradientTape() as t1:
        t1.watch(s)
        phi_s = tf.square(s) * tf.square(T - s)
    dphi = t1.gradient(phi_s, s)  # analytic: 2s(T-s)^2 - 2(T-s)s^2

    with tf.GradientTape() as t2:
        t2.watch(s)
        n_eval = bump(s)
    dn = t2.gradient(n_eval, s)

    dc = dc0 + dphi * n + phi * dn

    # Second derivative via tape
    with tf.GradientTape() as t3:
        t3.watch(s)
        # recompute pieces inside tape to get derivative right
        c0_tmp, dc0_tmp = hermite_baseline(s)
        phi_tmp  = tf.square(s) * tf.square(T - s)
        dphi_tmp = (2*s * tf.square(T - s) - 2*(T - s) * tf.square(s))
        n_tmp = bump(s)
        with tf.GradientTape() as t4:
            t4.watch(s)
            n_eval2 = bump(s)
        dn_tmp  = t4.gradient(n_eval2, s)
        dc_tmp  = dc0_tmp + dphi_tmp * n_tmp + phi_tmp * dn_tmp
    d2c = t3.gradient(dc_tmp, s)

    return c, dc, d2c

# --- Optimizer (smaller LR + clipping) ---
opt = keras.optimizers.Adam(learning_rate=2e-4, clipnorm=1.0)

# --- Training hyperparameters ---
EPOCHS = 4000
N_s    = 256

# Loss weights
w_bend  = 1.0   # ∫ ||c''||^2
w_speed = 30.0   # unit-speed penalty
w_pos   = 1e-6  # tiny positional regularizer

@tf.function
def train_step():
    s_batch = tf.random.uniform((N_s,1), 0.0, T, dtype=tf.float64)

    with tf.GradientTape() as tape:
        c, dc, d2c = embedded_curve_and_derivs(s_batch)

        # Bending energy ~ mean(||c''||^2) * T
        bend_loss = tf.reduce_mean(tf.reduce_sum(tf.square(d2c), axis=1, keepdims=True)) * T

        # Unit-speed penalty ~ mean((||c'||^2 - 1)^2) * T
        speed_sq   = tf.reduce_sum(tf.square(dc), axis=1, keepdims=True)
        speed_loss = tf.reduce_mean(tf.square(speed_sq - 1.0)) * T

        # Tiny positional regularizer (keeps curve bounded if bump misbehaves)
        pos_reg = tf.reduce_mean(tf.reduce_sum(tf.square(c), axis=1))

        total = w_bend * bend_loss + w_speed * speed_loss + w_pos * pos_reg

    grads = tape.gradient(total, bump.trainable_variables)
    opt.apply_gradients(zip(grads, bump.trainable_variables))
    return bend_loss, speed_loss, pos_reg, total

# --- Train ---
hist = {"epoch":[], "bend":[], "speed":[], "pos":[], "tot":[]}
for ep in range(1, EPOCHS+1):
    Lb, Ls, Lp, Lt = train_step()
    if ep % 500 == 0:
        print(f"ep {ep:5d} | bend {Lb.numpy():.3e} | speed {Ls.numpy():.3e} | posreg {Lp.numpy():.3e} | tot {Lt.numpy():.3e}")
    hist["epoch"].append(ep)
    hist["bend"].append(Lb.numpy())
    hist["speed"].append(Ls.numpy())
    hist["pos"].append(Lp.numpy())
    hist["tot"].append(Lt.numpy())

# --- Evaluate & plot ---
s_grid = np.linspace(0.0, T, 600)[:,None]
c0_grid, _ = hermite_baseline(tf.constant(s_grid, tf.float64))
c_nn, dc_nn, _ = embedded_curve_and_derivs(tf.constant(s_grid, tf.float64))

c0_np  = c0_grid.numpy()
cNN_np = c_nn.numpy()
spd_np = np.sum(dc_nn.numpy()**2, axis=1)

plt.figure(figsize=(7,5))
plt.plot(c0_np[:,0],  c0_np[:,1], '--', label='Hermite baseline', lw=2)
plt.plot(cNN_np[:,0], cNN_np[:,1],  '-', label=r'$c_\theta(s)$', lw=2)
plt.scatter([A[0]],[A[1]], c='C2', s=60, zorder=3, label='A')
plt.scatter([B[0]],[B[1]], c='C3', s=60, zorder=3, label='B')
# velocity arrows
scale=0.25
plt.quiver(A[0],A[1], v_A[0],v_A[1], angles='xy', scale_units='xy', scale=1/scale,
           color='C2', width=0.004, label=r'$v_A$')
plt.quiver(B[0],B[1], v_B[0],v_B[1], angles='xy', scale_units='xy', scale=1/scale,
           color='C3', width=0.004, label=r'$v_B$')
plt.axis('equal'); plt.grid(alpha=0.3)
plt.xlabel('x'); plt.ylabel('y')
plt.title('Hard endpoint pos/vel + bending energy + unit-speed')
plt.legend(); plt.tight_layout(); plt.show()

# --- Loss plot ---
plt.figure(figsize=(7,4))
plt.plot(hist["epoch"], hist["tot"],   label='Total', lw=2)
plt.plot(hist["epoch"], hist["bend"], '--', label='Bending')
plt.plot(hist["epoch"], hist["speed"],'--', label='Unit-speed')
plt.plot(hist["epoch"], hist["pos"],  '--', label='Pos. reg (tiny)')
plt.yscale('log'); plt.xlabel('Epoch'); plt.ylabel('Loss')
plt.grid(alpha=0.3); plt.legend(); plt.tight_layout(); plt.show()

# --- Speed profile ---
plt.figure(figsize=(7,3.3))
plt.plot(s_grid[:,0], spd_np, lw=2)
plt.axhline(1.0, color='k', ls='--', alpha=0.6)
plt.xlabel('s'); plt.ylabel(r'$\|c^\prime(s)\|^2$')
plt.title('Speed-squared along the learned curve')
plt.grid(alpha=0.3); plt.tight_layout(); plt.show()
```

**Soft Constraints Approach**

In this case we define the total loss
$$
L_{\text{total}}(\theta) = L_{\text{bend}}(\theta)
+ \lambda_{\text{BC}} L_{\text{BC}}(\theta)
+ \lambda_{\text{unit}} L_{\text{unit}}(\theta),
$$
where each component is given by
$$
L_{\text{bend}}(\theta)
= \int_0^T \kappa_{\theta}(s)^2 \, ds
\;\; \approx \;\; \frac{T}{N} \sum_{j=1}^N \kappa_{\theta}(s_j)^2,
$$
$$
L_{\text{BC}}(\theta)
= \|c_{\theta}(0) - A\|^2
+ \|c_{\theta}(T) - B\|^2
+ \|\dot{c}_{\theta}(0) - v_A\|^2
+ \|\dot{c}_{\theta}(T) - v_B\|^2,
$$
$$
L_{\text{unit}}(\theta)
= \int_0^T \big( \|\dot{c}_{\theta}(s)\|^2 - 1 \big)^2 \, ds
\;\; \approx \;\; \frac{T}{N} \sum_{j=1}^N
\left( \big(\dot{c}^1_{\theta}(s_j)\big)^2
+ \big(\dot{c}^2_{\theta}(s_j)\big)^2 - 1 \right)^2,
$$
where $\{s_j\}_{j=1}^N$ is a set of equidistant collocation points in the interval $[0, T]$.

```python
# ============================================================
# PINN for planar Euler elastica with soft BCs + unit-speed
# A = (0,0), B = (3,2), vA = (1,0), vB = (0,1)
# L_total = ∫ ||c''||^2 ds  + λ_BC*(pos & vel at s=0,T) + λ_unit*∫ (||c'||^2-1)^2 ds
# ============================================================

import numpy as np
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import matplotlib.pyplot as plt

# -------------------------
# Precision & seeds
# -------------------------
tf.keras.backend.set_floatx("float64")
np.random.seed(0)
tf.random.set_seed(0)

# -------------------------
# Problem setup
# -------------------------
A   = np.array([0.0, 0.0], dtype=np.float64)
B   = np.array([3.0, 0.0], dtype=np.float64)
vA  = np.array([1.0, 0.0], dtype=np.float64)   # unit
vB  = np.array([-1.0, 0.0], dtype=np.float64)   # unit

r = np.linalg.norm(B - A)             # distance
T = float(r + 2.0)                    # choose T > r, e.g. r+1

A_tf  = tf.constant(A[None, :],  tf.float64)    # shape (1,2)
B_tf  = tf.constant(B[None, :],  tf.float64)
vA_tf = tf.constant(vA[None, :], tf.float64)
vB_tf = tf.constant(vB[None, :], tf.float64)

# -------------------------
# Collocation points (equidistant)
# -------------------------
N = 256
s_grid = np.linspace(0.0, T, N).reshape(-1, 1)   # equidistant
s_tf   = tf.constant(s_grid, dtype=tf.float64)

# -------------------------
# Model: c_theta(s) -> R^2
# -------------------------
def build_mlp(width=64, depth=3, act="tanh"):
    s_in = keras.Input(shape=(1,))
    x = s_in
    for _ in range(depth):
        x = layers.Dense(width, activation=act)(x)
    out = layers.Dense(2, activation=None)(x)
    return keras.Model(s_in, out)

curve = build_mlp(width=64, depth=3, act="tanh")

# -------------------------
# Optimizer & weights
# -------------------------
opt = keras.optimizers.Adam(learning_rate=5e-4, clipnorm=1.0)

lambda_BC   = 100.0    # weight for boundary constraints (pos & vel)
lambda_unit = 10.0    # weight for unit-speed penalty

EPOCHS = 4000
PRINT_EVERY = 500

# -------------------------
# Loss computation helper
# -------------------------
@tf.function
def compute_losses():
    # Interior derivatives wrt s (for ∫ ||c''||^2 and unit-speed)
    with tf.GradientTape() as t2:
        t2.watch(s_tf)
        with tf.GradientTape() as t1:
            t1.watch(s_tf)
            c = curve(s_tf)                          # (N,2)
        dc  = t1.gradient(c, s_tf)                   # (N,2)
    d2c = t2.gradient(dc, s_tf)                      # (N,2)

    # Bending energy density: ||c''||^2
    bend_density = tf.reduce_sum(tf.square(d2c), axis=1, keepdims=True)  # (N,1)
    L_bend = tf.reduce_mean(bend_density) * T

    # Unit-speed penalty: (||c'||^2 - 1)^2
    speed_sq   = tf.reduce_sum(tf.square(dc), axis=1, keepdims=True)     # (N,1)
    unit_res   = speed_sq - 1.0
    L_unit     = tf.reduce_mean(tf.square(unit_res)) * T

    # Endpoint position penalties
    s0 = tf.zeros((1,1), dtype=tf.float64)
    sT = tf.fill((1,1), tf.cast(T, tf.float64))
    c0 = curve(s0)                                     # (1,2)
    cT = curve(sT)                                     # (1,2)
    pos_loss = tf.reduce_sum(tf.square(c0 - A_tf)) + tf.reduce_sum(tf.square(cT - B_tf))

    # Endpoint velocity penalties
    with tf.GradientTape() as tv0:
        tv0.watch(s0)
        c0_for_vel = curve(s0)
    dc0 = tv0.gradient(c0_for_vel, s0)                 # (1,2)

    with tf.GradientTape() as tvT:
        tvT.watch(sT)
        cT_for_vel = curve(sT)
    dcT = tvT.gradient(cT_for_vel, sT)                 # (1,2)

    vel_loss = tf.reduce_sum(tf.square(dc0 - vA_tf)) + tf.reduce_sum(tf.square(dcT - vB_tf))

    L_BC = pos_loss + vel_loss

    # Total
    L_total = L_bend + lambda_BC * L_BC + lambda_unit * L_unit
    return L_total, L_bend, L_BC, L_unit

# -------------------------
# Training loop
# -------------------------
hist = {"epoch": [], "total": [], "bend": [], "bc": [], "unit": []}

for ep in range(1, EPOCHS + 1):
    with tf.GradientTape() as tape:
        L_total, L_bend, L_BC, L_unit = compute_losses()
    grads = tape.gradient(L_total, curve.trainable_variables)
    opt.apply_gradients(zip(grads, curve.trainable_variables))

    # Logging
    if ep % PRINT_EVERY == 0:
        print(f"ep {ep:5d} | bend {L_bend.numpy():.3e} | BC {L_BC.numpy():.3e} | unit {L_unit.numpy():.3e} | total {L_total.numpy():.3e}")
    hist["epoch"].append(ep)
    hist["total"].append(L_total.numpy())
    hist["bend"].append(L_bend.numpy())
    hist["bc"].append(L_BC.numpy())
    hist["unit"].append(L_unit.numpy())

# -------------------------
# Evaluate & plot
# -------------------------
c_plot = curve(tf.constant(s_grid, dtype=tf.float64)).numpy()

plt.figure(figsize=(7,5))
plt.plot(c_plot[:,0], c_plot[:,1], lw=2, label=r"$c_\theta(s)$")
plt.scatter([A[0]], [A[1]], c="C2", s=60, zorder=3, label="A")
plt.scatter([B[0]], [B[1]], c="C3", s=60, zorder=3, label="B")
# velocity arrows (scaled for display)
scale = 0.5
plt.quiver(A[0], A[1], vA[0], vA[1], angles='xy', scale_units='xy', scale=1/scale,
           color="C2", width=0.004, label=r"$v_A$")
plt.quiver(B[0], B[1], vB[0], vB[1], angles='xy', scale_units='xy', scale=1/scale,
           color="C3", width=0.004, label=r"$v_B$")
plt.axis('equal'); plt.grid(alpha=0.3)
plt.xlabel("x"); plt.ylabel("y")
plt.title("Elastica via PINN with soft BCs + unit-speed")
plt.legend(); plt.tight_layout(); plt.show()

# Loss curves
plt.figure(figsize=(7,4))
plt.plot(hist["epoch"], hist["total"],  label="Total", lw=2)
plt.plot(hist["epoch"], hist["bend"],  "--", label="Bend")
plt.plot(hist["epoch"], hist["bc"],    "--", label="Boundary")
plt.plot(hist["epoch"], hist["unit"],  "--", label="Unit-speed")
plt.yscale("log"); plt.xlabel("Epoch"); plt.ylabel("Loss")
plt.grid(alpha=0.3); plt.legend(); plt.tight_layout(); plt.show()


```

**More Hard Embedding**




We introduce a one-dimensional free neural network $\rho_{\theta}(s)$ and define

$$
\varphi_{\theta}(s) = \frac{T-s}{T}\,\alpha + \frac{s}{T}\,\beta + s(T-s)\,\rho_{\theta}(s).
$$

This parametrization ensures the boundary conditions $\varphi_{\theta}(0) = \alpha$ and $\varphi_{\theta}(T) = \beta$.

To enforce the unit-speed condition, we set
$$
\dot{c}_{\theta}(s) = \big( \cos \varphi_{\theta}(s), \, \sin \varphi_{\theta}(s) \big).
$$

By integration, the curve takes the form
$$
c_{\theta}(s) = \left( x_A + \int_0^s \cos \varphi_{\theta}(u)\,du,\; y_A + \int_0^s \sin \varphi_{\theta}(u)\,du \right).
$$
This construction guarantees the conditions
$$
c_{\theta}(0) = A, \quad  
\dot{c}_{\theta}(0) = v_A, \quad \dot{c}_{\theta}(T) = v_B.
$$
The final boundary condition $c_{\theta}(T) = B$ must be enforced as
\begin{eqnarray*}
  x_A + \int_0^T \cos \varphi_{\theta}(u)\, du &=& x_B, \\
  y_A + \int_0^T \sin \varphi_{\theta}(u)\, du &=& y_B.
\end{eqnarray*}
Since these conditions are not automatically satisfied, they have to be added in the loss as soft constraints.

Next, we note that
$$
\ddot c_{\theta}(s) =
\big( -\sin \varphi_{\theta}(s) \, \dot \varphi_{\theta}(s),\;
       \cos \varphi_{\theta}(s) \, \dot \varphi_{\theta}(s) \big),
$$
so that
$$
\|\ddot c_{\theta}(s)\|^2 = \dot \varphi_{\theta}(s)^2.
$$
Therefore, the bending loss becomes
$$
L_{\text{bend}}(\theta) = \int_0^T \|\ddot c_{\theta}(s)\|^2 \, ds
= \int_0^T \dot \varphi_{\theta}(s)^2 \, ds.
$$
The final condition is then enforced by minimizing the loss
$$
L_{\text{final}}(\theta) =
\Bigg( \int_0^T \cos \varphi_{\theta}(u)\, du - (x_B - x_A) \Bigg)^2
+ \Bigg( \int_0^T \sin \varphi_{\theta}(u)\, du - (y_B - y_A) \Bigg)^2.
$$
The total loss is thus the combination of the previous two losses:
$$
L_{\text{tot}}(\theta) = L_{\text{bend}}(\theta) + \lambda \, L_{\text{final}}(\theta).
$$

```python
# ============================================================
# Elastica via hard-embedded unit speed:
#  c'(s) = (cos φ(s), sin φ(s))   =>  ||c'(s)|| = 1 (hard)
#  φ(s) = α*(1 - s/T) + β*(s/T) + s(T-s) ρθ(s)  (angle NN)
#  c(s) = A + ∫_0^s (cos φ(u), sin φ(u)) du     (TF-trapezoid)
# Loss: L_total = ∫ (φ'(s))^2 ds  + λ * ||c(T) - B||^2
# A=(0,0), B=(3,2), vA=(1,0) (α=0), vB=(0,1) (β=π/2)
# ============================================================

import numpy as np
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import matplotlib.pyplot as plt

tf.keras.backend.set_floatx("float64")
np.random.seed(0); tf.random.set_seed(0)

# --- Problem data ---
A = np.array([0.0, 0.0], dtype=np.float64)
B = np.array([3.0, 0.0], dtype=np.float64)
alpha = 0.0                 # vA = (1,0)
beta  =  np.pi         # vB = (-1,0)

r = np.linalg.norm(B - A)
T = float(r + 2.0)

A_tf = tf.constant(A[None, :], dtype=tf.float64)
B_tf = tf.constant(B[None, :], dtype=tf.float64)

# --- Grid & trapezoid ---
N  = 512
s_grid = np.linspace(0.0, T, N).reshape(-1, 1)
s_tf   = tf.constant(s_grid, dtype=tf.float64)
ds     = T / (N - 1)

def cumtrapz_tf(f_vals, ds):
    inc  = 0.5 * (f_vals[1:] + f_vals[:-1]) * ds   # (N-1,d)
    zero = tf.zeros_like(f_vals[:1])
    return tf.concat([zero, tf.cumsum(inc, axis=0)], axis=0)

# --- Angle net ρ_θ(s) ---
def build_rho(width=64, depth=3, act="tanh"):
    s_in = keras.Input(shape=(1,))
    x = s_in
    for _ in range(depth):
        x = layers.Dense(width, activation=act)(x)
    out = layers.Dense(1, activation=None)(x)  # scalar ρ(s)
    return keras.Model(s_in, out)

rho_net = build_rho(64, 3, "tanh")

# --- φ_θ, velocity, curve ---
def phi_theta(s):
    s_over_T = s / T
    base = (1.0 - s_over_T) * alpha + s_over_T * beta
    bump = s * (T - s) * rho_net(s)
    return base + bump

def vel_from_phi(phi):
    return tf.concat([tf.cos(phi), tf.sin(phi)], axis=1)  # (N,2)

def curve_from_phi(phi):
    v = vel_from_phi(phi)
    disp = cumtrapz_tf(v, ds)
    return A_tf + disp

# --- Losses ---
@tf.function
def compute_losses():
    with tf.GradientTape() as tape:
        tape.watch(s_tf)
        phi = phi_theta(s_tf)            # (N,1)
    phi_s = tape.gradient(phi, s_tf)     # dφ/ds

    L_bend = tf.reduce_mean(tf.square(phi_s)) * T

    c_vals = curve_from_phi(phi)         # (N,2)
    c_T    = c_vals[-1:]                 # (1,2)
    L_final = tf.reduce_sum(tf.square(c_T - B_tf))

    return L_bend, L_final, phi, c_vals

opt = keras.optimizers.Adam(learning_rate=3e-4, clipnorm=1.0)
lambda_final = 50.0
EPOCHS = 4000; PRINT_EVERY = 500

hist = {"epoch": [], "bend": [], "final": [], "total": []}

for ep in range(1, EPOCHS + 1):
    with tf.GradientTape() as tape:
        L_bend, L_final, phi, c_vals = compute_losses()
        L_total = L_bend + lambda_final * L_final
    grads = tape.gradient(L_total, rho_net.trainable_variables)
    opt.apply_gradients(zip(grads, rho_net.trainable_variables))

    if ep % PRINT_EVERY == 0:
        print(f"ep {ep:5d} | bend {L_bend.numpy():.3e} | final {L_final.numpy():.3e} | total {L_total.numpy():.3e}")

    hist["epoch"].append(ep)
    hist["bend"].append(L_bend.numpy())
    hist["final"].append(L_final.numpy())
    hist["total"].append(L_total.numpy())

# --- Evaluate & plots ---
phi_np = phi.numpy().ravel()
c_np   = c_vals.numpy()

# ====== SPEED CHECK (two safe options) ======

# Option A (uses definition): unit speed from velocity field directly
v_direct = vel_from_phi(phi).numpy()
speed_sq_A = np.sum(v_direct**2, axis=1)  # should be exactly 1 (up to fp error)

# Option B (autograd): recompute inside tape to keep graph
with tf.GradientTape() as tchk:
    tchk.watch(s_tf)
    phi_chk = phi_theta(s_tf)             # recompute inside the tape
    c_chk   = curve_from_phi(phi_chk)     # recompute curve inside the tape
dc_ds = tchk.gradient(c_chk, s_tf).numpy()  # (N,2)
speed_sq_B = np.sum(dc_ds**2, axis=1)

# --- Trajectory plot ---
plt.figure(figsize=(7,5))
plt.plot(c_np[:,0], c_np[:,1], lw=2, label=r"$c_\theta(s)$")
plt.scatter([A[0]],[A[1]], c='C2', s=60, label='A')
plt.scatter([B[0]],[B[1]], c='C3', s=60, label='B')
scale = 0.7
vA = np.array([np.cos(alpha), np.sin(alpha)])
vB = np.array([np.cos(beta),  np.sin(beta)])
plt.quiver(A[0], A[1], vA[0], vA[1], angles='xy', scale_units='xy', scale=1/scale,
           color='C2', width=0.004, label=r'$v_A$')
plt.quiver(B[0], B[1], vB[0], vB[1], angles='xy', scale_units='xy', scale=1/scale,
           color='C3', width=0.004, label=r'$v_B$')
plt.axis('equal'); plt.grid(alpha=0.3)
plt.xlabel('x'); plt.ylabel('y')
plt.title('Elastica with hard unit speed via angle network')
plt.legend(); plt.tight_layout(); plt.show()

# --- φ(s) and speed plots ---
s_plot = s_grid[:,0]
plt.figure(figsize=(7,5))
plt.plot(s_plot, phi_np, lw=2)
plt.xlabel('s'); plt.ylabel(r'$\varphi_\theta(s)$')
plt.grid(alpha=0.3); plt.tight_layout(); plt.show()

plt.figure(figsize=(7,5))
plt.plot(s_plot, speed_sq_A, lw=2, label='speed^2 (direct)')
plt.plot(s_plot, speed_sq_B, '--', lw=1.5, label='speed^2 (autograd)')
plt.axhline(1.0, color='k', ls='--', alpha=0.6)
plt.xlabel('s'); plt.ylabel(r'$\|c^\prime(s)\|^2$')
plt.title('Speed-squared (should be ~1)')
plt.grid(alpha=0.3); plt.legend(); plt.tight_layout(); plt.show()

# --- Loss curves ---
plt.figure(figsize=(7,5))
plt.plot(hist["epoch"], hist["total"],  label='Total', lw=2)
plt.plot(hist["epoch"], hist["bend"],  '--', label='Bending')
plt.plot(hist["epoch"], hist["final"], '--', label='Final-pos')
plt.yscale('log'); plt.xlabel('Epoch'); plt.ylabel('Loss')
plt.grid(alpha=0.3); plt.legend(); plt.tight_layout(); plt.show()
```

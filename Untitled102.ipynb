# **Estimation of the decay constant in Netwon's cooling law**

We consider the problem of estimating the thermal decay constant $k$ when the governing model is Newton's law of cooling. The corresponding ordinary differential equation  is
\begin{eqnarray*}
  u'(t) &=& -k^2 \,\big(u(t) - T\big), \\
  u(0)  &=& u_0,
\end{eqnarray*}
where $u(t)$ denotes the temperature of the body at time $t$, $u_0$ is its initial temperature, and $T$ is the constant ambient  temperature.

In addition to the model, we assume that experimental data are available in the form of $N$ temperature measurements,
$$
u_j = u(t_j), \qquad j=1,\dots,N,
$$
taken at discrete times $\{t_j\}$. The objective is to infer the value of $k^2$ from these observations using a neural network framework.


1.The first variant is to solve the differential equation explicitly. Its solution is
$$
u(t) = T + (u_0 - T) e^{-k^2 t}.
$$
Substituting $t = t_j$ into this expression gives $N$ relations:
$$
u_j = T + (u_0 - T) e^{-k^2 t_j}, \quad j = 1, \dots, N.
$$
If $k_j^2$ denotes the value obtained from the $j$-th relation, then
$$
k_j^2 = \frac{1}{t_j} \ln (u_0 - T) \;-\; \frac{1}{t_j} \ln (u_j - T).
$$
Thus, each measurement $u_j$ produces an explicit estimate $k_j^2$ for the decay constant.
Since all partial estimates are equally important, we weight them uniformly. The final estimate of the constant is then taken as the average:
$$
k^2 = \frac{1}{N} \sum_{j=1}^N k_j^2
= \frac{\ln (u_0 - T)}{N} - \frac{1}{N} \sum_{j=1}^N \frac{\ln (u_j - T)}{t_j}.
$$
The code is as in the following:

```python
import numpy as np
import matplotlib.pyplot as plt

# ---------------------------
# Data (t_j, u_j)
# ---------------------------
t = np.array([0.00, 1.11, 2.22, 3.33, 4.44, 5.56, 6.67, 7.78, 8.89, 10.00])
u = np.array([90.00, 64.42, 49.75, 41.33, 36.50, 33.73, 32.14, 31.22, 30.70, 30.40])

# Known ambient and initial temperatures (as per your "exact" curve request)
T = 30.0
u0 = 90.0

# ---------------------------
# Estimate k^2 using the formula
# k_j^2 = (1/t_j) * ln(u0 - T) - (1/t_j) * ln(u_j - T)
# Skip t=0
# ---------------------------
k2_values = []
for tj, uj in zip(t[1:], u[1:]):
    if (uj - T) <= 0:
        # Guard: log must have positive argument; (uj - T) should be > 0
        continue
    k2_j = (np.log(u0 - T) - np.log(uj - T)) / tj
    k2_values.append(k2_j)

k2_values = np.array(k2_values)
k2_est = float(np.mean(k2_values))
print(f"Estimated k^2 (average): {k2_est:.4f}")
print("Individual k^2 estimates:", np.round(k2_values, 4))

# ---------------------------
# Define the two model curves
# 1) exact with k^2 = 0.5
# 2) comparison with k^2 = 0.53
# ---------------------------
def u_model(t, k2, T=30.0, u0=90.0):
    return T + (u0 - T) * np.exp(-k2 * t)

t_plot = np.linspace(0, 10, 400)
u_exact_05  = u_model(t_plot, 0.5,  T, u0)
u_exact_05005 = u_model(t_plot, 0.5005, T, u0)

# ---------------------------
# Plot
# ---------------------------
plt.figure(figsize=(7, 4.5))
plt.scatter(t, u, color='C0', label='Data', zorder=3)
plt.plot(t_plot, u_exact_05,  label=r'$u_{\mathrm{exact}}(t)=30+(60)e^{-0.5t}$', linewidth=2)
plt.plot(t_plot, u_exact_05005, label=r'$u(t)=30+(60)e^{-0.5005t}$', linewidth=2, linestyle='--')

plt.xlabel('t')
plt.ylabel('u(t)')
plt.title('Newton Cooling: Data vs. Two Exponential Fits')
plt.grid(alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()
```

2. In this variant we consider a two-branch neural network with input $t$,
$$
NN(\theta; t) = \big( u_{\theta}(t), \; k_{\theta}(t) \big),
$$
where the first branch learns the solution $u(t)$ of Newton's cooling law, and the second branch learns the parameter $k$.


The ODE loss enforces the differential equation:
\begin{eqnarray*}
L_{\text{ODE}}(\theta) &=& \int_0^{\tau}
\left( \frac{d}{dt} u_{\theta}(t) + k_{\theta}(t)^2 \,\big(u_{\theta}(t) - T\big) \right)^2 \, dt \\[6pt]
&\approx& \frac{\tau}{N} \sum_{j=1}^N
\left( \frac{du_{\theta}(t_j)}{dt} + k_{\theta}(t_j)^2 \,\big(u_{\theta}(t_j) - T\big) \right)^2,
\end{eqnarray*}
where $\tau = \max\{t_j\}$.


The initial condition is enforced through
$$
L_{\text{IC}}(\theta) = \big(u_{\theta}(0) - u_0\big)^2,
$$
while the data loss is given by
$$
L_{\text{data}}(\theta) = \frac{1}{N} \sum_{j=1}^N \big(u_{\theta}(t_j) - u_j\big)^2.
$$
To enforce that the second branch learns a constant parameter, we minimize the variance of $k_{\theta}(t)$ over the training points:
$$
L_k(\theta) = \frac{1}{N} \sum_{j=1}^N \Big( k_{\theta}(t_j) - \bar{k}_{\theta} \Big)^2,
\qquad
\bar{k}_{\theta} = \frac{1}{N} \sum_{j=1}^N k_{\theta}(t_j).
$$
The composite total loss is then
$$
L_{\text{tot}}(\theta) = L_{\text{data}}(\theta) + L_{\text{ODE}}(\theta) + L_{\text{IC}}(\theta) + L_k(\theta).
$$
Finally, if
$$
\theta^* = \arg \min_{\theta} L_{\text{tot}}(\theta),
$$
then the estimated decay  constant is $k \approx k_{\theta^*}$.


```python
# ============================================================
# Newton cooling: two-branch NN for u(t) and k(t)
# Loss = data + ODE + IC + variance(k)
# Plots:
#   (1) Data + u_NN(t) + exact curve (k^2=0.5) on SAME figure
#   (2) Learned k^2 vs epochs
# ============================================================

import numpy as np
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import matplotlib.pyplot as plt

tf.keras.backend.set_floatx("float64")
np.random.seed(0); tf.random.set_seed(0)

# ---------------------------
# Data (from the table)
# ---------------------------
t_data = np.array([0.00, 1.11, 2.22, 3.33, 4.44, 5.56, 6.67, 7.78, 8.89, 10.00])
u_data = np.array([90.00, 64.42, 49.75, 41.33, 36.50, 33.73, 32.14, 31.22, 30.70, 30.40])

# Treat ambient T and initial u0 as known constants (can be adapted as needed)
T_const  = 30.40
u0_const = 90.00
tau = float(t_data.max())

t_tf = tf.constant(t_data[:, None], dtype=tf.float64)  # shape (N,1)
u_tf = tf.constant(u_data[:, None], dtype=tf.float64)  # shape (N,1)
T_tf = tf.constant(T_const, dtype=tf.float64)
u0_tf = tf.constant(u0_const, dtype=tf.float64)

N = t_data.size

# ---------------------------
# Two-branch network builder
# ---------------------------
def mlp(in_dim, width=64, depth=3, act="swish", out_dim=1):
    x = inp = keras.Input(shape=(in_dim,))
    for _ in range(depth):
        x = layers.Dense(width, activation=act)(x)
    out = layers.Dense(out_dim, activation=None)(x)
    return keras.Model(inp, out)

# Branch for u(t)
u_net_raw = mlp(in_dim=1, width=64, depth=3, act="swish", out_dim=1)

# Branch for k(t); we will pass through softplus to keep it nonnegative
k_net_raw = mlp(in_dim=1, width=32, depth=3, act="swish", out_dim=1)

softplus = tf.nn.softplus

# Convenience wrappers that return u_theta(t) and k_theta(t)
def u_theta(t):
    # You can also hard-embed IC if desired: u0 + t * n(t). Here we keep it flexible
    return u_net_raw(t)

def k_theta(t):
    # positive parameterization; small epsilon to avoid exact zero
    return softplus(k_net_raw(t)) + 1e-6

# ---------------------------
# Loss pieces
# ---------------------------
def losses():
    # ODE residual at data times (you can also sample extra collocation points)
    with tf.GradientTape() as tape:
        tape.watch(t_tf)
        u_pred = u_theta(t_tf)                      # (N,1)
    du_dt = tape.gradient(u_pred, t_tf)             # (N,1)

    k_pred = k_theta(t_tf)                          # (N,1)

    # ODE loss: u'(t) + k(t)^2 (u(t) - T) = 0
    ode_res = du_dt + tf.square(k_pred) * (u_pred - T_tf)
    L_ode = (tau / N) * tf.reduce_mean(tf.square(ode_res))

    # Data MSE
    L_data = tf.reduce_mean(tf.square(u_pred - u_tf))

    # Initial condition (soft, but you may hard-embed instead)
    u0_pred = u_theta(tf.zeros_like(t_tf[:1]))      # t=0
    L_ic = tf.reduce_mean(tf.square(u0_pred - u0_tf))

    # Variance penalty so that k_theta learns a constant across the sample
    k_mean = tf.reduce_mean(k_pred)
    L_k = tf.reduce_mean(tf.square(k_pred - k_mean))

    return L_data, L_ode, L_ic, L_k, u_pred, k_pred, du_dt

# ---------------------------
# Training
# ---------------------------
opt = keras.optimizers.Adam(1e-2)

# Loss weights (tune as needed)
w_data = 100.0
w_ode  = 1.0
w_ic   = 100.0
w_kvar = 200.0

EPOCHS = 400
PRINT_EVERY = 60

hist = {"epoch": [], "data": [], "ode": [], "ic": [], "kvar": [], "total": [], "k": [], "k2": []}

for ep in range(1, EPOCHS+1):
    with tf.GradientTape() as tape:
        L_data, L_ode, L_ic, L_k, u_pred, k_pred, du_dt = losses()
        L_tot = w_data*L_data + w_ode*L_ode + w_ic*L_ic + w_kvar*L_k
    vars_all = u_net_raw.trainable_variables + k_net_raw.trainable_variables
    grads = tape.gradient(L_tot, vars_all)
    opt.apply_gradients(zip(grads, vars_all))

    # Track mean k and k^2 for plotting
    k_mean_val = float(tf.reduce_mean(k_pred).numpy())
    k2_mean_val = k_mean_val**2

    if ep % PRINT_EVERY == 0:
        print(f"ep {ep:5d} | data {L_data.numpy():.3e} | ode {L_ode.numpy():.3e} "
              f"| ic {L_ic.numpy():.3e} | kvar {L_k.numpy():.3e} | tot {L_tot.numpy():.3e} "
              f"| mean k={k_mean_val:.4f}  (mean k^2={k2_mean_val:.4f})")

    hist["epoch"].append(ep)
    hist["data"].append(L_data.numpy())
    hist["ode"].append(L_ode.numpy())
    hist["ic"].append(L_ic.numpy())
    hist["kvar"].append(L_k.numpy())
    hist["total"].append(L_tot.numpy())
    hist["k"].append(k_mean_val)
    hist["k2"].append(k2_mean_val)

# ---------------------------
# Report k estimate and plots
# ---------------------------
# Take the average of k_theta over data times as the estimate (variance minimized)
k_vals = k_theta(t_tf).numpy().ravel()
k_est  = float(k_vals.mean())
k2_est = k_est**2
print(f"\nEstimated k (mean over data): {k_est:.6f}")
print(f"Estimated k^2: {k2_est:.6f}")

# Helper: closed-form model
def u_model(t, k2, T=T_const, u0=u0_const):
    return T + (u0 - T) * np.exp(-k2 * t)

# ---- (1) ONE figure with Data + NN + Exact (k^2=0.5) ----
t_plot = np.linspace(0, tau, 400)[:, None]
u_plot = u_theta(tf.constant(t_plot, dtype=tf.float64)).numpy()
u_exact = u_model(t_plot, 0.5)  # exact curve with k^2 = 0.5

plt.figure(figsize=(7,4.5))
plt.scatter(t_data, u_data, c='C0', label='Data', zorder=3)
plt.plot(t_plot, u_plot,  'C1', lw=2, label=r'$u_\theta(t)$ (NN)')
plt.plot(t_plot, u_exact, 'C2', lw=2, linestyle='--', label=r'Exact $k^2=0.5$')
plt.xlabel('t'); plt.ylabel('u(t)')
plt.title('Data, NN solution, and exact curve on the same plot')
plt.grid(alpha=0.3); plt.legend(); plt.tight_layout(); plt.show()

# Loss curves (optional but useful)
plt.figure(figsize=(7,4.5))
plt.plot(hist["epoch"], hist["total"],  label='Total', lw=2)
plt.plot(hist["epoch"], hist["data"],   '--', label='Data')
plt.plot(hist["epoch"], hist["ode"],    '--', label='ODE')
plt.plot(hist["epoch"], hist["ic"],     '--', label='IC')
plt.plot(hist["epoch"], hist["kvar"],   '--', label='Var(k)')
plt.yscale('log'); plt.xlabel('Epoch'); plt.ylabel('Loss')
plt.title('Loss vs Epochs')
plt.grid(alpha=0.3); plt.legend(); plt.tight_layout(); plt.show()

# ---- (2) Plot learned k^2 vs epochs ----
plt.figure(figsize=(7,4.5))
plt.plot(hist["epoch"], hist["k2"], lw=2, label=r'Learned $\overline{k^2}$')
plt.axhline(0.5, color='C2', linestyle='--', label=r'Ref $k^2=0.5$')
plt.xlabel('Epoch'); plt.ylabel(r'Learned $k^2$')
plt.title(r'$\hat{k}^2$ vs Epochs')
plt.grid(alpha=0.3); plt.legend(); plt.tight_layout(); plt.show()
```

3.
In this approach, we treat the unknown parameter $k$ as an internal
trainable parameter of the network. Let $w=(\theta,k)$ denote the
augmented parameter set, where $\theta$ are the standard weights of
the neural network and $k$ is the additional scalar parameter to be learned.


The network is constructed so that it automatically satisfies the
initial condition, by embedding it directly into the architecture as
$$u_w(t) \;=\; u_0 + t \, nn_w(t),$$
where $nn_w(t)$ is an unconstrained neural network.


To enforce the  governing Newton's law, we define the total loss as the squared residual of
the ODE:
$$
L_{\text{tot}}(w) \;=\; \int_0^{\tau}
\Bigg( \frac{d}{dt} u_w(t) + k^2 \big(u_w(t)-T\big) \Bigg)^2 \, dt.
$$

```python
# ============================================================
# Newton cooling: Variant 3 (hard-IC u(0)=u0, trainable k^2)
# ODE residual loss (+ optional data loss)
# Adds:
#   (1) data + u_NN(t) + u_exact(t) on same plot
#   (2) plot of learned k^2 vs epochs
# ============================================================

import numpy as np
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import matplotlib.pyplot as plt

# Use float64 for stability on small datasets
tf.keras.backend.set_floatx("float64")
np.random.seed(0); tf.random.set_seed(0)

# ---------------------------
# Data (t_j, u_j)
# ---------------------------
t_data = np.array([0.00, 1.11, 2.22, 3.33, 4.44, 5.56, 6.67, 7.78, 8.89, 10.00])
u_data = np.array([90.00, 64.42, 49.75, 41.33, 36.50, 33.73, 32.14, 31.22, 30.70, 30.40])

# Constants
u0 = 90.0
T  = 30.40     # ambient (use 30.0 if you prefer)
tau = float(t_data.max())

# Reference exact curve to overlay (change if desired)
k2_ref = 0.50
def u_exact_ref(t):
    return T + (u0 - T) * np.exp(-k2_ref * t)

# Tensors
t_tf = tf.constant(t_data[:, None], dtype=tf.float64)
u_tf = tf.constant(u_data[:, None], dtype=tf.float64)
T_tf = tf.constant(T, dtype=tf.float64)
u0_tf = tf.constant(u0, dtype=tf.float64)

# ---------------------------
# Network for nn_w(t) in u_w(t) = u0 + t * nn_w(t)
# ---------------------------
def mlp(in_dim=1, width=64, depth=3, act="tanh", out_dim=1):
    x = inp = keras.Input(shape=(in_dim,))
    for _ in range(depth):
        x = layers.Dense(width, activation=act)(x)
    out = layers.Dense(out_dim, activation=None)(x)
    return keras.Model(inp, out)

nn_w = mlp(in_dim=1, width=64, depth=3, act="tanh", out_dim=1)

# Trainable scalar raw parameter -> k^2 via softplus
k2_raw = tf.Variable(0.1, dtype=tf.float64, trainable=True, name="k2_raw")
def k2_param():
    return tf.nn.softplus(k2_raw) + tf.cast(1e-9, tf.float64)

def u_w(t):
    # Hard-embed IC: u(0) = u0
    return u0_tf + t * nn_w(t)

# ---------------------------
# Losses
# ---------------------------
def compute_losses(t_colloc):
    with tf.GradientTape() as tape:
        tape.watch(t_colloc)
        u_pred = u_w(t_colloc)               # (M,1)
    du_dt = tape.gradient(u_pred, t_colloc)  # (M,1)

    k2 = k2_param()
    ode_res = du_dt + k2 * (u_pred - T_tf)   # ODE: u' + k^2 (u - T) = 0

    # Integral approx over [0, tau]
    L_ode = (tau / tf.cast(tf.shape(t_colloc)[0], tf.float64)) * tf.reduce_mean(tf.square(ode_res))

    # Optional: small data MSE (set weight to 0.0 to disable)
    u_pred_data = u_w(t_tf)
    L_data = tf.reduce_mean(tf.square(u_pred_data - u_tf))

    return L_ode, L_data, u_pred_data, k2

# ---------------------------
# Training
# ---------------------------
opt = keras.optimizers.Adam(1e-3)
EPOCHS = 6000
PRINT_EVERY = 400

# Collocation sampling: mix data times + random interior points
M_extra = 128
def sample_collocation():
    t_extra = np.random.rand(M_extra, 1) * tau
    t_all = np.vstack([t_data[:, None], t_extra])
    return tf.constant(t_all, dtype=tf.float64)

# Loss weights
w_ode  = 1.0
w_data = 0.1   # set to 0.0 for strictly physics-only training

hist = {"epoch": [], "ode": [], "data": [], "total": [], "k2": []}

for ep in range(1, EPOCHS + 1):
    t_colloc = sample_collocation()
    with tf.GradientTape() as tape:
        L_ode, L_data, u_pred_data, k2_val = compute_losses(t_colloc)
        L_tot = w_ode * L_ode + w_data * L_data
    vars_all = nn_w.trainable_variables + [k2_raw]
    grads = tape.gradient(L_tot, vars_all)
    opt.apply_gradients(zip(grads, vars_all))

    # Log history
    hist["epoch"].append(ep)
    hist["ode"].append(L_ode.numpy())
    hist["data"].append(L_data.numpy())
    hist["total"].append(L_tot.numpy())
    hist["k2"].append(k2_val.numpy())

    if ep % PRINT_EVERY == 0:
        print(f"ep {ep:5d} | ODE {L_ode.numpy():.3e} | DATA {L_data.numpy():.3e} "
              f"| TOT {L_tot.numpy():.3e} | k^2={k2_val.numpy():.5f}")

# ---------------------------
# Report & Plots
# ---------------------------
k2_learned = float(k2_param().numpy())
print(f"\nLearned k^2: {k2_learned:.6f}")

# 1) Plot data + u_NN(t) + u_exact(t) on the same graph
t_plot = np.linspace(0, tau, 400)[:, None]
u_plot = u_w(tf.constant(t_plot, dtype=tf.float64)).numpy()
u_ref  = u_exact_ref(t_plot)

plt.figure(figsize=(7, 4.5))
plt.scatter(t_data, u_data, c='C0', label='Data', zorder=3)
plt.plot(t_plot, u_plot, 'C1', lw=2, label=r'$u_w(t)$ (NN)')
plt.plot(t_plot, u_ref,  'C2', lw=2, linestyle='--',
         label=rf'$u_{{\mathrm{{exact}}}}(t)=T+(u_0-T)e^{{-{k2_ref} t}}$')
plt.xlabel('t'); plt.ylabel('u(t)')
plt.title(rf'Hard-IC with trainable $k^2$ (learned $k^2={k2_learned:.3f}$)')
plt.grid(alpha=0.3); plt.legend(); plt.tight_layout(); plt.show()

# 2) Plot learned k^2 vs epochs
plt.figure(figsize=(7, 4.5))
plt.plot(hist["epoch"], hist["k2"], lw=2, label=r'$\hat{k}^2$ (learned)')
plt.axhline(k2_ref, color='C2', linestyle='--', label=rf'Ref $k^2={k2_ref}$')
plt.xlabel('Epoch'); plt.ylabel(r'Learned $k^2$')
plt.title(r'Estimate of $k^2$ vs Epochs')
plt.grid(alpha=0.3); plt.legend(); plt.tight_layout(); plt.show()

# (Optional) Loss curves
plt.figure(figsize=(7, 4.5))
plt.plot(hist["epoch"], hist["total"], label='Total', lw=2)
plt.plot(hist["epoch"], hist["ode"],   '--', label='ODE')
plt.plot(hist["epoch"], hist["data"],  '--', label='Data (optional)')
plt.yscale('log'); plt.xlabel('Epoch'); plt.ylabel('Loss')
plt.title('Loss vs Epochs')
plt.grid(alpha=0.3); plt.legend(); plt.tight_layout(); plt.show()
```

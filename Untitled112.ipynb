# **The Muskat Problem**

The one-dimensional Muskat problem can be formulated as finding
$u_i(x,t)$ and the moving interface $s(t)$ such that


$$
\begin{aligned}
\partial_x^2 u_1(x,t) &= 0, \qquad  0<x<s(t), \; 0<t\leq T, \\
\partial_x^2 u_2(x,t) &= 0,  \qquad   s(t)<x<L, \; 0<t\leq T, \\[1ex]
u_1(0,t) &= \phi(t), \quad u_2(L,t) = \psi(t), \\[1ex]
s'(t) &= -\alpha_1\, \partial_x u_1(s(t),t)
      \;=\; -\alpha_2\, \partial_x u_2(s(t),t), \\[1ex]
u_1(s(t),t) &= u_2(s(t),t),  \qquad  0<t\leq T, \\[1ex]
s(0) &= 1,
\end{aligned}
$$


with $\alpha_i$ given positive constants.

```python
# ============================================================
# 1D Muskat PINN with embedded boundary conditions
# u1(x,t) =  phi(t) + x        * nn1(x,t)      -> u1(0,t) = phi(t)
# u2(x,t) =  psi(t) + (x-L)    * nn2(x,t)      -> u2(L,t) = psi(t)
# s(t)   =  1 + ∫_0^t r(τ) dτ                 -> s(0)=1 (no monotonicity enforced)
#
# PDEs (each phase harmonic):      ∂_x^2 u1 = 0 in 0<x<s(t);  ∂_x^2 u2 = 0 in s(t)<x<L
# Pressure continuity at interface: u1(s(t),t) = u2(s(t),t)
# Muskat kinematic/equilibrium law: s'(t) + α1 ∂_x u1(s(t),t) = 0  and  s'(t) + α2 ∂_x u2(s(t),t) = 0
#
# Loss: L_tot = w_pde L_PDE + w_bm L_BM + w_m L_M
# ============================================================

import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt

# ----------------------------
# Problem setup
# ----------------------------
T_MAX = 5.0          # final time
L_DOM = 2.0          # length of domain (x in [0,2])
alpha1 = 1.0         # given positive constants α1, α2
alpha2 = 2.0

# Boundary data (feel free to change)
def phi_bc(t):  # left pressure at x=0
    return 1.0 + 0.25*tf.sin(0.8*t)

def psi_bc(t):  # right pressure at x=L
    return 1.0 - 0.15*tf.cos(0.6*t + 0.3)

# ----------------------------
# Networks
# ----------------------------
def build_mlp(input_dim, hidden=3, width=64, activation="tanh", output_dim=1):
    i = tf.keras.Input(shape=(input_dim,))
    z = i
    for _ in range(hidden):
        z = tf.keras.layers.Dense(width, activation=activation)(z)
    o = tf.keras.layers.Dense(output_dim)(z)
    return tf.keras.Model(i, o)

# Residual NNs for u1 and u2, and the rate r(t)=s'(t) (unconstrained sign)
nn1 = build_mlp(input_dim=2, hidden=3, width=64, activation="tanh", output_dim=1)  # for u1
nn2 = build_mlp(input_dim=2, hidden=3, width=64, activation="tanh", output_dim=1)  # for u2
r_net = build_mlp(input_dim=1, hidden=2, width=32, activation="tanh", output_dim=1) # for s'(t)

# Quadrature sub-steps for s(t) integral
M_QUAD = 100

@tf.function
def s_rate_of_t(t):
    # No softplus -> Muskat interface can move either direction
    return r_net(t)

@tf.function
def s_of_t(t):
    # Midpoint rule: s(t) = 1 + ∫_0^t r(τ) dτ
    t = tf.convert_to_tensor(t)
    ks = tf.range(M_QUAD, dtype=t.dtype) + 0.5
    taus = tf.expand_dims(ks, axis=0) * t / tf.cast(M_QUAD, t.dtype)  # shape (N,M)
    r_vals = r_net(tf.reshape(taus, (-1,1)))
    r_vals = tf.reshape(r_vals, (-1, M_QUAD))
    integral = tf.reduce_mean(r_vals, axis=1, keepdims=True) * t
    return 1.0 + integral   # s(0)=1

@tf.function
def u1_embedded(x, t):
    xt = tf.concat([x, t], axis=1)
    return phi_bc(t) + x*nn1(xt)  # enforces u1(0,t)=phi(t)

@tf.function
def u2_embedded(x, t):
    xt = tf.concat([x, t], axis=1)
    return psi_bc(t) + (x - L_DOM)*nn2(xt)  # enforces u2(L,t)=psi(t)

def smooth_heaviside(z, k=40.0):
    return 0.5 * (1.0 + tf.tanh(k * z))  # differentiable mask

# ----------------------------
# Training configuration
# ----------------------------
tf.keras.utils.set_random_seed(2025)
optimizer = tf.keras.optimizers.Adam(learning_rate=1e-4)

# Loss weights
w_pde = 1.0
w_bm  = 20.0
w_m   = 20.0

# Batches/steps
N_STEPS = 10000
N_PDE = 1200      # PDE collocation points
N_BM  = 256       # interface collocation times
N_M   = 1000      # collocation for Muskat law (domain-masked)

def sample_uniform(n, low, high):
    return tf.random.uniform((n, 1), minval=low, maxval=high, dtype=tf.float32)

# ----------------------------
# One training step
# ----------------------------
@tf.function
def train_step():
    # PDE collocation (all domain then mask by s(t))
    x_pde = sample_uniform(N_PDE, 0.0, L_DOM)
    t_pde = sample_uniform(N_PDE, 0.0, T_MAX)

    # Interface times for continuity + interface derivative
    t_bm = sample_uniform(N_BM, 0.0, T_MAX)

    # Collocation for Muskat law (domain-masked integrals)
    x_m = sample_uniform(N_M, 0.0, L_DOM)
    t_m = sample_uniform(N_M, 0.0, T_MAX)

    with tf.GradientTape() as tape_vars:
        # ----- PDE residuals: u_xx ≈ 0 in each phase -----
        with tf.GradientTape(persistent=True) as tape2:
            tape2.watch([x_pde, t_pde])
            with tf.GradientTape(persistent=True) as tape1:
                tape1.watch([x_pde, t_pde])
                u1_p = u1_embedded(x_pde, t_pde)
                u2_p = u2_embedded(x_pde, t_pde)
            u1_x  = tape1.gradient(u1_p, x_pde)
            u2_x  = tape1.gradient(u2_p, x_pde)
            del tape1
            u1_xx = tape2.gradient(u1_x, x_pde)
            u2_xx = tape2.gradient(u2_x, x_pde)
            del tape2

        s_p = s_of_t(t_pde)
        mask1 = smooth_heaviside(s_p - x_pde)         # ~1 on 0<x<s(t)
        mask2 = smooth_heaviside(x_pde - s_p)         # ~1 on s(t)<x<L
        # normalize masks to avoid scale drift if regions are small
        eps = 1e-8
        L_pde_1 = tf.reduce_sum(mask1 * tf.square(u1_xx)) / (tf.reduce_sum(mask1) + eps)
        L_pde_2 = tf.reduce_sum(mask2 * tf.square(u2_xx)) / (tf.reduce_sum(mask2) + eps)
        L_pde = 0.5*(L_pde_1 + L_pde_2)

        # ----- Boundary matching at interface: u1(s(t),t) = u2(s(t),t) -----
        s_bm = s_of_t(t_bm)
        u1_b = u1_embedded(s_bm, t_bm)
        u2_b = u2_embedded(s_bm, t_bm)
        L_bm = tf.reduce_mean(tf.square(u1_b - u2_b))

        # ----- Muskat equilibrium (domain integrals, masked) -----
        with tf.GradientTape(persistent=True) as tape_m:
            tape_m.watch([x_m, t_m])
            u1_m = u1_embedded(x_m, t_m)
            u2_m = u2_embedded(x_m, t_m)
        u1_x_m = tape_m.gradient(u1_m, x_m)
        u2_x_m = tape_m.gradient(u2_m, x_m)
        del tape_m

        s_m = s_of_t(t_m)
        dsdt_m = s_rate_of_t(t_m)

        mask1_m = smooth_heaviside(s_m - x_m)     # in 0<x<s(t)
        mask2_m = smooth_heaviside(x_m - s_m)     # in s(t)<x<L

        # Residuals extended over domains (as requested)
        r1 = dsdt_m + alpha1 * u1_x_m
        r2 = dsdt_m + alpha2 * u2_x_m

        L_m_1 = tf.reduce_sum(mask1_m * tf.square(r1)) / (tf.reduce_sum(mask1_m) + eps)
        L_m_2 = tf.reduce_sum(mask2_m * tf.square(r2)) / (tf.reduce_sum(mask2_m) + eps)
        L_m = 0.5*(L_m_1 + L_m_2)

        total = w_pde*L_pde + w_bm*L_bm + w_m*L_m

    vars_all = nn1.trainable_variables + nn2.trainable_variables + r_net.trainable_variables
    grads = tape_vars.gradient(total, vars_all)
    optimizer.apply_gradients(zip(grads, vars_all))

    return total, L_pde, L_bm, L_m

# ----------------------------
# Train + record history
# ----------------------------
loss_hist = {"total": [], "pde": [], "bm": [], "m": []}

for step in range(1, N_STEPS+1):
    total, L_pde, L_bm, L_m = train_step()
    for k, v in zip(loss_hist.keys(), [total, L_pde, L_bm, L_m]):
        loss_hist[k].append(v.numpy().item())
    if step % 100 == 0:
        print(f"Step {step:4d} | total={loss_hist['total'][-1]:.4e} "
              f"pde={loss_hist['pde'][-1]:.2e} bm={loss_hist['bm'][-1]:.2e} "
              f"m={loss_hist['m'][-1]:.2e}")

# ----------------------------
# Evaluate & plots
# ----------------------------
t_eval = np.linspace(0.0, T_MAX, 300, dtype=np.float32).reshape(-1,1)
s_eval = s_of_t(tf.convert_to_tensor(t_eval)).numpy().ravel()

plt.figure(figsize=(6,4))
plt.plot(t_eval.ravel(), s_eval, linewidth=2)
plt.xlabel("t"); plt.ylabel("s(t)")
plt.title("Learned Free Boundary s(t) — Muskat PINN")
plt.tight_layout(); plt.show()

# Total loss
plt.figure(figsize=(6,4))
plt.semilogy(loss_hist["total"], linewidth=2)
plt.xlabel("Epoch"); plt.ylabel("Total Loss")
plt.title("Total Loss vs Epochs")
plt.tight_layout(); plt.show()

# Component losses
plt.figure(figsize=(6,4))
for key in ["pde","bm","m"]:
    plt.semilogy(loss_hist[key], label=key)
plt.xlabel("Epoch"); plt.ylabel("Loss")
plt.title("Loss Components vs Epochs")
plt.legend(); plt.tight_layout(); plt.show()
```

**Exact formula**



I plotted the exact interface using the formula with the parameters
$\alpha_1=1$, $\alpha_2=2$, and $L=1$ (so
$A = \tfrac{1}{\alpha_1} - \tfrac{1}{\alpha_2} = 0.5$,
$B = \tfrac{L}{\alpha_2} = 0.5$).

The integral can be evaluated in closed form:
$$
\int_0^t \big(\phi(z)-\psi(z)\big)\,dz
= 0.3125\,(1-\cos(0.8t))
+ 0.25\big(\sin(0.6t+0.3) - \sin(0.3)\big).
$$

Then
$$
D(t) = (A+B)^2 + 2A \int_0^t \big(\phi(z)-\psi(z)\big)\,dz,
\qquad
s(t) = \frac{-B + \sqrt{D(t)}}{A}.
$$

I have plotted $s(t)$ for $0<t<5$ using these expressions.



```python
# Recompute and plot s(t) for L = 2 over 0 < t < 5 (rerun after reset)
import numpy as np
import matplotlib.pyplot as plt

alpha1 = 1.0
alpha2 = 2.0
L = 2.0  # updated as requested

A = 1/alpha1 - 1/alpha2  # 0.5
B = L/alpha2             # 1.0

def I(t):
    return 0.3125*(1 - np.cos(0.8*t)) + 0.25*(np.sin(0.6*t + 0.3) - np.sin(0.3))

def D(t):
    return (A + B)**2 + 2*A*I(t)

def s_exact(t):
    return (-B + np.sqrt(D(t)))/A

t = np.linspace(0, 5, 600)
s = s_exact(t)

plt.figure(figsize=(6,4))
plt.plot(t, s, linewidth=2)
plt.xlabel("t")
plt.ylabel("s(t)")
plt.title("Exact interface position s(t) for 0 < t < 5 (L = 2)")
plt.tight_layout()
plt.show()
```

We overlay next the learned NN solution for $s_{\theta}(t)$ with the exact solution $s_{exact}(t)$. We note a perfect overlap.

```python
# --------------------------------------
# Compare NN solution and exact solution
# --------------------------------------

# Parameters
alpha1 = 1.0
alpha2 = 2.0
L = 2.0

A = 1/alpha1 - 1/alpha2   # = 0.5
B = L/alpha2              # = 1.0

# Exact integral ∫0^t (phi - psi)
def I(t):
    return 0.3125*(1 - np.cos(0.8*t)) + 0.25*(np.sin(0.6*t + 0.3) - np.sin(0.3))

def D(t):
    return (A + B)**2 + 2*A*I(t)

def s_exact(t):
    return (-B + np.sqrt(D(t)))/A

# Evaluation grid
t_eval = np.linspace(0, 5, 300, dtype=np.float32).reshape(-1,1)

# NN prediction
s_nn = s_of_t(tf.convert_to_tensor(t_eval)).numpy().ravel()

# Exact solution
s_ex = s_exact(t_eval.ravel())

# Plot
plt.figure(figsize=(6,4))
plt.plot(t_eval, s_ex, 'k--', linewidth=2, label="Exact $s(t)$ (L=2)")
plt.plot(t_eval, s_nn, 'r-', linewidth=2, label="NN $s_\\theta(t)$")
plt.xlabel("t")
plt.ylabel("s(t)")
plt.title("NN vs Exact Interface Position $s(t)$, L=2")
plt.legend()
plt.tight_layout()
plt.show()
```

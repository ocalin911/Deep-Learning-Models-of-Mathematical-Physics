# **First Order ODE Discovery**

We aim to recover the first-order ODE
$$
u'(t) \;=\; -\tfrac{1}{2}\,\big(u(t) - 50\big),
$$
using 15 measurements $(t_i, u_i)$ given in th etable:


$t$  $u(t)$

0.000   100.0

0.714   84.984  
1.429   74.477  
2.143   67.126  
2.857   61.983  
3.571   58.384  
4.286   55.866  
5.000   54.104  
5.714   52.872  
6.429   52.009  
7.143   51.406  
7.857   50.984  
8.571   50.688

10.000  50.337  

The next code used Swish activation function instead of tanh.

```python
# ============================================================
# Two-network learning of autonomous ODE u'(t) = f(u)
# Networks: u_theta(t) and f_theta(u)
# Data: 15 samples in [0,10] for u(t) from u'(t) = -0.5 (u-50), u(0)=100
# Hard embedding: u_theta(t) = u0 + t * nn_t(t)  => u_theta(0) = u0
# Loss: L_tot = L_data + L_ODE
# ============================================================

import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, models
import matplotlib.pyplot as plt

# ----------------------------
# 0) Dataset (15 points on [0,10])
# ----------------------------
u0 = 100.0

def u_exact(t):
    # exact for reference: u(t) = 50 + 50*exp(-0.5 t)
    return 50.0 + 50.0 * np.exp(-0.5 * t)

# 15 equally spaced t in [0,10]
t_data = np.linspace(0.0, 10.0, 15, dtype=np.float32).reshape(-1, 1)
u_data = u_exact(t_data).astype(np.float32)

# Optional: normalize t to [0,1] for smoother training
t_min, t_max = float(t_data.min()), float(t_data.max())
def t_norm(t):
    return (t - t_min) / (t_max - t_min + 1e-12)

# ----------------------------
# 1) Networks
# ----------------------------
def build_u_net(hidden=3, width=64, act="tanh"):
    """u network core (produces residual nn_t). Hard embedding applied outside."""
    inp = tf.keras.Input(shape=(1,))    # input is t (we use normalized t)
    z = inp
    for _ in range(hidden):
        z = layers.Dense(width, activation=act)(z)
    out = layers.Dense(1)(z)
    return models.Model(inp, out, name="nn_t")

def build_f_net(hidden=3, width=64, act="swish"):
    """f network: f_theta(u)"""
    inp = tf.keras.Input(shape=(1,))    # input is u
    z = inp
    for _ in range(hidden):
        z = layers.Dense(width, activation=act)(z)
    out = layers.Dense(1)(z)
    return models.Model(inp, out, name="f_theta")

nn_t = build_u_net(hidden=3, width=64, act="swish")
f_theta = build_f_net(hidden=3, width=64, act="swish")

@tf.function
def u_theta(t):
    """Hard-embedded trajectory: u(t) = u0 + t * nn_t( t_normalized )."""
    t = tf.convert_to_tensor(t, dtype=tf.float32)
    tn = (t - t_min) / (t_max - t_min + 1e-12)
    return u0 + t * nn_t(tn)

# ----------------------------
# 2) Loss components
# ----------------------------
lambda_data = tf.constant(1.0, tf.float32)
lambda_ode  = tf.constant(1.0, tf.float32)

# choose M extra ODE collocation points (optional)
M = 64
tau = np.linspace(0.0, 10.0, M, dtype=np.float32).reshape(-1, 1)

@tf.function
def compute_losses(t_batch, u_batch, tau_batch):
    """Compute L_data and L_ODE per your formulas."""
    t_batch = tf.convert_to_tensor(t_batch, dtype=tf.float32)
    u_batch = tf.convert_to_tensor(u_batch, dtype=tf.float32)
    tau_batch = tf.convert_to_tensor(tau_batch, dtype=tf.float32)

    # ----- Data loss -----
    u_pred = u_theta(t_batch)
    L_data = tf.reduce_mean(tf.square(u_pred - u_batch))

    # u'_theta(t) via autodiff
    with tf.GradientTape() as tape1:
        tape1.watch(t_batch)
        u_for_grad = u_theta(t_batch)
    du_dt_at_data = tape1.gradient(u_for_grad, t_batch)

    # First ODE term: at data points, f_theta(u_i)
    f_at_data = f_theta(u_batch)   # note: uses ground-truth u_i
    term1 = tf.reduce_mean(tf.square(du_dt_at_data - f_at_data))

    # Second ODE term: at additional collocation times tau_j
    with tf.GradientTape() as tape2:
        tape2.watch(tau_batch)
        u_tau = u_theta(tau_batch)
    du_dt_at_tau = tape2.gradient(u_tau, tau_batch)
    f_at_tau = f_theta(u_tau)
    term2 = tf.reduce_mean(tf.square(du_dt_at_tau - f_at_tau))

    L_ODE = term1 + term2
    L_total = lambda_data * L_data + lambda_ode * L_ODE
    return L_total, L_data, L_ODE, term1, term2

# ----------------------------
# 3) Training loop
# ----------------------------
optimizer = tf.keras.optimizers.Adam(learning_rate=1e-3)

t_train = tf.convert_to_tensor(t_data, dtype=tf.float32)
u_train = tf.convert_to_tensor(u_data, dtype=tf.float32)
tau_train = tf.convert_to_tensor(tau, dtype=tf.float32)

loss_hist, data_hist, ode_hist = [], [], []
ode1_hist, ode2_hist = [], []

EPOCHS = 12000
for ep in range(1, EPOCHS+1):
    with tf.GradientTape() as tape:
        L_total, L_data, L_ODE, term1, term2 = compute_losses(t_train, u_train, tau_train)
    vars_all = nn_t.trainable_variables + f_theta.trainable_variables
    grads = tape.gradient(L_total, vars_all)
    optimizer.apply_gradients(zip(grads, vars_all))

    loss_hist.append(float(L_total.numpy()))
    data_hist.append(float(L_data.numpy()))
    ode_hist.append(float(L_ODE.numpy()))
    ode1_hist.append(float(term1.numpy()))
    ode2_hist.append(float(term2.numpy()))

    if ep % 500 == 0:
        print(f"Epoch {ep:5d} | L_tot={loss_hist[-1]:.3e}  "
              f"L_data={data_hist[-1]:.3e}  L_ODE={ode_hist[-1]:.3e}  "
              f"[data ODE={ode1_hist[-1]:.3e}, colloc ODE={ode2_hist[-1]:.3e}]")

# ----------------------------
# 4) Evaluation & plots
# ----------------------------
t_eval = np.linspace(0, 10, 200, dtype=np.float32).reshape(-1,1)
u_eval = u_theta(t_eval).numpy()
u_true = u_exact(t_eval)

# Plot u(t): data vs NN vs exact
plt.figure(figsize=(6.2,4.2))
plt.plot(t_eval, u_true, 'k--', lw=2, label='Exact $u(t)$')
plt.plot(t_eval, u_eval, 'r-', lw=2, label='NN $u_\\theta(t)$')
plt.plot(t_data, u_data, 'bo', ms=5, label='Data')
plt.xlabel('t'); plt.ylabel('u(t)')
plt.title('Trajectory fit')
plt.grid(True, ls='--', alpha=0.5); plt.legend()
plt.tight_layout(); plt.show()

# Learned f(u): sample u-range from data
u_min = float(u_data.min()); u_max = float(u_data.max())
u_grid = np.linspace(u_min, u_max, 200, dtype=np.float32).reshape(-1,1)
f_learned = f_theta(u_grid).numpy()
# Exact f(u) = -0.5 (u-50) shown for reference
f_true = -0.5*(u_grid - 50.0)

plt.figure(figsize=(6.2,4.2))
plt.plot(u_grid, f_true, 'k--', lw=2, label='Exact $f(u)$')
plt.plot(u_grid, f_learned, 'm-', lw=2, label='NN $f_\\theta(u)$')
plt.xlabel('u'); plt.ylabel('f(u)')
plt.title('Learned dynamics $f(u)$')
plt.grid(True, ls='--', alpha=0.5); plt.legend()
plt.tight_layout(); plt.show()

# Loss curves
plt.figure(figsize=(6.2,4.2))
plt.semilogy(loss_hist, 'm-', label='L_total')
plt.semilogy(data_hist, 'g--', label='L_data')
plt.semilogy(ode_hist, 'c--', label='L_ODE')
plt.semilogy(ode1_hist, 'b:', label='ODE@data')
plt.semilogy(ode2_hist, 'r:', label='ODE@colloc')
plt.xlabel('epoch'); plt.ylabel('loss')
plt.title('Training losses')
plt.grid(True, ls='--', alpha=0.5); plt.legend()
plt.tight_layout(); plt.show()
```

 **Larangian Neural Network**

We consider the case of the Harmonic oscillator.




Consider the Lagrangian
$$
L(q,\dot q)=\tfrac12\,\dot q^{\,2}-\tfrac12\,q^{2}.
$$
The associated Euler--Lagrange equation is
$\ddot q + q = 0$.
With initial conditions $q(0)=q_0$ and $\dot q(0)=v_0$, the solution is


$$
q(t)=q_0\cos t + v_0\sin t,\quad
\dot q(t)=-\,q_0\sin t + v_0\cos t,\quad
\ddot q(t)=-\,q_0\cos t - v_0\sin t.
$$


Along each solution, the energy is conserved:
$$
E(q_0,v_0)
= \tfrac12\,\dot q(t)^2 + \tfrac12\,q(t)^2
= \tfrac12\,(v_0^2 + q_0^2).
$$


We generate $25$ pairs of initial conditions $(q_0,v_0)$ with $E(q_0,v_0) \in[0.2,1]$.
For each trajectory, we uniformly sample $N=30$ time points and record the triplets
$(q(t_i),\,\dot q(t_i),\,\ddot q(t_i))$, then add Gaussian noise with variance
$\sigma^2=0.1$ to the data. These noisy triplets constitute the training set:
the first two components are the inputs to the network, and the last one
(the acceleration) is the target variable.

```python
# -*- coding: utf-8 -*-
# Harmonic oscillator dataset, Baseline vs. LNN (Softplus activations)
# Plots: energy vs time, and TRAINING LOSSES IN LOG SCALE.

import numpy as np
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import matplotlib.pyplot as plt

# -----------------------------
# 0) Repro + TF config
# -----------------------------
np.random.seed(42)
tf.random.set_seed(42)

# -----------------------------
# 1) Ground-truth dynamics & data generation
# -----------------------------
# q'' + q = 0  (m = ω = 1)
def exact_solution(t, q0, v0):
    q = q0*np.cos(t) + v0*np.sin(t)
    v = -q0*np.sin(t) + v0*np.cos(t)
    a = -q
    return q, v, a

def energy(q, v):
    return 0.5*(v**2 + q**2)

T = 10.0
N = 30
num_ic = 25
sigma2 = 0.1
sigma = np.sqrt(sigma2)

def sample_initial_conditions(num, E_min=0.2, E_max=1.0):
    q0s, v0s = [], []
    while len(q0s) < num:
        E = np.random.uniform(E_min, E_max)
        r = np.sqrt(2*E)  # since E = 0.5*(q0^2+v0^2)
        theta = np.random.uniform(0, 2*np.pi)
        q0, v0 = r*np.cos(theta), r*np.sin(theta)
        q0s.append(q0); v0s.append(v0)
    return np.array(q0s), np.array(v0s)

q0s, v0s = sample_initial_conditions(num_ic)
t_grid = np.linspace(0.0, T, N)

X_list, y_list = [], []
for q0, v0 in zip(q0s, v0s):
    q, v, a = exact_solution(t_grid, q0, v0)
    # Add Gaussian noise
    qn = q + np.random.normal(0.0, sigma, size=q.shape)
    vn = v + np.random.normal(0.0, sigma, size=v.shape)
    an = a + np.random.normal(0.0, sigma, size=a.shape)
    X_list.append(np.stack([qn, vn], axis=1))
    y_list.append(an[:, None])

X = np.concatenate(X_list, axis=0).astype(np.float32)  # (num_ic*N, 2)
y = np.concatenate(y_list, axis=0).astype(np.float32)  # (num_ic*N, 1)

# Train/val split
perm = np.random.permutation(len(X))
split = int(0.8*len(X))
idx_train, idx_val = perm[:split], perm[split:]
X_train, y_train = X[idx_train], y[idx_train]
X_val, y_val     = X[idx_val], y[idx_val]

# -----------------------------
# 2) Baseline model: (q, v) -> a  (Softplus activations)
# -----------------------------
def build_baseline():
    inp = layers.Input(shape=(2,))
    x = layers.Dense(164, activation="softplus")(inp)
    x = layers.Dense(164, activation="softplus")(x)
    out = layers.Dense(1, activation=None)(x)
    return keras.Model(inp, out, name="baseline_a_of_qv")

baseline = build_baseline()
baseline.compile(optimizer=keras.optimizers.Adam(1e-3), loss="mse")

hist_baseline = baseline.fit(
    X_train, y_train,
    validation_data=(X_val, y_val),
    epochs=4000,
    batch_size=128,
    verbose=1,
)

# -----------------------------
# 3) LNN: learn L(q, v), then a via Euler–Lagrange (Softplus activations)
# -----------------------------
class LagrangianNet(keras.Model):
    """Scalar L(q, v) network with Softplus."""
    def __init__(self):
        super().__init__()
        self.mlp = keras.Sequential([
            layers.Input(shape=(2,)),
            layers.Dense(164, activation="softplus"),
            layers.Dense(164, activation="softplus"),
            layers.Dense(1, activation=None),
        ])

    def call(self, qv):
        return self.mlp(qv)

def lnn_acceleration(L_net, qv):
    """
    Given a batch of (q, v), compute a = (L_vv)^{-1} (L_q - L_qv * v).
    Shapes are all (B, 1).
    """
    qv = tf.convert_to_tensor(qv)
    qv = tf.reshape(qv, (-1, 2))
    with tf.GradientTape(persistent=True) as tape2:
        tape2.watch(qv)
        with tf.GradientTape(persistent=True) as tape1:
            tape1.watch(qv)
            L = L_net(qv)                         # (B,1)
        dL_dqv = tape1.gradient(L, qv)            # (B,2): [dL/dq, dL/dv]
        L_q = dL_dqv[:, 0:1]
        L_v = dL_dqv[:, 1:2]
    # Second derivatives
    L_vv = tape2.gradient(L_v, qv)[:, 1:2]        # d/dv(L_v)
    L_qv = tape2.gradient(L_q, qv)[:, 1:2]        # d/dv(L_q)
    del tape1, tape2

    v = qv[:, 1:2]
    eps = 1e-6
    a = (L_q - L_qv * v) / (L_vv + eps)
    return a

class LNN_Model(keras.Model):
    """Wraps L(q,v) and exposes a(q,v) as output for supervised training on acceleration."""
    def __init__(self):
        super().__init__()
        self.L_net = LagrangianNet()

    def call(self, qv):
        return lnn_acceleration(self.L_net, qv)

lnn = LNN_Model()
opt = keras.optimizers.Adam(1e-3)

# Custom training loop for LNN (higher-order grads)
train_losses, val_losses = [], []
batch_size = 128
epochs = 4000

train_ds = tf.data.Dataset.from_tensor_slices((X_train, y_train)).shuffle(4096).batch(batch_size)
val_ds   = tf.data.Dataset.from_tensor_slices((X_val,   y_val)).batch(batch_size)

mse = keras.losses.MeanSquaredError()

for ep in range(epochs):
    # Train
    tl = []
    for xb, yb in train_ds:
        with tf.GradientTape() as tape:
            yhat = lnn(xb)
            loss = mse(yb, yhat)
        grads = tape.gradient(loss, lnn.trainable_variables)
        opt.apply_gradients(zip(grads, lnn.trainable_variables))
        tl.append(loss.numpy())
    train_losses.append(np.mean(tl))
    # Val
    vl = []
    for xb, yb in val_ds:
        yhat = lnn(xb)
        vl.append(mse(yb, yhat).numpy())
    val_losses.append(np.mean(vl))

hist_lnn = {"loss": train_losses, "val_loss": val_losses}

# -----------------------------
# 4) Rollout integrator (RK4) using each model
# -----------------------------
def rk4_rollout(accel_fn, q0, v0, dt=0.05, T=20.0):
    steps = int(T/dt)
    qs = np.zeros(steps+1); vs = np.zeros(steps+1); ts = np.linspace(0, T, steps+1)
    qs[0], vs[0] = q0, v0

    def a_np(q, v):
        inp = np.array([[q, v]], dtype=np.float32)
        a = accel_fn(inp).numpy().squeeze()
        return float(a)

    for k in range(steps):
        q, v = qs[k], vs[k]
        k1_q = v
        k1_v = a_np(q, v)

        k2_q = v + 0.5*dt*k1_v
        k2_v = a_np(q + 0.5*dt*k1_q, v + 0.5*dt*k1_v)

        k3_q = v + 0.5*dt*k2_v
        k3_v = a_np(q + 0.5*dt*k2_q, v + 0.5*dt*k2_v)

        k4_q = v + dt*k3_v
        k4_v = a_np(q + dt*k3_q, v + dt*k3_v)

        qs[k+1] = q + (dt/6.0)*(k1_q + 2*k2_q + 2*k3_q + k4_q)
        vs[k+1] = v + (dt/6.0)*(k1_v + 2*k2_v + 2*k3_v + k4_v)

    return ts, qs, vs

def accel_baseline(qv_batch):
    return baseline(qv_batch)

def accel_lnn(qv_batch):
    return lnn(qv_batch)

# Choose a validation initial condition and compare rollouts
q0_test, v0_test = 0.8, -0.2
dt, T_roll = 0.05, 20.0

t_ex = np.linspace(0, T_roll, int(T_roll/dt)+1)
q_ex, v_ex, a_ex = exact_solution(t_ex, q0_test, v0_test)

t_b, q_b, v_b = rk4_rollout(accel_baseline, q0_test, v0_test, dt=dt, T=T_roll)
t_l, q_l, v_l = rk4_rollout(accel_lnn,      q0_test, v0_test, dt=dt, T=T_roll)

E_ex = energy(q_ex, v_ex)
E_b  = energy(q_b, v_b)
E_l  = energy(q_l, v_l)

# -----------------------------
# 5) Plots
# -----------------------------
# (A) Energy vs time
plt.figure(figsize=(7,4))
plt.plot(t_ex, E_ex, label="Ground truth")
plt.plot(t_b,  E_b,  label="Baseline (rollout)")
plt.plot(t_l,  E_l,  label="LNN (rollout)")
plt.xlabel("time")
plt.ylabel("energy")
plt.title("Energy vs. time")
plt.legend()
plt.tight_layout()
plt.show()

# (B) Training loss curves — LOG SCALE
b_tr = np.clip(np.array(hist_baseline.history["loss"]),     1e-12, None)
b_va = np.clip(np.array(hist_baseline.history["val_loss"]), 1e-12, None)
l_tr = np.clip(np.array(hist_lnn["loss"]),                  1e-12, None)
l_va = np.clip(np.array(hist_lnn["val_loss"]),              1e-12, None)

plt.figure(figsize=(7,4))
plt.semilogy(b_tr, label="Baseline train")
plt.semilogy(b_va, label="Baseline val")
plt.semilogy(l_tr, label="LNN train")
plt.semilogy(l_va, label="LNN val")
plt.xlabel("epoch")
plt.ylabel("MSE loss (log scale)")
plt.title("Training losses (log y-scale)")
plt.legend()
plt.tight_layout()
plt.show()

# (C) Optional: trajectory q(t) comparison
plt.figure(figsize=(7,4))
plt.plot(t_ex, q_ex, label="q exact")
plt.plot(t_b,  q_b,  label="q baseline")
plt.plot(t_l,  q_l,  label="q LNN")
plt.xlabel("time")
plt.ylabel("q(t)")
plt.title("Trajectory comparison")
plt.legend()
plt.tight_layout()
plt.show()
```

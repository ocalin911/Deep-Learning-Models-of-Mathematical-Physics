# **The Mixed value problem for the 1D wave equation**


We consider the mixed homogeneous mixed problem:

\begin{cases}
\partial_t^2 v-\partial_x^2 v=\sin(\pi x) \cos(\pi t), & (t,x)\in(0,\infty)\times(0,1),\\[2pt]
v(0,x)= - \frac{1}{2\pi^2} \sin (\pi x) , & x\in[0,1],\\[2pt]
\partial_t v(0,x)= \pi \sin(\pi x), & x\in[0,1],\\[2pt]
v(t,0)=v(t,1)=0, & t\in[0,1],
\end{cases}

Since all the hypotheses of the reflection method are satisfied, then d'Alembert's formula provides the following classical solution
$$v(t,x)=\Bigl[-\frac{1}{2\pi^2}\cos(\pi t)\;+\;\sin(\pi t)\;+\;\frac{t}{2\pi}\,\sin(\pi t)\Bigr]\sin(\pi x),
\quad (t,x)\in[0,\infty)\times[0,1]. $$

To solve this problem using neural networks, we use the following solution embedding



\begin{eqnarray*}
 v_{\theta}(t, x) &=& v(0, x) + t \partial_t v(0, x) + t^2 x(1-x)\,  {\rm nn}(\theta; t,x)\\
&=&  - \frac{1}{2\pi^2} \sin (\pi x) + \pi t \sin(\pi x)+ t^2 x(1-x) \, {\rm nn}(\theta; t,x).
\end{eqnarray*}

```python
# PINN for 1D wave mixed problem on (0,1):
#   v_tt - v_xx = sin(pi x) cos(pi t)
# ICs: v(0,x) = -(1/(2*pi^2)) sin(pi x),   v_t(0,x) = pi sin(pi x)
# BCs: v(t,0)=v(t,1)=0
# Ansatz (hard-embedding ICs + BCs):
#   v_theta(t,x) = v0(x) + t * v1(x) + t^2 * x*(1-x) * nn(t,x)

import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, Model
import matplotlib.pyplot as plt

# ----------------------------
# Problem setup
# ----------------------------
pi = np.pi
T_MAX = 2.0
X_MIN, X_MAX = 0.0, 1.0

# Exact solution
def v_exact_np(t, x):
    a = -(1/(2*pi**2))*np.cos(pi*t) + np.sin(pi*t) + (t/(2*pi))*np.sin(pi*t)
    return a * np.sin(pi*x)

# Initial data
def v0_np(x):
    return -(1/(2*pi**2)) * np.sin(pi*x)
def v1_np(x):
    return  pi * np.sin(pi*x)

# ----------------------------
# Collocation points
# ----------------------------
NT, NX = 128, 256
t_col = np.linspace(0.0, T_MAX, NT, dtype=np.float32)[:,None]
x_col = np.linspace(X_MIN, X_MAX, NX, dtype=np.float32)[None,:]
TT, XX = np.meshgrid(t_col.flatten(), x_col.flatten(), indexing='ij')
t_train = TT.reshape(-1,1).astype(np.float32)
x_train = XX.reshape(-1,1).astype(np.float32)

t_tf = tf.constant(t_train, dtype=tf.float32)
x_tf = tf.constant(x_train, dtype=tf.float32)

# ----------------------------
# Network (for the residual correction)
# ----------------------------
def build_mlp(widths=(64,64,64), activation='tanh'):
    inp = layers.Input(shape=(2,))
    z = inp
    for w in widths:
        z = layers.Dense(w, activation=activation)(z)
    out = layers.Dense(1, activation=None)(z)
    return Model(inp, out, name="nn_core")

nn = build_mlp((64,64,64), 'tanh')
optimizer = tf.keras.optimizers.Adam(1e-3)
EPOCHS = 3000

# ----------------------------
# Ansatz with hard embedding
# v_theta(t,x) = v0(x) + t*v1(x) + t^2 * x*(1-x) * nn(t,x)
# ----------------------------
@tf.function
def v_theta(t, x):
    v0 = -(1.0/(2.0*(np.pi**2))) * tf.sin(np.pi * x)
    v1 =  (np.pi) * tf.sin(np.pi * x)
    z   = tf.concat([t, x], axis=1)
    corr = nn(z)
    return v0 + t*v1 + (t**2) * x*(1.0-x) * corr

# RHS f(t,x) = sin(pi x) cos(pi t)
@tf.function
def rhs(t, x):
    return tf.sin(np.pi * x) * tf.cos(np.pi * t)

# Residual: v_tt - v_xx - f(t,x)
@tf.function
def residual(t, x):
    with tf.GradientTape(persistent=True) as t2:
        t2.watch([t, x])
        with tf.GradientTape(persistent=True) as t1:
            t1.watch([t, x])
            v = v_theta(t, x)
        v_t  = t1.gradient(v, t)
        v_x  = t1.gradient(v, x)
        del t1
    v_tt = t2.gradient(v_t, t)
    v_xx = t2.gradient(v_x, x)
    del t2
    return v_tt - v_xx - rhs(t, x)

@tf.function
def loss_pde(t, x):
    r = residual(t, x)
    return tf.reduce_mean(tf.square(r))

@tf.function
def train_step():
    with tf.GradientTape() as tape:
        Lp = loss_pde(t_tf, x_tf)
    grads = tape.gradient(Lp, nn.trainable_variables)
    optimizer.apply_gradients(zip(grads, nn.trainable_variables))
    return Lp

# ----------------------------
# Train
# ----------------------------
hist = []
for ep in range(1, EPOCHS+1):
    Lp = train_step()
    hist.append(float(Lp.numpy()))
    if ep % 300 == 0:
        print(f"Epoch {ep:4d} | PDE loss {hist[-1]:.3e}")

# ----------------------------
# Plot training loss
# ----------------------------
plt.figure(figsize=(6,4))
plt.plot(hist)
plt.yscale('log')
plt.xlabel('Epoch'); plt.ylabel('PDE loss'); plt.title('Training loss')
plt.tight_layout(); plt.show()

# ----------------------------
# Compare profiles at several times
# ----------------------------
t_list = [0.1, 0.5, 1.0, 1.5, 2.0]
x_plot = np.linspace(0.0, 1.0, 400, dtype=np.float32).reshape(-1,1)

plt.figure(figsize=(9,6))
for t0 in t_list:
    t_arr = np.full_like(x_plot, t0, dtype=np.float32)
    v_nn = v_theta(tf.constant(t_arr), tf.constant(x_plot)).numpy().flatten()
    v_ex = v_exact_np(t0, x_plot.flatten())
    plt.plot(x_plot, v_ex, label=f'Exact t={t0}')
    plt.plot(x_plot, v_nn, '--', label=f'NN t={t0}')
plt.xlabel('x'); plt.ylabel('v(t,x)')
plt.title('Exact vs NN (all times on one graph)')
plt.legend(ncol=2)
plt.tight_layout(); plt.show()
```

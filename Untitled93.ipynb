**The Simple Pendulum**

The dynamics of a simple planar pendulum of unit mass and unit length is governed by the Hamiltonian
$$
H(q,p)=\frac12\,p^{2}-\kappa\cos q,
\qquad
\kappa>0.
$$

The Hamiltonian system is
$$
\dot q = p,
\qquad
\dot p = -\,\kappa\sin q,
$$
together with the initial conditions $q(0)=q_{0}$ and $p(0)=p_{0}$.

```python
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt
from  scipy.integrate import solve_ivp

# --- problem parameters --------------------------------------------------
kappa = 1.0               # pendulum constant   (appears in  -kappa sin q)
q0, p0 = 1.0, 0.0         # initial conditions q(0), p(0)
T       = 1.0             # final time
m       = 100             # collocation points
epochs  = 100             # training epochs
lr      = 5e-3            # Adam learning rate
np.random.seed(0)
tf.random.set_seed(0)

# --- collocation grid ----------------------------------------------------
t_colloc = tf.constant(np.linspace(0.0, T, m)[:, None], dtype=tf.float32)

# --- network  ------------------------------------------------------------
model = tf.keras.Sequential([
    tf.keras.layers.InputLayer(input_shape=(1,)),
    tf.keras.layers.Dense(64, activation='tanh'),
    tf.keras.layers.Dense(64, activation='tanh'),
    tf.keras.layers.Dense(64, activation='tanh'),
    tf.keras.layers.Dense(2,  activation=None)          # nn¹, nn²
])

opt = tf.keras.optimizers.Adam(lr)
loss_history = []

# --- physics loss --------------------------------------------------------
@tf.function
def physics_loss():
    with tf.GradientTape(persistent=True) as tape:
        tape.watch(t_colloc)
        nn_out  = model(t_colloc)              # (m,2)
        nn1, nn2 = nn_out[:, 0:1], nn_out[:, 1:2]

        q_pred = q0 + t_colloc * nn1
        p_pred = p0 + t_colloc * nn2

    dq_dt = tape.gradient(q_pred, t_colloc)
    dp_dt = tape.gradient(p_pred, t_colloc)
    del tape

    res_q = dq_dt - p_pred
    res_p = dp_dt + kappa * tf.sin(q_pred)
    return tf.reduce_mean(res_q**2) + tf.reduce_mean(res_p**2)

@tf.function
def train_step():
    with tf.GradientTape() as tape:
        loss = physics_loss()
    grads = tape.gradient(loss, model.trainable_variables)
    opt.apply_gradients(zip(grads, model.trainable_variables))
    return loss

# --- training loop -------------------------------------------------------
for epoch in range(1, epochs + 1):
    loss_val = train_step().numpy()
    loss_history.append(loss_val)
    if epoch % 10 == 0 or epoch == 1:
        print(f"Epoch {epoch:3d}: loss = {loss_val:.3e}")

# --- evaluate network ----------------------------------------------------
t_plot  = np.linspace(0.0, T, 400)[:, None].astype(np.float32)
nn_out  = model(t_plot).numpy()
q_pred  = q0 + t_plot * nn_out[:, 0:1]

# --- reference solution (RK45) ------------------------------------------
def pendulum_ode(t, y):
    q, v = y
    return [v, -np.sin(q)]

sol   = solve_ivp(pendulum_ode, [0, T], [q0, p0],
                  t_eval=t_plot.flatten(), method='RK45')
q_ref = sol.y[0]

# --- plots ---------------------------------------------------------------
plt.figure(figsize=(7, 3.5))
plt.plot(t_plot, q_ref,  label='RK45 reference',  c='tab:blue')
plt.plot(t_plot, q_pred, '--', label='PINN prediction', c='tab:orange')
plt.xlabel('t'); plt.ylabel('q(t)')
plt.title('Simple pendulum: PINN vs Runge–Kutta')
plt.legend(); plt.grid(True); plt.tight_layout()

plt.figure(figsize=(7, 3.5))
plt.semilogy(loss_history, c='tab:green')
plt.xlabel('epoch'); plt.ylabel('physics loss')
plt.title('Training loss'); plt.tight_layout()
plt.show()
```

The next code implements a two-boundary condition problem $q(0)=1, q(1)=0.6$ for the simple pendulum. It uses conservation of energy as a regularizer.

```python
"""
Simple pendulum
Boundary values: q(0)=1,  q(T)=0.6.
Three hidden layers (64 tanh) → two outputs (nn¹, nn²).
Boundary conditions are embedded; energy variance is used as a
soft constraint.  Trains for 100 epochs.
"""

import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, Model
import matplotlib.pyplot as plt

# ------------------------ problem parameters ---------------------------
kappa   = 1.0       # pendulum constant
T       = 1.0       # final time
q0, qT  = 1.0, 0.6  # q(0), q(T)
m       = 1000     # collocation points
epochs  = 500
lam     = 1.0       # weight for energy penalty

t_col = tf.constant(np.linspace(0.0, T, m)[:, None], dtype=tf.float32)

# ------------------------ network: 3×64 tanh, 2 outputs ----------------
inp  = layers.Input(shape=(1,))
x    = layers.Dense(64, activation='tanh')(inp)
x    = layers.Dense(64, activation='tanh')(x)
x    = layers.Dense(64, activation='tanh')(x)
out  = layers.Dense(2)(x)               # (nn¹, nn²)
model = Model(inp, out)

opt = tf.keras.optimizers.Adam(1e-3)
loss_history = []

# ------------------------ helper functions -----------------------------
@tf.function
def H(q, p):
    return 0.5 * p**2 - kappa * tf.cos(q)

@tf.function
def physics_loss(q, p, q_dot, p_dot):
    """q̇ = p,    ṗ = −κ sin q"""
    return (tf.reduce_mean((q_dot - p)**2) +
            tf.reduce_mean((p_dot + kappa * tf.sin(q))**2))

@tf.function
def energy_variance(q, p):
    e_bar = tf.reduce_mean(H(q, p))
    return tf.reduce_mean((H(q, p) - e_bar)**2)

# ------------------------ training loop -------------------------------
for epoch in range(1, epochs + 1):
    with tf.GradientTape(persistent=True) as tape:
        tape.watch(t_col)

        nn1, nn2 = tf.split(model(t_col), 2, axis=1)

        # embed boundary conditions on [0, T]
        q_pred = (T - t_col) / T * q0 + t_col / T * qT + t_col * (T - t_col) * nn1
        p_pred = nn2

        # automatic derivatives wrt time
        q_dot = tape.gradient(q_pred, t_col)
        p_dot = tape.gradient(p_pred, t_col)

        loss_phy = physics_loss(q_pred, p_pred, q_dot, p_dot)
        loss_eng = energy_variance(q_pred, p_pred)
        total    = loss_phy + lam * loss_eng

    grads = tape.gradient(total, model.trainable_variables)
    opt.apply_gradients(zip(grads, model.trainable_variables))
    del tape
    loss_history.append(total.numpy())

    if epoch % 20 == 0 or epoch == 1:
        print(f"Epoch {epoch:3d} | total loss = {total.numpy():.3e}")

# ------------------------ evaluation ----------------------------------
t_plot = np.linspace(0.0, T, 500)[:, None].astype(np.float32)
nn1_p, nn2_p = tf.split(model(t_plot), 2, axis=1)
q_plot = (T - t_plot) / T * q0 + t_plot / T * qT + t_plot * (T - t_plot) * nn1_p
p_plot = nn2_p

# ------------------------ plots ---------------------------------------
fig, axs = plt.subplots(1, 3, figsize=(14, 4))

axs[0].plot(t_plot, q_plot.numpy())
axs[0].set_title(r'Position $q(t)$')
axs[0].set_xlabel('time $t$')

axs[1].plot(t_plot, p_plot.numpy(), color='orange')
axs[1].set_title(r'Momentum $p(t)$')
axs[1].set_xlabel('time $t$')

axs[2].semilogy(loss_history)
axs[2].set_title('Training loss')
axs[2].set_xlabel('epoch')

plt.tight_layout()
plt.show()
```

**Reconstructing a Vector Field**

Consider the vector field $(x+y)\frac{\partial}{\partial x} + (x-y)\frac{\partial}{\partial y} $ in $R^2$, which has a saddle singularity at the origin, $(0, 0)$. The next code snippet generates this vector field:

```python
import numpy as np
import matplotlib.pyplot as plt

# 1) Create a grid over R^2
x = np.linspace(-2, 2, 20)
y = np.linspace(-2, 2, 20)
X, Y = np.meshgrid(x, y)

# 2) Define a vector field, e.g. F(x,y) = (-y, x) (a rotation around the origin)
U = X+Y
V = X-Y

# 3) Plot with small arrows using quiver
plt.figure(figsize=(6,6))
plt.quiver(X, Y, U, V,
           angles='xy', scale_units='xy', scale=10, width=0.005)
plt.title("Vector Field $F(x,y)=(x+y,\,x-y)$ on $\mathbb{R}^2$")
plt.xlabel("$x$")
plt.ylabel("$y$")
plt.xlim(-2,2)
plt.ylim(-2,2)
plt.gca().set_aspect('equal', 'box')
plt.grid(True)
plt.show()
```

We select randomly $N=100$ vectors and we reconstruct the vector field by trainng a feed-forward net, $f_{\theta}(x, y) =\big(f^1_{\theta}(x, y), f^2_{\theta}(x, y)\big) $, on these given vectors. Then we plot the reconstructed vector field in red and the original vector field in blue. The learning is pretty fast.

```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Input

# 1) Build the original vector field on a 20×20 grid
x = np.linspace(-2, 2, 20)
y = np.linspace(-2, 2, 20)
X, Y = np.meshgrid(x, y)
U = X + Y
V = X - Y

# 2) Sample N=100 data points (x_i, y_i, u_i, v_i)
all_data = np.column_stack([X.ravel(), Y.ravel(), U.ravel(), V.ravel()])
np.random.seed(42)
indices = np.random.choice(all_data.shape[0], size=100, replace=False)
sample = all_data[indices]
x_train = sample[:, :2].astype(np.float32)   # shape (100,2)
y_train = sample[:, 2:].astype(np.float32)   # shape (100,2)

# 3) Build the neural network f_theta: R^2 -> R^2
model = Sequential([
    Input(shape=(2,)),
    Dense(32, activation='tanh'),
    Dense(32, activation='tanh'),
    Dense(2, activation='linear')
])
model.compile(optimizer='adam', loss='mse')

# 4) Train on the sampled data
history = model.fit(
    x_train, y_train,
    epochs=500,
    batch_size=16,
    verbose=1
)

# 5) Evaluate the learned field on the same grid
points = np.column_stack([X.ravel(), Y.ravel()]).astype(np.float32)  # (400,2)
pred = model.predict(points)
U_pred = pred[:, 0].reshape(X.shape)
V_pred = pred[:, 1].reshape(Y.shape)

# 6) Plot original (blue) vs learned (red) vector fields
plt.figure(figsize=(6,6))
plt.quiver(X, Y, U,   V,   color='blue', scale=20, width=0.006, label='Original')
plt.quiver(X, Y, U_pred, V_pred, color='red',  scale=20, width=0.003, label='Learned')
plt.title("Original vs Learned Vector Field")
plt.xlabel("x")
plt.ylabel("y")
plt.legend(loc='upper right')
plt.xlim(-2,2)
plt.ylim(-2,2)
plt.gca().set_aspect('equal', 'box')
plt.grid(True)
plt.show()
```

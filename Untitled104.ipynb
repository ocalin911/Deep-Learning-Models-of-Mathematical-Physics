# **Estimating the Length of a Pendulum String**

The equation of motion for a simple pendulum of unit mass is

$$
\frac{d^2 u(t)}{dt^2} + \frac{g}{L} \sin u(t) \;=\; 0,
$$

where $g$ is the gravitational acceleration and $L$ is the length of the pendulum.

The objective is to infer the pendulum length $L$ from the experimental data presented in the given table.
For convenience, we introduce the parameter
$$
\kappa^2 = \frac{g}{L},
$$
so that the problem reduces to estimating $\kappa^2$ using neural networks.


```python
# ============================================================
# Nonlinear pendulum: u''(t) + kappa^2 * sin(u(t)) = 0
# Learn u(t) on [0, 5] using a NN + internal trainable kappa^2
# Loss = ODE residual + data misfit
# Hard-embed u(0)=0 via u(t) = t * n(t)
# Plots:
#   (1) RK data (points + polyline) vs NN prediction
#   (2) Loss vs epochs (total, ODE, data)
#   (3) Learned kappa^2 vs epochs (with true line at 2.0)
# ============================================================

import numpy as np
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import matplotlib.pyplot as plt

# ---------------------------
# Precision & seeds
# ---------------------------
tf.keras.backend.set_floatx("float64")
np.random.seed(0)
tf.random.set_seed(0)

# ---------------------------
# Domain & "true" parameter
# ---------------------------
t_min, t_max = 0.0, 5.0
T_END = t_max
kappa2_true = 2.0  # the table below was generated for u'' + 2 sin(u) = 0

# ---------------------------
# Data from the provided RK table
# ---------------------------
t_data = np.array([0.00000, 0.55556, 1.11111, 1.66667, 2.22222,
                   2.77778, 3.33333, 3.88889, 4.44444, 5.00000], dtype=np.float64)
u_data = np.array([0.00000, 0.50088, 0.72183, 0.55094, 0.07385,
                  -0.49472, -0.72328, -0.55320, -0.08436, 0.38555], dtype=np.float64)

t_tf = tf.constant(t_data[:, None], dtype=tf.float64)  # shape (N,1)
u_tf = tf.constant(u_data[:, None], dtype=tf.float64)  # shape (N,1)
N_data = t_data.size

# ---------------------------
# Helper: small MLP
# ---------------------------
def mlp(in_dim=1, width=64, depth=3, act="tanh", out_dim=1):
    x = inp = keras.Input(shape=(in_dim,))
    for _ in range(depth):
        x = layers.Dense(width, activation=act)(x)
    out = layers.Dense(out_dim, activation=None)(x)
    return keras.Model(inp, out)

# Free network n(t) used inside the hard embedding u(t) = t * n(t)
n_net = mlp(in_dim=1, width=64, depth=3, act="tanh", out_dim=1)

# Trainable scalar for kappa^2 (constrained positive via softplus)
kappa2_raw = tf.Variable(0.1, dtype=tf.float64, trainable=True, name="kappa2_raw")
def kappa2_param():
    # softplus to ensure positivity + tiny epsilon to avoid exact zero
    return tf.nn.softplus(kappa2_raw) + tf.cast(1e-9, tf.float64)

# Hard-embed u(0)=0:
# u(t) = t * n(t)
def u_nn(t):
    return t * n_net(t)

# ---------------------------
# Collocation sampling for ODE residual
# ---------------------------
def sample_collocation(M=256):
    # uniform sample in (0, T_END]
    t = np.random.rand(M, 1) * (T_END - t_min) + t_min
    return tf.constant(t, dtype=tf.float64)

# ODE residual r(t) = u''(t) + kappa^2 * sin(u(t))
def ode_residual(t_batch):
    with tf.GradientTape(persistent=True) as tape2:
        tape2.watch(t_batch)
        u_vals = u_nn(t_batch)                 # (M,1)
        du_dt  = tape2.gradient(u_vals, t_batch)
    d2u_dt2 = tape2.gradient(du_dt, t_batch)
    del tape2
    k2 = kappa2_param()
    return d2u_dt2 + k2 * tf.sin(u_vals)

# Compute losses
def compute_losses(t_colloc):
    # ODE residual loss (mean squared)
    res = ode_residual(t_colloc)
    L_ode = tf.reduce_mean(tf.square(res))

    # Data loss at measurement points
    u_pred_data = u_nn(t_tf)
    L_data = tf.reduce_mean(tf.square(u_pred_data - u_tf))

    return L_ode, L_data, u_pred_data

# ---------------------------
# Training setup
# ---------------------------
opt = keras.optimizers.Adam(5e-4)

w_ode  = 1.0     # physics consistency
w_data = 10.0    # data fit (increase if needed)

EPOCHS = 5000
PRINT_EVERY = 500
M_colloc = 256

hist = {
    "epoch": [],
    "L_ode": [],
    "L_data": [],
    "L_tot": [],
    "kappa2": [],
}

# ---------------------------
# Train
# ---------------------------
for ep in range(1, EPOCHS + 1):
    t_colloc = sample_collocation(M_colloc)

    with tf.GradientTape() as tape:
        L_ode, L_data, u_pred_data = compute_losses(t_colloc)
        L_tot = w_ode * L_ode + w_data * L_data

    vars_all = n_net.trainable_variables + [kappa2_raw]
    grads = tape.gradient(L_tot, vars_all)
    opt.apply_gradients(zip(grads, vars_all))

    # Log
    k2_val = float(kappa2_param().numpy())
    hist["epoch"].append(ep)
    hist["L_ode"].append(float(L_ode.numpy()))
    hist["L_data"].append(float(L_data.numpy()))
    hist["L_tot"].append(float(L_tot.numpy()))
    hist["kappa2"].append(k2_val)

    if ep % PRINT_EVERY == 0:
        print(f"ep {ep:5d} | L_ode {L_ode.numpy():.3e} | L_data {L_data.numpy():.3e} "
              f"| L_tot {L_tot.numpy():.3e} | kappa^2={k2_val:.6f}")

print(f"\nFinal learned kappa^2: {hist['kappa2'][-1]:.6f} (true {kappa2_true:.6f})")

# ---------------------------
# Plots
# ---------------------------
# 1) RK data points + polyline, vs NN prediction
t_plot = np.linspace(t_min, t_max, 600)[:, None]
u_plot = u_nn(tf.constant(t_plot, dtype=tf.float64)).numpy()

plt.figure(figsize=(7,4.5))
plt.scatter(t_data, u_data, c='C0', label='RK data (points)', zorder=3)
plt.plot(t_data, u_data, 'C0', alpha=0.5, label='RK polyline')
plt.plot(t_plot, u_plot, 'C1', lw=2, label=r'$u_\theta(t)$ (NN)')
plt.xlabel('t'); plt.ylabel('u(t)')
plt.title('Nonlinear Pendulum: RK data vs NN solution')
plt.grid(alpha=0.3); plt.legend(); plt.tight_layout(); plt.show()

# 2) Loss vs epochs (log scale)
plt.figure(figsize=(7,4.5))
plt.plot(hist["epoch"], hist["L_tot"],  label='Total', lw=2)
plt.plot(hist["epoch"], hist["L_ode"],  '--', label='ODE')
plt.plot(hist["epoch"], hist["L_data"], '--', label='Data')
plt.yscale('log'); plt.xlabel('Epoch'); plt.ylabel('Loss')
plt.title('Loss vs Epochs')
plt.grid(alpha=0.3); plt.legend(); plt.tight_layout(); plt.show()

# 3) kappa^2 vs epochs
plt.figure(figsize=(7,4.5))
plt.plot(hist["epoch"], hist["kappa2"], lw=2, label=r'Learned $\kappa^2$')
plt.axhline(kappa2_true, color='C2', linestyle='--', label=rf'True $\kappa^2={kappa2_true}$')
plt.xlabel('Epoch'); plt.ylabel(r'$\kappa^2$')
plt.title(r'Learned $\kappa^2$ vs Epochs')
plt.grid(alpha=0.3); plt.legend(); plt.tight_layout(); plt.show()
```

```python
# ============================================================
# Nonlinear pendulum: u''(t) + kappa^2 * sin(u(t)) = 0
# Learn u(t) on [0, 5] using a NN + internal trainable kappa^2
# Loss = ODE residual + data misfit
# Hard-embed u(0)=0 via u(t) = t * n(t)
# Plots:
#   (1) RK4 reference solution (dense) vs NN prediction and data points
#   (2) Loss vs epochs (total, ODE, data)
#   (3) Learned kappa^2 vs epochs (with true line)
# ============================================================

import numpy as np
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import matplotlib.pyplot as plt

# ---------------------------
# Precision & seeds
# ---------------------------
tf.keras.backend.set_floatx("float64")
np.random.seed(0)
tf.random.set_seed(0)

# ---------------------------
# Domain & "true" parameter
# ---------------------------
t_min, t_max = 0.0, 5.0
T_END = t_max
kappa2_true = 2.0  # target ODE: u'' + 2 sin(u) = 0

# ---------------------------
# Data from the provided RK table (10 points)
# ---------------------------
t_data = np.array([0.00000, 0.55556, 1.11111, 1.66667, 2.22222,
                   2.77778, 3.33333, 3.88889, 4.44444, 5.00000], dtype=np.float64)
u_data = np.array([0.00000, 0.50088, 0.72183, 0.55094, 0.07385,
                  -0.49472, -0.72328, -0.55320, -0.08436, 0.38555], dtype=np.float64)

t_tf = tf.constant(t_data[:, None], dtype=tf.float64)  # shape (N,1)
u_tf = tf.constant(u_data[:, None], dtype=tf.float64)  # shape (N,1)
N_data = t_data.size

# ---------------------------
# Reference RK4 solver for u'' + kappa^2 sin(u) = 0
# State y = [u, v], with v = u'
# y' = [v, -kappa^2 sin(u)]
# ---------------------------
def rk4_pendulum(kappa2, t0, t1, h, u0=0.0, v0=1.0):
    n_steps = int(np.ceil((t1 - t0) / h))
    t = np.linspace(t0, t1, n_steps + 1)
    u = np.zeros_like(t)
    v = np.zeros_like(t)
    u[0], v[0] = u0, v0

    def f_u(u, v): return v
    def f_v(u):     return -kappa2 * np.sin(u)

    for i in range(n_steps):
        ui, vi = u[i], v[i]

        k1_u = f_u(ui, vi)
        k1_v = f_v(ui)

        u2 = ui + 0.5 * h * k1_u
        v2 = vi + 0.5 * h * k1_v
        k2_u = f_u(u2, v2)
        k2_v = f_v(u2)

        u3 = ui + 0.5 * h * k2_u
        v3 = vi + 0.5 * h * k2_v
        k3_u = f_u(u3, v3)
        k3_v = f_v(u3)

        u4 = ui + h * k3_u
        v4 = vi + h * k3_v
        k4_u = f_u(u4, v4)
        k4_v = f_v(u4)

        u[i+1] = ui + (h/6.0) * (k1_u + 2*k2_u + 2*k3_u + k4_u)
        v[i+1] = vi + (h/6.0) * (k1_v + 2*k2_v + 2*k3_v + k4_v)

    return t, u

# Dense RK4 reference for plotting
t_rk, u_rk = rk4_pendulum(kappa2_true, t_min, t_max, h=0.002, u0=0.0, v0=1.0)

# ---------------------------
# Helper: small MLP
# ---------------------------
def mlp(in_dim=1, width=64, depth=3, act="tanh", out_dim=1):
    x = inp = keras.Input(shape=(in_dim,))
    for _ in range(depth):
        x = layers.Dense(width, activation=act)(x)
    out = layers.Dense(out_dim, activation=None)(x)
    return keras.Model(inp, out)

# Free network n(t) used inside the hard embedding u(t) = t * n(t)
n_net = mlp(in_dim=1, width=64, depth=3, act="tanh", out_dim=1)

# Trainable scalar for kappa^2 (constrained positive via softplus)
kappa2_raw = tf.Variable(0.1, dtype=tf.float64, trainable=True, name="kappa2_raw")
def kappa2_param():
    # softplus to ensure positivity + tiny epsilon to avoid exact zero
    return tf.nn.softplus(kappa2_raw) + tf.cast(1e-9, tf.float64)

# Hard-embed u(0)=0:
# u(t) = t * n(t)
def u_nn(t):
    return t * n_net(t)

# ---------------------------
# Collocation sampling for ODE residual
# ---------------------------
def sample_collocation(M=256):
    # uniform sample in [0, T_END]
    t = np.random.rand(M, 1) * (T_END - t_min) + t_min
    return tf.constant(t, dtype=tf.float64)

# ODE residual r(t) = u''(t) + kappa^2 * sin(u(t))
def ode_residual(t_batch):
    with tf.GradientTape(persistent=True) as tape2:
        tape2.watch(t_batch)
        u_vals = u_nn(t_batch)                 # (M,1)
        du_dt  = tape2.gradient(u_vals, t_batch)
    d2u_dt2 = tape2.gradient(du_dt, t_batch)
    del tape2
    k2 = kappa2_param()
    return d2u_dt2 + k2 * tf.sin(u_vals)

# Compute losses
def compute_losses(t_colloc):
    # ODE residual loss (mean squared)
    res = ode_residual(t_colloc)
    L_ode = tf.reduce_mean(tf.square(res))

    # Data loss at measurement points
    u_pred_data = u_nn(t_tf)
    L_data = tf.reduce_mean(tf.square(u_pred_data - u_tf))

    return L_ode, L_data, u_pred_data

# ---------------------------
# Training setup
# ---------------------------
opt = keras.optimizers.Adam(1e-3)

w_ode  = 1.0     # physics consistency
w_data = 10.0    # data fit

EPOCHS = 3000
PRINT_EVERY = 500
M_colloc = 256

hist = {
    "epoch": [],
    "L_ode": [],
    "L_data": [],
    "L_tot": [],
    "kappa2": [],
}

# ---------------------------
# Train
# ---------------------------
for ep in range(1, EPOCHS + 1):
    t_colloc = sample_collocation(M_colloc)

    with tf.GradientTape() as tape:
        L_ode, L_data, u_pred_data = compute_losses(t_colloc)
        L_tot = w_ode * L_ode + w_data * L_data

    vars_all = n_net.trainable_variables + [kappa2_raw]
    grads = tape.gradient(L_tot, vars_all)
    opt.apply_gradients(zip(grads, vars_all))

    # Log
    k2_val = float(kappa2_param().numpy())
    hist["epoch"].append(ep)
    hist["L_ode"].append(float(L_ode.numpy()))
    hist["L_data"].append(float(L_data.numpy()))
    hist["L_tot"].append(float(L_tot.numpy()))
    hist["kappa2"].append(k2_val)

    if ep % PRINT_EVERY == 0:
        print(f"ep {ep:5d} | L_ode {L_ode.numpy():.3e} | L_data {L_data.numpy():.3e} "
              f"| L_tot {L_tot.numpy():.3e} | kappa^2={k2_val:.6f}")

print(f"\nFinal learned kappa^2: {hist['kappa2'][-1]:.6f} (true {kappa2_true:.6f})")

# ---------------------------
# Plots
# ---------------------------
# Dense NN prediction for plotting
t_plot = np.linspace(t_min, t_max, 600)[:, None]
u_plot = u_nn(tf.constant(t_plot, dtype=tf.float64)).numpy()

# 1) RK4 reference (dense), NN, and data points
plt.figure(figsize=(7,4.5))
plt.plot(t_rk, u_rk, 'C2', lw=2, label='RK4 reference')
plt.plot(t_plot, u_plot, 'C1', lw=2, label=r'$u_\theta(t)$ (NN)')
plt.scatter(t_data, u_data, c='C0', zorder=3, label='Data (10 pts)')
plt.xlabel('t'); plt.ylabel('u(t)')
plt.title('Nonlinear Pendulum: RK4 vs NN vs Data')
plt.grid(alpha=0.3); plt.legend(); plt.tight_layout(); plt.show()

# 2) Loss vs epochs (log scale)
plt.figure(figsize=(7,4.5))
plt.plot(hist["epoch"], hist["L_tot"],  label='Total', lw=2)
plt.plot(hist["epoch"], hist["L_ode"],  '--', label='ODE')
plt.plot(hist["epoch"], hist["L_data"], '--', label='Data')
plt.yscale('log'); plt.xlabel('Epoch'); plt.ylabel('Loss')
plt.title('Loss vs Epochs')
plt.grid(alpha=0.3); plt.legend(); plt.tight_layout(); plt.show()

# 3) kappa^2 vs epochs
plt.figure(figsize=(7,4.5))
plt.plot(hist["epoch"], hist["kappa2"], lw=2, label=r'Learned $\kappa^2$')
plt.axhline(kappa2_true, color='C2', linestyle='--', label=rf'True $\kappa^2={kappa2_true}$')
plt.xlabel('Epoch'); plt.ylabel(r'$\kappa^2$')
plt.title(r'Learned $\kappa^2$ vs. Epochs')
plt.grid(alpha=0.3); plt.legend(); plt.tight_layout(); plt.show()
```

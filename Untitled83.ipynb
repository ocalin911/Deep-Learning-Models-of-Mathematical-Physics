**Free Falling from a high Altitude**



We define the trial solution
$
x_{\theta}(t)\;=\;x_{0} + v_{0}\,t \;+\;\tfrac12\,t^{2}\,f_{\theta}(t),
$
which by construction satisfies the initial conditions $x(0)=x_{0}$ (above the Earth) and $\dot x(0)=v_{0}$.  We then enforce the ODE residual
$$
x_{\theta}''(t) \;+\; g\,\frac{r_{0}^{2}}{\bigl(r_{0}+x_{\theta}(t)\bigr)^{2}}
\;=\;0
$$
at a set of collocation points $\{t_{i}\}$.  No separate boundary-condition loss is needed, since the ansatz already embeds the initial data.  Training proceeds by minimizing the mean-square residual over these points, yielding a neural-network approximation $x_{\theta^*}(t)$ to the true trajectory.


```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras import Sequential, Input
from tensorflow.keras.layers import Dense
from tensorflow.keras.optimizers import Adam

# Physical parameters
g  = 9.81           # gravitational acceleration
r0 = 6.371e6        # Earth radius (m)

# Initial conditions
x0, v0 = 10000.0, 16.0

# Collocation points
N_phys = 1000
t_phys = np.linspace(0, 10, N_phys).reshape(-1,1).astype(np.float32)

# Build f_theta(t)
model = Sequential([
    Input(shape=(1,)),
    Dense(62, activation='tanh'),
    Dense(32, activation='tanh'),
    Dense(16, activation='tanh'),
    Dense(1, activation='linear'),
])

optimizer = Adam(learning_rate=1e-3)

@tf.function
def pinn_loss():
    t = tf.constant(t_phys)  # (N_phys,1)
    with tf.GradientTape(persistent=True) as tape2:
        tape2.watch(t)
        f = model(t)
        # trial solution and derivatives
        x_t = x0 + v0*t + 0.5*t**2 * f
        x_tprime = tape2.gradient(x_t, t)
    x_tsec = tape2.gradient(x_tprime, t)
    del tape2

    rhs = -g * r0**2 / (r0 + x_t)**2
    return tf.reduce_mean((x_tsec - rhs)**2)

# Training loop
epochs = 4000
loss_history = []

for ep in range(epochs):
    with tf.GradientTape() as tape:
        loss = pinn_loss()
    grads = tape.gradient(loss, model.trainable_variables)
    optimizer.apply_gradients(zip(grads, model.trainable_variables))
    loss_history.append(loss.numpy())
    if ep % 200 == 0:
        print(f"Epoch {ep:4d}, loss = {loss.numpy():.3e}")

# Learned trajectory
t_test = np.linspace(0,10,300).reshape(-1,1).astype(np.float32)
f_test = model.predict(t_test)
x_pred = x0 + v0*t_test + 0.5*t_test**2 * f_test

plt.figure(figsize=(8,5))
plt.plot(t_test, x_pred, 'r-', label='PINN $x_{\\theta}(t)$')
plt.xlabel('t (s)')
plt.ylabel('x(t) (m)')
plt.title('Learned trajectory via PINN')
plt.legend()
plt.grid(True)
plt.show()

# Neural‐network contribution f_theta
plt.figure(figsize=(8,5))
plt.plot(t_test, f_test, 'b-', label='PINN $f_{\\theta}(t)$')
plt.xlabel('t (s)')
plt.ylabel('$f_{\\theta}(t)$')
plt.title('Neural‐network correction term')
plt.legend()
plt.grid(True)
plt.show()

# --- New: Loss history plot ---
plt.figure(figsize=(8,5))
plt.plot(np.arange(epochs), loss_history, 'k-')
plt.yscale('log')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.title('PINN Training Loss vs. Epoch')
plt.grid(True, which='both', ls='--')
plt.show()
```



The following script implements an energy-conserving PINN to approximate the solution of the inverse-square ODE
$$
x''(t) \;=\;-\,g\,\frac{r_{0}^{2}}{\bigl(r_{0}+x(t)\bigr)^{2}},
$$
with initial data $x(0)=x_{0}$, $\dot x(0)=v_{0}$.

This approach enforces both the exact boundary data and the physical conservation law
 $$ \frac{1}{2} f'_{\theta}(t) + U(f_{\theta}(t) ) = \tfrac12\,v_{0}^{2}
\;-\;
g\,\frac{r_{0}^{2}}{r_{0}+x_{0}}=E, $$
 yielding a mesh-free, continuous approximation $x_{\theta^*}(t)$ that automatically satisfies the energy constraint of the inverse-square law. We note that we do not get to the correct solution $x(t)$, which is supposed to be decreasing.


```python
import numpy as np
import tensorflow as tf
from tensorflow.keras import Sequential, Input
from tensorflow.keras.layers import Dense
from tensorflow.keras.optimizers import Adam
import matplotlib.pyplot as plt

# Physical constants
g  = 9.81           # gravitational acceleration (m/s²)
r0 = 6.371e6        # Earth's radius (m)

# Initial conditions
x0, v0 = 10000.0, 0.0
# Total energy constant
E = 0.5 * v0**2 - g * r0**2 / (r0 + x0)

# Collocation grid for enforcing the conservation law
t_colloc = np.linspace(0, 10, 500).reshape(-1, 1).astype(np.float32)

# Build f_theta: t → x_theta(t)
model = Sequential([
    Input(shape=(1,)),
    Dense(64, activation='tanh'),
    Dense(32, activation='tanh'),
    Dense(16, activation='tanh'),
    Dense(1, activation='linear'),
])

optimizer = Adam(learning_rate=1e-3)

@tf.function
def loss_fn():
    t = tf.constant(t_colloc)           # shape (500,1)
    # compute x_theta(t) and its derivative
    with tf.GradientTape() as tape:
        tape.watch(t)
        x_pred = model(t)               # xθ(t)
    x_t = tape.gradient(x_pred, t)      # xθ'(t)

    # 1) Boundary‐condition loss at t = 0
    t0 = tf.constant([[0.0]], dtype=tf.float32)
    x0_pred = model(t0)                 # xθ(0)
    xprime0 = x_t[0]                    # xθ'(0)
    bd_loss = tf.square(x0_pred - x0) + tf.square(xprime0 - v0)

    # 2) Conservation loss at collocation points
    U = - g * r0**2 / (r0 + x_pred)    # U(xθ(t))
    energy_residual = 0.5*tf.square(x_t) + U - E
    cons_loss = tf.reduce_mean(tf.square(energy_residual))

    return bd_loss + cons_loss

# Training loop
epochs = 200
for epoch in range(epochs+1):
    with tf.GradientTape() as tape:
        L = loss_fn()
    grads = tape.gradient(L, model.trainable_variables)
    optimizer.apply_gradients(zip(grads, model.trainable_variables))
    if epoch % 200 == 0:
        print(f"Epoch {epoch:4d}, loss = {L.numpy().item():.3e}")

# Evaluate and plot the learned trajectory
t_test = np.linspace(0, 10, 300).reshape(-1, 1).astype(np.float32)
x_nn   = model.predict(t_test)

plt.figure(figsize=(8,5))
plt.plot(t_test, x_nn, 'r-', label=r'PINN $x_{\theta}(t)$')
plt.xlabel('t (s)')
plt.ylabel('x(t) (m)')
plt.title('Variant (ii): Energy‐Conserving PINN Solution')
plt.legend()
plt.grid(True)
plt.show()
```

However,  the conservation loss is still useful as it can be used as a regularization term. Thus, the total loss in this case becomes
 $$L_{\rm tot} =L_{\rm ODE}(\theta)  + \lambda   L_{\rm cons}(\theta).$$
 The hyperparameter $\lambda$ is chosen small enough such that no oscillatory effects are introduced.


```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras import Sequential, Input
from tensorflow.keras.layers import Dense
from tensorflow.keras.optimizers import Adam

# Physical parameters
g  = 9.81           # gravitational acceleration
r0 = 6.371e6        # Earth radius (m)

# Initial conditions
x0, v0 = 10000.0, 16.0

# Total energy constant
E = 0.5 * v0**2 - g * r0**2 / (r0 + x0)

# Collocation points
N_phys = 1000
t_phys = np.linspace(0, 10, N_phys).reshape(-1,1).astype(np.float32)

# Regularization weight
lambda_cons = 0.005

# Build f_theta(t)
model = Sequential([
    Input(shape=(1,)),
    Dense(62, activation='tanh'),
    Dense(32, activation='tanh'),
    Dense(16, activation='tanh'),
    Dense(1, activation='linear'),
])

optimizer = Adam(learning_rate=1e-3)

@tf.function
def total_loss():
    t = tf.constant(t_phys)  # (N_phys,1)
    with tf.GradientTape(persistent=True) as tape2:
        tape2.watch(t)
        f = model(t)
        # trial solution x_theta and its first derivative
        x_t = x0 + v0*t + 0.5*t**2 * f
        x_tprime = tape2.gradient(x_t, t)
    x_tsec = tape2.gradient(x_tprime, t)
    del tape2

    # ODE residual loss
    rhs = -g * r0**2 / tf.square(r0 + x_t)
    ode_loss = tf.reduce_mean(tf.square(x_tsec - rhs))

    # energy–conservation residual: ½(x')² + U(x) – E
    U = -g * r0**2 / (r0 + x_t)
    cons_res = 0.5 * tf.square(x_tprime) + U - E
    cons_loss = tf.reduce_mean(tf.square(cons_res))

    # total
    return ode_loss + lambda_cons * cons_loss

# Training loop
epochs = 4000
loss_history = []

for ep in range(epochs):
    with tf.GradientTape() as tape:
        L = total_loss()
    grads = tape.gradient(L, model.trainable_variables)
    optimizer.apply_gradients(zip(grads, model.trainable_variables))
    loss_history.append(L.numpy())
    if ep % 200 == 0:
        print(f"Epoch {ep:4d}, loss = {L.numpy():.3e}")

# Plot learned trajectory
t_test = np.linspace(0, 10, 300).reshape(-1,1).astype(np.float32)
f_test = model.predict(t_test)
x_pred = x0 + v0*t_test + 0.5*t_test**2 * f_test

plt.figure(figsize=(8,5))
plt.plot(t_test, x_pred, 'r-', label='PINN $x_{\\theta}(t)$')
plt.xlabel('t (s)')
plt.ylabel('x(t) (m)')
plt.title('Trajectory with ODE + Energy Regularizer')
plt.legend()
plt.grid(True)
plt.show()

# Plot PINN correction f_theta
plt.figure(figsize=(8,5))
plt.plot(t_test, f_test, 'b-', label='PINN $f_{\\theta}(t)$')
plt.xlabel('t (s)')
plt.ylabel('$f_{\\theta}(t)$')
plt.title('Neural‐network correction term')
plt.legend()
plt.grid(True)
plt.show()

# --- Loss history ---
plt.figure(figsize=(8,5))
plt.plot(loss_history, 'k-')
plt.yscale('log')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.title('Training Loss (ODE + Energy Reg.)')
plt.grid(True, which='both', ls='--')
plt.show()
```

Next we enclose a direct method using Runge-Kuta method.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp

# Physical parameters
g  = 9.81           # gravitational acceleration (m/s^2)
r0 = 6.371e6        # Earth radius (m)

# Initial conditions
x0, v0 = 10000.0, 16.0

# Define the second‐order ODE as a first‐order system:
# Let y = [x, x'], then y' = [ x',  -g*r0^2/(r0 + x)^2 ].
def ode_system(t, y):
    x, xdot = y
    xdd = -g * r0**2 / (r0 + x)**2
    return [xdot, xdd]

# Time span and evaluation points
t_span = (0.0, 10.0)
t_eval = np.linspace(*t_span, 500)

# Solve with an explicit Runge–Kutta method (RK45)
sol = solve_ivp(
    ode_system,
    t_span,
    y0=[x0, v0],
    t_eval=t_eval,
    method='RK45',
    atol=1e-8,
    rtol=1e-8,
)

# Extract solution
t_num = sol.t
x_num = sol.y[0]

# Plot
plt.figure(figsize=(8,5))
plt.plot(t_num, x_num, 'b-', linewidth=2, label='RK45 solution')
plt.xlabel('t (s)')
plt.ylabel('x(t) (m)')
plt.title('Numerical Solution of $x\'\'=-g r_0^2/(r_0+x)^2$')
plt.legend()
plt.grid(True)
plt.show()
```

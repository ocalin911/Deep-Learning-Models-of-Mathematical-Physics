# **The Wave Equation in 2D**

Consider the Cauchy problem
\begin{align*}
\partial_t^2 u(t,x) - \Delta u(t,x) &= t\,\|x\|^2, && (t,x)\in (0,\infty)\times R^2,\\
u(0,x) &= 1, && x\in R^2,\\
\partial_t u(0,x) &= 0, && x\in R^2.
\end{align*}


By radial symmetry, with $r=\|x\|$ and $w(t,r)=u(t,x)$, this reduces to
\begin{align*}
\partial_t^2 w(t,r) - \partial_r^2 w(t,r) - \frac{1}{r}\,\partial_r w(t,r) &= t\,r^2, && (t,r)\in (0,\infty)\times[0,\infty),\\
w(0,r) &= 1, && r\ge 0,\\
\partial_t w(0,r) &= 0, && r\ge 0.
\end{align*}

We approximate the solution with the embedded ansatz
$$
w_{\theta}(t,r) \;=\; 1 \;+\; t^2(s_{\theta}(t)+  r^2\,\mathrm{nn}(\theta; t,r),
$$
which enforces the initial conditions and guarantees that $\partial_r w_\theta(t,0)=0$.

```python
# Fixed PINN for radial 2D wave:
#   w_tt - w_rr - (1/r) w_r = t r^2
# IC: w(0,r)=1, w_t(0,r)=0
# Ansatz (fixed): w = 1 + t^2 * ( s(t) + r^2 * n(t,r) )  -> can represent t^5/30 term
# ABC at r=R: (w_r + w_t) ~ 0 (c = 1)

import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, Model
import matplotlib.pyplot as plt

# ----------------------------
# Domain and settings (same ranges as your example figure)
# ----------------------------
T_MAX = 1.0
R_MAX = 2.0
NT = 128           # time collocation
NR = 256           # radial collocation
NB = 128           # boundary-time samples for ABC
EPOCHS = 3000
LR = 1e-3
LAMBDA_ABS = 1.0   # weight for absorbing penalty
R_EPS = 1e-6       # tiny >0 to include the center in sampling

tf.random.set_seed(0)
np.random.seed(0)

# ----------------------------
# Collocation sampling (uniform grid)
# ----------------------------
t_col = np.linspace(0.0, T_MAX, NT, dtype=np.float32)[:, None]
r_col = np.linspace(R_EPS, R_MAX, NR, dtype=np.float32)[None, :]
TT, RR = np.meshgrid(t_col.flatten(), r_col.flatten(), indexing='ij')
t_train = TT.reshape(-1, 1).astype(np.float32)
r_train = RR.reshape(-1, 1).astype(np.float32)

# Boundary times for ABC at r=R
t_b = np.linspace(0.0, T_MAX, NB, dtype=np.float32)[:, None]
r_b = R_MAX * np.ones_like(t_b, dtype=np.float32)

# ----------------------------
# Exact solution (for validation / plots)
# ----------------------------
def w_exact_np(t, r):
    return 1.0 + (t**3)*(r**2)/6.0 + (t**5)/30.0

# ----------------------------
# Networks
# ----------------------------
def build_space_time_net(widths=(64, 64, 64), activation='tanh'):
    """2D MLP for n(t,r)"""
    inp = layers.Input(shape=(2,))
    z = inp
    for w in widths:
        z = layers.Dense(w, activation=activation)(z)
    out = layers.Dense(1, activation=None)(z)
    return Model(inp, out, name="n_theta")

def build_time_net(widths=(32, 32), activation='tanh'):
    """1D MLP for s(t)"""
    inp = layers.Input(shape=(1,))
    z = inp
    for w in widths:
        z = layers.Dense(w, activation=activation)(z)
    out = layers.Dense(1, activation=None)(z)
    return Model(inp, out, name="s_phi")

n_net = build_space_time_net((64, 64, 64), 'tanh')
s_net = build_time_net((32, 32), 'tanh')
optimizer = tf.keras.optimizers.Adam(LR)

# ----------------------------
# Helper: scale inputs to [-1,1] (autodiff handles chain rule)
# ----------------------------
def scale_tr(t, r):
    t_s = 2.0 * t / T_MAX - 1.0
    r_s = 2.0 * r / R_MAX - 1.0
    return t_s, r_s

# Ansatz: w = 1 + t^2 * ( s(t) + r^2 * n(t,r) )
def w_theta(t, r):
    t_s, r_s = scale_tr(t, r)
    s = s_net(t_s)                   # (N,1)
    n = n_net(tf.concat([t_s, r_s], axis=1))  # (N,1)
    return 1.0 + (t**2) * ( s + (r**2) * n )

# PDE residual: w_tt - w_rr - (1/r) w_r - t r^2
def residual(t, r):
    with tf.GradientTape(persistent=True) as t2:
        t2.watch([t, r])
        with tf.GradientTape(persistent=True) as t1:
            t1.watch([t, r])
            w = w_theta(t, r)
        w_t = t1.gradient(w, t)
        w_r = t1.gradient(w, r)
        del t1
    w_tt = t2.gradient(w_t, t)
    w_rr = t2.gradient(w_r, r)
    del t2
    return w_tt - w_rr - (w_r / r) - (t * (r**2))

# Losses
def loss_pde(t, r):
    res = residual(t, r)
    return tf.reduce_mean(tf.square(res))

def loss_abs(t_b, r_b):
    # Sommerfeld (first-order) at r=R with c=1: w_r + w_t ~ 0
    with tf.GradientTape() as tape_r:
        tape_r.watch(r_b)
        w_b = w_theta(t_b, r_b)
    w_r_b = tape_r.gradient(w_b, r_b)
    with tf.GradientTape() as tape_t:
        tape_t.watch(t_b)
        w_b2 = w_theta(t_b, r_b)
    w_t_b = tape_t.gradient(w_b2, t_b)
    return tf.reduce_mean(tf.square(w_r_b + w_t_b))

# Tensors
t_tf  = tf.constant(t_train, dtype=tf.float32)
r_tf  = tf.constant(r_train, dtype=tf.float32)
t_btf = tf.constant(t_b,     dtype=tf.float32)
r_btf = tf.constant(r_b,     dtype=tf.float32)

@tf.function
def train_step():
    with tf.GradientTape() as tape:
        lp = loss_pde(t_tf, r_tf)
        la = loss_abs(t_btf, r_btf)
        L  = lp + LAMBDA_ABS * la
    vars_all = n_net.trainable_variables + s_net.trainable_variables
    grads = tape.gradient(L, vars_all)
    optimizer.apply_gradients(zip(grads, vars_all))
    return L, lp, la

# ----------------------------
# Training loop
# ----------------------------
hist_tot, hist_pde, hist_abs = [], [], []
for ep in range(1, EPOCHS + 1):
    L, Lp, La = train_step()
    hist_tot.append(float(L.numpy()))
    hist_pde.append(float(Lp.numpy()))
    hist_abs.append(float(La.numpy()))
    if ep % 300 == 0:
        print(f"Epoch {ep:4d} | total {hist_tot[-1]:.3e} | pde {hist_pde[-1]:.3e} | abs {hist_abs[-1]:.3e}")

# ----------------------------
# Plots
# ----------------------------
# 1) Loss curves
plt.figure(figsize=(6,4))
plt.plot(hist_tot, label='total')
plt.plot(hist_pde, label='PDE', alpha=0.85)
plt.plot(hist_abs, label='ABS', alpha=0.85)
plt.yscale('log')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.title('Training losses')
plt.legend()
plt.tight_layout()
plt.show()

# 2) Profiles at t = {0.1, 0.5, 1.0}
t_list = [0.1, 0.5, 1.0]
r_plot = np.linspace(0.0, R_MAX, 800, dtype=np.float32)
r_plot_safe = np.maximum(r_plot, R_EPS).reshape(-1,1).astype(np.float32)

plt.figure(figsize=(9,6))
for t0 in t_list:
    t_arr = np.full_like(r_plot_safe, t0, dtype=np.float32)
    # NN
    w_nn = w_theta(tf.constant(t_arr), tf.constant(r_plot_safe)).numpy().flatten()
    # Exact
    w_ex = w_exact_np(t0, r_plot)
    plt.plot(r_plot, w_ex, label=f'Exact t={t0}')
    plt.plot(r_plot, w_nn, '--', label=f'NN t={t0}')

plt.xlabel('r')
plt.ylabel('w(t,r)')
plt.title('Exact vs NN (all times on one graph)')
plt.legend(ncol=2)
plt.tight_layout()
plt.show()
```

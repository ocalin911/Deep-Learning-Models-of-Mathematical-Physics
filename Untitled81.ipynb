# **Eigenfunctions and Eigenvalues**

The next code finds the first eigenvalue $\lambda_1=1$ and the first eigenfunction $f_1(t) = C\sin t$ of the boundary-value problem $$f''(t) = -\lambda f(t), \qquad f(0) = f(\pi) =0$$
with the normalization $\int_0^{\pi} f(t)^2 \, dt=1$.

```python
import numpy as np
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

# Domain boundaries
T_MIN, T_MAX = 0.0, np.pi

# Training parameters
N_RESIDUAL = 100   # number of interior points for ODE residual
N_BC = 2           # boundary points (0 and pi)
N_NORM = 100       # number of points to approximate normalization integral
N_EPOCHS = 1500
LEARNING_RATE = 1e-3

# Generate points
t_res = np.linspace(T_MIN, T_MAX, N_RESIDUAL + 2)[1:-1].reshape(-1, 1)  # interior (0,pi)
t_bc = np.array([[T_MIN], [T_MAX]])  # boundary points
t_norm = np.linspace(T_MIN, T_MAX, N_NORM).reshape(-1, 1)  # for normalization

# Convert to TensorFlow constants
t_res_tf = tf.constant(t_res, dtype=tf.float32)
t_bc_tf = tf.constant(t_bc, dtype=tf.float32)
t_norm_tf = tf.constant(t_norm, dtype=tf.float32)


def build_eigen_net(hidden_units=50, n_hidden_layers=3):
    """
    Builds a Keras model that takes input t and outputs [lambda(t), f(t)].
    We'll enforce lambda(t) is constant and normalize f(t).
    """
    t_input = keras.Input(shape=(1,), name="t")
    x = t_input
    for _ in range(n_hidden_layers):
        x = layers.Dense(hidden_units, activation="tanh")(x)
    f_out = layers.Dense(1, activation=None, name="f_out")(x)
    lambda_out = layers.Dense(1, activation=None, name="lambda_out")(x)
    return keras.Model(inputs=t_input, outputs=[lambda_out, f_out])


def compute_second_derivative(f_model, t):
    """
    Computes f''(t) using nested GradientTapes.
    """
    with tf.GradientTape() as tape2:
        tape2.watch(t)
        with tf.GradientTape() as tape1:
            tape1.watch(t)
            _, f_val = f_model(t)
        f_t = tape1.gradient(f_val, t)
    f_tt = tape2.gradient(f_t, t)
    return f_tt


def first_eigen_loss(model, t_res, t_bc, t_norm, alpha_bc=10.0, alpha_const=1.0, alpha_norm=10.0):
    """
    Loss enforcing:
      1) f'' + lambda*f = 0 on t_res
      2) f(0)=0, f(pi)=0
      3) lambda(t) ≈ constant on t_res
      4) normalization ∫_0^π [f(t)]^2 dt = 1
    """
    # Predict lambda and f at residual points
    lambda_res, f_res = model(t_res)         # (N_RESIDUAL, 1)
    f_tt_res = compute_second_derivative(model, t_res)  # (N_RESIDUAL, 1)

    # 1) ODE residual
    ode_residual = f_tt_res + lambda_res * f_res
    loss_ode = tf.reduce_mean(tf.square(ode_residual))

    # 2) Boundary conditions
    _, f_bc = model(t_bc)  # (2,1)
    loss_bc = tf.reduce_mean(tf.square(f_bc))

    # 3) lambda(t) ≈ constant
    lambda_mean = tf.reduce_mean(lambda_res)
    loss_const = tf.reduce_mean(tf.square(lambda_res - lambda_mean))

    # 4) normalization: approximate ∫_0^π f(t)^2 dt ≈ (π / N_NORM) * sum f_norm^2
    _, f_norm = model(t_norm)  # (N_NORM,1)
    sq_f = tf.square(f_norm)
    integral_approx = (np.pi / N_NORM) * tf.reduce_sum(sq_f)
    loss_norm = tf.square(integral_approx - 1.0)

    # total loss
    loss = loss_ode + alpha_bc * loss_bc + alpha_const * loss_const + alpha_norm * loss_norm
    return loss, lambda_mean


# Training the first eigenpair
model = build_eigen_net()
optimizer = keras.optimizers.Adam(learning_rate=LEARNING_RATE)

for epoch in range(N_EPOCHS):
    with tf.GradientTape() as tape:
        loss_value, lambda_mean = first_eigen_loss(
            model, t_res_tf, t_bc_tf, t_norm_tf
        )
    grads = tape.gradient(loss_value, model.trainable_variables)
    optimizer.apply_gradients(zip(grads, model.trainable_variables))

    if epoch % 300 == 0:
        print(f"Epoch {epoch}, Loss = {loss_value:.6f}, λ ≈ {lambda_mean.numpy():.6f}")

# Extract eigenvalue and eigenfunction
lambda_1 = lambda_mean.numpy()  # approximate constant eigenvalue

# Eigenfunction as a Python callable
def eigenfunction(t):
    t_arr = np.array(t).reshape(-1, 1)
    _, f_arr = model.predict(t_arr)
    return f_arr

# Display results
print(f"\nFirst eigenvalue: λ₁ ≈ {lambda_1:.6f}")

import matplotlib.pyplot as plt

t_plot = np.linspace(0, np.pi, 200).reshape(-1, 1)
f_plot = eigenfunction(t_plot)
# Normalize for plotting
f_plot /= np.max(np.abs(f_plot))

plt.figure(figsize=(6, 4))
plt.plot(t_plot, f_plot, label=f"f₁(t), λ≈{lambda_1:.3f}")
plt.xlabel("t")
plt.ylabel("f₁(t)")
plt.title("First eigenfunction (normalized)")
plt.legend()
plt.grid(True)
plt.show()
```

In the following we shall find the first two eigenfunctions and eigenvalues.

```python
import numpy as np
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import matplotlib.pyplot as plt

# Domain boundaries
T_MIN, T_MAX = 0.0, np.pi

# Training parameters
N_RESIDUAL = 100   # interior points for ODE residual
N_BC = 2           # boundary points (0 and π)
N_NORM = 150       # points to approximate normalization integral
N_EPOCHS = 2000
LEARNING_RATE = 1e-3

# Generate points
t_res = np.linspace(T_MIN, T_MAX, N_RESIDUAL + 2)[1:-1].reshape(-1, 1)  # (0, π)
t_bc = np.array([[T_MIN], [T_MAX]])  # {0, π}
t_norm = np.linspace(T_MIN, T_MAX, N_NORM).reshape(-1, 1)

# Convert to TensorFlow constants
t_res_tf = tf.constant(t_res, dtype=tf.float32)
t_bc_tf = tf.constant(t_bc, dtype=tf.float32)
t_norm_tf = tf.constant(t_norm, dtype=tf.float32)


def build_eigen_net(hidden_units=50, n_hidden_layers=3):
    """
    Builds a Keras model mapping t -> [lambda(t), f(t)].
    """
    t_input = keras.Input(shape=(1,))
    x = t_input
    for _ in range(n_hidden_layers):
        x = layers.Dense(hidden_units, activation="tanh")(x)
    f_out = layers.Dense(1, activation=None, name="f_out")(x)
    lambda_out = layers.Dense(1, activation=None, name="lambda_out")(x)
    return keras.Model(inputs=t_input, outputs=[lambda_out, f_out])


def compute_second_derivative(f_model, t):
    """
    Computes f''(t) via nested GradientTapes.
    """
    with tf.GradientTape() as tape2:
        tape2.watch(t)
        with tf.GradientTape() as tape1:
            tape1.watch(t)
            _, f_val = f_model(t)
        f_t = tape1.gradient(f_val, t)
    f_tt = tape2.gradient(f_t, t)
    return f_tt


def eigen_loss(model, t_res, t_bc, t_norm, prev_funcs=None,
               alpha_bc=10.0, alpha_const=1.0, alpha_norm=10.0, alpha_orth=300.0):
    """
    Loss enforcing:
      1) f'' + lambda*f = 0 on t_res
      2) f(0)=0, f(pi)=0
      3) lambda(t) ≈ constant on t_res
      4) normalization ∫_0^π f^2 dt = 1
      5) orthogonality to prev_funcs (list of callables)
    """
    # Residual: f'' + λ f ≈ 0
    lambda_res, f_res = model(t_res)
    f_tt_res = compute_second_derivative(model, t_res)
    ode_res = f_tt_res + lambda_res * f_res
    loss_ode = tf.reduce_mean(tf.square(ode_res))

    # Boundary conditions: f(0)=0, f(π)=0
    _, f_bc = model(t_bc)  # shape (2, 1)
    loss_bc = tf.reduce_mean(tf.square(f_bc))

    # Enforce λ nearly constant
    lambda_mean = tf.reduce_mean(lambda_res)
    loss_const = tf.reduce_mean(tf.square(lambda_res - lambda_mean))

    # Normalization: ∫ f^2 dt = 1
    _, f_norm = model(t_norm)
    integral_approx = (np.pi / N_NORM) * tf.reduce_sum(tf.square(f_norm))
    loss_norm = tf.square(integral_approx - 1.0)

    # Orthogonality to previous functions
    loss_orth = 0.0
    if prev_funcs:
        for prev in prev_funcs:
            f_prev = prev(t_res_tf)  # shape (N_RESIDUAL, 1)
            inner_prod = tf.reduce_mean(f_res * f_prev) * np.pi
            loss_orth += tf.square(inner_prod)

    loss = (loss_ode
            + alpha_bc * loss_bc
            + alpha_const * loss_const
            + alpha_norm * loss_norm
            + alpha_orth * loss_orth)
    return loss, lambda_mean


def train_eigen_pair(prev_models=None, epochs=N_EPOCHS):
    """
    Trains one eigen-network, orthogonal to prev_models if provided.
    Returns model, λ estimate, loss history list.
    """
    model = build_eigen_net()
    optimizer = keras.optimizers.Adam(learning_rate=LEARNING_RATE)

    loss_hist = []
    for epoch in range(epochs):
        with tf.GradientTape() as tape:
            loss_value, lambda_mean = eigen_loss(
                model,
                t_res_tf,
                t_bc_tf,
                t_norm_tf,
                prev_funcs=[m.predict for m in prev_models] if prev_models else None
            )
        grads = tape.gradient(loss_value, model.trainable_variables)
        optimizer.apply_gradients(zip(grads, model.trainable_variables))
        loss_hist.append(loss_value.numpy())

        if epoch % 100 == 0:
            print(f"Epoch {epoch}, Loss = {loss_value:.6f}, λ ≈ {lambda_mean.numpy():.6f}")

    return model, lambda_mean.numpy(), loss_hist


# === Main: Compute first 3 eigenpairs ===

NUM_EIGS = 2
trained_models = []
eigenvalues = []
eigenfunctions = []
loss_histories = []

for n in range(NUM_EIGS):
    print(f"\nTraining eigenpair #{n+1}...")
    model_n, lam_n, loss_hist_n = train_eigen_pair(
        prev_models=trained_models if n > 0 else None,
        epochs=N_EPOCHS
    )
    trained_models.append(model_n)
    eigenvalues.append(lam_n)
    loss_histories.append(loss_hist_n)

    # Define Python-callable for f_n(t)
    def make_func(m):
        return lambda t: m.predict(np.array(t).reshape(-1, 1))[1]
    eigenfunctions.append(make_func(model_n))

# === Plot loss curves ===

plt.figure(figsize=(8, 6))
for i in range(NUM_EIGS):
    plt.plot(loss_histories[i], label=f"Eigen #{i+1}")
plt.yscale('log')
plt.xlabel("Epoch")
plt.ylabel("Loss (log scale)")
plt.title("Training Loss for First Two Eigenpairs")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()

# === Plot eigenfunctions ===

t_plot = np.linspace(0, np.pi, 300).reshape(-1, 1)
plt.figure(figsize=(8, 6))
for i, f_n in enumerate(eigenfunctions):
    f_vals = f_n(t_plot)
    f_vals /= np.max(np.abs(f_vals))  # normalize max to 1
    plt.plot(t_plot, f_vals, label=f"n={i+1}, λ≈{eigenvalues[i]:.3f}")
plt.xlabel("t")
plt.ylabel("f_n(t)")
plt.title("First Two Eigenfunctions (Normalized)")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()

# === Print eigenvalues ===

print("\nApproximate eigenvalues:")
for i, lam in enumerate(eigenvalues):
    print(f"  λ_{i+1} ≈ {lam:.6f}")
```

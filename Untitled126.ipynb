# **Cauchy's Problem for Heat Operator**

Consider the Cauchy problem
\begin{eqnarray}
  \partial_t u(t, x) - \partial_x^2 u(t, x) &=& \sin t, \\
  u(0, x) &=& e^{-x^2}, \quad x \in R.
\end{eqnarray}
A straightforward computation shows that a classical solution is
$$
u(t, x) = 1 - \cos t \;+\; \frac{1}{\sqrt{1+4t}} \exp\!\left(-\tfrac{x^2}{1+4t}\right).
$$

To approximate this solution with a neural network, we embed the initial condition into the ansatz
$$
u_{\theta}(t, x) = e^{-x^2} + t \,\mathrm{nn}(\theta; t, x),
$$
where $\mathrm{nn}(\theta; t, x)$ denotes the neural network output.

We provide next several variants with a different number oftraining epochs and time instances t for validation against the exact solution.

```python
# PINN for 1D Heat Equation with source: u_t - u_xx = sin(t)
# IC: u(0,x) = exp(-x^2) on R  (we truncate x in [-XMAX, XMAX])

import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, Model
import matplotlib.pyplot as plt

# ----------------------------
# Problem setup
# ----------------------------
XMAX = 4.0          # spatial truncation |x| <= XMAX
TMIN, TMAX = 0.0, 0.7
NT, NX = 128, 128    # collocation grid
BATCH = NT * NX

# Exact solution for comparison
def u_exact(t, x):
    t = np.asarray(t)
    x = np.asarray(x)
    return (1.0 - np.cos(t)) + (1.0 / np.sqrt(1.0 + 4.0*t)) * np.exp(-x**2 / (1.0 + 4.0*t))

# Build collocation points
t_grid = np.linspace(TMIN, TMAX, NT)
x_grid = np.linspace(-XMAX, XMAX, NX)
TT, XX = np.meshgrid(t_grid, x_grid, indexing='ij')
t_train = TT.reshape(-1, 1).astype(np.float32)
x_train = XX.reshape(-1, 1).astype(np.float32)

# ----------------------------
# Neural network (nn)
# ----------------------------
def build_mlp(widths=(64, 64, 64), activation='tanh'):
    inputs = layers.Input(shape=(2,))
    z = inputs
    for w in widths:
        z = layers.Dense(w, activation=activation)(z)
    out = layers.Dense(1, activation=None)(z)
    return Model(inputs, out, name="nn_core")

nn = build_mlp(widths=(64, 64, 64), activation='tanh')

# Ansatz: u_theta(t,x) = exp(-x^2) + t * nn(t,x)
@tf.function
def u_theta(t, x):
    # t, x: (N,1) tensors
    nn_in = tf.concat([t, x], axis=1)  # (N,2)
    nn_out = nn(nn_in)                 # (N,1)
    return tf.exp(-x**2) + t * nn_out  # (N,1)

# PDE residual r = u_t - u_xx - sin(t)
@tf.function
def pde_residual(t, x):
    with tf.GradientTape(persistent=True) as tape2:
        tape2.watch([t, x])
        with tf.GradientTape(persistent=True) as tape1:
            tape1.watch([t, x])
            u = u_theta(t, x)
        u_t = tape1.gradient(u, t)
        u_x = tape1.gradient(u, x)
        del tape1
    u_xx = tape2.gradient(u_x, x)
    del tape2
    return u_t - u_xx - tf.sin(t)

# Loss = mean squared residual over collocation
@tf.function
def loss_fn(t, x):
    r = pde_residual(t, x)
    return tf.reduce_mean(tf.square(r))

optimizer = tf.keras.optimizers.Adam(1e-3)

# ----------------------------
# Training loop
# ----------------------------
EPOCHS = 2000
loss_history = []

t_tf = tf.constant(t_train)  # (B,1)
x_tf = tf.constant(x_train)  # (B,1)

for epoch in range(1, EPOCHS+1):
    with tf.GradientTape() as tape:
        loss = loss_fn(t_tf, x_tf)
    grads = tape.gradient(loss, nn.trainable_variables)
    optimizer.apply_gradients(zip(grads, nn.trainable_variables))
    loss_history.append(float(loss.numpy()))
    if epoch % 100 == 0:
        print(f"Epoch {epoch:4d} | PDE loss: {loss.numpy():.3e}")

# ----------------------------
# Evaluation & plots
# ----------------------------
def u_theta_np(t, x):
    t_tf = tf.convert_to_tensor(t.astype(np.float32).reshape(-1,1))
    x_tf = tf.convert_to_tensor(x.astype(np.float32).reshape(-1,1))
    with tf.no_grad() if hasattr(tf, "no_grad") else tf.device("/CPU:0"):
        u_val = u_theta(t_tf, x_tf).numpy().reshape(t.shape)
    return u_val

ts_to_plot = [0.1, 0.2, 0.5, 0.7]
x_plot = np.linspace(-XMAX, XMAX, 400)

plt.figure(figsize=(6,4))
plt.plot(loss_history)
plt.yscale('log')
plt.xlabel('Epoch')
plt.ylabel('PDE loss')
plt.title('Training Loss (log scale)')
plt.tight_layout()
plt.show()

for t0 in ts_to_plot:
    T_plot = np.full_like(x_plot, t0)
    u_ex = u_exact(T_plot, x_plot)
    u_nn = u_theta_np(T_plot, x_plot)

    plt.figure(figsize=(7,4))
    plt.plot(x_plot, u_ex, label='Exact', linewidth=2)
    plt.plot(x_plot, u_nn, '--', label='NN (PINN)')
    plt.title(f'Profiles at t = {t0}')
    plt.xlabel('x')
    plt.ylabel('u(t,x)')
    plt.legend()
    plt.tight_layout()
    plt.show()
```

```python
# PINN for 1D Heat Equation with source: u_t - u_xx = sin(t)
# IC: u(0,x) = exp(-x^2) on R  (train on truncated x in [-XMAX, XMAX])
# Plots: (1) Loss vs. epochs, (2) All profiles (t=0.1,0.2,0.5,0.7) on one graph

import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, Model
import matplotlib.pyplot as plt

# ----------------------------
# Problem setup
# ----------------------------
XMAX = 4.0
TMIN, TMAX = 0.0, 0.7
NT, NX = 128, 128

def u_exact(t, x):
    t = np.asarray(t)
    x = np.asarray(x)
    return (1.0 - np.cos(t)) + (1.0 / np.sqrt(1.0 + 4.0*t)) * np.exp(-x**2 / (1.0 + 4.0*t))

# Collocation points
t_grid = np.linspace(TMIN, TMAX, NT)
x_grid = np.linspace(-XMAX, XMAX, NX)
TT, XX = np.meshgrid(t_grid, x_grid, indexing='ij')
t_train = TT.reshape(-1, 1).astype(np.float32)
x_train = XX.reshape(-1, 1).astype(np.float32)

# ----------------------------
# Neural network core
# ----------------------------
def build_mlp(widths=(64, 64, 64), activation='tanh'):
    inputs = layers.Input(shape=(2,))
    z = inputs
    for w in widths:
        z = layers.Dense(w, activation=activation)(z)
    out = layers.Dense(1, activation=None)(z)
    return Model(inputs, out, name="nn_core")

nn = build_mlp(widths=(64, 64, 64), activation='tanh')

# Ansatz: u_theta(t,x) = exp(-x^2) + t * nn(t,x)
@tf.function
def u_theta(t, x):
    nn_in = tf.concat([t, x], axis=1)
    nn_out = nn(nn_in)
    return tf.exp(-x**2) + t * nn_out

# PDE residual r = u_t - u_xx - sin(t)
@tf.function
def pde_residual(t, x):
    with tf.GradientTape(persistent=True) as tape2:
        tape2.watch([t, x])
        with tf.GradientTape(persistent=True) as tape1:
            tape1.watch([t, x])
            u = u_theta(t, x)
        u_t = tape1.gradient(u, t)
        u_x = tape1.gradient(u, x)
        del tape1
    u_xx = tape2.gradient(u_x, x)
    del tape2
    return u_t - u_xx - tf.sin(t)

# Loss = mean squared residual
@tf.function
def loss_fn(t, x):
    r = pde_residual(t, x)
    return tf.reduce_mean(tf.square(r))

optimizer = tf.keras.optimizers.Adam(1e-3)

# ----------------------------
# Training
# ----------------------------
EPOCHS = 1000
loss_history = []

t_tf = tf.constant(t_train)
x_tf = tf.constant(x_train)

for epoch in range(1, EPOCHS+1):
    with tf.GradientTape() as tape:
        loss = loss_fn(t_tf, x_tf)
    grads = tape.gradient(loss, nn.trainable_variables)
    optimizer.apply_gradients(zip(grads, nn.trainable_variables))
    loss_history.append(float(loss.numpy()))
    if epoch % 100 == 0:
        print(f"Epoch {epoch:4d} | PDE loss: {loss.numpy():.3e}")

# ----------------------------
# Evaluation helpers
# ----------------------------
def u_theta_np(t_vec, x_vec):
    # t_vec, x_vec are 1D arrays of same length or one scalar & one array (broadcast)
    T, X = np.meshgrid(np.atleast_1d(t_vec), np.atleast_1d(x_vec), indexing='ij')
    t_flat = T.reshape(-1, 1).astype(np.float32)
    x_flat = X.reshape(-1, 1).astype(np.float32)
    u_flat = u_theta(tf.constant(t_flat), tf.constant(x_flat)).numpy().reshape(T.shape)
    return T, X, u_flat

# ----------------------------
# Plots
# ----------------------------
# 1) Loss curve
plt.figure(figsize=(6,4))
plt.plot(loss_history)
plt.yscale('log')
plt.xlabel('Epoch')
plt.ylabel('PDE loss')
plt.title('Training Loss (log scale)')
plt.tight_layout()
plt.show()

# 2) All profiles on one graph
ts_to_plot = [0.1, 0.2, 0.5, 0.7]
x_plot = np.linspace(-XMAX, XMAX, 600)

plt.figure(figsize=(8,5))
for t0 in ts_to_plot:
    T_plot = np.full_like(x_plot, t0)
    u_ex = u_exact(T_plot, x_plot)
    _, _, U_nn = u_theta_np(np.array([t0], dtype=np.float32), x_plot)
    u_nn = U_nn[0]  # shape (len(x_plot),)

    # Plot both for each time
    plt.plot(x_plot, u_ex, label=f'Exact t={t0}', linewidth=2)
    plt.plot(x_plot, u_nn, '--', label=f'NN t={t0}')

plt.xlabel('x')
plt.ylabel('u(t,x)')
plt.title('Exact vs. PINN profiles at t=0.1, 0.2, 0.5, 0.7')
plt.legend(ncol=2)
plt.tight_layout()
plt.show()
```

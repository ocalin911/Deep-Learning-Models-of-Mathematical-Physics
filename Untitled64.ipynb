**The Harmonic Oscillator**

For the harmonic oscillator the potential energy is
$U(q)=\tfrac12\,q^{2}$,
so the Hamiltonian function takes the form
$$
H(q,p)=\tfrac12\,p^{2}+\tfrac12\,q^{2}.
$$
With initial conditions $q(0)=q_0$ and $p(0)=p_0$, the  Hamiltonian equations
\begin{align*}
\dot q(t)&=\;\frac{\partial H}{\partial p}=p(t)\\
\dot p(t)&=-\frac{\partial H}{\partial q}=-\,q(t).
\end{align*}
have an exact solution given by the circular curve
\begin{align*}
q(t) &= q_0 \cos t + p_0 \sin t\\
p(t) &= p_0 \cos t - q_0 \sin t.
\end{align*}

The next code shows that a neural network can learn this exact solution.
It uses a boundary loss and an ODE (physics) loss.


```python
import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, models
import matplotlib.pyplot as plt

# Define the Hamiltonian
@tf.function
def hamiltonian(q, p):
    return 0.5 * p ** 2 + 0.5 * q ** 2

# Physics-informed loss
@tf.function
def physics_informed_loss(t, q, p, q_dot, p_dot):
    dq_dt = p
    dp_dt = -q
    loss_q = tf.reduce_mean((q_dot - dq_dt) ** 2)
    loss_p = tf.reduce_mean((p_dot - dp_dt) ** 2)
    return loss_q + loss_p

# Conservation loss
@tf.function
def conservation_loss(q, p, energy):
    current_energy = hamiltonian(q, p)
    return tf.reduce_mean((current_energy - energy) ** 2)

# Neural network model
class HamiltonianNN(tf.keras.Model):
    def __init__(self):
        super(HamiltonianNN, self).__init__()
        self.shared = models.Sequential([
            layers.Input(shape=(1,)),
            layers.Dense(64, activation='tanh'),
            layers.Dense(64, activation='tanh'),
            layers.Dense(64, activation='tanh')
        ])
        self.q_net = layers.Dense(1)
        self.p_net = layers.Dense(1)

    def call(self, t):
        shared_out = self.shared(t)
        q = self.q_net(shared_out)
        p = self.p_net(shared_out)
        return q, p

# Generate training data
n_samples = 1000
t = tf.convert_to_tensor(np.linspace(0, 6.29, n_samples).reshape(-1, 1), dtype=tf.float32)

# Initial and boundary conditions
q0 = 1.0
p0 = 0.0
energy = hamiltonian(q0, p0)

# Model and optimizer
model = HamiltonianNN()
optimizer = tf.keras.optimizers.Adam(learning_rate=0.001)

# Training loop
epochs = 5002
for epoch in range(epochs):
    with tf.GradientTape() as tape:
        with tf.GradientTape() as tape_q:
            tape_q.watch(t)
            q, p = model(t)
            q_dot = tape_q.gradient(q, t)

        with tf.GradientTape() as tape_p:
            tape_p.watch(t)
            _, p = model(t)
            p_dot = tape_p.gradient(p, t)

        loss_bd = (q[0] - q0) ** 2 + (p[0] - p0) ** 2
        loss_con = conservation_loss(q, p, energy)
        loss_phy = physics_informed_loss(t, q, p, q_dot, p_dot)
        total_loss = loss_bd + 0*loss_con + loss_phy

    grads = tape.gradient(total_loss, model.trainable_variables)
    optimizer.apply_gradients(zip(grads, model.trainable_variables))

    if epoch % 500 == 0:
        print(f"Epoch {epoch}, Loss: {total_loss.numpy()[0]:.4f}")

# Predict and plot results
q, p = model(t)
q = q.numpy().flatten()
p = p.numpy().flatten()

# Plot q(t) and p(t)
plt.figure(figsize=(12, 4))
plt.subplot(1, 3, 1)
plt.plot(t, q, label='q(t)')
plt.title("Position q(t)")
plt.xlabel("Time t")
plt.ylabel("q(t)")
plt.legend()

plt.subplot(1, 3, 2)
plt.plot(t, p, label='p(t)', color='orange')
plt.title("Momentum p(t)")
plt.xlabel("Time t")
plt.ylabel("p(t)")
plt.legend()

# Phase space plot (q(t), p(t))
plt.subplot(1, 3, 3)
plt.plot(q, p, label='(q(t), p(t))')
plt.title("Phase Space")
plt.xlabel("q(t)")
plt.ylabel("p(t)")
plt.legend()

plt.tight_layout()
plt.show()
```

The next code uses both the ODE loss and the conservation of energy loss.

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, models
import matplotlib.pyplot as plt

# ------------------------------------------------------------------ #
#  Hamiltonian and helper losses
# ------------------------------------------------------------------ #
@tf.function
def hamiltonian(q, p):
    """H = ½(p² + q²) for the 1-D harmonic oscillator."""
    return 0.5 * p**2 + 0.5 * q**2

@tf.function
def physics_informed_loss(t, q, p, q_dot, p_dot):
    """Enforce q̇ = p and ṗ = −q."""
    loss_q = tf.reduce_mean((q_dot - p) ** 2)
    loss_p = tf.reduce_mean((p_dot + q) ** 2)
    return loss_q + loss_p

@tf.function
def conservation_loss(q, p, E0):
    """Keep the energy constant."""
    return tf.reduce_mean((hamiltonian(q, p) - E0) ** 2)

# ------------------------------------------------------------------ #
#  Neural-network model
# ------------------------------------------------------------------ #
class HamiltonianNN(tf.keras.Model):
    def __init__(self):
        super().__init__()
        self.shared = models.Sequential([
            layers.Input(shape=(1,)),
            layers.Dense(64, activation="tanh"),
            layers.Dense(64, activation="tanh"),
            layers.Dense(64, activation="tanh"),
        ])
        self.q_head = layers.Dense(1)
        self.p_head = layers.Dense(1)

    def call(self, t):
        z = self.shared(t)
        return self.q_head(z), self.p_head(z)

# ------------------------------------------------------------------ #
#  Training data
# ------------------------------------------------------------------ #
n_pts   = 1000
t_grid  = tf.constant(np.linspace(0, 2*np.pi, n_pts)[:, None], dtype=tf.float32)

q0, p0  = 1.0, 0.0
E0      = hamiltonian(q0, p0)

# ------------------------------------------------------------------ #
#  Model, optimiser, and training loop
# ------------------------------------------------------------------ #
model = HamiltonianNN()
opt   = tf.keras.optimizers.Adam(1e-4)

epochs      = 5001
loss_curve  = []

for epoch in range(epochs):
    with tf.GradientTape() as tape:
        # forward pass and automatic time-derivatives
        with tf.GradientTape() as tape_q:
            tape_q.watch(t_grid)
            q_pred, p_pred = model(t_grid)
        q_dot = tape_q.gradient(q_pred, t_grid)

        with tf.GradientTape() as tape_p:
            tape_p.watch(t_grid)
            _, p_tmp = model(t_grid)
        p_dot = tape_p.gradient(p_tmp, t_grid)

        # loss terms
        loss_bd  = (q_pred[0] - q0)**2 + (p_pred[0] - p0)**2
        loss_con = conservation_loss(q_pred, p_pred, E0)
        loss_phy = physics_informed_loss(t_grid, q_pred, p_pred, q_dot, p_dot)
        total    = loss_bd + 0*loss_con + loss_phy

    grads = tape.gradient(total, model.trainable_variables)
    opt.apply_gradients(zip(grads, model.trainable_variables))

    loss_curve.append(total.numpy().item())   # <-- record loss
    if epoch % 500 == 0:
        print(f"Epoch {epoch:4d} | total loss = {total.numpy().item():.3e}")

# ------------------------------------------------------------------ #
#  Evaluation and plots
# ------------------------------------------------------------------ #
q_pred, p_pred = model(t_grid)
q_pred = q_pred.numpy().flatten()
p_pred = p_pred.numpy().flatten()

fig, axs = plt.subplots(1, 4, figsize=(16, 4))

# q(t)
axs[0].plot(t_grid, q_pred)
axs[0].set_title(r'Position $q(t)$')
axs[0].set_xlabel('time $t$')

# p(t)
axs[1].plot(t_grid, p_pred, color='orange')
axs[1].set_title(r'Momentum $p(t)$')
axs[1].set_xlabel('time $t$')

# Phase space
axs[2].plot(q_pred, p_pred)
axs[2].set_title('Phase space')
axs[2].set_xlabel('$q$')
axs[2].set_ylabel('$p$')

# Loss curve
axs[3].semilogy(loss_curve)
axs[3].set_title('Training loss')
axs[3].set_xlabel('epoch')
axs[3].set_ylabel('loss')

plt.tight_layout()
plt.show()
```

# **Finding Potential for Hyperbolic Oscillator**

We consider a particle of unit mass moving in a one-dimensional potential field $U(q)$.
Given $10$ measurements of the particle’s state $(t_i, q(t_i), q'(t_i))$, as listed in the next table, our objectives are:


$(i)$ To recover the potential field $U(q)$;


$(ii)$ To reconstruct the trajectory of the particle $q(t)$.


$t$   $q(t)$   $q'(t)$

0.000  1.000  -0.750

0.556  1.046  -0.376

1.111  1.209  0.416

1.667  1.534  1.623

2.222  2.085  3.303

2.778  2.965  5.565

3.333  4.330  8.594

3.889  6.404  12.664

4.444  9.514  18.170

5.000  14.141  25.663


```python
# ============================================================
# PINN for Euler-Lagrange system with natural Lagrangian
# Hard embedding of initial conditions q(0), q'(0)
# Example: U(q) = -0.5 q^2, trajectory q(t) = cosh(t) - 0.75 sinh(t)
# ============================================================

import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, models
import matplotlib.pyplot as plt

# ----------------------------
# True solution and data
# ----------------------------
def q_true(t):
    return np.cosh(t) - 0.75*np.sinh(t)

def qp_true(t):
    return np.sinh(t) - 0.75*np.cosh(t)

q0 = q_true(0.0)
qp0 = qp_true(0.0)

t_data = np.linspace(0, 5, 10, dtype=np.float32).reshape(-1,1)
q_data = q_true(t_data).reshape(-1,1).astype(np.float32)
qp_data = qp_true(t_data).reshape(-1,1).astype(np.float32)

tau = np.linspace(0, 5, 50, dtype=np.float32).reshape(-1,1)

# ----------------------------
# Neural networks
# ----------------------------
def build_mlp(inp_dim=1, out_dim=1, hidden=3, width=64, act="swish"):
    inp = tf.keras.Input(shape=(inp_dim,))
    z = inp
    for _ in range(hidden):
        z = layers.Dense(width, activation=act)(z)
    out = layers.Dense(out_dim)(z)
    return models.Model(inp, out)

# small NN for correction term
nn_core = build_mlp(inp_dim=1, out_dim=1, hidden=3, width=64, act="swish")
# potential network
U_net = build_mlp(inp_dim=1, out_dim=1, hidden=3, width=64, act="swish")

# Hard embedding of IC: q(t) = q0 + t*qp0 + t^2 * nn_core(t)
@tf.function
def q_net(t):
    return q0 + qp0*t + t**2 * nn_core(t)

# ----------------------------
# Loss function
# ----------------------------
@tf.function
def compute_loss(t_batch, q_batch, qp_batch, tau_batch):
    # q predictions
    with tf.GradientTape(persistent=True) as t2:
        t2.watch(t_batch)
        with tf.GradientTape(persistent=True) as t1:
            t1.watch(t_batch)
            q_pred = q_net(t_batch)
        qp_pred = t1.gradient(q_pred, t_batch)
    qpp_pred = t2.gradient(qp_pred, t_batch)
    del t1, t2

    # Potential derivatives
    with tf.GradientTape() as tapeU:
        tapeU.watch(q_batch)
        U_pred = U_net(q_batch)
    Uq_pred = tapeU.gradient(U_pred, q_batch)

    # Data loss
    L_data = tf.reduce_mean((q_pred - q_batch)**2) + tf.reduce_mean((qp_pred - qp_batch)**2)

    # EL loss at data points
    L_EL_data = tf.reduce_mean((qpp_pred + Uq_pred)**2)

    # EL loss at collocation points
    with tf.GradientTape(persistent=True) as t2c:
        t2c.watch(tau_batch)
        with tf.GradientTape(persistent=True) as t1c:
            t1c.watch(tau_batch)
            q_tau = q_net(tau_batch)
        qp_tau = t1c.gradient(q_tau, tau_batch)
    qpp_tau = t2c.gradient(qp_tau, tau_batch)
    del t1c, t2c

    with tf.GradientTape() as tapeUc:
        tapeUc.watch(q_tau)
        U_tau = U_net(q_tau)
    Uq_tau = tapeUc.gradient(U_tau, q_tau)

    L_EL_tau = tf.reduce_mean((qpp_tau + Uq_tau)**2)

    # Total loss
    L_total = L_data + L_EL_data + L_EL_tau
    return L_total, L_data, L_EL_data, L_EL_tau

# ----------------------------
# Training
# ----------------------------
optimizer = tf.keras.optimizers.Adam(5e-4)
EPOCHS = 5000

loss_hist = []
for ep in range(1, EPOCHS+1):
    with tf.GradientTape() as tape:
        L_total, L_data, L_EL_data, L_EL_tau = compute_loss(
            tf.convert_to_tensor(t_data),
            tf.convert_to_tensor(q_data),
            tf.convert_to_tensor(qp_data),
            tf.convert_to_tensor(tau)
        )
    vars_all = nn_core.trainable_variables + U_net.trainable_variables
    grads = tape.gradient(L_total, vars_all)
    optimizer.apply_gradients(zip(grads, vars_all))

    loss_hist.append(float(L_total.numpy()))

    if ep % 1000 == 0:
        print(f"Epoch {ep}: L_tot={L_total.numpy():.3e}, "
              f"L_data={L_data.numpy():.3e}, L_EL_data={L_EL_data.numpy():.3e}")


# ----------------------------
# Evaluation
# ----------------------------
t_test = np.linspace(0, 5, 200).reshape(-1,1).astype(np.float32)
t_test_tf = tf.convert_to_tensor(t_test)

# q(t) prediction
q_pred = q_net(t_test_tf).numpy()
q_exact = q_true(t_test)

# Compute q'(t) with autograd
with tf.GradientTape() as tapeq:
    tapeq.watch(t_test_tf)
    q_pred_tf = q_net(t_test_tf)
qp_pred = tapeq.gradient(q_pred_tf, t_test_tf).numpy()
qp_exact = qp_true(t_test)


# Potential vs true
q_vals = np.linspace(-2, 15, 200).reshape(-1,1).astype(np.float32)
U_pred = U_net(q_vals).numpy()
U_true = -0.5*q_vals**2

# ----------------------------
# Plots
# ----------------------------
plt.figure(figsize=(6,4))
plt.plot(t_test, q_exact, 'k--', lw=2, label="True $q(t)$")
plt.plot(t_test, q_pred, 'r-', lw=2, label="NN $q_\\theta(t)$")
plt.scatter(t_data, q_data, c='b', marker='o', label="Data")
plt.xlabel("t"); plt.ylabel("q(t)")
plt.legend(); plt.grid(True)
plt.title("Trajectory $q(t)$: true vs learned")
plt.show()

plt.figure(figsize=(6,4))
plt.plot(q_vals, U_true, 'k--', lw=2, label="True $U(q)=-0.5q^2$")
plt.plot(q_vals, U_pred, 'r-', lw=2, label="NN $U_\\theta(q)$")
plt.xlabel("q"); plt.ylabel("U(q)")
plt.legend(); plt.grid(True)
plt.title("Potential: true vs learned")
plt.show()

plt.figure(figsize=(6,4))
plt.semilogy(loss_hist, 'm-')
plt.xlabel("epoch"); plt.ylabel("Total Loss")
plt.title("Training loss vs epochs")
plt.grid(True)
plt.show()

# Phase-space plot
plt.figure(figsize=(6,4))
plt.plot(q_exact, qp_exact, 'k--', lw=2, label="True trajectory")
plt.plot(q_pred, qp_pred, 'r-', lw=2, label="NN trajectory")
plt.xlabel("q"); plt.ylabel("q'(t)")
plt.legend(); plt.grid(True)
plt.title("Phase-space trajectory")
plt.show()
```

In the next code we use a learning rate schedule to remove oscillations in the loss function.

```python
# ============================================================
# PINN for Euler-Lagrange system with natural Lagrangian
# Hard embedding of initial conditions q(s0), q'(0)
# Example: U(q) = -0.5 q^2, trajectory q(t) = cosh(t) - 0.75 sinh(t)
# Learning rate schedule:
#   0-1000 epochs -> 1e-3
#   1001-3000     -> 5e-4
#   3001-5000     -> 1e-4
#   5001-7000     -> 5e-5
# ============================================================

import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, models
import matplotlib.pyplot as plt

# ----------------------------
# True solution and data
# ----------------------------
def q_true(t):
    return np.cosh(t) - 0.75*np.sinh(t)

def qp_true(t):
    return np.sinh(t) - 0.75*np.cosh(t)

q0 = q_true(0.0)
qp0 = qp_true(0.0)

t_data = np.linspace(0, 5, 10, dtype=np.float32).reshape(-1,1)
q_data = q_true(t_data).reshape(-1,1).astype(np.float32)
qp_data = qp_true(t_data).reshape(-1,1).astype(np.float32)

tau = np.linspace(0, 5, 50, dtype=np.float32).reshape(-1,1)

# ----------------------------
# Neural networks
# ----------------------------
def build_mlp(inp_dim=1, out_dim=1, hidden=3, width=64, act="swish"):
    inp = tf.keras.Input(shape=(inp_dim,))
    z = inp
    for _ in range(hidden):
        z = layers.Dense(width, activation=act)(z)
    out = layers.Dense(out_dim)(z)
    return models.Model(inp, out)

# small NN for correction term
nn_core = build_mlp(inp_dim=1, out_dim=1, hidden=3, width=64, act="swish")
# potential network
U_net = build_mlp(inp_dim=1, out_dim=1, hidden=3, width=64, act="swish")

# Hard embedding of IC: q(t) = q0 + t*qp0 + t^2 * nn_core(t)
@tf.function
def q_net(t):
    return q0 + qp0*t + t**2 * nn_core(t)

# ----------------------------
# Loss function
# ----------------------------
@tf.function
def compute_loss(t_batch, q_batch, qp_batch, tau_batch):
    # q predictions
    with tf.GradientTape(persistent=True) as t2:
        t2.watch(t_batch)
        with tf.GradientTape(persistent=True) as t1:
            t1.watch(t_batch)
            q_pred = q_net(t_batch)
        qp_pred = t1.gradient(q_pred, t_batch)
    qpp_pred = t2.gradient(qp_pred, t_batch)
    del t1, t2

    # Potential derivatives
    with tf.GradientTape() as tapeU:
        tapeU.watch(q_batch)
        U_pred = U_net(q_batch)
    Uq_pred = tapeU.gradient(U_pred, q_batch)

    # Data loss
    L_data = tf.reduce_mean((q_pred - q_batch)**2) + tf.reduce_mean((qp_pred - qp_batch)**2)

    # EL loss at data points
    L_EL_data = tf.reduce_mean((qpp_pred + Uq_pred)**2)

    # EL loss at collocation points
    with tf.GradientTape(persistent=True) as t2c:
        t2c.watch(tau_batch)
        with tf.GradientTape(persistent=True) as t1c:
            t1c.watch(tau_batch)
            q_tau = q_net(tau_batch)
        qp_tau = t1c.gradient(q_tau, tau_batch)
    qpp_tau = t2c.gradient(qp_tau, tau_batch)
    del t1c, t2c

    with tf.GradientTape() as tapeUc:
        tapeUc.watch(q_tau)
        U_tau = U_net(q_tau)
    Uq_tau = tapeUc.gradient(U_tau, q_tau)

    L_EL_tau = tf.reduce_mean((qpp_tau + Uq_tau)**2)

    # Total loss
    L_total = L_data + L_EL_data + L_EL_tau
    return L_total, L_data, L_EL_data, L_EL_tau

# ----------------------------
# Training with LR schedule
# ----------------------------
boundaries = [1000, 3000, 5000]
values = [1e-3, 5e-4, 1e-4, 5e-5]
lr_schedule = tf.keras.optimizers.schedules.PiecewiseConstantDecay(boundaries, values)
optimizer = tf.keras.optimizers.Adam(learning_rate=lr_schedule)

EPOCHS = 7000
loss_hist = []

for ep in range(1, EPOCHS+1):
    with tf.GradientTape() as tape:
        L_total, L_data, L_EL_data, L_EL_tau = compute_loss(
            tf.convert_to_tensor(t_data),
            tf.convert_to_tensor(q_data),
            tf.convert_to_tensor(qp_data),
            tf.convert_to_tensor(tau)
        )
    vars_all = nn_core.trainable_variables + U_net.trainable_variables
    grads = tape.gradient(L_total, vars_all)
    optimizer.apply_gradients(zip(grads, vars_all))

    loss_hist.append(float(L_total.numpy()))

    if ep % 1000 == 0:
        step = int(optimizer.iterations.numpy())
        if callable(optimizer.learning_rate):
            current_lr = float(optimizer.learning_rate(step).numpy())
        else:
            current_lr = float(tf.convert_to_tensor(optimizer.learning_rate).numpy())
        print(f"Epoch {ep}: LR={current_lr:.1e}, "
              f"L_tot={L_total.numpy():.3e}, L_data={L_data.numpy():.3e}, L_EL_data={L_EL_data.numpy():.3e}")

# ----------------------------
# Evaluation
# ----------------------------
t_test = np.linspace(0, 5, 200).reshape(-1,1).astype(np.float32)
t_test_tf = tf.convert_to_tensor(t_test)

# q(t) prediction
q_pred = q_net(t_test_tf).numpy()
q_exact = q_true(t_test)

# Compute q'(t) with autograd
with tf.GradientTape() as tapeq:
    tapeq.watch(t_test_tf)
    q_pred_tf = q_net(t_test_tf)
qp_pred = tapeq.gradient(q_pred_tf, t_test_tf).numpy()
qp_exact = qp_true(t_test)

# Potential vs true
q_vals = np.linspace(-2, 15, 200).reshape(-1,1).astype(np.float32)
U_pred = U_net(q_vals).numpy()
U_true = -0.5*q_vals**2

# ----------------------------
# Plots
# ----------------------------
plt.figure(figsize=(6,4))
plt.plot(t_test, q_exact, 'k--', lw=2, label="True $q(t)$")
plt.plot(t_test, q_pred, 'r-', lw=2, label="NN $q_\\theta(t)$")
plt.scatter(t_data, q_data, c='b', marker='o', label="Data")
plt.xlabel("t"); plt.ylabel("q(t)")
plt.legend(); plt.grid(True)
plt.title("Trajectory $q(t)$: true vs learned")
plt.show()

plt.figure(figsize=(6,4))
plt.plot(q_vals, U_true, 'k--', lw=2, label="True $U(q)=-0.5q^2$")
plt.plot(q_vals, U_pred, 'r-', lw=2, label="NN $U_\\theta(q)$")
plt.xlabel("q"); plt.ylabel("U(q)")
plt.legend(); plt.grid(True)
plt.title("Potential: true vs learned")
plt.show()

plt.figure(figsize=(6,4))
plt.semilogy(loss_hist, 'm-')
plt.xlabel("epoch"); plt.ylabel("Total Loss")
plt.title("Training loss vs epochs")
plt.grid(True)
plt.show()

# Phase-space plot
plt.figure(figsize=(6,4))
plt.plot(q_exact, qp_exact, 'k--', lw=2, label="True trajectory")
plt.plot(q_pred, qp_pred, 'r-', lw=2, label="NN trajectory")
plt.xlabel("q"); plt.ylabel("q'(t)")
plt.legend(); plt.grid(True)
plt.title("Phase-space trajectory")
plt.show()
```

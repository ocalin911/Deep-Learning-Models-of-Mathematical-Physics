# **Neural ODEs**

The following system models diffusive coupling on a network together with a uniform
relaxation toward a reference, encompassing many physical settings (heat flow with sinks,
overdamped coupled elements, resistor networks, etc.):
$$
\dot{x}_i(t)\;=\;-k\,x_i(t)\;+\;\sum_{j=1}^N b_{ij}\,\big(x_i(t)-x_j(t)\big),
\qquad i=1,\dots,N,\ \ t\in[0,T].
$$

To collect data syntheitically, we
set $N=5$, $k=1$, $T=2$, and choose symmetric couplings
$$
b_{ij} \;=\;
\begin{cases}
1, & i\neq j,\\[4pt]
-(N-1), & i=j,
\end{cases}
\qquad i,j=1,\dots,N.
$$
 Partition $[0,T]$ into $n=20$ steps with
$t_j=j\Delta t$, $\Delta t=T/n$. For each $i$ and $t_j$, record the state $x_i(t_j)$ and
its velocity $\dot{x}_i(t_j)$ obtained  via a
high-order numerical solver. The resulting pairs
$\big\{(x_i(t_j),\dot{x}_i(t_j))\big\}_{i=1,\dots,N}^{j=0,\dots,n}$ constitute the
training dataset.

Then we use a neural ODE to recover the trajectories.



```python
# -*- coding: utf-8 -*-
# Neural ODE with Attention: synthetic linear diffusion + leak, training on dX/dt
# TensorFlow / Keras implementation (single-file, ready for Colab / local)

import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

# -----------------------------
# Repro
# -----------------------------
np.random.seed(42)
tf.random.set_seed(42)

# -----------------------------
# Synthetic system definition
# -----------------------------
N = 5          # number of particles
k_leak = 1.0     # leak coefficient k
T = 1.0          # total time
n_samples = 20   # number of sampled times for training
tau_attn = 0.85   # attention temperature (for the model)

# Complete-graph couplings (uniform)
# b_ij = 1 for i != j; b_ii = -(N-1)
B = np.ones((N, N), dtype=float)
np.fill_diagonal(B, -(N - 1.0))  # diagonal

def rhs_true(x):
    """
    True RHS: dx/dt = -k * x + sum_j b_ij (x_i - x_j)
    Vectorized per i: (sum_j b_ij)*x_i - (B @ x)_i
    """
    # Ensure 1-D
    x = np.asarray(x).reshape(-1)
    row_sum = np.sum(B, axis=1)          # (N,)
    term = row_sum * x - B @ x           # (N,)
    return -k_leak * x + term            # (N,)

# High-accuracy reference integrator (RK4)
def rk4_step(f, x, dt):
    k1 = f(x)
    k2 = f(x + 0.5 * dt * k1)
    k3 = f(x + 0.5 * dt * k2)
    k4 = f(x + dt * k3)
    return x + (dt / 6.0) * (k1 + 2*k2 + 2*k3 + k4)

def integrate_rk4(f, x0, t_grid):
    x = x0.copy().reshape(-1)
    traj = [x.copy()]
    for i in range(1, len(t_grid)):
        dt = t_grid[i] - t_grid[i-1]
        # sub-steps for accuracy if coarse grid (optional)
        sub = max(1, int(np.ceil(dt / 0.002)))
        dts = dt / sub
        for _ in range(sub):
            x = rk4_step(f, x, dts)
        traj.append(x.copy())
    return np.stack(traj, axis=0)  # [T, N]

# -----------------------------
# Generate training data
# -----------------------------
# Initial condition (random)
x0 = np.random.randn(N) * 0.5

# Fine time grid for GT rollout
Tfine = 401
t_fine = np.linspace(0.0, T, Tfine)
X_fine = integrate_rk4(rhs_true, x0, t_fine)      # [Tfine, N]

# Sampled times for training (subset)
t_samp = np.linspace(0.0, T, n_samples+1)         # n+1 points (include t=0)
idx_samp = np.searchsorted(t_fine, t_samp)
idx_samp = np.clip(idx_samp, 0, Tfine-1)

X_samp = X_fine[idx_samp]                         # [n+1, N]
# Velocities at sampled points (use true RHS on the sampled states)
dX_samp = np.stack([rhs_true(x) for x in X_samp], axis=0)  # [n+1, N]

# Training tensors
X_train = X_samp[..., None].astype(np.float32)     # [n+1, N, 1]
Y_train = dX_samp[..., None].astype(np.float32)    # [n+1, N, 1]

# -----------------------------
# Attention + vector field model (Keras)
# -----------------------------
# We build a model that takes [N,1] state and outputs [N,1] derivative.
# Attention is computed over the N "tokens" (particles).
c_attn = 8

class ParticleAttention(layers.Layer):
    """
    Particle-wise attention over N particles for scalar states.
    Inputs: X [batch, N, 1]
    Outputs: context [batch, N, 1], attn_weights [batch, N, N]
    """
    def __init__(self, c_attn=8, tau=0.85, **kwargs):
        super().__init__(**kwargs)
        self.c = c_attn
        self.tau = tau
        self.WQ = layers.Dense(c_attn, use_bias=False)
        self.WK = layers.Dense(c_attn, use_bias=False)
        self.WV = layers.Dense(1,      use_bias=False)  # values to 1-dim context

    def call(self, X):
        # X: [B, N, 1]
        Q = self.WQ(X)            # [B, N, c]
        K = self.WK(X)            # [B, N, c]
        V = self.WV(X)            # [B, N, 1]

        # scores: [B, N, N] = Q @ K^T / (sqrt(c)*tau)
        scores = tf.einsum('bnd,bmd->bnm', Q, K) / (tf.sqrt(tf.cast(self.c, tf.float32)) * self.tau)
        A = tf.nn.softmax(scores, axis=-1)  # along j (last dim)
        ctx = tf.einsum('bnm,bmj->bnj', A, V)  # [B, N, 1]
        return ctx, A

class VectorField(layers.Layer):
    """
    Shared particle-wise vector field f_theta(x_i, ctx_i) -> dx_i/dt
    Implemented pointwise with a small MLP.
    """
    def __init__(self, hidden=128, **kwargs):
        super().__init__(**kwargs)
        self.mlp = keras.Sequential([
            layers.Dense(hidden, activation='tanh'),
            layers.Dense(hidden, activation='tanh'),
            layers.Dense(1)  # derivative scalar per particle
        ])

    def call(self, X, ctx):
        # X, ctx: [B, N, 1]
        Z = tf.concat([X, ctx], axis=-1)  # [B, N, 2]
        return self.mlp(Z)                # [B, N, 1]

# Build the Keras model for derivative prediction
inp = keras.Input(shape=(N, 1), name='x_state')       # [B, N, 1]
ctx, A = ParticleAttention(c_attn=c_attn, tau=tau_attn, name='attn')(inp)
dX_pred = VectorField(hidden=128, name='vf')(inp, ctx)

model = keras.Model(inputs=inp, outputs=dX_pred, name='NeuralODE_Attn_RHS')
model.compile(optimizer=keras.optimizers.Adam(1e-3), loss='mse')

# -----------------------------
# Train on derivative matching
# -----------------------------
history = model.fit(X_train, Y_train,
                    epochs=3600,
                    batch_size=8,
                    verbose=0)
print("Final training MSE:", float(history.history['loss'][-1]))

# -----------------------------
# Roll out the learned Neural ODE (RK4) and compare
# -----------------------------
@tf.function
def rhs_model_tf(x_batch):
    # x_batch: [B, N] -> return [B, N]
    xb = tf.expand_dims(x_batch, -1)        # [B, N, 1]
    dxb = model(xb, training=False)         # [B, N, 1]
    return tf.squeeze(dxb, -1)              # [B, N]

def rhs_model_np(x):
    # x: [N] -> [N]
    xb = tf.convert_to_tensor(x[None, :], dtype=tf.float32)  # [1, N]
    dx = rhs_model_tf(xb)                                     # [1, N]
    return dx.numpy()[0]

# Learned rollout from same x0 on same fine grid
X_learn = integrate_rk4(rhs_model_np, x0, t_fine)  # [Tfine, N]

# -----------------------------
# Plot: Trajectories & data
# -----------------------------
plt.figure(figsize=(10, 6))
sel = [0, 1, 2]  # plot a few particles
for i in sel:
    plt.plot(t_fine, X_fine[:, i], label=f'True x[{i}]', lw=2)
    plt.plot(t_fine, X_learn[:, i], '--', label=f'Learned x[{i}]', lw=2)
    # overlay sampled data points
    plt.plot(t_samp, X_samp[:, i], 'o', ms=5, alpha=0.6)

plt.xlabel('t')
plt.ylabel('state x_i(t)')
plt.title('Trajectories: true vs learned (Neural ODE + attention)\nMarkers: sampled data points')
plt.legend(ncol=2, fontsize=9)
plt.tight_layout()
plt.show()

# Optional: plot all particles as a carpet
plt.figure(figsize=(10, 4))
plt.imshow(X_fine.T, aspect='auto', extent=[t_fine[0], t_fine[-1], 0, N], origin='lower', cmap='viridis')
plt.colorbar(label='x_i(t)')
plt.title('True trajectories (all particles)')
plt.xlabel('t'); plt.ylabel('particle index')
plt.tight_layout()
plt.show()

plt.figure(figsize=(10, 4))
plt.imshow(X_learn.T, aspect='auto', extent=[t_fine[0], t_fine[-1], 0, N], origin='lower', cmap='viridis')
plt.colorbar(label='x_i(t)')
plt.title('Learned trajectories (all particles)')
plt.xlabel('t'); plt.ylabel('particle index')
plt.tight_layout()
plt.show()
```

```python
# -*- coding: utf-8 -*-
# Neural ODE with Attention: synthetic linear diffusion + leak, training on dX/dt
# Now includes a loss-vs-epochs plot.

import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

# -----------------------------
# Repro
# -----------------------------
np.random.seed(42)
tf.random.set_seed(42)

# -----------------------------
# Synthetic system definition
# -----------------------------
N = 5          # number of particles
k_leak = 1.0     # leak coefficient k
T = 2.0          # total time
n_samples = 20   # number of sampled times for training
tau_attn = 0.5   # attention temperature (for the model)

# Complete-graph couplings (uniform)
# b_ij = 1 for i != j; b_ii = -(N-1)
B = np.ones((N, N), dtype=float)
np.fill_diagonal(B, -(N - 1.0))  # diagonal

def rhs_true(x):
    """
    True RHS: dx/dt = -k * x + sum_j b_ij (x_i - x_j)
    Vectorized per i: (sum_j b_ij)*x_i - (B @ x)_i
    """
    x = np.asarray(x).reshape(-1)
    row_sum = np.sum(B, axis=1)          # (N,)
    term = row_sum * x - B @ x           # (N,)
    return -k_leak * x + term            # (N,)

# High-accuracy reference integrator (RK4)
def rk4_step(f, x, dt):
    k1 = f(x)
    k2 = f(x + 0.5 * dt * k1)
    k3 = f(x + 0.5 * dt * k2)
    k4 = f(x + dt * k3)
    return x + (dt / 6.0) * (k1 + 2*k2 + 2*k3 + k4)

def integrate_rk4(f, x0, t_grid):
    x = x0.copy().reshape(-1)
    traj = [x.copy()]
    for i in range(1, len(t_grid)):
        dt = t_grid[i] - t_grid[i-1]
        sub = max(1, int(np.ceil(dt / 0.002)))  # sub-steps for accuracy
        dts = dt / sub
        for _ in range(sub):
            x = rk4_step(f, x, dts)
        traj.append(x.copy())
    return np.stack(traj, axis=0)  # [T, N]

# -----------------------------
# Generate training data
# -----------------------------
x0 = np.random.randn(N) * 0.5

Tfine = 401
t_fine = np.linspace(0.0, T, Tfine)
X_fine = integrate_rk4(rhs_true, x0, t_fine)      # [Tfine, N]

t_samp = np.linspace(0.0, T, n_samples+1)         # n+1 points (include t=0)
idx_samp = np.searchsorted(t_fine, t_samp)
idx_samp = np.clip(idx_samp, 0, Tfine-1)

X_samp = X_fine[idx_samp]                         # [n+1, N]
dX_samp = np.stack([rhs_true(x) for x in X_samp], axis=0)  # [n+1, N]

X_train = X_samp[..., None].astype(np.float32)     # [n+1, N, 1]
Y_train = dX_samp[..., None].astype(np.float32)    # [n+1, N, 1]

# -----------------------------
# Attention + vector field model (Keras)
# -----------------------------
c_attn = 8

class ParticleAttention(layers.Layer):
    def __init__(self, c_attn=8, tau=0.5, **kwargs):
        super().__init__(**kwargs)
        self.c = c_attn
        self.tau = tau
        self.WQ = layers.Dense(c_attn, use_bias=False)
        self.WK = layers.Dense(c_attn, use_bias=False)
        self.WV = layers.Dense(1,      use_bias=False)

    def call(self, X):
        Q = self.WQ(X)            # [B, N, c]
        K = self.WK(X)            # [B, N, c]
        V = self.WV(X)            # [B, N, 1]
        scores = tf.einsum('bnd,bmd->bnm', Q, K) / (tf.sqrt(tf.cast(self.c, tf.float32)) * self.tau)
        A = tf.nn.softmax(scores, axis=-1)        # [B, N, N]
        ctx = tf.einsum('bnm,bmj->bnj', A, V)     # [B, N, 1]
        return ctx, A

class VectorField(layers.Layer):
    def __init__(self, hidden=128, **kwargs):
        super().__init__(**kwargs)
        self.mlp = keras.Sequential([
            layers.Dense(hidden, activation='tanh'),
            layers.Dense(hidden, activation='tanh'),
            layers.Dense(1)
        ])
    def call(self, X, ctx):
        Z = tf.concat([X, ctx], axis=-1)          # [B, N, 2]
        return self.mlp(Z)                        # [B, N, 1]

inp = keras.Input(shape=(N, 1), name='x_state')       # [B, N, 1]
ctx, A = ParticleAttention(c_attn=c_attn, tau=tau_attn, name='attn')(inp)
dX_pred = VectorField(hidden=128, name='vf')(inp, ctx)

model = keras.Model(inputs=inp, outputs=dX_pred, name='NeuralODE_Attn_RHS')
model.compile(optimizer=keras.optimizers.Adam(1e-3), loss='mse')

# -----------------------------
# Train on derivative matching
# -----------------------------
history = model.fit(X_train, Y_train, epochs=10000, batch_size=8, verbose=0)
print("Final training MSE:", float(history.history['loss'][-1]))

# NEW: plot loss vs epochs
plt.figure(figsize=(7.0, 3.2))
plt.plot(history.history['loss'], lw=2)
plt.xlabel('Epoch')
plt.ylabel('Training loss (MSE)')
plt.title('Derivative-matching loss vs epochs')
plt.tight_layout()
plt.show()

# -----------------------------
# Roll out the learned Neural ODE (RK4) and compare
# -----------------------------
@tf.function
def rhs_model_tf(x_batch):
    xb = tf.expand_dims(x_batch, -1)        # [B, N, 1]
    dxb = model(xb, training=False)         # [B, N, 1]
    return tf.squeeze(dxb, -1)              # [B, N]

def rhs_model_np(x):
    xb = tf.convert_to_tensor(x[None, :], dtype=tf.float32)  # [1, N]
    dx = rhs_model_tf(xb)                                     # [1, N]
    return dx.numpy()[0]

X_learn = integrate_rk4(rhs_model_np, x0, t_fine)  # [Tfine, N]

# -----------------------------
# Plot: Trajectories & data
# -----------------------------
plt.figure(figsize=(10, 6))
sel = [0, 1, 2]  # plot a few particles
for i in sel:
    plt.plot(t_fine, X_fine[:, i], label=f'True x[{i}]', lw=2)
    plt.plot(t_fine, X_learn[:, i], '--', label=f'Learned x[{i}]', lw=2)
    plt.plot(t_samp, X_samp[:, i], 'o', ms=5, alpha=0.6)  # sampled data

plt.xlabel('t')
plt.ylabel('state x_i(t)')
plt.title('Trajectories: true vs learned (Neural ODE + attention)\nMarkers: sampled data points')
plt.legend(ncol=2, fontsize=9)
plt.tight_layout()
plt.show()

# Optional: carpet plots
plt.figure(figsize=(10, 4))
plt.imshow(X_fine.T, aspect='auto', extent=[t_fine[0], t_fine[-1], 0, N],
           origin='lower', cmap='viridis')
plt.colorbar(label='x_i(t)')
plt.title('True trajectories (all particles)')
plt.xlabel('t'); plt.ylabel('particle index')
plt.tight_layout()
plt.show()

plt.figure(figsize=(10, 4))
plt.imshow(X_learn.T, aspect='auto', extent=[t_fine[0], t_fine[-1], 0, N],
           origin='lower', cmap='viridis')
plt.colorbar(label='x_i(t)')
plt.title('Learned trajectories (all particles)')
plt.xlabel('t'); plt.ylabel('particle index')
plt.tight_layout()
plt.show()
```

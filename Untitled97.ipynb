**Pendulum via Generating Function**


 In the next code we consider the case of the simple pendulum with the Hamiltonian function
$$H(q,p) = \frac{1}{2} p^2 - \cos q.$$
In this case, the Hamiltonian system of equations cannot be solved explicitly using elementary functions, so we employ the method of generating functions to approximate the solution. Specifically, we learn the solution of the pendulum system via a generating function  $G(q_0, p_0, t)$ , which is approximated a priori by a neural network.

The network outputs the generating function
$G$, and then uses automatic differentiation to extract the time-dependent solution components
$q(t)$ and $p(t)$.

We plot the pendulum solution $q(t)$ and compare with the numerical solution given by Runge--Kutta method.

```python
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp

# Reproducibility
np.random.seed(42)
tf.random.set_seed(42)

# Initial conditions
q0_val = 1.0
p0_val = 0.0

# Time domain
t_min, t_max = 0.0, 1.0
N_t = 200
t_vals = np.linspace(t_min, t_max, N_t).reshape(-1, 1)
t_vals_tf = tf.convert_to_tensor(t_vals, dtype=tf.float32)

# Define the neural network G_theta(q0, p0, t)
def create_model():
    return tf.keras.Sequential([
        tf.keras.layers.InputLayer(input_shape=(3,)),  # (q0, p0, t)
        tf.keras.layers.Dense(64, activation='tanh'),
        tf.keras.layers.Dense(64, activation='tanh'),
        tf.keras.layers.Dense(64, activation='tanh'),
        tf.keras.layers.Dense(1)  # scalar output G
    ])

model = create_model()
optimizer = tf.keras.optimizers.Adam(learning_rate=0.001)

# Hamiltonian and derivatives
def H(q, p):
    return 0.5 * p**2 - tf.cos(q)

def dH_dq(q, p):
    return tf.sin(q)

def dH_dp(q, p):
    return p

# Loss function
def compute_loss():
    with tf.GradientTape(persistent=True) as tape2:
        tape2.watch(t_vals_tf)
        q0 = tf.ones_like(t_vals_tf) * q0_val
        p0 = tf.ones_like(t_vals_tf) * p0_val
        inputs = tf.concat([q0, p0, t_vals_tf], axis=1)
        G = model(inputs)

        with tf.GradientTape(persistent=True) as tape1:
            tape1.watch([q0, p0, t_vals_tf])
            inputs = tf.concat([q0, p0, t_vals_tf], axis=1)
            G = model(inputs)

        dq = tape1.gradient(G, p0)
        dp = tape1.gradient(G, q0)
        dq_dt = tape2.gradient(dq, t_vals_tf)
        dp_dt = tape2.gradient(dp, t_vals_tf)
        del tape1, tape2

    res_q = dq_dt - dH_dp(dq, dp)
    res_p = dp_dt + dH_dq(dq, dp)
    loss_ode = tf.reduce_mean(tf.square(res_q)) + tf.reduce_mean(tf.square(res_p))

    # Initial conditions
    q0_tensor = tf.constant([[q0_val]], dtype=tf.float32)
    p0_tensor = tf.constant([[p0_val]], dtype=tf.float32)
    t0 = tf.constant([[0.0]], dtype=tf.float32)
    with tf.GradientTape(persistent=True) as tape_ic:
        tape_ic.watch([q0_tensor, p0_tensor])
        input0 = tf.concat([q0_tensor, p0_tensor, t0], axis=1)
        G0 = model(input0)
    q_ic = tape_ic.gradient(G0, p0_tensor)
    p_ic = tape_ic.gradient(G0, q0_tensor)
    del tape_ic

    loss_ic = tf.square(q_ic - q0_val) + tf.square(p_ic - p0_val)
    total_loss = loss_ode + loss_ic
    return total_loss, loss_ode, loss_ic, dq

# Training loop
epochs = 1000
loss_history = []

for epoch in range(epochs):
    with tf.GradientTape() as tape:
        total_loss, loss_ode, loss_ic, _ = compute_loss()
    grads = tape.gradient(total_loss, model.trainable_variables)
    optimizer.apply_gradients(zip(grads, model.trainable_variables))

    loss_history.append(total_loss.numpy().item())
    if epoch % 100 == 0:
        print(f"Epoch {epoch}: Total Loss = {total_loss.numpy().item():.6f}")

# Predict learned trajectory
q0_eval = tf.constant([[q0_val]] * N_t, dtype=tf.float32)
p0_eval = tf.constant([[p0_val]] * N_t, dtype=tf.float32)
with tf.GradientTape(persistent=True) as tape_eval:
    tape_eval.watch([q0_eval, p0_eval])
    t_vals_tf = tf.convert_to_tensor(t_vals, dtype=tf.float32)
    input_eval = tf.concat([q0_eval, p0_eval, t_vals_tf], axis=1)
    G_eval = model(input_eval)

q_pred = tape_eval.gradient(G_eval, p0_eval).numpy().flatten()
del tape_eval

# Numerical (Runge–Kutta) reference solution
def pendulum_ode(t, y):
    q, p = y
    dqdt = p
    dpdt = -np.sin(q)
    return [dqdt, dpdt]

sol = solve_ivp(pendulum_ode, [t_min, t_max], [q0_val, p0_val], t_eval=t_vals.flatten(), method='RK45')
q_rk = sol.y[0]

# Plot NN solution vs Runge–Kutta
plt.figure(figsize=(10, 5))
plt.plot(t_vals, q_pred, label='NN solution $q_\\theta(t)$', lw=2)
plt.plot(t_vals, q_rk, 'r--', label='Runge--Kutta $q(t)$', lw=2)
plt.xlabel("Time $t$")
plt.ylabel("$q(t)$")
plt.title("Pendulum Equation: Neural Network vs Runge--Kutta")
plt.legend()
plt.grid(True)
plt.show()

# Plot loss
plt.figure(figsize=(10, 5))
plt.plot(loss_history)
plt.yscale('log')
plt.xlabel("Epoch")
plt.ylabel("Total Loss")
plt.title("Training Loss")
plt.grid(True)
plt.show()
```

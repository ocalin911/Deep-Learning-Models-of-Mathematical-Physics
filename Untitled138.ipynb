# **Generating-Function Network (GFNN)**

 We shall use the same example as before -- the harmonic oscillator. Under the same data collection as in the previous section, and the same architecture where we used swish function instead of softmax and performed training for only 1,000 epochs with Adam with lr=0.01 and batch size of 81.

```python
# GFNN (Generating-Function Network) vs Baseline for Harmonic Oscillator
# ---------------------------------------------------------------------
# Requirements: tensorflow>=2.9, numpy, matplotlib (plots optional)
import math
import numpy as np
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

tf.random.set_seed(42)
np.random.seed(42)
rng = np.random.default_rng(42)

# ---------------------------
# 1) Exact dynamics & dataset
# ---------------------------
def exact_traj(t, q0, p0):
    q = q0*np.cos(t) + p0*np.sin(t)
    p = p0*np.cos(t) - q0*np.sin(t)
    return q, p

def make_dataset(n_ic=25, times_per_ic=30, T=10.0, sigma=0.0, E_range=(0.2,1.0)):
    """Return:
       X = [t, q, E], y = p, Anchors = [t=0, q0, E] for per-trajectory anchoring."""
    q0s, p0s, Es = [], [], []
    while len(q0s) < n_ic:
        q0 = rng.uniform(-1.5, 1.5)
        p0 = rng.uniform(-1.5, 1.5)
        E  = 0.5*(q0*q0 + p0*p0)
        if E_range[0] <= E <= E_range[1]:
            q0s.append(q0); p0s.append(p0); Es.append(E)

    t_all, q_all, p_all, E_all, q0_all = [], [], [], [], []
    for q0, p0, E in zip(q0s, p0s, Es):
        t = np.linspace(0.0, T, times_per_ic, dtype=np.float32)
        q, p = exact_traj(t, q0, p0)
        if sigma > 0:
            q = q + rng.normal(0.0, sigma, size=q.shape)
            p = p + rng.normal(0.0, sigma, size=p.shape)
        t_all.append(t.astype(np.float32))
        q_all.append(q.astype(np.float32))
        p_all.append(p.astype(np.float32))
        E_all.append(np.full_like(t, E, dtype=np.float32))
        q0_all.append(np.full_like(t, q0, dtype=np.float32))

    t_all = np.concatenate(t_all)
    q_all = np.concatenate(q_all)
    p_all = np.concatenate(p_all)
    E_all = np.concatenate(E_all)
    q0_all = np.concatenate(q0_all)

    X = np.stack([t_all, q_all, E_all], axis=1).astype(np.float32)  # inputs: (t, q, E)
    y = p_all[:, None].astype(np.float32)                            # target: p
    anchors = np.stack([np.zeros_like(E_all, dtype=np.float32), q0_all, E_all], axis=1).astype(np.float32)  # (0,q0,E)
    return X, y, anchors

# Hyperparams
N_IC, N_T, T_H = 25, 30, 10.0
SIGMA = 0.0  # set to 0.1 for noisy data if desired

X_all, y_all, anchors_all = make_dataset(n_ic=N_IC, times_per_ic=N_T, T=T_H, sigma=SIGMA)

# Train/val split
N = X_all.shape[0]
perm = rng.permutation(N)
split = int(0.8*N)
tr_idx, va_idx = perm[:split], perm[split:]

Xtr, ytr, Anch_tr = X_all[tr_idx], y_all[tr_idx], anchors_all[tr_idx]
Xva, yva = X_all[va_idx], y_all[va_idx]

BATCH = 81
EPOCHS = 1000
LR = 1e-2

train_ds_gfnn = tf.data.Dataset.from_tensor_slices((Xtr, ytr, Anch_tr)).shuffle(4096).batch(BATCH).prefetch(2)
train_ds_base = tf.data.Dataset.from_tensor_slices((Xtr, ytr)).shuffle(4096).batch(BATCH).prefetch(2)
val_ds_base   = tf.data.Dataset.from_tensor_slices((Xva, yva)).batch(BATCH).prefetch(2)

# ---------------------------
# 2) Baseline: (t,q,E) -> p
# ---------------------------
def build_baseline(hidden=164, depth=2, act="swish"):
    inp = keras.Input(shape=(3,), name="t_q_E")
    x = inp
    for _ in range(depth):
        x = layers.Dense(hidden, activation=act)(x)
    out = layers.Dense(1, activation=None, name="p_pred")(x)
    return keras.Model(inp, out, name="Baseline")

baseline = build_baseline()
baseline.compile(optimizer=keras.optimizers.Adam(LR), loss="mse")
baseline.fit(train_ds_base, validation_data=val_ds_base, epochs=EPOCHS, verbose=0)

# ------------------------------------------
# 3) GFNN: learn G(t,q,E); p = ∂G/∂q; HJ resid
# ------------------------------------------
class GFNN(keras.Model):
    def __init__(self, hidden=164, depth=2, act="swish"):
        super().__init__()
        self.layers_ = [layers.Dense(hidden, activation=act) for _ in range(depth)]
        self.out = layers.Dense(1, activation=None)  # scalar G(t,q,E)
    def call(self, x):
        z = x
        for L in self.layers_:
            z = L(z)
        return self.out(z)

gfnn = GFNN()
_ = gfnn(tf.zeros((1,3)))  # build
opt = keras.optimizers.Adam(LR)

# Loss weights
w_hj, w_mom, w_anchor = 1.0, 1.0, 1e-3

@tf.function
def train_step_gfnn(xb, yb, anchors_b):
    # xb: (B,3) = (t,q,E); yb: (B,1) = p target; anchors_b: (B,3) = (0,q0,E)
    with tf.GradientTape() as tape:
        t = xb[:, :1]; q = xb[:, 1:2]; E = xb[:, 2:3]

        # G(t,q,E) and its partials wrt t and q
        with tf.GradientTape(persistent=True) as g:
            g.watch([t, q])
            G = gfnn(tf.concat([t, q, E], axis=1))
        dG_dt = g.gradient(G, t)
        dG_dq = g.gradient(G, q)
        del g

        p_pred = dG_dq
        # Harmonic-oscillator HJ residual: ∂t G + 1/2 p^2 + 1/2 q^2 = 0
        HJ_res = dG_dt + 0.5*(p_pred**2) + 0.5*(q**2)
        loss_hj  = tf.reduce_mean(tf.square(HJ_res))
        loss_mom = tf.reduce_mean(tf.square(p_pred - yb))

        # Per-trajectory anchor: G(0, q0, E) = 0
        G_anchor = gfnn(anchors_b)
        loss_an  = tf.reduce_mean(tf.square(G_anchor - 0.0))

        loss = w_hj*loss_hj + w_mom*loss_mom + w_anchor*loss_an

    grads = tape.gradient(loss, gfnn.trainable_variables)
    opt.apply_gradients(zip(grads, gfnn.trainable_variables))
    return loss

# Train GFNN
for epoch in range(EPOCHS):
    for xb, yb, ab in train_ds_gfnn:
        _ = train_step_gfnn(xb, yb, ab)

# ---------------------------
# 4) Evaluation on (q0,p0)
# ---------------------------
def evaluate_models(q0, p0, T=10.0, M=400):
    t = np.linspace(0., T, M, dtype=np.float32)
    q_true, p_true = exact_traj(t, q0, p0)
    E = np.full_like(t, 0.5*(q0*q0 + p0*p0), dtype=np.float32)

    X_eval = np.stack([t, q_true.astype(np.float32), E], axis=1)

    # Baseline prediction p(t,q,E)
    p_base = baseline.predict(X_eval, verbose=0).squeeze()

    # GFNN prediction p = ∂G/∂q(t,q,E)
    t_tf = tf.constant(t[:,None], dtype=tf.float32)
    q_tf = tf.constant(q_true[:,None], dtype=tf.float32)
    E_tf = tf.constant(E[:,None], dtype=tf.float32)
    with tf.GradientTape() as g:
        g.watch(q_tf)
        G = gfnn(tf.concat([t_tf, q_tf, E_tf], axis=1))
    p_gfnn = g.gradient(G, q_tf).numpy().squeeze()

    # Energies (for reference)
    E_true = 0.5*(p_true**2 + q_true**2)
    E_base = 0.5*(p_base**2 + q_true**2)
    E_gf   = 0.5*(p_gfnn**2 + q_true**2)

    rmse_base = math.sqrt(np.mean((p_base - p_true)**2))
    rmse_gfnn = math.sqrt(np.mean((p_gfnn - p_true)**2))
    return dict(t=t, q=q_true, p_true=p_true, p_base=p_base, p_gfnn=p_gfnn,
                E_true=E_true, E_base=E_base, E_gf=E_gf,
                rmse_base=rmse_base, rmse_gfnn=rmse_gfnn)

# Fixed test IC
q0_test, p0_test = 0.8, -0.4
res = evaluate_models(q0_test, p0_test, T=T_H)

print(f"Evaluation on (q0={q0_test}, p0={p0_test}):")
print(f"  RMSE Baseline : {res['rmse_base']:.6f}")
print(f"  RMSE GFNN     : {res['rmse_gfnn']:.6f}")

#Optional: quick plots (uncomment if you want to visualize)
import matplotlib.pyplot as plt
t = res['t']
plt.figure(); plt.plot(t, res['p_true'], label='Exact p(t)')
plt.plot(t, res['p_base'], '--', label='Baseline'); plt.plot(t, res['p_gfnn'], ':', label='GFNN')
plt.xlabel('t'); plt.ylabel('p(t)'); plt.legend(); plt.title('Momentum vs Time'); plt.show()
plt.figure(); plt.plot(t, res['E_true'], label='Exact E'); plt.plot(t, res['E_base'], '--', label='Baseline E'); plt.plot(t, res['E_gf'], ':', label='GFNN E')
plt.xlabel('t'); plt.ylabel('Energy'); plt.legend(); plt.title('Energy vs Time'); plt.show()
```

# **Finding Conservation Laws I**

**Explicit formulas**

We sampled data along the trajectory of a harmonic oscillator
$$(x(t), \dot x(t) ) = (2\sin t, 2\cos t). $$
and use a NN to obtain $\dot x = f( t, x)$.

```python
# ============================================================
# Learn \dot{x} = f_\theta(t, x) from given (t_j, x_j, dotx_j)
# and plot \dot{x} vs x, time trace, and training loss.
# Data provided are samples from x(t)=2 sin t, \dot x(t)=2 cos t
# ============================================================

import numpy as np
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import matplotlib.pyplot as plt

# Use float64 for stability/precision
tf.keras.backend.set_floatx("float64")
np.random.seed(0); tf.random.set_seed(0)

# ------------------------------------------------------------
# 1) Provided data table (t_j, x(t_j), dot{x}(t_j))
# ------------------------------------------------------------
t_data = np.array([0.0, 0.3, 0.6, 0.9, 1.2,
                   1.5, 1.8, 2.1, 2.4, 2.7, 3.0], dtype=np.float64)

x_data = np.array([0.0000, 0.5910, 1.1280, 1.5656, 1.8632,
                   1.9940, 1.9468, 1.7270, 1.3624, 0.8962, 0.2822], dtype=np.float64)

dotx_data = np.array([ 2.0000,  1.9107,  1.6209,  1.2449,  0.7247,
                       0.1411, -0.5817, -1.0009, -1.4560, -1.8226, -1.9799], dtype=np.float64)

# Inputs (t, x) and labels dot{x}
X_in = np.stack([t_data, x_data], axis=1)   # shape (N, 2)
y_out = dotx_data[:, None]                  # shape (N, 1)

# ------------------------------------------------------------
# 2) Define f_theta(t, x) network
# ------------------------------------------------------------
def build_mlp(in_dim=2, width=32, depth=2, act="swish", out_dim=1):
    inp = keras.Input(shape=(in_dim,))
    x = inp
    for _ in range(depth):
        x = layers.Dense(width, activation=act)(x)
    out = layers.Dense(out_dim, activation=None)(x)
    return keras.Model(inp, out, name="f_theta")

f_theta = build_mlp(in_dim=2, width=64, depth=4, act="tanh", out_dim=1)
f_theta.summary()

# ------------------------------------------------------------
# 3) Train with MSE loss: L = mean( f_theta(t_i,x_i) - dot{x}_i )^2
# ------------------------------------------------------------
f_theta.compile(optimizer=keras.optimizers.Adam(1e-3),
                loss="mse")

EPOCHS = 300
history = f_theta.fit(
    X_in, y_out,
    batch_size=len(t_data),   # small dataset; full-batch works well
    epochs=EPOCHS,
    verbose=0
)

print(f"Final training MSE: {history.history['loss'][-1]:.3e}")

# ------------------------------------------------------------
# 4) Evaluate on a dense grid and plot \dot{x} vs x
#    For comparison, generate the ground-truth curve as well
# ------------------------------------------------------------
t_grid = np.linspace(0.0, 3.0, 400)
x_true = 2.0 * np.sin(t_grid)
dotx_true = 2.0 * np.cos(t_grid)

X_grid = np.stack([t_grid, x_true], axis=1)
dotx_pred = f_theta.predict(X_grid, verbose=0).ravel()

plt.figure(figsize=(4.5,4.5))
plt.plot(x_true, dotx_true, label="Ground truth: $2\\cos t$ vs $2\\sin t$", lw=2)
plt.scatter(x_data, dotx_data, color='C3', s=35, zorder=4, label="Data")
plt.scatter(x_true, dotx_pred, s=3, color="green", alpha=0.5, label=r"NN: $f_{\theta^*}(t,x)$", zorder=3)
plt.xlabel("$x$")
plt.ylabel(r"$\dot{x}$")
plt.title(r"Learned $\dot{x}\approx f_{\theta^*}(t,x)$ plotted against $x$")
plt.grid(alpha=0.3); plt.legend(); plt.tight_layout(); plt.show()

# ------------------------------------------------------------
# 5) Also visualize \dot{x}(t): prediction vs truth, with data
# ------------------------------------------------------------
# To get NN's time trace, feed (t, x(t)) into f_theta
X_time = np.stack([t_grid, x_true], axis=1)
dotx_pred_time = f_theta.predict(X_time, verbose=0).ravel()

plt.figure(figsize=(4.5,4.5))
plt.plot(t_grid, dotx_true, lw=2, label="Ground truth $\\dot{x}(t)=2\\cos t$")
plt.plot(t_grid, dotx_pred_time, "--", color="green",  lw=2, label=r"NN $f_{\theta^*}(t,x(t))$")
plt.scatter(t_data, dotx_data, c='C3', s=35, label="Data", zorder=4)
plt.xlabel("$t$")
plt.ylabel(r"$\dot{x}(t)$")
plt.title(r"Time trace of $\dot{x}(t)$: NN vs ground truth")
plt.grid(alpha=0.3); plt.legend(); plt.tight_layout(); plt.show()

# ------------------------------------------------------------
# 6) Plot training loss (log scale)
# ------------------------------------------------------------
plt.figure(figsize=(4.5,4.5))
plt.plot(history.history["loss"])
plt.yscale("log")
plt.xlabel("Epoch"); plt.ylabel("MSE loss (log)")
plt.title("Training loss")
plt.grid(alpha=0.3); plt.tight_layout(); plt.show()
```

We added noise to the data and recovered $\dot x$ in terms of $x$.

```python
# ============================================================
# Learn \dot{x} = f_\theta(t, x) from noisy (t_j, x_j, dotx_j)
# and plot results
# ============================================================

import numpy as np
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import matplotlib.pyplot as plt

# Use float64 for stability/precision
tf.keras.backend.set_floatx("float64")
np.random.seed(0); tf.random.set_seed(0)

# ------------------------------------------------------------
# 1) Provided data table (t_j, x(t_j), dot{x}(t_j)) with noise
# ------------------------------------------------------------
t_data = np.array([0.0, 0.3, 0.6, 0.9, 1.2,
                   1.5, 1.8, 2.1, 2.4, 2.7, 3.0], dtype=np.float64)

x_true = np.array([0.0000, 0.5910, 1.1280, 1.5656, 1.8632,
                   1.9940, 1.9468, 1.7270, 1.3624, 0.8962, 0.2822], dtype=np.float64)

dotx_true = np.array([ 2.0000,  1.9107,  1.6209,  1.2449,  0.7247,
                       0.1411, -0.5817, -1.0009, -1.4560, -1.8226, -1.9799], dtype=np.float64)

# Add Gaussian noise
noise_level_x = 0.05   # adjust as needed
noise_level_dx = 0.05
x_data = x_true + noise_level_x * np.random.randn(*x_true.shape)
dotx_data = dotx_true + noise_level_dx * np.random.randn(*dotx_true.shape)

# Inputs (t, x) and labels dot{x}
X_in = np.stack([t_data, x_data], axis=1)
y_out = dotx_data[:, None]

# ------------------------------------------------------------
# 2) Define f_theta(t, x) network
# ------------------------------------------------------------
def build_mlp(in_dim=2, width=64, depth=2, act="tanh", out_dim=1):
    inp = keras.Input(shape=(in_dim,))
    x = inp
    for _ in range(depth):
        x = layers.Dense(width, activation=act)(x)
    out = layers.Dense(out_dim, activation=None)(x)
    return keras.Model(inp, out, name="f_theta")

f_theta = build_mlp(in_dim=2, width=64, depth=4, act="tanh", out_dim=1)
f_theta.summary()

# ------------------------------------------------------------
# 3) Train with MSE loss
# ------------------------------------------------------------
f_theta.compile(optimizer=keras.optimizers.Adam(1e-3),
                loss="mse")

EPOCHS = 800
history = f_theta.fit(
    X_in, y_out,
    batch_size=len(t_data),
    epochs=EPOCHS,
    verbose=0
)

print(f"Final training MSE: {history.history['loss'][-1]:.3e}")

# ------------------------------------------------------------
# 4) Evaluate on a dense grid and plot \dot{x} vs x
# ------------------------------------------------------------
t_grid = np.linspace(0.0, 3.0, 400)
x_grid = 2.0 * np.sin(t_grid)
dotx_grid = 2.0 * np.cos(t_grid)

X_grid = np.stack([t_grid, x_grid], axis=1)
dotx_pred = f_theta.predict(X_grid, verbose=0).ravel()

plt.figure(figsize=(4.5,4.5))
plt.plot(x_grid, dotx_grid, label="Ground truth: $2\\cos t$ vs $2\\sin t$", lw=2)
plt.scatter(x_data, dotx_data, color='C3', s=35, zorder=4, label="Noisy data")
plt.scatter(x_grid, dotx_pred, s=3, color="green", alpha=0.5, label=r"NN: $f_{\theta^*}(t,x)$", zorder=3)
plt.xlabel("$x$")
plt.ylabel(r"$\dot{x}$")
plt.title(r"Learned $\dot{x}\approx f_{\theta^*}(t,x)$ (with noisy data)")
plt.grid(alpha=0.3); plt.legend(); plt.tight_layout(); plt.show()

# ------------------------------------------------------------
# 5) Time trace: prediction vs truth
# ------------------------------------------------------------
X_time = np.stack([t_grid, x_grid], axis=1)
dotx_pred_time = f_theta.predict(X_time, verbose=0).ravel()

plt.figure(figsize=(4.5,4.5))
plt.plot(t_grid, dotx_grid, lw=2, label="Ground truth $\\dot{x}(t)=2\\cos t$")
plt.plot(t_grid, dotx_pred_time, "--", lw=2, color="green",label=r"NN $f_{\theta^*}(t,x(t))$")
plt.scatter(t_data, dotx_data, c='C3', s=35, label="Noisy data", zorder=4)
plt.xlabel("$t$")
plt.ylabel(r"$\dot{x}(t)$")
plt.title(r"Time trace with noisy data")
plt.grid(alpha=0.3); plt.legend(); plt.tight_layout(); plt.show()

# ------------------------------------------------------------
# 6) Training loss
# ------------------------------------------------------------
plt.figure(figsize=(4.5,4.5))
plt.plot(history.history["loss"])
plt.yscale("log")
plt.xlabel("Epoch"); plt.ylabel("MSE loss (log)")
plt.title("Training loss (with noisy data)")
plt.grid(alpha=0.3); plt.tight_layout(); plt.show()
```

The next code tries to find a conservation law of the type $F(t, x, \dot x) = C$, but does not succeed.

```python
# ============================================================
# Discover an implicit conservation law F_theta(t, x, dx) ~ const
# from (t_j, x_j, dx_j) for x=2 sin t, dx=2 cos t
# ============================================================

import numpy as np
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import matplotlib.pyplot as plt

# High precision
tf.keras.backend.set_floatx("float64")
np.random.seed(0); tf.random.set_seed(0)

# ------------------------------------------------------------
# Data: same as before (NOISELESS)
# ------------------------------------------------------------
t_data = np.array([0.0, 0.3, 0.6, 0.9, 1.2,
                   1.5, 1.8, 2.1, 2.4, 2.7, 3.0], dtype=np.float64)

x_data   = np.array([0.0000, 0.5910, 1.1280, 1.5656, 1.8632,
                     1.9940, 1.9468, 1.7270, 1.3624, 0.8962, 0.2822], dtype=np.float64)

dx_data  = np.array([ 2.0000,  1.9107,  1.6209,  1.2449,  0.7247,
                      0.1411, -0.5817, -1.0009, -1.4560, -1.8226, -1.9799], dtype=np.float64)

# Input to F_theta is (t, x, dx)
X_train = np.stack([t_data, x_data, dx_data], axis=1).astype(np.float64)

# ------------------------------------------------------------
# Model: F_theta(t, x, dx) -> R
# ------------------------------------------------------------
def build_mlp(in_dim=3, width=64, depth=1, act="tanh", out_dim=1):
    inp = keras.Input(shape=(in_dim,))
    x = inp
    for _ in range(depth):
        x = layers.Dense(width, activation=act)(x)
    out = layers.Dense(out_dim, activation=None)(x)
    return keras.Model(inp, out, name="F_theta")

F_theta = build_mlp(in_dim=3, width=64, depth=3, act="tanh", out_dim=1)
F_theta.summary()

# ------------------------------------------------------------
# Training: minimize variance-type loss so F is (approximately) constant
# L = mean( (F - mean(F))^2 )
# ------------------------------------------------------------
optimizer = keras.optimizers.Adam(1e-4)

EPOCHS = 10000
PRINT_EVERY = 200

hist_loss = []

X_tf = tf.constant(X_train, dtype=tf.float64)

@tf.function
def train_step():
    with tf.GradientTape() as tape:
        F_vals = F_theta(X_tf)                # (N,1)
        F_mean = tf.reduce_mean(F_vals)
        loss = tf.reduce_mean(tf.square(F_vals - F_mean))
    grads = tape.gradient(loss, F_theta.trainable_variables)
    optimizer.apply_gradients(zip(grads, F_theta.trainable_variables))
    return loss

for ep in range(1, EPOCHS+1):
    L = train_step()
    hist_loss.append(L.numpy())
    if ep % PRINT_EVERY == 0:
        print(f"epoch {ep:4d} | variance loss {L.numpy():.3e}")

# ------------------------------------------------------------
# 1) Show that F is ~ constant along the true trajectory
# ------------------------------------------------------------
t_grid = np.linspace(0.0, 3.0, 400)
x_grid = 2.0*np.sin(t_grid)
dx_grid = 2.0*np.cos(t_grid)
X_grid = np.stack([t_grid, x_grid, dx_grid], axis=1)

F_vals_grid = F_theta(tf.constant(X_grid, dtype=tf.float64)).numpy().ravel()
F_const = F_theta(X_tf).numpy().mean()  # learned constant level

plt.figure(figsize=(4.5, 4.5))
plt.plot(t_grid, F_vals_grid, lw=2, label=r"$F_\theta(t, x(t), \dot x(t))$")
plt.axhline(F_const, color="C3", ls="--", lw=2, label=r"Learned constant $\bar F$")
plt.xlabel("$t$")
plt.ylabel(r"$F_\theta$")
plt.title("Implicit law evaluated along the true trajectory")
plt.grid(alpha=0.3); plt.legend(); plt.tight_layout(); plt.show()

# ------------------------------------------------------------
# 2) Implicit plot in the (x, dx) plane at a representative t0
#    Plot the level set F(t0, x, dx) = \bar F and overlay true circle x^2+dx^2=4
# ------------------------------------------------------------
t0 = 3
xv = np.linspace(-2.5, 2.5, 300)
dv = np.linspace(-2.5, 2.5, 300)
XX, DD = np.meshgrid(xv, dv)
TT = np.full_like(XX, t0)

X_slice = np.stack([TT.ravel(), XX.ravel(), DD.ravel()], axis=1)
F_slice = F_theta(tf.constant(X_slice, dtype=tf.float64)).numpy().reshape(XX.shape)

plt.figure(figsize=(4.5, 4.5))
CS = plt.contour(XX, DD, F_slice - F_const, levels=[0.0], colors=["C1"], linewidths=2)
plt.clabel(CS, inline=True, fmt={0.0: r"$F_\theta(t_0,x,\dot x)=\bar F$"}, fontsize=10)

# Overlay true trajectory (x(t), dx(t)) for t in [0,3]
plt.plot(x_grid, dx_grid, 'C0', lw=2, label=r"True $(x(t),\dot x(t))$")
# Reference circle x^2 + dx^2 = 4
theta = np.linspace(0, 2*np.pi, 400)
plt.plot(2*np.cos(theta), 2*np.sin(theta), 'k--', alpha=0.5, label=r"$x^2+\dot x^2=4$")

plt.scatter(x_data, dx_data, c='C3', s=35, zorder=4, label="Data")
plt.xlabel("$x$")
plt.ylabel(r"$\dot x$")
plt.title(rf"Implicit level set at $t_0={t0}$")
plt.axis("equal"); plt.grid(alpha=0.3); plt.legend(); plt.tight_layout(); plt.show()

# ------------------------------------------------------------
# 3) Training loss history
# ------------------------------------------------------------
plt.figure(figsize=(4.5, 4.5))
plt.plot(hist_loss, lw=2)
plt.yscale("log")
plt.xlabel("Epoch"); plt.ylabel("Variance loss (log)")
plt.title("Training loss")
plt.grid(alpha=0.3); plt.tight_layout(); plt.show()
```

In the next code we find a law $F(x, \dot x) =C$, which does not ontain explicitly $t$ as an argument.

```python
# ============================================================
# Learn an x–dx implicit conservation law: F_theta(x, dx) ~ const
# from (x_j, dx_j) of x=2 sin t, dx=2 cos t
# ============================================================

import numpy as np
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import matplotlib.pyplot as plt

# High precision & seeds
tf.keras.backend.set_floatx("float64")
np.random.seed(0); tf.random.set_seed(0)

# ------------------------------------------------------------
# Data (no time used here): (x_j, dx_j) from x=2 sin t, dx=2 cos t
# ------------------------------------------------------------
t_data = np.array([0.0, 0.3, 0.6, 0.9, 1.2,
                   1.5, 1.8, 2.1, 2.4, 2.7, 3.0], dtype=np.float64)

x_data  = np.array([0.0000, 0.5910, 1.1280, 1.5656, 1.8632,
                    1.9940, 1.9468, 1.7270, 1.3624, 0.8962, 0.2822], dtype=np.float64)

dx_data = np.array([ 2.0000,  1.9107,  1.6209,  1.2449,  0.7247,
                     0.1411, -0.5817, -1.0009, -1.4560, -1.8226, -1.9799], dtype=np.float64)

X_train = np.stack([x_data, dx_data], axis=1).astype(np.float64)  # shape (N,2)
X_tf = tf.constant(X_train, dtype=tf.float64)

# ------------------------------------------------------------
# Model: F_theta(x, dx) -> R
# ------------------------------------------------------------
def build_mlp(in_dim=2, width=64, depth=1, act="tanh", out_dim=1):
    inp = keras.Input(shape=(in_dim,))
    x = inp
    for _ in range(depth):
        x = layers.Dense(width, activation=act)(x)
    out = layers.Dense(out_dim, activation=None)(x)
    return keras.Model(inp, out, name="F_theta")

F_theta = build_mlp(in_dim=2, width=64, depth=1, act="tanh", out_dim=1)
F_theta.summary()

# ------------------------------------------------------------
# Training: minimize variance so F is (approximately) constant on data
# L = mean( (F - mean(F))^2 )
# ------------------------------------------------------------
optimizer = keras.optimizers.Adam(1e-4)
EPOCHS = 10000
PRINT_EVERY = 200
hist_loss = []

@tf.function
def train_step():
    with tf.GradientTape() as tape:
        F_vals = F_theta(X_tf)                  # (N,1)
        F_mean = tf.reduce_mean(F_vals)
        loss = tf.reduce_mean(tf.square(F_vals - F_mean))
    grads = tape.gradient(loss, F_theta.trainable_variables)
    optimizer.apply_gradients(zip(grads, F_theta.trainable_variables))
    return loss

for ep in range(1, EPOCHS+1):
    L = train_step()
    hist_loss.append(L.numpy())
    if ep % PRINT_EVERY == 0:
        print(f"epoch {ep:4d} | variance loss {L.numpy():.3e}")

# ------------------------------------------------------------
# 1) Evaluate constancy along the true trajectory (for reference)
# ------------------------------------------------------------
t_grid = np.linspace(0.0, 3.0, 400)
x_grid  = 2.0*np.sin(t_grid)
dx_grid = 2.0*np.cos(t_grid)
X_path = np.stack([x_grid, dx_grid], axis=1)

F_vals_path = F_theta(tf.constant(X_path, dtype=tf.float64)).numpy().ravel()
F_const = float(F_theta(X_tf).numpy().mean())

plt.figure(figsize=(4.5, 4.5))
plt.plot(t_grid, F_vals_path, lw=2, label=r"$F_\theta(x(t), \dot x(t))$")
plt.axhline(F_const, color="C3", ls="--", lw=2, label=r"Learned constant $\bar F$")
plt.xlabel("$t$")
plt.ylabel(r"$F_\theta$")
plt.title("Implicit law evaluated along the true trajectory")
plt.grid(alpha=0.3); plt.legend(); plt.tight_layout(); plt.show()

# ------------------------------------------------------------
# 2) Implicit level set in (x, dx)-plane: F_theta(x,dx) = \bar F
#    and overlay the true circle x^2 + dx^2 = 4
# ------------------------------------------------------------
xv = np.linspace(-2.5, 2.5, 350)
dv = np.linspace(-2.5, 2.5, 350)
XX, DD = np.meshgrid(xv, dv)
grid_pairs = np.stack([XX.ravel(), DD.ravel()], axis=1)
F_slice = F_theta(tf.constant(grid_pairs, dtype=tf.float64)).numpy().reshape(XX.shape)

plt.figure(figsize=(4.5, 4.5))
CS = plt.contour(XX, DD, F_slice - F_const, levels=[0.0], colors=["C1"], linewidths=2)
plt.clabel(CS, inline=True, fmt={0.0: r"$F_\theta(x,\dot x)=\bar F$"}, fontsize=10)

# True trajectory and reference circle
theta = np.linspace(0, 2*np.pi, 400)
plt.plot(2*np.cos(theta), 2*np.sin(theta), 'k--', alpha=0.5, label=r"$x^2+\dot x^2=4$")
plt.plot(x_grid, dx_grid, 'C0', lw=2, label=r"True $(x(t),\dot x(t))$")
plt.scatter(x_data, dx_data, c='C3', s=35, zorder=4, label="Data")

plt.xlabel("$x$")
plt.ylabel(r"$\dot x$")
plt.title("Implicit level set in the phase plane")
plt.axis("equal"); plt.grid(alpha=0.3); plt.legend(); plt.tight_layout(); plt.show()

# ------------------------------------------------------------
# 3) Training loss history
# ------------------------------------------------------------
plt.figure(figsize=(4.5, 4.5))
plt.plot(hist_loss, lw=2)
plt.yscale("log")
plt.xlabel("Epoch"); plt.ylabel("Variance loss (log)")
plt.title("Training loss")
plt.grid(alpha=0.3); plt.tight_layout(); plt.show()
```

We added some L2 regularization to make the interpolation curve more ridgid.

```python
# ============================================================
# Learn an x–dx implicit conservation law: F_theta(x, dx) ~ const
# from (x_j, dx_j) of x=2 sin t, dx=2 cos t
# (with L2 regularization)
# ============================================================

import numpy as np
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import matplotlib.pyplot as plt

# High precision & seeds
tf.keras.backend.set_floatx("float64")
np.random.seed(0); tf.random.set_seed(0)

# ------------------------------------------------------------
# Data (no time used here): (x_j, dx_j) from x=2 sin t, dx=2 cos t
# ------------------------------------------------------------
t_data = np.array([0.0, 0.3, 0.6, 0.9, 1.2,
                   1.5, 1.8, 2.1, 2.4, 2.7, 3.0], dtype=np.float64)

x_data  = np.array([0.0000, 0.5910, 1.1280, 1.5656, 1.8632,
                    1.9940, 1.9468, 1.7270, 1.3624, 0.8962, 0.2822], dtype=np.float64)

dx_data = np.array([ 2.0000,  1.9107,  1.6209,  1.2449,  0.7247,
                     0.1411, -0.5817, -1.0009, -1.4560, -1.8226, -1.9799], dtype=np.float64)

X_train = np.stack([x_data, dx_data], axis=1).astype(np.float64)  # shape (N,2)
X_tf = tf.constant(X_train, dtype=tf.float64)

# ------------------------------------------------------------
# L2 regularization strength
# ------------------------------------------------------------
L2_LAMBDA = 1e-4  # try 1e-6..1e-4 depending on desired smoothness

# ------------------------------------------------------------
# Model: F_theta(x, dx) -> R  (Dense layers with L2)
# ------------------------------------------------------------
def build_mlp(in_dim=2, width=64, depth=1, act="tanh", out_dim=1):
    reg = keras.regularizers.l2(L2_LAMBDA)
    inp = keras.Input(shape=(in_dim,))
    x = inp
    for _ in range(depth):
        x = layers.Dense(width, activation=act, kernel_regularizer=reg)(x)
    out = layers.Dense(out_dim, activation=None, kernel_regularizer=reg)(x)
    return keras.Model(inp, out, name="F_theta")

# You had depth=1 originally; keep that, but L2 applies regardless
F_theta = build_mlp(in_dim=2, width=64, depth=1, act="tanh", out_dim=1)
F_theta.summary()

# ------------------------------------------------------------
# Training: minimize variance so F is (approximately) constant on data
# L_total = mean( (F - mean(F))^2 ) + sum(L2 penalties)
# ------------------------------------------------------------
optimizer = keras.optimizers.Adam(1e-4)
EPOCHS = 10000
PRINT_EVERY = 200
hist_loss = []
hist_base = []
hist_reg  = []

@tf.function
def train_step():
    with tf.GradientTape() as tape:
        F_vals = F_theta(X_tf)                      # (N,1)
        F_mean = tf.reduce_mean(F_vals)
        base_loss = tf.reduce_mean(tf.square(F_vals - F_mean))
        reg_loss = tf.add_n(F_theta.losses) if F_theta.losses else tf.constant(0.0, dtype=tf.float64)
        loss = base_loss + reg_loss
    grads = tape.gradient(loss, F_theta.trainable_variables)
    optimizer.apply_gradients(zip(grads, F_theta.trainable_variables))
    return loss, base_loss, reg_loss

for ep in range(1, EPOCHS+1):
    L, L_base, L_reg = train_step()
    hist_loss.append(float(L.numpy()))
    hist_base.append(float(L_base.numpy()))
    hist_reg.append(float(L_reg.numpy()))
    if ep % PRINT_EVERY == 0:
        print(f"epoch {ep:4d} | total {L.numpy():.3e} | base {L_base.numpy():.3e} | L2 {L_reg.numpy():.3e}")

# ------------------------------------------------------------
# 1) Evaluate constancy along the true trajectory (for reference)
# ------------------------------------------------------------
t_grid = np.linspace(0.0, 3.0, 400)
x_grid  = 2.0*np.sin(t_grid)
dx_grid = 2.0*np.cos(t_grid)
X_path = np.stack([x_grid, dx_grid], axis=1)

F_vals_path = F_theta(tf.constant(X_path, dtype=tf.float64)).numpy().ravel()
F_const = float(F_theta(X_tf).numpy().mean())

plt.figure(figsize=(4.8, 4.6))
plt.plot(t_grid, F_vals_path, lw=2, label=r"$F_\theta(x(t), \dot x(t))$")
plt.axhline(F_const, color="C3", ls="--", lw=2, label=r"Learned constant $\bar F$")
plt.xlabel("$t$")
plt.ylabel(r"$F_\theta$")
plt.title("Implicit law evaluated along the true trajectory")
plt.grid(alpha=0.3); plt.legend(); plt.tight_layout(); plt.show()

# ------------------------------------------------------------
# 2) Implicit level set in (x, dx)-plane: F_theta(x,dx) = \bar F
#    and overlay the true circle x^2 + dx^2 = 4
# ------------------------------------------------------------
xv = np.linspace(-2.5, 2.5, 350)
dv = np.linspace(-2.5, 2.5, 350)
XX, DD = np.meshgrid(xv, dv)
grid_pairs = np.stack([XX.ravel(), DD.ravel()], axis=1)
F_slice = F_theta(tf.constant(grid_pairs, dtype=tf.float64)).numpy().reshape(XX.shape)

plt.figure(figsize=(4.8, 4.6))
CS = plt.contour(XX, DD, F_slice - F_const, levels=[0.0], colors=["C1"], linewidths=2)
plt.clabel(CS, inline=True, fmt={0.0: r"$F_\theta(x,\dot x)=\bar F$"}, fontsize=10)

# True trajectory and reference circle
theta = np.linspace(0, 2*np.pi, 400)
plt.plot(2*np.cos(theta), 2*np.sin(theta), 'k--', alpha=0.5, label=r"$x^2+\dot x^2=4$")
plt.plot(x_grid, dx_grid, 'C0', lw=2, label=r"True $(x(t),\dot x(t))$")
plt.scatter(x_data, dx_data, c='C3', s=35, zorder=4, label="Data")

plt.xlabel("$x$")
plt.ylabel(r"$\dot x$")
plt.title("Implicit level set in the phase plane")
plt.axis("equal"); plt.grid(alpha=0.3); plt.legend(); plt.tight_layout(); plt.show()

# ------------------------------------------------------------
# 3) Training loss history (total, base, L2)
# ------------------------------------------------------------
plt.figure(figsize=(4.8, 4.6))
plt.plot(hist_loss, lw=2, label="total")
plt.plot(hist_base, lw=1.5, label="base variance")
plt.plot(hist_reg,  lw=1.5, label="L2")
plt.yscale("log")
plt.xlabel("Epoch"); plt.ylabel("Loss (log)")
plt.title("Training loss (with L2 regularization)")
plt.grid(alpha=0.3); plt.legend(); plt.tight_layout(); plt.show()
```

**Learning a Velocity Field from Position Data**

 We shall generate our own synthetic data by collecting $30$ measurements $\{(t_j, x_j)\}$ from the solution $x(t) = \tanh(x+\pi/4)$ of the equation
$$ \dot x(t)  = x^2 + 1,\quad  x(0)=1. $$
Then we use deep learning to recover the velocity function $f(t,x)=x^2+1$.

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, Model, optimizers
import matplotlib.pyplot as plt

# 1) Generate N=200 data from dx/dt = x^2 + 1, x(0)=1, for t in [0, 0.75]
N = 30
t_data = np.linspace(0, 0.75, N, dtype=np.float32).reshape(-1,1)
# Analytical solution: x(t) = tan(t + pi/4)
x_data = np.tan(t_data + np.pi/4)

# 2) Plot the data
plt.figure(figsize=(6,4))
plt.scatter(t_data, x_data, c='black', s=20, label='Data $(t_j, x_j)$')
plt.xlabel('t')
plt.ylabel('x(t)')
plt.title('Data from $dx/dt = x^2 + 1$')
plt.grid(True)
plt.legend()
plt.show()

# 3) Collocation points for physics loss
M = 200
t_colloc = np.linspace(0, 0.75, M, dtype=np.float32).reshape(-1,1)

# Convert to TF tensors
t_train_tf  = tf.convert_to_tensor(t_data)
x_train_tf  = tf.convert_to_tensor(x_data)
t_colloc_tf = tf.convert_to_tensor(t_colloc)

# 4) Build model_sol: t -> x_pred
def build_model_sol():
    inp = layers.Input(shape=(1,))
    x = layers.Dense(64, activation='swish')(inp)
    x = layers.Dense(64, activation='swish')(x)
    out = layers.Dense(1)(x)
    return Model(inp, out, name='model_sol')

# 5) Build model_f: x_pred -> f_pred
def build_model_f():
    inp = layers.Input(shape=(1,))
    x = layers.Dense(64, activation='swish')(inp)
    x = layers.Dense(64, activation='swish')(x)
    out = layers.Dense(1)(x)
    return Model(inp, out, name='model_f')

model_sol = build_model_sol()
model_f   = build_model_f()


# 6) Optimizer and weight
optimizer = optimizers.Adam(learning_rate=0.7e-4)
w_phys    = 0.1

# 7) Training step
@tf.function
def train_step():
    with tf.GradientTape(persistent=True) as tape:
        # Data loss
        x_pred = model_sol(t_train_tf, training=True)        # (N,1)
        loss_data = tf.reduce_mean((x_pred - x_train_tf)**2)

        # Physics loss: compute dx/dt and compare to f(x)
        with tf.GradientTape() as g:
            g.watch(t_colloc_tf)
            x_coll = model_sol(t_colloc_tf, training=True)    # (M,1)
        dx_dt = g.gradient(x_coll, t_colloc_tf)             # (M,1)

        f_pred = model_f(x_coll, training=True)             # (M,1)
        loss_phys = tf.reduce_mean((dx_dt - f_pred)**2)

        loss_total = loss_data + w_phys * loss_phys

    grads_sol = tape.gradient(loss_total, model_sol.trainable_variables)
    grads_f   = tape.gradient(loss_total, model_f.trainable_variables)
    optimizer.apply_gradients(zip(grads_sol + grads_f,
                                  model_sol.trainable_variables + model_f.trainable_variables))
    return loss_total, loss_data, loss_phys

# 8) Training loop
epochs = 300000
history = {'total':[], 'data':[], 'phys':[]}
for epoch in range(1, epochs+1):
    Lt, Ld, Lp = train_step()
    history['total'].append(Lt.numpy().item())
    history['data'].append(Ld.numpy().item())
    history['phys'].append(Lp.numpy().item())
    if epoch % 500 == 0:
        print(f"Epoch {epoch}: total={Lt.numpy().item():.4e}, data={Ld.numpy().item():.4e}, phys={Lp.numpy().item():.4e}")

# 9) Plot loss evolution
plt.figure(figsize=(6,4))
plt.plot(history['total'], label='Total Loss')
#plt.plot(history['data'], '--', label='Data Loss')
#plt.plot(history['phys'], ':', label='Physics Loss')
plt.yscale('log')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.title('Training Loss vs. Epoch')
plt.legend()
plt.grid(True)
plt.show()

# 10) Plot learned f(x) vs true x^2+1
x_vals = np.linspace(float(x_data.min()), float(x_data.max()), 200, dtype=np.float32).reshape(-1,1)
f_pred_vals = model_f(x_vals).numpy()
f_true_vals = x_vals**2 + 1

plt.figure(figsize=(6,4))
plt.plot(x_vals, f_pred_vals, 'r-', label='Learned $f_\\theta(x)$')
plt.plot(x_vals, f_true_vals, 'b--', label='$x^2 + 1$ (true)')
plt.xlabel('$x$')
plt.ylabel('$f(x)$')
plt.title('Learned vs True Velocity Field')
plt.legend()
plt.grid(True)
plt.show()

```

**Learning a Force Field from Position Data**

We artificially generate $N=20$ data points $(t_j, x_j)$ from the trajectroy of the ODE $$\ddot x = - x, \quad x(0)=1, \quad x'(0) =0. $$
We use these data points to train a neural network to learn the acceleration field $-x$ and the solution.

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, Model, optimizers
import matplotlib.pyplot as plt

# Step 1: Generate data
N = 20
t_data = np.linspace(0, 2*np.pi, N, dtype=np.float32).reshape(-1,1)
x_data = np.cos(t_data)

# Plot data
plt.figure(figsize=(6,4))
plt.scatter(t_data, x_data, c='black', label='Data (t_j, x_j)')
plt.plot(t_data, x_data, 'k--', alpha=0.3)
plt.xlabel('t'); plt.ylabel('x(t)')
plt.title('Data from x(t)=cos(t)')
plt.legend(); plt.grid(True)
plt.show()

# Step 2: Build PINN to learn x(t) and f(x)
# Convert data to tensors
t_train = tf.convert_to_tensor(t_data)
x_train = tf.convert_to_tensor(x_data)

# Collocation points
M = 100
t_colloc = np.linspace(0, 2*np.pi, M, dtype=np.float32).reshape(-1,1)
t_colloc_tf = tf.convert_to_tensor(t_colloc)

# Model for solution x(t)
def build_model_sol():
    inp = layers.Input(shape=(1,))
    x = layers.Dense(32, activation='swish')(inp)
    x = layers.Dense(32, activation='swish')(x)
    out = layers.Dense(1)(x)
    return Model(inp, out, name='model_sol')

# Model for f(x)
def build_model_f():
    inp = layers.Input(shape=(1,))
    x = layers.Dense(64, activation='swish')(inp)
    x = layers.Dense(64, activation='swish')(x)
    out = layers.Dense(1)(x)
    return Model(inp, out, name='model_f')

model_sol = build_model_sol()
model_f   = build_model_f()

optimizer = optimizers.Adam(learning_rate=1e-3)
w_phys = 5.0

# Training step
@tf.function
def train_step():
    with tf.GradientTape(persistent=True) as tape:
        # Data loss
        x_pred_data = model_sol(t_train, training=True)
        loss_data = tf.reduce_mean((x_pred_data - x_train)**2)

        # Physics loss: second derivative d2x/dt2 = f(x)
        with tf.GradientTape() as tape2:
            tape2.watch(t_colloc_tf)
            with tf.GradientTape() as tape1:
                tape1.watch(t_colloc_tf)
                x_coll = model_sol(t_colloc_tf, training=True)
            dx_dt = tape1.gradient(x_coll, t_colloc_tf)
        d2x_dt2 = tape2.gradient(dx_dt, t_colloc_tf)

        f_pred = model_f(x_coll, training=True)
        loss_phys = tf.reduce_mean((d2x_dt2 - f_pred)**2)

        loss_total = loss_data + w_phys * loss_phys

    grads_sol = tape.gradient(loss_total, model_sol.trainable_variables)
    grads_f   = tape.gradient(loss_total, model_f.trainable_variables)
    optimizer.apply_gradients(zip(grads_sol + grads_f,
                                  model_sol.trainable_variables + model_f.trainable_variables))
    return loss_total, loss_data, loss_phys

# Training loop
epochs = 50000
history = {'total':[], 'data':[], 'phys':[]}
for epoch in range(1, epochs+1):
    Lt, Ld, Lp = train_step()
    history['total'].append(Lt.numpy().item())
    history['data'].append(Ld.numpy().item())
    history['phys'].append(Lp.numpy().item())
    if epoch % 500 == 0:
        print(f"Epoch {epoch}: total={Lt.numpy().item():.4e}, data={Ld.numpy().item():.4e}, phys={Lp.numpy().item():.4e}")

# Step 3: Plot learned vs exact solution
t_plot = np.linspace(0, 2*np.pi, 200, dtype=np.float32).reshape(-1,1)
x_pred_plot = model_sol(t_plot).numpy()
x_exact = np.cos(t_plot)

plt.figure(figsize=(6,4))
plt.plot(t_plot, x_pred_plot, 'r-', label='Learned $x_\\theta(t)$')
plt.plot(t_plot, x_exact, 'b--', label='Exact $\\cos t$')
plt.scatter(t_data, x_data, c='black', s=20, label='Training Data')
plt.xlabel('t'); plt.ylabel('x(t)')
plt.title('Learned vs Exact Solution')
plt.legend(); plt.grid(True)
plt.show()

# Plot learned f(x) vs exact -x
x_vals = np.linspace(-1, 1, 200, dtype=np.float32).reshape(-1,1)
f_pred_vals = model_f(x_vals).numpy()
f_exact_vals = -x_vals

plt.figure(figsize=(6,4))
plt.plot(x_vals, f_pred_vals, 'r-', label='Learned $f_\\theta(x)$')
plt.plot(x_vals, f_exact_vals, 'b--', label='Exact $-x$')
plt.xlabel('$x$'); plt.ylabel('$f(x)$')
plt.title('Learned vs Exact Force Field')
plt.legend(); plt.grid(True)
plt.show()

```

Next we solve a similar problem for the ODE $$\ddot x = - x^2, \quad x(0)=1, \quad x'(0)=0.$$

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, Model, optimizers
import matplotlib.pyplot as plt

# Step 1: Synthetic data via RK4 for x'' = -x^2
T = 2.0
N = 20
t_data = np.linspace(0, T, N, dtype=np.float32)
# High-resolution RK4 integration
dt = 1e-3
ts = np.arange(0, T + dt, dt, dtype=np.float32)
x, v = 1.0, 0.0
x_hist = [x]
for _ in ts[:-1]:
    def deriv(x, v):
        return v, -x**2
    k1x, k1v = deriv(x, v)
    k2x, k2v = deriv(x + 0.5*dt*k1x, v + 0.5*dt*k1v)
    k3x, k3v = deriv(x + 0.5*dt*k2x, v + 0.5*dt*k2v)
    k4x, k4v = deriv(x + dt*k3x,  v + dt*k3v)
    x += dt*(k1x + 2*k2x + 2*k3x + k4x)/6
    v += dt*(k1v + 2*k2v + 2*k3v + k4v)/6
    x_hist.append(x)
x_hist = np.array(x_hist, dtype=np.float32)

# sample synthetic data at t_data
x_data = np.interp(t_data, ts, x_hist).astype(np.float32).reshape(-1,1)

# Plot the synthetic data
plt.figure(figsize=(6,4))
plt.scatter(t_data, x_data, c='black', label='Data $(t_j,x_j)$')
plt.plot(ts, x_hist, 'k--', alpha=0.3, label='Reference trajectory')
plt.xlabel('t'); plt.ylabel('x(t)')
plt.title("Data for $x''=-x^2$, $x(0)=1$, $x'(0)=0$")
plt.legend(); plt.grid(True)
plt.show()

# Step 2: Build PINN models

# Convert data to tf tensors
t_train = tf.convert_to_tensor(t_data.reshape(-1,1), dtype=tf.float32)
x_train = tf.convert_to_tensor(x_data, dtype=tf.float32)

# Collocation points for physics loss
M =100
t_colloc = np.linspace(0, T, M, dtype=np.float32).reshape(-1,1)
t_colloc_tf = tf.convert_to_tensor(t_colloc, dtype=tf.float32)

# Solution network: t -> [x_pred, v_pred]
def build_model_sol():
    inp = layers.Input(shape=(1,))
    y   = layers.Dense(32, activation='swish')(inp)
    y   = layers.Dense(32, activation='swish')(y)
    out = layers.Dense(2)(y)
    return Model(inp, out)

# Force network: x_pred -> f_pred
def build_model_f():
    inp = layers.Input(shape=(1,))
    y   = layers.Dense(64, activation='swish')(inp)
    y   = layers.Dense(64, activation='swish')(y)
    out = layers.Dense(1)(y)
    return Model(inp, out)

model_sol = build_model_sol()
model_f   = build_model_f()

optimizer = optimizers.Adam(learning_rate=1e-3)
w_phys    = 1

# Training step
@tf.function
def train_step():
    with tf.GradientTape(persistent=True) as tape:
        # (a) Data loss
        sol_pred = model_sol(t_train, training=True)
        x_pred   = sol_pred[:,0:1]
        v_pred   = sol_pred[:,1:2]
        loss_data = tf.reduce_mean((x_pred - x_train)**2)
        # enforce initial conditions x(0)=1, v(0)=0
        loss_ic   = tf.square(x_pred[0] - 1.0) + tf.square(v_pred[0] - 0.0)

        # (b) Physics loss
        sol_coll = model_sol(t_colloc_tf, training=True)
        x_coll   = sol_coll[:,0:1]
        v_coll   = sol_coll[:,1:2]

        # dx/dt ≈ v_coll
        with tf.GradientTape() as g1:
            g1.watch(t_colloc_tf)
            xg = model_sol(t_colloc_tf)[:,0:1]
        dx_dt = g1.gradient(xg, t_colloc_tf)
        loss_phys1 = tf.reduce_mean((dx_dt - v_coll)**2)

        # dv/dt ≈ f_pred(x_coll)
        with tf.GradientTape() as g2:
            g2.watch(t_colloc_tf)
            vg = model_sol(t_colloc_tf)[:,1:2]
        dv_dt = g2.gradient(vg, t_colloc_tf)
        f_pred = model_f(x_coll, training=True)
        loss_phys2 = tf.reduce_mean((dv_dt - f_pred)**2)

        loss_total = loss_data + loss_ic + w_phys*(loss_phys1 + loss_phys2)

    grads_sol = tape.gradient(loss_total, model_sol.trainable_variables)
    grads_f   = tape.gradient(loss_total, model_f.trainable_variables)
    optimizer.apply_gradients(zip(grads_sol + grads_f,
                                  model_sol.trainable_variables + model_f.trainable_variables))
    return loss_total, loss_data, (loss_phys1 + loss_phys2)

# Training loop
epochs = 50000
for ep in range(1, epochs+1):
    Lt, Ld, Lp = train_step()
    if ep % 500 == 0:
        print(f"Epoch {ep}: total={Lt.numpy().item():.4e}, "
              f"data={Ld.numpy().item():.4e}, phys={Lp.numpy().item():.4e}")

# Step 3: Plot results

# Learned vs reference solution
t_plot = np.linspace(0, T, 200, dtype=np.float32).reshape(-1,1)
x_pred_plot = model_sol(t_plot)[:,0:1].numpy()
x_ref       = np.interp(t_plot.flatten(), ts, x_hist).reshape(-1,1)

plt.figure(figsize=(6,4))
plt.plot(t_plot, x_pred_plot, 'r-', label='Learned $x_\\theta(t)$')
plt.plot(t_plot, x_ref,       'b--', label='Reference')
plt.scatter(t_data, x_data,   c='k',  s=30, label='Data')
plt.xlabel('t'); plt.ylabel('x(t)')
plt.title('Learned vs Reference Solution')
plt.legend(); plt.grid(True)
plt.show()

# Learned f(x) vs true -x^2
x_vals      = np.linspace(x_data.min(), x_data.max(), 200, dtype=np.float32).reshape(-1,1)
f_pred_vals = model_f(x_vals).numpy()
f_true_vals = -x_vals**2

plt.figure(figsize=(6,4))
plt.plot(x_vals, f_pred_vals, 'r-',  label='Learned $f_\\theta(x)$')
plt.plot(x_vals, f_true_vals, 'b--', label='True $-x^2$')
plt.xlabel('$x$'); plt.ylabel('$f(x)$')
plt.title('Learned vs True Force Field')
plt.legend(); plt.grid(True)
plt.show()
```

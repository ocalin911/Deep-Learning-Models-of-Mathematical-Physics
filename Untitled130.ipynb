**The Heat Kernel on the Unit Circle**

We seek the heat kernel on the unit circle ${\mathbb S}^{1}$, parameterized by the angle $s\in[0,2\pi)$.
By rotational invariance we may take the source at $s_{0}=0$, so we look for a $2\pi$-periodic in $s$ function $K(t;s)$ such that


\begin{align*}
(\partial_{t}-\partial_{s}^{2})\,K(t;s)&=0, \quad  t>0,\ s\in[0,2\pi),\\[2pt]
K(0;s)&=\delta(s)\\
K(t;0)&=K(t;2\pi)\\
  \partial_{s}K(t;0)&=\partial_{s}K(t;2\pi)\\
\partial_{s}^{2}K(t;0)&=\partial_{s}^{2}K(t;2\pi).
\end{align*}
Then the heat kernel from $s_0$ to $s$ will be
$$ K(t; s_0, s) = K(t; 0, s-s_0) = K(0; s-s_0). $$

The analytic formula  of the heat kernel is
$$ K(t; s) =\frac{1}{\sqrt{4\pi t}}
\sum_{k\in\mathbb{Z}} \exp\!\left(-\frac{(s+2\pi k)^{2}}{4t}\right),
\qquad s\in\mathbb{R}$$
The next code plots the profile of th eheat kernel for $t=0.1, 0.3, 0.5, 1.0, 2.0, 4.0$  on the interval $[0, 2\pi]$.


```python
import numpy as np
import matplotlib.pyplot as plt

def K_circle(t, s, K=60):
    """
    Heat kernel on the circle via method of images:
      K(t; s) = (1 / sqrt(4π t)) * sum_{k=-K..K} exp(- (s + 2πk)^2 / (4t))
    Parameters
    ----------
    t : float, > 0
    s : array-like, angles in radians
    K : int, truncation level for the image sum
    """
    s = np.asarray(s)
    k = np.arange(-K, K + 1)[:, None]       # (2K+1) x 1 column
    shifts = s[None, :] + 2*np.pi * k       # broadcast over s
    terms = np.exp(-(shifts**2) / (4.0 * t))
    return (1.0 / np.sqrt(4.0 * np.pi * t)) * np.sum(terms, axis=0)

# Grid on [0, 2π]
s = np.linspace(0.0, 2*np.pi, 1000)

# Times to plot
times = [0.1, 0.3, 0.5, 1.0, 2.0, 4.0]

plt.figure(figsize=(8, 5))
for t in times:
    plt.plot(s, K_circle(t, s, K=60), label=f"t = {t}")
plt.xlabel("s")
plt.ylabel("K(t; s)")
plt.title("Heat kernel on the circle $S^1$ via image sum")
plt.legend()
plt.tight_layout()
plt.show()
```

The next code learns the heat kernel and validates it against the analytic formula.

```python
# Neural Heat Kernel on the Circle S^1 with Hard Embedding + Conservation Loss
# ----------------------------------------------------------------------------
# K_theta(t; s) = δ_ε(s) + t * NN(t, cos s, sin s)
# PDE residual: (∂_t - ∂_s^2) K = 0
# Initial trace: δ_ε(s) using the specified piecewise periodic mollifier
# Conservation: ∫_0^{2π} K(t, s) ds = 1  (enforced at multiple times)

import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, Model
import matplotlib.pyplot as plt

# ----------------------------
# Geometry & exact kernel
# ----------------------------
PI = np.pi
TWOPI = 2.0 * np.pi

def heat_kernel_circle_image(t, s, K=60):
    """
    Closed-form heat kernel on S^1 via method of images (numerically stable):
      K(t; s) = (1 / sqrt(4π t)) * Σ_{k=-K..K} exp(- (s + 2πk)^2 / (4t))
    """
    s = np.asarray(s)
    k = np.arange(-K, K+1)[:, None]            # (2K+1) x 1
    shifts = s[None, :] + TWOPI * k
    terms = np.exp(-(shifts**2) / (4.0 * t))
    return (1.0 / np.sqrt(4.0 * np.pi * t)) * np.sum(terms, axis=0)

# ----------------------------
# Piecewise periodic mollifier δ_ε(s)
# δ_ε(s) = (1/√(4π)ε) * exp(- d(s)^2 / (4ε^2)),
# where d(s) = min(s_mod, 2π - s_mod), s_mod = s mod 2π
# ----------------------------
def piecewise_mollifier_np(s, eps=0.08):
    s = np.asarray(s)
    s_mod = np.mod(s, TWOPI)
    d = np.minimum(s_mod, TWOPI - s_mod)
    return (1.0 / (np.sqrt(4.0*np.pi) * eps)) * np.exp(-(d**2) / (4.0 * eps**2))

def piecewise_mollifier_tf(s, eps=0.08):
    # s: (N,1) tensor
    s_mod = tf.math.floormod(s, TWOPI)  # in [0, 2π)
    d = tf.minimum(s_mod, TWOPI - s_mod)
    return (1.0 / (np.sqrt(4.0*np.pi) * eps)) * tf.exp(-(d**2) / (4.0 * eps**2))

# ----------------------------
# Collocation grids & params
# ----------------------------
T_MAX = 2.0
NT = 128             # time collocation for PDE
NS = 256             # spatial collocation for PDE
EPS0 = 0.15       # mollifier width
SEED = 0

tf.random.set_seed(SEED)
np.random.seed(SEED)

# Uniform grid in (t,s), t>0 for PDE residual
t_grid = np.linspace(1e-3, T_MAX, NT, dtype=np.float32)[:, None]
s_grid = np.linspace(0.0, TWOPI, NS, dtype=np.float32)[None, :]
TT, SS = np.meshgrid(t_grid.flatten(), s_grid.flatten(), indexing='ij')
t_train = TT.reshape(-1, 1).astype(np.float32)   # (NT*NS, 1)
s_train = SS.reshape(-1, 1).astype(np.float32)   # (NT*NS, 1)

# Initial-trace samples (optional)
s0 = np.linspace(0.0, TWOPI, NS, endpoint=False, dtype=np.float32).reshape(-1, 1)
delta_eps_np = piecewise_mollifier_np(s0.flatten(), eps=EPS0).astype(np.float32)

# Conservation-loss samples (time + spatial quadrature)
NTC = 64           # number of time samples for conservation
NSC = 512          # number of spatial quadrature nodes (periodic)
t_cons_np = np.linspace(1e-3, T_MAX, NTC, dtype=np.float32).reshape(-1, 1)
s_cons_np = np.linspace(0.0, TWOPI, NSC, endpoint=False, dtype=np.float32).reshape(-1, 1)
delta_s = TWOPI / NSC

# ----------------------------
# Neural network (periodic inputs)
# ----------------------------
def build_nn(widths=(64, 64), activation='tanh'):
    inp = layers.Input(shape=(3,), name='inp_(t,cos,sin)')
    z = inp
    for w in widths:
        z = layers.Dense(w, activation=activation)(z)
    out = layers.Dense(1, activation=None)(z)
    return Model(inp, out, name='nn_core')

nn = build_nn((64, 64), 'tanh')
optimizer = tf.keras.optimizers.Adam(5e-4)

# K_theta(t; s) = δ_ε(s) + t * nn(t, cos s, sin s)
def K_theta(t, s):
    x = tf.concat([t, tf.cos(s), tf.sin(s)], axis=1)  # (N,3)
    return piecewise_mollifier_tf(s, eps=EPS0) + t * nn(x)

# PDE residual r = ∂_t K - ∂_s^2 K
def residual(t, s):
    with tf.GradientTape(persistent=True) as t2:
        t2.watch([t, s])
        with tf.GradientTape(persistent=True) as t1:
            t1.watch([t, s])
            K = K_theta(t, s)
        K_t = t1.gradient(K, t)
        K_s = t1.gradient(K, s)
        del t1
    K_ss = t2.gradient(K_s, s)
    del t2
    return K_t - K_ss

# ----------------------------
# Losses
# ----------------------------
t_tf = tf.constant(t_train, dtype=tf.float32)
s_tf = tf.constant(s_train, dtype=tf.float32)
s0_tf = tf.constant(s0, dtype=tf.float32)
delta0_tf = tf.constant(delta_eps_np.reshape(-1, 1), dtype=np.float32)

t_cons_tf = tf.constant(t_cons_np, dtype=tf.float32)      # (NTC,1)
s_cons_tf = tf.constant(s_cons_np, dtype=tf.float32)      # (NSC,1)
delta_s_tf = tf.constant(delta_s, dtype=tf.float32)

LAMBDA_IC   = 1e-2   # initial-trace weight (small)
LAMBDA_CONS = 1e-1   # conservation weight

def conservation_loss():
    """
    Enforce ∫_0^{2π} K(t, s) ds = 1 for times in t_cons_tf.
    """
    t_rep = tf.repeat(t_cons_tf, repeats=tf.shape(s_cons_tf)[0], axis=0)  # (NTC*NSC,1)
    s_tile = tf.tile(s_cons_tf, [tf.shape(t_cons_tf)[0], 1])              # (NTC*NSC,1)
    K_vals = K_theta(t_rep, s_tile)                                       # (NTC*NSC,1)
    K_vals = tf.reshape(K_vals, (tf.shape(t_cons_tf)[0], tf.shape(s_cons_tf)[0]))  # (NTC,NSC)
    integrals = tf.reduce_sum(K_vals, axis=1) * delta_s_tf                # (NTC,)
    return tf.reduce_mean(tf.square(integrals - 1.0))

@tf.function
def train_step():
    with tf.GradientTape() as tape:
        # PDE loss on interior points
        r = residual(t_tf, s_tf)
        L_pde = tf.reduce_mean(tf.square(r))

        # Initial-trace penalty at t=0 (optional)
        K_init = K_theta(tf.zeros_like(s0_tf), s0_tf)
        L_ic = tf.reduce_mean(tf.square(K_init - delta0_tf))

        # Conservation loss
        L_cons = conservation_loss()

        L = L_pde + LAMBDA_IC * L_ic + LAMBDA_CONS * L_cons
    grads = tape.gradient(L, nn.trainable_variables)
    optimizer.apply_gradients(zip(grads, nn.trainable_variables))
    return L, L_pde, L_ic, L_cons

# ----------------------------
# Training
# ----------------------------
EPOCHS = 5000
hist_total, hist_pde, hist_ic, hist_cons = [], [], [], []

for ep in range(1, EPOCHS+1):
    L, Lp, Lic, Lc = train_step()
    hist_total.append(float(L.numpy()))
    hist_pde.append(float(Lp.numpy()))
    hist_ic.append(float(Lic.numpy()))
    hist_cons.append(float(Lc.numpy()))
    if ep % 250 == 0:
        print(f"Epoch {ep:4d} | total {hist_total[-1]:.3e} | PDE {hist_pde[-1]:.3e} | IC {hist_ic[-1]:.3e} | CONS {hist_cons[-1]:.3e}")

# ----------------------------
# Evaluation & plots
# ----------------------------
# Grid for plotting
s_plot = np.linspace(0.0, TWOPI, 1000, dtype=np.float32).reshape(-1, 1)

def K_theta_np(t, s_vec):
    t_arr = np.full_like(s_vec, t, dtype=np.float32)
    out = K_theta(tf.constant(t_arr), tf.constant(s_vec)).numpy().flatten()
    return out

times = [0.1, 0.3, 1.0]

# 1) Loss curves
plt.figure(figsize=(7,4))
plt.plot(hist_total, label='total')
plt.plot(hist_pde, label='PDE')
plt.plot(hist_ic, label='IC (t=0)', alpha=0.8)
plt.plot(hist_cons, label='Conservation', alpha=0.8)
plt.yscale('log')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.title('Training losses')
plt.legend()
plt.tight_layout()
plt.show()

# 2) Compare NN kernel vs. analytic kernel
plt.figure(figsize=(9,5))
for t in times:
    # NN
    y_nn = K_theta_np(t, s_plot)
    # Analytic (image-sum; equivalent to the cosh-series)
    y_ex = heat_kernel_circle_image(t, s_plot.flatten(), K=100)
    plt.plot(s_plot, y_ex, label=f'Exact t={t}')
    plt.plot(s_plot, y_nn, '--', label=f'NN t={t}')
plt.xlabel('s')
plt.ylabel(r'$K(t;\,s - s_0)$ with $s_0=0$')
plt.title('Heat kernel on $S^1$: NN vs. closed form (piecewise mollifier)')
plt.legend(ncol=2)
plt.tight_layout()
plt.show()
```

# **Second Order ODE Discovery**

We aim to recover the second-order ODE
$$
u''(t) \;=\; 3u'(t) - 2u(t),
$$
from a finite set of measurements. For validation purposes, we use the exact analytical solution
$$
u_{\mathrm{exact}}(t) \;=\; e^t - e^{2t}.
$$

In the next code we train the network for 3,000 epochs. This is not enough, since the exact solution $u_{\rm exact}(t)$ is not recovered well.

```python
# ============================================================
# Two-network discovery for a 2nd-order ODE: u''(t) = f(u,u')
# Networks: u_theta(t) and f_theta(u,u')
# Data: 10 samples in [-2, 0.5] from u(t) = e^t - e^{2t}
# Plots:
#  - u_theta(t) vs data (and exact curve)
#  - Heatmap of f_theta(u,u')
#  - Loss vs epochs
# ============================================================

import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, models
import matplotlib.pyplot as plt

# ----------------------------
# 1) Dataset (t, u, u') — decimals
# ----------------------------
t_data = np.array([-2.000, -1.722, -1.444, -1.167, -0.889,
                   -0.611, -0.333, -0.056,  0.222,  0.500], dtype=np.float32).reshape(-1,1)

u_data = np.array([ 0.11702,  0.14675,  0.18024,  0.21443,  0.24210,
                    0.24817,  0.20311,  0.05112, -0.31077, -1.06956], dtype=np.float32).reshape(-1,1)

up_data = np.array([ 0.09870,  0.11482,  0.12460,  0.11746,  0.07309,
                   -0.04640, -0.31030, -0.84372, -1.87040, -3.78784], dtype=np.float32).reshape(-1,1)

# Time normalization to stabilize training
t_min, t_max = float(t_data.min()), float(t_data.max())
def t_norm(t):
    return (t - t_min) / (t_max - t_min + 1e-12)

# ----------------------------
# 2) Build networks
# ----------------------------
def build_u_net(hidden=3, width=64, act="swish"):
    inp = tf.keras.Input(shape=(1,))
    z = inp
    for _ in range(hidden):
        z = layers.Dense(width, activation=act)(z)
    out = layers.Dense(1)(z)
    return models.Model(inp, out, name="u_net")

def build_f_net(hidden=3, width=64, act="swish"):
    inp = tf.keras.Input(shape=(2,))  # (u, u')
    z = inp
    for _ in range(hidden):
        z = layers.Dense(width, activation=act)(z)
    out = layers.Dense(1)(z)
    return models.Model(inp, out, name="f_net")

u_net = build_u_net(hidden=3, width=64, act="swish")
f_net = build_f_net(hidden=3, width=64, act="swish")

@tf.function
def u_theta(t):
    """Trial solution network; t is unnormalized, we normalize inside."""
    tn = (t - t_min) / (t_max - t_min + 1e-12)
    return u_net(tn)

# ----------------------------
# 3) Loss (data + ODE with extra collocation points)
# ----------------------------
lambda_data = tf.constant(1.0, tf.float32)
lambda_ode  = tf.constant(1.0, tf.float32)

# Collocation points for ODE consistency (optional but helpful)
M = 64
tau = np.linspace(t_min, t_max, M, dtype=np.float32).reshape(-1,1)

@tf.function
def compute_losses(t_batch, u_batch, up_batch, tau_batch):
    t_batch = tf.convert_to_tensor(t_batch, dtype=tf.float32)
    u_batch = tf.convert_to_tensor(u_batch, dtype=tf.float32)
    up_batch = tf.convert_to_tensor(up_batch, dtype=tf.float32)
    tau_batch = tf.convert_to_tensor(tau_batch, dtype=tf.float32)

    # Data fit: u and u'
    with tf.GradientTape() as tape_u1:
        tape_u1.watch(t_batch)
        u_pred = u_theta(t_batch)
    up_pred = tape_u1.gradient(u_pred, t_batch) / (t_max - t_min + 1e-12)

    L_u  = tf.reduce_mean(tf.square(u_pred - u_batch))
    L_up = tf.reduce_mean(tf.square(up_pred - up_batch))

    # ODE term at data points
    with tf.GradientTape(persistent=True) as tape_ode_data: # Use persistent=True for multiple gradients
        tape_ode_data.watch(t_batch)
        u_for_ode_data = u_theta(t_batch)
        up_for_ode_data = tape_ode_data.gradient(u_for_ode_data, t_batch) / (t_max - t_min + 1e-12)
    upp_at_data = tape_ode_data.gradient(up_for_ode_data, t_batch) / (t_max - t_min + 1e-12)
    del tape_ode_data # Delete the tape to release memory

    f_at_data = f_net(tf.concat([u_batch, up_batch], axis=1))
    L_ode_data = tf.reduce_mean(tf.square(upp_at_data - f_at_data))

    # ODE term at extra collocation points tau
    with tf.GradientTape(persistent=True) as tape_ode_tau: # Use persistent=True for multiple gradients
        tape_ode_tau.watch(tau_batch)
        u_tau = u_theta(tau_batch)
        up_tau = tape_ode_tau.gradient(u_tau, tau_batch) / (t_max - t_min + 1e-12)
    upp_tau = tape_ode_tau.gradient(up_tau, tau_batch) / (t_max - t_min + 1e-12)
    del tape_ode_tau # Delete the tape to release memory


    f_at_tau = f_net(tf.concat([u_tau, up_tau], axis=1))
    L_ode_tau = tf.reduce_mean(tf.square(upp_tau - f_at_tau))

    L_data = L_u + L_up
    L_ODE  = L_ode_data + L_ode_tau
    L_total = lambda_data * L_data + lambda_ode * L_ODE
    return L_total, L_data, L_ODE

# ----------------------------
# 4) Training
# ----------------------------
# Modest schedule; feel free to increase EPOCHS for tighter fits
boundaries = [1000, 2000]
values = [1e-3, 5e-4, 1e-4]
lr_schedule = tf.keras.optimizers.schedules.PiecewiseConstantDecay(boundaries, values)
optimizer = tf.keras.optimizers.Adam(learning_rate=lr_schedule)

EPOCHS = 3000
hist_tot, hist_data, hist_ode = [], [], []

for ep in range(1, EPOCHS+1):
    with tf.GradientTape() as tape:
        L_total, L_data, L_ODE = compute_losses(t_data, u_data, up_data, tau)
    vars_all = u_net.trainable_variables + f_net.trainable_variables
    grads = tape.gradient(L_total, vars_all)
    optimizer.apply_gradients(zip(grads, vars_all))

    hist_tot.append(float(L_total.numpy()))
    hist_data.append(float(L_data.numpy()))
    hist_ode.append(float(L_ODE.numpy()))

    if ep % 500 == 0:
        print(f"Epoch {ep:4d} | L_tot={hist_tot[-1]:.3e}  L_data={hist_data[-1]:.3e}  L_ODE={hist_ode[-1]:.3e}")

# ----------------------------
# 5) Evaluation & Plots
# ----------------------------
t_eval = np.linspace(t_min, t_max, 400, dtype=np.float32).reshape(-1,1)
u_pred_eval = u_theta(tf.convert_to_tensor(t_eval)).numpy()

# Exact curve for visual reference
u_exact = np.exp(t_eval) - np.exp(2*t_eval)

# (A) u(t): NN vs data and exact
plt.figure(figsize=(6.4,4.3))
plt.plot(t_eval, u_pred_eval, 'r-', label=r'NN $u_\theta(t)$', linewidth=2)
plt.plot(t_data, u_data, 'ko', label='data', markersize=5)
plt.plot(t_eval, u_exact, 'b--', label=r'exact $u(t)$', linewidth=1.5)
plt.xlabel('t'); plt.ylabel('u(t)')
plt.title('Learned solution $u_\\theta(t)$ vs data and exact')
plt.grid(True, ls='--', alpha=0.5); plt.legend()
plt.tight_layout(); plt.show()

# (B) Heatmap of f(u,u')
u_min, u_max = float(u_data.min()), float(u_data.max())
up_min, up_max = float(up_data.min()), float(up_data.max())
pad_u, pad_up = 0.05, 0.05
u_grid = np.linspace(u_min - pad_u, u_max + pad_u, 160, dtype=np.float32)
up_grid = np.linspace(up_min - pad_up, up_max + pad_up, 160, dtype=np.float32)
UU, UUP = np.meshgrid(u_grid, up_grid, indexing='xy')
f_grid = f_net(np.stack([UU.ravel(), UUP.ravel()], axis=1)).numpy().reshape(UU.shape)

plt.figure(figsize=(6.4,4.8))
im = plt.imshow(f_grid.T, origin='lower',
                extent=[u_grid[0], u_grid[-1], up_grid[0], up_grid[-1]],
                aspect='auto')
plt.colorbar(im, label=r'$f_\theta(u,u^\prime)$')
plt.scatter(u_data, up_data, c='w', edgecolors='k', s=40, label='data (u,u\')')
plt.xlabel('u'); plt.ylabel("u'")
plt.title(r'Learned $f_\theta(u,u^\prime)$ (heatmap)')
plt.legend(loc='upper left')
plt.tight_layout(); plt.show()

# (C) Loss curves
plt.figure(figsize=(6.4,4.3))
plt.semilogy(hist_tot, 'm-', label='L_total')
plt.semilogy(hist_data, 'g--', label='L_data')
plt.semilogy(hist_ode, 'c--', label='L_ODE')
plt.xlabel('epoch'); plt.ylabel('loss')
plt.title('Training losses')
plt.grid(True, ls='--', alpha=0.5); plt.legend()
plt.tight_layout(); plt.show()
```

The next code uses 15,000 epochs of training and a well-refined learning rate schedule. The exact solution $u_{\rm exact}(t)$ is recovered perfectly.

```python
# ============================================================
# Two-network discovery for a 2nd-order ODE: u''(t) = f(u,u')
# Networks: residual NN r_theta(t) inside hard-embedded u_theta(t)
# Data: 10 samples in [-2, 0.5] from u(t) = e^t - e^{2t}
# Plots:
#  - u_theta(t) vs data (and exact curve)
#  - Heatmap of f_theta(u,u')
#  - Loss vs epochs
# ============================================================

import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, models
import matplotlib.pyplot as plt

# ----------------------------
# 1) Dataset (t, u, u') — decimals
# ----------------------------
t_data = np.array([-2.000, -1.722, -1.444, -1.167, -0.889,
                   -0.611, -0.333, -0.056,  0.222,  0.500], dtype=np.float32).reshape(-1,1)

u_data = np.array([ 0.11702,  0.14675,  0.18024,  0.21443,  0.24210,
                    0.24817,  0.20311,  0.05112, -0.31077, -1.06956], dtype=np.float32).reshape(-1,1)

up_data = np.array([ 0.09870,  0.11482,  0.12460,  0.11746,  0.07309,
                   -0.04640, -0.31030, -0.84372, -1.87040, -3.78784], dtype=np.float32).reshape(-1,1)

# Normalize time for stable NN input (only for the NN argument)
t_min, t_max = float(t_data.min()), float(t_data.max())
TSPAN = (t_max - t_min + 1e-12)

# Embedding anchor from the table (t0 = -2)
t0  = tf.constant(-2.0, dtype=tf.float32)
u0  = tf.constant(0.11702, dtype=tf.float32)   # u(-2)
up0 = tf.constant(0.09870, dtype=tf.float32)   # u'(-2)

# ----------------------------
# 2) Build networks
# ----------------------------
def build_residual_u_net(hidden=3, width=64, act="swish"):
    """Residual network r_theta(t) used in the hard embedding."""
    inp = tf.keras.Input(shape=(1,))  # will feed normalized time
    z = inp
    for _ in range(hidden):
        z = layers.Dense(width, activation=act)(z)
    out = layers.Dense(1)(z)  # scalar residual
    return models.Model(inp, out, name="r_u_net")

def build_f_net(hidden=3, width=64, act="swish"):
    """Right-hand side f_theta(u, u')."""
    inp = tf.keras.Input(shape=(2,))  # (u, u')
    z = inp
    for _ in range(hidden):
        z = layers.Dense(width, activation=act)(z)
    out = layers.Dense(1)(z)
    return models.Model(inp, out, name="f_net")

r_u_net = build_residual_u_net(hidden=3, width=64, act="swish")
f_net   = build_f_net(hidden=3, width=64, act="swish")

@tf.function
def u_theta(t):
    """
    Hard-embedded trial solution:
      u_theta(t) = u0 + up0*(t - t0) + (t - t0)^2 * r_u_net( (t - t_min)/TSPAN )
    """
    t = tf.cast(t, tf.float32)
    tn = (t - t_min) / TSPAN
    r  = r_u_net(tn)
    dt = (t - t0)
    return u0 + up0 * dt + tf.square(dt) * r

# ----------------------------
# 3) Loss (data + ODE with extra collocation points)
# ----------------------------
lambda_data = tf.constant(1.0, tf.float32)
lambda_ode  = tf.constant(1.0, tf.float32)

# Collocation points for ODE consistency
M = 64
tau = np.linspace(t_min, t_max, M, dtype=np.float32).reshape(-1,1)

@tf.function
def compute_losses(t_batch, u_batch, up_batch, tau_batch):
    t_batch  = tf.convert_to_tensor(t_batch,  dtype=tf.float32)
    u_batch  = tf.convert_to_tensor(u_batch,  dtype=tf.float32)
    up_batch = tf.convert_to_tensor(up_batch, dtype=tf.float32)
    tau_batch= tf.convert_to_tensor(tau_batch, dtype=tf.float32)

    # ----- Data losses (u, u')
    with tf.GradientTape(persistent=True) as tape2:
        tape2.watch(t_batch)
        with tf.GradientTape() as tape1:
            tape1.watch(t_batch)
            u_pred = u_theta(t_batch)          # u(t)
        up_pred = tape1.gradient(u_pred, t_batch)     # u'(t) (no manual /TSPAN needed)
    upp_at_data = tape2.gradient(up_pred, t_batch)    # u''(t)
    del tape1, tape2

    L_u  = tf.reduce_mean(tf.square(u_pred - u_batch))
    L_up = tf.reduce_mean(tf.square(up_pred - up_batch))

    # ODE at data points
    f_at_data = f_net(tf.concat([u_batch, up_batch], axis=1))
    L_ode_data = tf.reduce_mean(tf.square(upp_at_data - f_at_data))

    # ----- ODE at extra collocation points tau
    with tf.GradientTape(persistent=True) as tapeT2:
        tapeT2.watch(tau_batch)
        with tf.GradientTape() as tapeT1:
            tapeT1.watch(tau_batch)
            u_tau = u_theta(tau_batch)         # u(tau)
        up_tau = tapeT1.gradient(u_tau, tau_batch)    # u'(tau)
    upp_tau = tapeT2.gradient(up_tau, tau_batch)      # u''(tau)
    del tapeT1, tapeT2

    f_at_tau = f_net(tf.concat([u_tau, up_tau], axis=1))
    L_ode_tau = tf.reduce_mean(tf.square(upp_tau - f_at_tau))

    L_data  = L_u + L_up
    L_ODE   = L_ode_data + L_ode_tau
    L_total = lambda_data * L_data + lambda_ode * L_ODE
    return L_total, L_data, L_ODE

# ----------------------------
# 4) Training (piecewise-constant LR)
# ----------------------------
boundaries = [2000, 5000, 8000, 10000,]
values = [1e-3, 6e-4, 1e-4, 5e-5, 1e-5]
lr_schedule = tf.keras.optimizers.schedules.PiecewiseConstantDecay(boundaries, values)
optimizer = tf.keras.optimizers.Adam(learning_rate=lr_schedule)

EPOCHS = 15000
hist_tot, hist_data, hist_ode = [], [], []

for ep in range(1, EPOCHS+1):
    with tf.GradientTape() as tape:
        L_total, L_data, L_ODE = compute_losses(t_data, u_data, up_data, tau)
    vars_all = r_u_net.trainable_variables + f_net.trainable_variables
    grads = tape.gradient(L_total, vars_all)
    optimizer.apply_gradients(zip(grads, vars_all))

    hist_tot.append(float(L_total.numpy()))
    hist_data.append(float(L_data.numpy()))
    hist_ode.append(float(L_ODE.numpy()))

    if ep % 500 == 0:
        print(f"Epoch {ep:4d} | L_tot={hist_tot[-1]:.3e}  L_data={hist_data[-1]:.3e}  L_ODE={hist_ode[-1]:.3e}")

# ----------------------------
# 5) Evaluation & Plots
# ----------------------------
t_eval = np.linspace(t_min, t_max, 400, dtype=np.float32).reshape(-1,1)
u_pred_eval = u_theta(tf.convert_to_tensor(t_eval)).numpy()

# Exact curve for reference
u_exact = np.exp(t_eval) - np.exp(2*t_eval)

# (A) u(t): NN vs data and exact
plt.figure(figsize=(6.4,4.3))
plt.plot(t_eval, u_pred_eval, label=r'NN $u_\theta(t)$', linewidth=2)
plt.plot(t_data, u_data, 'o', label='data', markersize=5)
plt.plot(t_eval, u_exact, '--', label=r'exact $u(t)$', linewidth=1.5)
plt.xlabel('t'); plt.ylabel('u(t)')
plt.title('Learned solution $u_\\theta(t)$ vs data and exact')
plt.grid(True, ls='--', alpha=0.5); plt.legend()
plt.tight_layout(); plt.show()

# (B) Heatmap of f(u,u')
u_min, u_max = float(u_data.min()), float(u_data.max())
up_min, up_max = float(up_data.min()), float(up_data.max())
pad_u, pad_up = 0.05, 0.05
u_grid  = np.linspace(u_min - pad_u,  u_max + pad_u,  160, dtype=np.float32)
up_grid = np.linspace(up_min - pad_up, up_max + pad_up, 160, dtype=np.float32)
UU, UUP = np.meshgrid(u_grid, up_grid, indexing='xy')
f_grid = f_net(np.stack([UU.ravel(), UUP.ravel()], axis=1)).numpy().reshape(UU.shape)

plt.figure(figsize=(6.4,4.8))
im = plt.imshow(f_grid.T, origin='lower',
                extent=[u_grid[0], u_grid[-1], up_grid[0], up_grid[-1]],
                aspect='auto')
plt.colorbar(im, label=r'$f_\theta(u,u^\prime)$')
plt.scatter(u_data, up_data, c='w', edgecolors='k', s=40, label='data (u,u\')')
plt.xlabel('u'); plt.ylabel("u'")
plt.title(r'Learned $f_\theta(u,u^\prime)$ (heatmap)')
plt.legend(loc='upper left')
plt.tight_layout(); plt.show()

# (C) Loss curves
plt.figure(figsize=(6.4,4.3))
plt.semilogy(hist_tot,  label='L_total')
plt.semilogy(hist_data, label='L_data')
plt.semilogy(hist_ode,  label='L_ODE')
plt.xlabel('epoch'); plt.ylabel('loss')
plt.title('Training losses')
plt.grid(True, ls='--', alpha=0.5); plt.legend()
plt.tight_layout(); plt.show()
```

We run several linear approximations of the surface $(u, u') \rightarrow f_{\theta}(u, u')$:

```python
# === Linear approximations of f_theta(u,u') at multiple times ================
# Run this AFTER training. Uses the trained u_theta (for u0, up0) and f_net.

def linearize_f_at_t(t_scalar):
    """Return (u0, up0, f0, a00, a10, a01) for the linearization at time t."""
    # Evaluate (u0, up0) from u_theta
    t_tf = tf.constant([[t_scalar]], dtype=tf.float32)
    with tf.GradientTape() as tape_t:
        tape_t.watch(t_tf)
        u0_tf = u_theta(t_tf)                      # shape (1,1)
    up0_tf = tape_t.gradient(u0_tf, t_tf)          # shape (1,1)

    u0  = float(u0_tf.numpy()[0,0])
    up0 = float(up0_tf.numpy()[0,0])

    # Autodiff of f_net wrt inputs x=[u, u'] at x0=(u0, up0)
    x0 = tf.constant([u0, up0], dtype=tf.float32)
    with tf.GradientTape() as tape_x:
        tape_x.watch(x0)
        f0_tf = f_net(x0[None, :])[0,0]            # scalar
    g = tape_x.gradient(f0_tf, x0)                 # [∂f/∂u, ∂f/∂u']

    f0  = float(f0_tf.numpy())
    gu  = float(g[0].numpy())
    gup = float(g[1].numpy())

    # Non-centered coefficients
    a10, a01 = gu, gup
    a00 = f0 - a10*u0 - a01*up0

    return dict(t=float(t_scalar), u0=u0, up0=up0, f0=f0, a00=a00, a10=a10, a01=a01)

# Times of interest
times = [-2.0, -1.75, -1.5,  -1.167, -1.0, -0.7, -0.5, -0.25, 0.0, 0.25, 0.5]

# Compute and print
coeffs_by_t = {}
print("Linear approximations  f(u,u') ≈ a00 + a10*u + a01*u'  (non-centered)\n")
for t in times:
    res = linearize_f_at_t(t)
    coeffs_by_t[t] = res
    print(f"t = {t:>4.1f}:")
    print(f"  center (u0, u0') = ({res['u0']:.6g}, {res['up0']:.6g})")
    print(f"  f0 = {res['f0']:.6g}")
    print(f"  a00 = {res['a00']:.6g},  a10 = {res['a10']:.6g},  a01 = {res['a01']:.6g}\n")

# 'coeffs_by_t' now holds the coefficients for later use, e.g.:
# coeffs_by_t[-1.0]['a00'], coeffs_by_t[-1.0]['a10'], coeffs_by_t[-1.0]['a01']
```

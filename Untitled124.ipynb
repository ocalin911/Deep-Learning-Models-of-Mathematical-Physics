# **Mixed Value Problem for the Heat Operator**

We solve the following problem:\begin{cases}
\partial_t u - \partial_x^2 u  = 0, & \text{in } (0, \pi)\times (0, T), \\[3pt]
u(0,x) = \sin x, & x \in (0, \pi), \\[3pt]
u(t, 0) = 0,\\
u(t, L) = 0,\quad t\geq 0,
\end{cases}

The first code trains the network $u_{\theta}(t, x)$ for 3,000 epochs. Perfect match with the exact solution $u_{\rm exact}(t, x) = e^{-t} \sin x$ is obtained. The loss starts oscillating. Decreasing the learning rate usually fixes the problem.

```python
# heat_pinn_dirichlet_sin.py
# Requirements: tensorflow>=2.10, numpy, matplotlib

import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

# -----------------------
# Problem setup
# -----------------------
pi = np.pi
T  = 1.0
X_MIN, X_MAX = 0.0, pi

# Exact solution u(t,x) = e^{-t} sin x
def u_exact(t, x):
    return np.exp(-t) * np.sin(x)

# -----------------------
# Neural network (N_theta)
# -----------------------
def make_model(width=64, depth=4, activation="tanh"):
    inputs = keras.Input(shape=(2,), dtype=tf.float32)  # (t, x)
    z = inputs
    for _ in range(depth):
        z = layers.Dense(width, activation=activation)(z)
    z = layers.Dense(1, activation=None)(z)
    return keras.Model(inputs, z, name="core_net")

# Hard embedding: u_theta(t,x) = sin x + t * x*(pi-x) * N_theta(t,x)
@tf.function
def u_theta(model, t, x):
    tx = tf.concat([t, x], axis=1)  # shape (N,2)
    N = model(tx)                    # (N,1)
    return tf.sin(x) + t * x * (pi - x) * N

# PDE residual R = u_t - u_xx
@tf.function
def pde_residual(model, t, x):
    with tf.GradientTape(persistent=True, watch_accessed_variables=False) as tape2:
        tape2.watch([t, x])
        with tf.GradientTape(persistent=True, watch_accessed_variables=False) as tape1:
            tape1.watch([t, x])
            u = u_theta(model, t, x)            # (N,1)
        u_t = tape1.gradient(u, t)               # (N,1)
        u_x = tape1.gradient(u, x)               # (N,1)
    u_xx = tape2.gradient(u_x, x)                # (N,1)
    del tape1, tape2
    return u_t - u_xx

# -----------------------
# Collocation sampling
# -----------------------
def sample_collocation(Nt=128, Nx=256):
    # Uniform random in (0,T] x (0,pi)
    t = tf.random.uniform((Nt*Nx, 1), minval=1e-6, maxval=T, dtype=tf.float32)
    x = tf.random.uniform((Nt*Nx, 1), minval=X_MIN, maxval=X_MAX, dtype=tf.float32)
    return t, x

# -----------------------
# Training setup
# -----------------------
model = make_model(width=64, depth=4, activation="tanh")
optimizer = keras.optimizers.Adam(1e-3)

@tf.function
def train_step(model, batch_size_t=64, batch_size_x=128):
    t, x = sample_collocation(batch_size_t, batch_size_x)  # ~8192 points
    with tf.GradientTape() as tape:
        R = pde_residual(model, t, x)
        loss = tf.reduce_mean(tf.square(R))
    grads = tape.gradient(loss, model.trainable_variables)
    optimizer.apply_gradients(zip(grads, model.trainable_variables))
    return loss

# -----------------------
# Train
# -----------------------
EPOCHS = 3000
print_interval = 200
hist = []
for ep in range(1, EPOCHS+1):
    loss = train_step(model)
    hist.append(float(loss.numpy()))
    if ep % print_interval == 0 or ep == 1:
        print(f"Epoch {ep:4d} | PDE loss: {hist[-1]:.6e}")

# -----------------------
# Plot: training loss
# -----------------------
plt.figure()
plt.plot(hist)
plt.yscale("log")
plt.xlabel("Epoch")
plt.ylabel("PDE residual loss")
plt.title("Training loss")
plt.tight_layout()
plt.show()

# -----------------------
# Compare NN vs Exact at requested times
# -----------------------
times = [0.1, 0.2, 0.5, 0.7, 1.0]
x_plot = np.linspace(X_MIN, X_MAX, 400).astype(np.float32).reshape(-1,1)

plt.figure(figsize=(8,6))
for tval in times:
    t_arr = np.full_like(x_plot, tval, dtype=np.float32)
    t_tf = tf.constant(t_arr, dtype=tf.float32)
    x_tf = tf.constant(x_plot, dtype=tf.float32)

    u_nn = u_theta(model, t_tf, x_tf).numpy().reshape(-1)
    u_ex = u_exact(tval, x_plot.reshape(-1))

    # Plot both on same axes
    plt.plot(x_plot.reshape(-1), u_ex,  linestyle="--", label=f"exact t={tval}")
    plt.plot(x_plot.reshape(-1), u_nn,  linestyle="-",  label=f"NN t={tval}")

plt.xlabel("x")
plt.ylabel("u(t,x)")
plt.title("Heat equation: exact vs NN (hard-embedded IC & BC)")
plt.legend(ncol=2, fontsize=9)
plt.tight_layout()
plt.show()
```

**The Simple Pendulum**

***(i) Action-minimization:***

 We minimize the discretized action
  $$
    S[\phi_\theta]
    \;=\;
    \int_{0}^{T}\!L\bigl(\phi_\theta(t),\dot\phi_\theta(t)\bigr)\,dt,
  $$
  where $L=\tfrac12\dot\phi^2+\kappa\cos\phi$ is the Lagrangian.

  We assume the initial conditions $\phi(0)=1$ and $\dot \phi(0) =0$. We compare the solution with a reference solution obtained by the method Runge--Kutta 4.

```python
import numpy as np
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import matplotlib.pyplot as plt

# ─── 1) PINN TRAINING ─────────────────────────────────────
np.random.seed(0)
tf.random.set_seed(0)

# Discretize t ∈ [0,1]
N = 6000
t_vals = np.linspace(0, 1.0, N + 1).reshape(-1,1).astype(np.float32)
dt = t_vals[1,0] - t_vals[0,0]
t_tf = tf.constant(t_vals)

# Build NN model for x(t)
def build_model():
    inp = keras.Input(shape=(1,))
    h = layers.Dense(32, activation='tanh')(inp)
    h = layers.Dense(32, activation='tanh')(h)
    h = layers.Dense(32, activation='tanh')(h)
    h = layers.Dense(32, activation='tanh')(h)
    out = layers.Dense(1)(h)
    return keras.Model(inputs=inp, outputs=out)

model = build_model()
optimizer = keras.optimizers.Adam(1e-3)

# Compute x and x'
def x_and_dx(t):
    with tf.GradientTape() as tape:
        tape.watch(t)
        x_t = model(t)
    dx_dt = tape.gradient(x_t, t)
    return x_t, dx_dt

# Lagrangian loss via trapezoidal rule
def lagrangian_loss(x_t, dx_dt):
    integrand = 0.5*dx_dt**2 + tf.cos(x_t)
    f0, f1 = integrand[:-1], integrand[1:]
    return tf.reduce_sum(0.5*(f0+f1)) * dt

# Boundary loss at t=0
def boundary_loss(x_t, dx_dt):
    x0, dx0 = x_t[0], dx_dt[0]
    return tf.square(x0 - 1.0) + tf.square(dx0 - 0.0)

@tf.function
def compute_loss():
    x_t, dx_dt = x_and_dx(t_tf)
    return lagrangian_loss(x_t, dx_dt) + 1000*boundary_loss(x_t, dx_dt)

@tf.function
def train_step():
    with tf.GradientTape() as tape:
        loss = compute_loss()
    grads = tape.gradient(loss, model.trainable_variables)
    optimizer.apply_gradients(zip(grads, model.trainable_variables))
    return loss

# Train
epochs = 500
loss_history = []
for epoch in range(epochs):
    L = train_step().numpy().item()
    loss_history.append(L)
    if epoch % 50 == 0:
        print(f"Epoch {epoch}: Loss = {L:.6f}")

# PINN solution
x_pred_tf, _ = x_and_dx(t_tf)
x_pinn = x_pred_tf.numpy().flatten()

# ─── 2) RK4 NUMERICAL SOLUTION ────────────────────────────
dt_num = dt    # match PINN grid
N_num  = N
t_num  = t_vals.flatten()
x_num  = np.zeros(N+1)
v_num  = np.zeros(N+1)
x_num[0], v_num[0] = 1.0, 0.0

def f(state):
    xi, vi = state
    return np.array([vi, -np.sin(xi)])

for i in range(N_num):
    state = np.array([x_num[i], v_num[i]])
    k1 = dt_num * f(state)
    k2 = dt_num * f(state + 0.5*k1)
    k3 = dt_num * f(state + 0.5*k2)
    k4 = dt_num * f(state + k3)
    delta = (k1 + 2*k2 + 2*k3 + k4)/6.0
    x_num[i+1] = x_num[i] + delta[0]
    v_num[i+1] = v_num[i] + delta[1]

# ─── 3) PLOT OVERLAY AND LOSS ────────────────────────────
plt.figure(figsize=(12,5))

# (a) overlay PINN & RK4
plt.subplot(1,2,1)
plt.plot(t_vals, x_pinn, '-', color='blue', lw=2, label='PINN')
plt.plot(t_num, x_num,   '--', color='orange', lw=1, label='RK4')
plt.scatter([0],[1], color='red', label='x(0)=1')
plt.title("NN vs. RK4 Solution of $x''+\sin x=0$")
plt.xlabel('t'); plt.ylabel('x(t)')
plt.legend(); plt.grid(True)

# (b) training loss
plt.subplot(1,2,2)
plt.plot(loss_history, '-g', lw=2)
plt.yscale('log')
plt.title("PINN Loss vs. Epoch")
plt.xlabel('Epoch'); plt.ylabel('Loss')
plt.grid(True)

plt.tight_layout()
plt.show()
```

We note that the NN solution diverges from the RK-classical solution. To fix this problem we introduce some data loss using two data points, at t=0.5 and t=1.0. Pinning down the solution at these two points makes it to match the classical solution.

```python
import numpy as np
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import matplotlib.pyplot as plt

# ─── 1) Setup & RK4 Reference ────────────────────────────
np.random.seed(0)
tf.random.set_seed(0)

# Domain [0,1]
T = 1.0
dt_num = 0.01
N_num  = int(T / dt_num)
t_num  = np.linspace(0, 1, N_num+1)
x_num  = np.zeros(N_num+1)
v_num  = np.zeros(N_num+1)
x_num[0], v_num[0] = 1.0, 0.0

def f_rhs(state):
    xi, vi = state
    return np.array([vi, -np.sin(xi)])

# integrate with RK4
for i in range(N_num):
    s = np.array([x_num[i], v_num[i]])
    k1 = dt_num*f_rhs(s)
    k2 = dt_num*f_rhs(s + 0.5*k1)
    k3 = dt_num*f_rhs(s + 0.5*k2)
    k4 = dt_num*f_rhs(s + k3)
    d = (k1 + 2*k2 + 2*k3 + k4)/6
    x_num[i+1] = x_num[i] + d[0]
    v_num[i+1] = v_num[i] + d[1]

# pick 4 interior data points at t = 0.5, 1
fracs = [ 0.5,  1.0]
idxs  = [int(frac*N_num) for frac in fracs]
t_data = t_num[idxs].reshape(-1,1).astype(np.float32)
x_data = x_num[idxs].reshape(-1,1).astype(np.float32)
t_data_tf = tf.constant(t_data)

# ─── 2) PINN Setup ───────────────────────────────────────
N_pinn = 10000
t_pinn = np.linspace(0, 1, N_pinn+1).reshape(-1,1).astype(np.float32)
t_tf   = tf.constant(t_pinn)
dt_pinn = t_pinn[1,0] - t_pinn[0,0]

def build_pinn():
    inp = keras.Input(shape=(1,))
    h = layers.Dense(32, activation='tanh')(inp)
    h = layers.Dense(32, activation='tanh')(h)
    h = layers.Dense(32, activation='tanh')(h)
    h = layers.Dense(32, activation='tanh')(h)
    out = layers.Dense(1)(h)
    return keras.Model(inputs=inp, outputs=out)

model = build_pinn()
opt   = keras.optimizers.Adam(1e-3)

def x_and_derivatives(t):
    with tf.GradientTape(persistent=True) as tape:
        tape.watch(t)
        x = model(t)
        dx = tape.gradient(x, t)
    ddx = tape.gradient(dx, t)
    del tape
    return x, dx, ddx

# ─── 3) Loss components ──────────────────────────────────
def lagrangian_loss(x_t, dx_dt):
    I = 0.5*dx_dt**2 + tf.cos(x_t)
    f0, f1 = I[:-1], I[1:]
    return tf.reduce_sum(0.5*(f0+f1)) * dt_pinn

def boundary_loss(x_t, dx_dt):
    x0, dx0 = x_t[0], dx_dt[0]
    return tf.square(x0 - 1.0) + tf.square(dx0 - 0.0)

def data_loss(x_t):
    # gather at the 4 data indices (rescaled to PINN grid)
    # find nearest Pinn indices
    idxs_pinn = [int(frac*N_pinn) for frac in fracs]
    x_pred_data = tf.gather(x_t, idxs_pinn)
    x_true_data = tf.constant(x_data, dtype=tf.float32)
    return tf.reduce_mean((x_pred_data - x_true_data)**2)

@tf.function
def compute_loss():
    x_t, dx_t, _ = x_and_derivatives(t_tf)
    L1 = lagrangian_loss(x_t, dx_t)
    L2 = 1000.0 * boundary_loss(x_t, dx_t)
    L3 = 1000.0 * data_loss(x_t)
    return L1 + L2 + L3

@tf.function
def train_step():
    with tf.GradientTape() as tape:
        loss = compute_loss()
    grads = tape.gradient(loss, model.trainable_variables)
    opt.apply_gradients(zip(grads, model.trainable_variables))
    return loss

# ─── 4) Train PINN ───────────────────────────────────────
epochs = 2000
loss_hist = []
for ep in range(epochs):
    L = train_step().numpy().item()
    loss_hist.append(L)
    if ep % 200 == 0:
        print(f"Epoch {ep}: Loss = {L:.6e}")

# extract PINN solution
x_pred_tf, _, _ = x_and_derivatives(t_tf)
x_pinn = x_pred_tf.numpy().flatten()

# ─── 5) Plot overlay & loss ─────────────────────────────
plt.figure(figsize=(12,5))

# (a) PINN vs RK4
plt.subplot(1,2,1)
plt.plot(t_pinn, x_pinn, '-b', lw=2, label='PINN')
plt.scatter(t_data, x_data, color='green', s=50, marker='x', label='Data points')
plt.plot(t_num, x_num, '--', color='orange', lw=1, label='RK4')
plt.scatter([0],[1], color='red', label='IC')
plt.title("NN w/ Data Loss vs RK4")
plt.xlabel('t'); plt.ylabel('x(t)')
plt.legend(); plt.grid(True)

# (b) Loss curve
plt.subplot(1,2,2)
plt.plot(loss_hist, '-g', lw=2)
plt.yscale('log')
plt.title("Training Loss vs. Epoch")
plt.xlabel('Epoch'); plt.ylabel('Total Loss')
plt.grid(True)

plt.tight_layout()
plt.show()
```

***(ii) PINN (Euler--Lagrange) loss***

We enforce the ODE
  $$\ddot\phi + \kappa\sin\phi=0$$
   by minimizing
 $$
    \mathcal{L}_{\rm PINN}(\theta)
    \;=\;
    \frac{1}{N}\sum_{i=1}^N
    \bigl(\ddot\phi_\theta(t_i)+\kappa\sin\phi_\theta(t_i)\bigr)^2
 $$
  over a grid $\{t_i\}$. Comparing with the solution given by Runge--Kutta method we note that they coincide on the interval $[0, 1]$.


```python
import numpy as np
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import matplotlib.pyplot as plt

# ─── PINN SETUP ─────────────────────────────────────────
np.random.seed(0)
tf.random.set_seed(0)

# Domain [0,1]
N_pinn = 1000
t_pinn = np.linspace(0, 1, N_pinn+1).reshape(-1,1).astype(np.float32)
t_tf   = tf.constant(t_pinn)
dt_pinn = t_pinn[1,0] - t_pinn[0,0]

# Build PINN model: approximates x(t)
def build_pinn():
    inp = keras.Input(shape=(1,))
    h = layers.Dense(32, activation='tanh')(inp)
    h = layers.Dense(32, activation='tanh')(h)
    h = layers.Dense(32, activation='tanh')(h)
    out = layers.Dense(1)(h)
    return keras.Model(inputs=inp, outputs=out)

model = build_pinn()
optimizer = keras.optimizers.Adam(1e-3)

# Compute x and its derivatives
def x_and_derivatives(t):
    with tf.GradientTape(persistent=True) as tape:
        tape.watch(t)
        g = model(t)
        x = 1.0 + t**2 * g   # embed x(0)=1 and x'(0)=0
        dx = tape.gradient(x, t)
    ddx = tape.gradient(dx, t)
    del tape
    return x, dx, ddx

# Loss: enforce x'' + sin(x) = 0
@tf.function
def train_step():
    with tf.GradientTape() as tape:
        x_pred, _, ddx_pred = x_and_derivatives(t_tf)
        residual = ddx_pred + tf.sin(x_pred)
        loss = tf.reduce_mean(residual**2)
    grads = tape.gradient(loss, model.trainable_variables)
    optimizer.apply_gradients(zip(grads, model.trainable_variables))
    return loss

# Train PINN
epochs = 1000
loss_history = []
for epoch in range(epochs):
    L = train_step().numpy()
    loss_history.append(L)
    if epoch % 200 == 0:
        print(f"Epoch {epoch}: loss = {L:.6e}")

# Extract PINN solution
x_pinn = x_and_derivatives(t_tf)[0].numpy().flatten()

# ─── NUMERICAL RK4 SOLUTION ────────────────────────────
dt_num = 0.01
N_num  = int(1.0 / dt_num)
t_num  = np.linspace(0, 1, N_num+1)
x_num  = np.zeros(N_num+1)
v_num  = np.zeros(N_num+1)
x_num[0] = 1.0
v_num[0] = 0.0

def f(state):
    xi, vi = state
    return np.array([vi, -np.sin(xi)])

for i in range(N_num):
    state = np.array([x_num[i], v_num[i]])
    k1 = dt_num * f(state)
    k2 = dt_num * f(state + 0.5*k1)
    k3 = dt_num * f(state + 0.5*k2)
    k4 = dt_num * f(state + k3)
    delta = (k1 + 2*k2 + 2*k3 + k4) / 6.0
    x_num[i+1] = x_num[i] + delta[0]
    v_num[i+1] = v_num[i] + delta[1]

# ─── PLOT RESULTS ─────────────────────────────────────
plt.figure(figsize=(12,5))

# (a) Solutions overlay
plt.subplot(1,2,1)
plt.plot(t_pinn, x_pinn, '-b', lw=2, label='PINN solution')
plt.scatter(t_num, x_num, s=5, color='orange', label='RK4 solution')
plt.scatter([0], [1], color='red', label='Initial cond.')
plt.title("Solutions: PINN vs RK4")
plt.xlabel('t'); plt.ylabel('x(t)')
plt.legend(); plt.grid(True)

# (b) PINN loss
plt.subplot(1,2,2)
plt.plot(loss_history, '-g', lw=2)
plt.yscale('log')
plt.title("PINN Training Loss")
plt.xlabel('Epoch'); plt.ylabel('MSE residual')
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import matplotlib.pyplot as plt

# ─── 1) PINN Setup ─────────────────────────────────────────
np.random.seed(0)
tf.random.set_seed(0)

T = 5.0
N = 5000          # number of discretization points for PINN
t_vals = np.linspace(0, T, N+1).reshape(-1,1).astype(np.float32)
t_tf   = tf.constant(t_vals)
dt_pinn = t_vals[1,0] - t_vals[0,0]

# 2) Build the PINN model
def build_model():
    inp = keras.Input(shape=(1,))
    h = layers.Dense(32, activation='tanh')(inp)
    h = layers.Dense(32, activation='tanh')(h)
    h = layers.Dense(32, activation='tanh')(h)
    h = layers.Dense(32, activation='tanh')(h)
    out = layers.Dense(1)(h)
    return keras.Model(inputs=inp, outputs=out)

model = build_model()
optimizer = keras.optimizers.Adam(learning_rate=1e-4)

# 3) Parametrize x(t) with boundary x(0)=1, x'(0)=0
def x_and_derivatives(t):
    with tf.GradientTape(persistent=True) as tape:
        tape.watch(t)
        g = model(t)
        x = 1.0 + t**2 * g   # embed x(0)=1 and x'(0)=0
        dx = tape.gradient(x, t)
    ddx = tape.gradient(dx, t)
    del tape
    return x, dx, ddx

# 4) Training step enforcing x'' + sin(x) = 0
@tf.function
def train_step():
    with tf.GradientTape() as tape:
        x_pred, dx_pred, ddx_pred = x_and_derivatives(t_tf)
        resid = ddx_pred + tf.sin(x_pred)
        loss  = tf.reduce_mean(resid**2)
    grads = tape.gradient(loss, model.trainable_variables)
    optimizer.apply_gradients(zip(grads, model.trainable_variables))
    return loss

# 5) Train the PINN
epochs = 1000
loss_history = []
for epoch in range(epochs):
    L = train_step().numpy()
    loss_history.append(L)
    if epoch % 200 == 0:
        print(f"Epoch {epoch}: Loss = {L:.6f}")

# 6) PINN solution
x_pred_tf, _, _ = x_and_derivatives(t_tf)
x_pinn = x_pred_tf.numpy().flatten()

# ─── 7) Numerical RK4 Solution ───────────────────────────
dt_num = 0.005
N_num  = int(T / dt_num)
t_num  = np.linspace(0, T, N_num+1)
x_num  = np.zeros(N_num+1)
v_num  = np.zeros(N_num+1)
x_num[0] = 1.0
v_num[0] = 0.0

def f(state):
    xi, vi = state
    return np.array([vi, -np.sin(xi)])

for i in range(N_num):
    state = np.array([x_num[i], v_num[i]])
    k1 = dt_num * f(state)
    k2 = dt_num * f(state + 0.5 * k1)
    k3 = dt_num * f(state + 0.5 * k2)
    k4 = dt_num * f(state + k3)
    delta = (k1 + 2*k2 + 2*k3 + k4) / 6.0
    x_num[i+1] = x_num[i] + delta[0]
    v_num[i+1] = v_num[i] + delta[1]

# ─── 8) Plot PINN vs RK4 and Loss ────────────────────────
plt.figure(figsize=(12,5))

# Left: compare solutions
plt.subplot(1,2,1)
plt.plot(t_vals, x_pinn, '-b', lw=2, label='PINN solution')
plt.scatter(t_num, x_num, s=5, color='orange', label='RK4 solution')
plt.scatter([0], [1], color='red', zorder=5, label='Initial condition')
plt.xlabel('t'); plt.ylabel('x(t)')
plt.title("PINN vs. Numerical RK4 for $x'' + \sin x=0$")
plt.legend(); plt.grid(True)

# Right: training loss
plt.subplot(1,2,2)
plt.plot(loss_history, '-g', lw=2)
plt.xlabel('Epoch'); plt.ylabel('MSE Residual Loss')
plt.title('PINN Training Loss vs. Epoch')
plt.yscale('log'); plt.grid(True)

plt.tight_layout()
plt.show()
```

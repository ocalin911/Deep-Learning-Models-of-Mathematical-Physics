# **1-D Addiabatic Process**

We consider an adiabatic process for an ideal fluid confined in a long, thin tube,
so that the motion can be modeled using a single spatial coordinate $x$.
Let $v(t)$, $\rho(t,x)$, and $p(t,x)$ denote the velocity, density, and pressure,
respectively, at position $x$ and time $t$. The initial state of the gas is prescribed by given conditions
$$
v(0) = v_0, \qquad \rho(x,0) = \rho_0(x), \qquad p(x,0) = p_0(x).
$$
subject to certain compatibility conditions.

We shall use both supervised  and hybrid learning.

**Unsupervised learning**

The next code plots heat maps for the solutions  $v(t)$, $\rho(t,x)$, and $p(t,x)$ .

```python
import tensorflow as tf
from tensorflow.keras import layers, models
import numpy as np
import matplotlib.pyplot as plt

# ----------------------------
# Build Neural Network with branching
# ----------------------------
def build_model(hidden_units=64, hidden_layers=3):
    # Branch for velocity v(t) only
    t_input = tf.keras.Input(shape=(1,), name="t_input")
    h_t = t_input
    for _ in range(hidden_layers):
        h_t = layers.Dense(hidden_units, activation="swish")(h_t)
    v_out = layers.Dense(1, name="v_branch")(h_t)

    # Branch for rho(t,x) and p(t,x)
    tx_input = tf.keras.Input(shape=(2,), name="tx_input")  # (t, x)
    h_tx = tx_input
    for _ in range(hidden_layers):
        h_tx = layers.Dense(hidden_units, activation="swish")(h_tx)
    rho_out = layers.Dense(1, name="rho_branch")(h_tx)
    p_out   = layers.Dense(1, name="p_branch")(h_tx)

    model = models.Model(inputs=[t_input, tx_input], outputs=[v_out, rho_out, p_out])
    return model

model = build_model()

# ----------------------------
# PDE Residual Losses
# ----------------------------
@tf.function
def pinn_loss(model, t, x, params, v0, A, B):
    c, gamma = params
    eps = 1e-8
    t_input = t
    tx_input = tf.concat([t, x], axis=1)

    with tf.GradientTape(persistent=True) as tape:
        tape.watch([t, x])
        v_branch, rho_branch, p_branch = model([t_input, tx_input])

        # Embed initial conditions
        rho0 = tf.maximum(A - B*x, eps)**(1.0/(gamma-1.0))
        p0   = c * tf.maximum(A - B*x, eps)**(gamma/(gamma-1.0))

        v = v0 + t * v_branch
        rho = rho0 + t * rho_branch
        p = p0 + t * p_branch

    dv_dt   = tape.gradient(v, t)
    drho_dt = tape.gradient(rho, t)
    drho_dx = tape.gradient(rho, x)
    dp_dx   = tape.gradient(p, x)
    del tape

    rho_safe = tf.maximum(rho, eps)

    L_cont  = tf.reduce_mean((drho_dt + v * drho_dx)**2)
    L_euler = tf.reduce_mean((rho_safe * dv_dt + dp_dx)**2)
    L_state = tf.reduce_mean((p - c * rho_safe**gamma)**2)

    return L_cont, L_euler, L_state

# ----------------------------
# Training Step
# ----------------------------
optimizer = tf.keras.optimizers.Adam(3e-5)

@tf.function
def train_step(t, x, params, lambdas, v0, A, B):
    with tf.GradientTape() as tape:
        L_cont, L_euler, L_state = pinn_loss(model, t, x, params, v0, A, B)
        # Total loss
        L_total = (lambdas[0]*L_cont +
                   lambdas[1]*L_euler +
                   lambdas[2]*L_state)

    grads = tape.gradient(L_total, model.trainable_variables)
    grads = [tf.clip_by_norm(g, 1.0) for g in grads]
    optimizer.apply_gradients(zip(grads, model.trainable_variables))
    return L_total, L_cont, L_euler, L_state

# ----------------------------
# Example Usage
# ----------------------------
if __name__ == "__main__":
    c = 1.0
    gamma = 1.8
    params = (c, gamma)

    v0 = tf.constant(1.0, dtype=tf.float32)
    A = tf.constant(2.0, dtype=tf.float32)
    B = tf.constant(0.5, dtype=tf.float32)

    Nt, Nx = 50, 50
    t = tf.random.uniform((Nt*Nx, 1), minval=0, maxval=1)
    x = tf.random.uniform((Nt*Nx, 1), minval=0, maxval=(A.numpy()/B.numpy())*0.8)

    lambdas = (1.0, 1.0, 1.0)

    loss_hist, cont_hist, euler_hist, state_hist = [], [], [], []

    for epoch in range(5001):
        L_total, L_cont, L_euler, L_state = train_step(t, x, params, lambdas, v0, A, B)
        if epoch % 500 == 0:
            print(f"Epoch {epoch}: "
                  f"Total={L_total.numpy():.6f}, "
                  f"Cont={L_cont.numpy():.6f}, "
                  f"Euler={L_euler.numpy():.6f}, "
                  f"State={L_state.numpy():.6f}")
        loss_hist.append(L_total.numpy())
        cont_hist.append(L_cont.numpy())
        euler_hist.append(L_euler.numpy())
        state_hist.append(L_state.numpy())

    # ----------------------------
    # Evaluation and Plots
    # ----------------------------
    t_test = np.linspace(0, 1, 100)
    x_test = np.linspace(0, 2, 100)
    T, X = np.meshgrid(t_test, x_test)

    t_in = T.flatten()[:,None].astype(np.float32)
    x_in = X.flatten()[:,None].astype(np.float32)

    v_branch, rho_branch, p_branch = model([t_in, np.hstack([t_in, x_in])], training=False)

    rho0 = (A.numpy() - B.numpy()*x_in.flatten())**(1.0/(gamma-1.0))
    p0   = c*(A.numpy() - B.numpy()*x_in.flatten())**(gamma/(gamma-1.0))

    V   = (v0.numpy() + t_in.flatten()*v_branch.numpy().flatten()).reshape(100,100)
    RHO = (rho0 + t_in.flatten()*rho_branch.numpy().flatten()).reshape(100,100)
    P   = (p0 + t_in.flatten()*p_branch.numpy().flatten()).reshape(100,100)

    # Plot v
    plt.figure(figsize=(6,4))
    plt.contourf(T, X, V, levels=50)
    plt.colorbar(label="v(t)")
    plt.xlabel("t")
    plt.ylabel("x")
    plt.title("Velocity v(t,x)")
    plt.show()

    # Plot rho
    plt.figure(figsize=(6,4))
    plt.contourf(T, X, RHO, levels=50)
    plt.colorbar(label="ρ(t,x)")
    plt.xlabel("t")
    plt.ylabel("x")
    plt.title("Density ρ(t,x)")
    plt.show()

    # Plot p
    plt.figure(figsize=(6,4))
    plt.contourf(T, X, P, levels=50)
    plt.colorbar(label="p(t,x)")
    plt.xlabel("t")
    plt.ylabel("x")
    plt.title("Pressure p(t,x)")
    plt.show()

    # Plot losses
    plt.figure(figsize=(6,4))
    plt.plot(loss_hist, label="Total Loss")
    plt.plot(cont_hist, label="Continuity")
    plt.plot(euler_hist, label="Euler")
    plt.plot(state_hist, label="State")
    plt.yscale("log")
    plt.xlabel("Epoch")
    plt.ylabel("Loss")
    plt.legend()
    plt.title("Training Losses")
    plt.show()
```

The exact profile of the solution $v(t)$ is given below:

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
c = 1.0
gamma = 1.8
v0 = 1.0
A = 2.0
B = 0.5

# Initial density profile
def rho0(x):
    return (A - B*x)**(1.0/(gamma-1.0))

# Equation (CKdata): pick two x values to solve for C and K (linear relation)
x1, x2 = 0.0, 1.0
lhs1 = 0.5*v0**2 - (c*gamma/(gamma-1.0))*rho0(x1)**(gamma-1.0)
lhs2 = 0.5*v0**2 - (c*gamma/(gamma-1.0))*rho0(x2)**(gamma-1.0)
# System: lhs = C*x + K
C = (lhs2 - lhs1)/(x2 - x1)
K = lhs1 - C*x1

# Functions v(t), rho(t,x)
def v_t(t):
    return v0 + C*t

def rho_tx(t, x):
    val = ((gamma-1)/(c*gamma)) * (0.5*(v0+ C*t)**2 - C*x - K)
    return np.maximum(val, 0.0)**(1.0/(gamma-1.0))

# Plot v(t)
t_vals = np.linspace(0, 1, 200)
plt.figure(figsize=(6,4))
plt.plot(t_vals, v_t(t_vals))
plt.xlabel("t")
plt.ylabel("v(t)")
plt.title("Velocity profile v(t)")
plt.grid(True)
plt.show()

# Plot rho(t,x) profiles at fixed times
x_vals = np.linspace(0, 2, 200)
times = [0.0, 0.2, 0.5, 0.75, 1.0]
plt.figure(figsize=(6,4))
for t in times:
    plt.plot(x_vals, rho_tx(t, x_vals), label=f"t={t}")
plt.xlabel("x")
plt.ylabel(r"$\rho(t,x)$")
plt.title("Density profiles at different times")
plt.legend()
plt.grid(True)
plt.show()
```

The difference between the neural network solution for $v_{\theta}(t)$ and the exact $v(t)$ is given next:

```python
# ----------------------------
# Compare network vs exact v(t)
# ----------------------------
t_plot = np.linspace(0, 1, 200).reshape(-1,1).astype(np.float32)

# Network prediction (branch depends only on t)
v_branch, _, _ = model([t_plot, np.hstack([t_plot, np.zeros_like(t_plot)])], training=False)
v_net = float(v0) + t_plot.flatten() * v_branch.numpy().flatten()  # cast v0 to float

# Exact solution: v(t) = v0 + C t
v_exact = float(v0) + C * t_plot.flatten()

# Plot
plt.figure(figsize=(6,4))
plt.plot(t_plot, v_exact, 'k--', linewidth=2, label="Exact v(t)")
plt.plot(t_plot, v_net, 'r-', linewidth=2, label="Network v(t)")
plt.xlabel("t")
plt.ylabel("Velocity v(t)")
plt.title("Comparison: Network vs Exact v(t)")
plt.legend()
plt.grid(True)
plt.show()
```

**Hybrid learning**

We train again using one measurement at $t=1$. We obtain a solution which has error inside th einterval $(0,1)$.

```python
import tensorflow as tf
from tensorflow.keras import layers, models
import numpy as np
import matplotlib.pyplot as plt

# ----------------------------
# Build Neural Network with branching
# ----------------------------
def build_model(hidden_units=64, hidden_layers=3):
    # Branch for velocity v(t) only
    t_input = tf.keras.Input(shape=(1,), name="t_input")
    h_t = t_input
    for _ in range(hidden_layers):
        h_t = layers.Dense(hidden_units, activation="swish")(h_t)
    v_out = layers.Dense(1, name="v_branch")(h_t)

    # Branch for rho(t,x) and p(t,x)
    tx_input = tf.keras.Input(shape=(2,), name="tx_input")  # (t, x)
    h_tx = tx_input
    for _ in range(hidden_layers):
        h_tx = layers.Dense(hidden_units, activation="swish")(h_tx)
    rho_out = layers.Dense(1, name="rho_branch")(h_tx)
    p_out   = layers.Dense(1, name="p_branch")(h_tx)

    model = models.Model(inputs=[t_input, tx_input], outputs=[v_out, rho_out, p_out])
    return model

model = build_model()

# ----------------------------
# PDE Residual Losses
# ----------------------------
@tf.function
def pinn_loss(model, t, x, params, v0, A, B):
    c, gamma = params
    eps = 1e-8
    t_input = t
    tx_input = tf.concat([t, x], axis=1)

    with tf.GradientTape(persistent=True) as tape:
        tape.watch([t, x])
        v_branch, rho_branch, p_branch = model([t_input, tx_input])

        # Embed initial conditions
        rho0 = tf.maximum(A - B*x, eps)**(1.0/(gamma-1.0))
        p0   = c * tf.maximum(A - B*x, eps)**(gamma/(gamma-1.0))

        v = v0 + t * v_branch
        rho = rho0 + t * rho_branch
        p = p0 + t * p_branch

    dv_dt   = tape.gradient(v, t)
    drho_dt = tape.gradient(rho, t)
    drho_dx = tape.gradient(rho, x)
    dp_dx   = tape.gradient(p, x)
    del tape

    rho_safe = tf.maximum(rho, eps)

    L_cont  = tf.reduce_mean((drho_dt + v * drho_dx)**2)
    L_euler = tf.reduce_mean((rho_safe * dv_dt + dp_dx)**2)
    L_state = tf.reduce_mean((p - c * rho_safe**gamma)**2)

    return L_cont, L_euler, L_state

# ----------------------------
# Extra Observation Loss at t=1
# ----------------------------
def observation_loss(model, v0, C):
    t1 = tf.constant([[1.0]], dtype=tf.float32)
    x_dummy = tf.zeros_like(t1)  # dummy for rho/p branch
    v_branch, _, _ = model([t1, tf.concat([t1, x_dummy], axis=1)])
    v_pred = v0 + 1.0 * v_branch
    v_exact = v0 + C*1.0
    return tf.reduce_mean((v_pred - v_exact)**2)

# ----------------------------
# Training Step
# ----------------------------
optimizer = tf.keras.optimizers.Adam(3e-5)

@tf.function
def train_step(t, x, params, lambdas, v0, A, B, C):
    with tf.GradientTape() as tape:
        L_cont, L_euler, L_state = pinn_loss(model, t, x, params, v0, A, B)
        L_obs = observation_loss(model, v0, C)

        L_total = (lambdas[0]*L_cont +
                   lambdas[1]*L_euler +
                   lambdas[2]*L_state +
                   lambdas[3]*L_obs)

    grads = tape.gradient(L_total, model.trainable_variables)
    grads = [tf.clip_by_norm(g, 1.0) for g in grads]
    optimizer.apply_gradients(zip(grads, model.trainable_variables))
    return L_total, L_cont, L_euler, L_state, L_obs

# ----------------------------
# Example Usage
# ----------------------------
if __name__ == "__main__":
    c = 1.0
    gamma = 1.8
    params = (c, gamma)

    v0 = 1.0
    A = tf.constant(2.0, dtype=tf.float32)
    B = tf.constant(0.5, dtype=tf.float32)

    # Compute constants C and K from initial data relation
    def rho0(x):
        return (A.numpy() - B.numpy()*x)**(1.0/(gamma-1.0))
    x1, x2 = 0.0, 1.0
    lhs1 = 0.5*v0**2 - (c*gamma/(gamma-1.0))*rho0(x1)**(gamma-1.0)
    lhs2 = 0.5*v0**2 - (c*gamma/(gamma-1.0))*rho0(x2)**(gamma-1.0)
    C = (lhs2 - lhs1)/(x2 - x1)
    K = lhs1 - C*x1
    print("C =", C, "K =", K)

    # Training data
    Nt, Nx = 50, 50
    t = tf.random.uniform((Nt*Nx, 1), minval=0, maxval=1)
    x = tf.random.uniform((Nt*Nx, 1), minval=0, maxval=(A.numpy()/B.numpy())*0.8)

    lambdas = (1.0, 1.0, 1.0, 50.0)

    loss_hist = []

    for epoch in range(6001):
        L_total, L_cont, L_euler, L_state, L_obs = train_step(t, x, params, lambdas, v0, A, B, C)
        if epoch % 500 == 0:
            print(f"Epoch {epoch}: "
                  f"Total={L_total.numpy():.6f}, "
                  f"Cont={L_cont.numpy():.6f}, "
                  f"Euler={L_euler.numpy():.6f}, "
                  f"State={L_state.numpy():.6f}, "
                  f"Obs={L_obs.numpy():.6f}")
        loss_hist.append(L_total.numpy())

    # ----------------------------
    # Compare v_theta(t) vs exact v(t)
    # ----------------------------
    t_plot = np.linspace(0, 1, 200).reshape(-1,1).astype(np.float32)
    v_branch, _, _ = model([t_plot, np.hstack([t_plot, np.zeros_like(t_plot)])], training=False)
    v_net = float(v0) + t_plot.flatten() * v_branch.numpy().flatten()
    v_exact = float(v0) + C * t_plot.flatten()

    plt.figure(figsize=(6,4))
    plt.plot(t_plot, v_exact, 'k--', linewidth=2, label="Exact v(t)")
    plt.plot(t_plot, v_net, 'r-', linewidth=2, label="Network vθ(t)")
    plt.xlabel("t")
    plt.ylabel("Velocity")
    plt.title("Comparison of Network vs Exact Velocity")
    plt.legend()
    plt.grid(True)
    plt.show()

    # Plot loss history
    plt.figure(figsize=(6,4))
    plt.plot(loss_hist)
    plt.yscale("log")
    plt.xlabel("Epoch")
    plt.ylabel("Total Loss")
    plt.title("Training Loss History")
    plt.show()
```

```python
import tensorflow as tf
from tensorflow.keras import layers, models
import numpy as np
import matplotlib.pyplot as plt

# ----------------------------
# Build Neural Network with branching
# ----------------------------
def build_model(hidden_units=64, hidden_layers=3):
    # Branch for velocity v(t) only
    t_input = tf.keras.Input(shape=(1,), name="t_input")
    h_t = t_input
    for _ in range(hidden_layers):
        h_t = layers.Dense(hidden_units, activation="swish")(h_t)
    v_out = layers.Dense(1, name="v_branch")(h_t)

    # Branch for rho(t,x) and p(t,x)
    tx_input = tf.keras.Input(shape=(2,), name="tx_input")  # (t, x)
    h_tx = tx_input
    for _ in range(hidden_layers):
        h_tx = layers.Dense(hidden_units, activation="swish")(h_tx)
    rho_out = layers.Dense(1, name="rho_branch")(h_tx)
    p_out   = layers.Dense(1, name="p_branch")(h_tx)

    model = models.Model(inputs=[t_input, tx_input], outputs=[v_out, rho_out, p_out])
    return model

model = build_model()

# ----------------------------
# PDE Residual Losses
# ----------------------------
@tf.function
def pinn_loss(model, t, x, params, v0, A, B):
    c, gamma = params
    eps = 1e-8
    t_input = t
    tx_input = tf.concat([t, x], axis=1)

    with tf.GradientTape(persistent=True) as tape:
        tape.watch([t, x])
        v_branch, rho_branch, p_branch = model([t_input, tx_input])

        # Embed initial conditions
        rho0 = tf.maximum(A - B*x, eps)**(1.0/(gamma-1.0))
        p0   = c * tf.maximum(A - B*x, eps)**(gamma/(gamma-1.0))

        v = v0 + t * v_branch
        rho = rho0 + t * rho_branch
        p = p0 + t * p_branch

    dv_dt   = tape.gradient(v, t)
    drho_dt = tape.gradient(rho, t)
    drho_dx = tape.gradient(rho, x)
    dp_dx   = tape.gradient(p, x)
    del tape

    rho_safe = tf.maximum(rho, eps)

    L_cont  = tf.reduce_mean((drho_dt + v * drho_dx)**2)
    L_euler = tf.reduce_mean((rho_safe * dv_dt + dp_dx)**2)
    L_state = tf.reduce_mean((p - c * rho_safe**gamma)**2)

    return L_cont, L_euler, L_state

# ----------------------------
# Extra Observation Losses
# ----------------------------
def observation_loss(model, v0, C):
    obs_loss = 0.0
    for t_val in [0.5, 1.0]:
        t_in = tf.constant([[t_val]], dtype=tf.float32)
        x_dummy = tf.zeros_like(t_in)  # dummy for rho/p branch
        v_branch, _, _ = model([t_in, tf.concat([t_in, x_dummy], axis=1)])
        v_pred = v0 + t_val * v_branch
        v_exact = v0 + C*t_val
        obs_loss += tf.reduce_mean((v_pred - v_exact)**2)
    return obs_loss

# ----------------------------
# Training Step
# ----------------------------
optimizer = tf.keras.optimizers.Adam(3e-5)

@tf.function
def train_step(t, x, params, lambdas, v0, A, B, C):
    with tf.GradientTape() as tape:
        L_cont, L_euler, L_state = pinn_loss(model, t, x, params, v0, A, B)
        L_obs = observation_loss(model, v0, C)

        L_total = (lambdas[0]*L_cont +
                   lambdas[1]*L_euler +
                   lambdas[2]*L_state +
                   lambdas[3]*L_obs)

    grads = tape.gradient(L_total, model.trainable_variables)
    grads = [tf.clip_by_norm(g, 1.0) for g in grads]
    optimizer.apply_gradients(zip(grads, model.trainable_variables))
    return L_total, L_cont, L_euler, L_state, L_obs

# ----------------------------
# Example Usage
# ----------------------------
if __name__ == "__main__":
    c = 1.0
    gamma = 1.8
    params = (c, gamma)

    v0 = 1.0
    A = tf.constant(2.0, dtype=tf.float32)
    B = tf.constant(0.5, dtype=tf.float32)

    # Compute constants C and K from initial data relation
    def rho0(x):
        return (A.numpy() - B.numpy()*x)**(1.0/(gamma-1.0))
    x1, x2 = 0.0, 1.0
    lhs1 = 0.5*v0**2 - (c*gamma/(gamma-1.0))*rho0(x1)**(gamma-1.0)
    lhs2 = 0.5*v0**2 - (c*gamma/(gamma-1.0))*rho0(x2)**(gamma-1.0)
    C = (lhs2 - lhs1)/(x2 - x1)
    K = lhs1 - C*x1
    print("C =", C, "K =", K)

    # Training data
    Nt, Nx = 50, 50
    t = tf.random.uniform((Nt*Nx, 1), minval=0, maxval=1)
    x = tf.random.uniform((Nt*Nx, 1), minval=0, maxval=(A.numpy()/B.numpy())*0.8)

    lambdas = (1.0, 1.0, 1.0, 50.0)  # add weight for obs

    loss_hist = []

    for epoch in range(6001):
        L_total, L_cont, L_euler, L_state, L_obs = train_step(t, x, params, lambdas, v0, A, B, C)
        if epoch % 500 == 0:
            print(f"Epoch {epoch}: "
                  f"Total={L_total.numpy():.6f}, "
                  f"Cont={L_cont.numpy():.6f}, "
                  f"Euler={L_euler.numpy():.6f}, "
                  f"State={L_state.numpy():.6f}, "
                  f"Obs={L_obs.numpy():.6f}")
        loss_hist.append(L_total.numpy())

    # ----------------------------
    # Compare v_theta(t) vs exact v(t)
    # ----------------------------
    t_plot = np.linspace(0, 1, 200).reshape(-1,1).astype(np.float32)
    v_branch, _, _ = model([t_plot, np.hstack([t_plot, np.zeros_like(t_plot)])], training=False)
    v_net = float(v0) + t_plot.flatten() * v_branch.numpy().flatten()
    v_exact = float(v0) + C * t_plot.flatten()

    plt.figure(figsize=(6,4))
    plt.plot(t_plot, v_exact, 'k--', linewidth=2, label="Exact v(t)")
    plt.plot(t_plot, v_net, 'r-', linewidth=2, label="Network vθ(t)")
    plt.xlabel("t")
    plt.ylabel("Velocity")
    plt.title("Comparison of Network vs Exact Velocity")
    plt.legend()
    plt.grid(True)
    plt.show()

    # Plot loss history
    plt.figure(figsize=(6,4))
    plt.plot(loss_hist)
    plt.yscale("log")
    plt.xlabel("Epoch")
    plt.ylabel("Total Loss")
    plt.title("Training Loss History")
    plt.show()
```

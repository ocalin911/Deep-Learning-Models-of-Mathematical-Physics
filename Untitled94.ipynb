**Charged Particles in a Constant Magnetic Field**

Then  Hamiltonian is the following sum of squares
$$
 H(q,p)=\frac12\Bigl(p_{1}+\tfrac{b}{2}q_{2}\Bigr)^{2}
      +\frac12\Bigl(p_{2}-\tfrac{b}{2}q_{1}\Bigr)^{2}.
$$
The Hamiltonian system is
\begin{eqnarray*}
\dot q_1 &=& \frac{\partial H}{\partial p_1} = p_1 + \frac{b}{2} q_2\\
\dot q_2 &=& \frac{\partial H}{\partial p_2} = p_2 - \frac{b}{2} q_1\\
\dot p_1 &=& -\frac{\partial H}{\partial q_1} = \frac{b}{2} ( p_2 -  \frac{b}{2} q_1 ) =\frac{b}{2} \dot q_2\\
\dot p_2 &=& -\frac{\partial H}{\partial q_2} = -\frac{b}{2} ( p_1 +  \frac{b}{2} q_2 ) =-\frac{b}{2} \dot q_1.
\end{eqnarray*}
We use neural nets to solve the system withthe initial condition $q(0)=(1,0)$ and $p(0)=(0,1)$.

The next code uses a neural network which uses swish activation function.

```python
"""
PINN for a charged particle in a uniform magnetic field

Hamiltonian  H = ½ (p1 + ½ b q2)^2 + ½ (p2 − ½ b q1)^2
with   b = 1,  m = e = 1.

Four network outputs (nn¹, …, nn⁴) are used to embed the quadratic
Taylor expansions of q₁, q₂, p₁, p₂.  The physics-informed loss enforces
Hamilton’s equations at 1 000 collocation points on t ∈ [0,1].

The script trains for 100 epochs and plots:
  • the PINN trajectory (q₁, q₂);
  • the analytic circular orbit of radius ‖K‖ = ‖ẏ(0)‖/b;
  • the training-loss curve (semilog).
"""

import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt

# ─────────── problem parameters ─────────────────────────────────────────
b         = 1        # magnetic-field strength
T         = 6.29        # final time
m_points  = 200      # collocation points
epochs    = 8000       # training epochs
lr        = 5e-5       # Adam learning rate

# initial data (feel free to modify)
q10, q20 = 1.0, 0.0
p10, p20 = 1.0, 0.0

# first-time derivatives at t = 0
dq10 = p10 + 0.5 * b * q20
dq20 = p20 - 0.5 * b * q10
dp10 = 0.5 * b * (p20 - 0.5 * b * q10)
dp20 = -0.5 * b * (p10 + 0.5 * b * q20)

# ─────────── collocation grid ──────────────────────────────────────────
t_col = tf.constant(np.linspace(0.0, T, m_points)[:, None], dtype=tf.float32)

# ─────────── neural network (4 outputs) ────────────────────────────────
model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(1,)),          # (t,)
    tf.keras.layers.Dense(64, activation='swish'),
    tf.keras.layers.Dense(64, activation='swish'),
    tf.keras.layers.Dense(64, activation='swish'),
    tf.keras.layers.Dense(64, activation='swish'),
    tf.keras.layers.Dense(64, activation='swish'),
    tf.keras.layers.Dense(64, activation='swish'),
    tf.keras.layers.Dense(4,  activation=None)  # nn¹,…,nn⁴
])

optimizer = tf.keras.optimizers.Adam(lr)
loss_history = []

# ─────────── helper: embed initial conditions ──────────────────────────
def embed_qp(nn_out, t):
    """Return (q1,q2,p1,p2) with embedded ICs and quadratic tails."""
    nn1, nn2, nn3, nn4 = tf.split(nn_out, 4, axis=1)

    q1 = q10 + t * dq10 + 0.5 * t**2 * nn1
    q2 = q20 + t * dq20 + 0.5 * t**2 * nn2
    p1 = p10 + t * dp10 + 0.5 * t**2 * nn3
    p2 = p20 + t * dp20 + 0.5 * t**2 * nn4
    return q1, q2, p1, p2

# ─────────── physics-informed loss ─────────────────────────────────────
@tf.function
def physics_loss():
    with tf.GradientTape(persistent=True) as tape:
        tape.watch(t_col)
        nn_out = model(t_col)
        q1, q2, p1, p2 = embed_qp(nn_out, t_col)

    dq1_dt = tape.gradient(q1, t_col)
    dq2_dt = tape.gradient(q2, t_col)
    dp1_dt = tape.gradient(p1, t_col)
    dp2_dt = tape.gradient(p2, t_col)
    del tape

    res1 = dq1_dt - (p1 + 0.5 * b * q2)
    res2 = dq2_dt - (p2 - 0.5 * b * q1)
    res3 = dp1_dt - 0.5 * b * (p2 - 0.5 * b * q1)
    res4 = dp2_dt + 0.5 * b * (p1 + 0.5 * b * q2)

    return tf.reduce_mean(res1**2 + res2**2 + res3**2 + res4**2)

@tf.function
def train_step():
    with tf.GradientTape() as tape:
        loss = physics_loss()
    grads = tape.gradient(loss, model.trainable_variables)
    optimizer.apply_gradients(zip(grads, model.trainable_variables))
    return loss

# ─────────── training loop ─────────────────────────────────────────────
for epoch in range(1, epochs + 1):
    loss_val = train_step().numpy()
    loss_history.append(loss_val)
    if epoch % 100 == 0 or epoch == 1:
        print(f"Epoch {epoch:3d}: loss = {loss_val:.3e}")

# ─────────── evaluate the trained network ──────────────────────────────
t_plot = np.linspace(0.0, T, 600)[:, None].astype(np.float32)
nn_plot = model(t_plot).numpy()
q1_plot, q2_plot, _, _ = embed_qp(nn_plot, t_plot)

# ─────────── analytic circular orbit (component-wise) ──────────────────
Jinv = np.array([[0, -1],
                 [1,  0]])                    # J^{-1} = -J
Kvec = (1.0 / b) * (Jinv @ np.array([dq10, dq20]))
Cvec = np.array([q10, q20]) - Kvec

def exact_orbit(t_array):
    cos_bt = np.cos(b * t_array)
    sin_bt = np.sin(b * t_array)
    q1 = Cvec[0] + cos_bt * Kvec[0] + sin_bt * Kvec[1]
    q2 = Cvec[1] - sin_bt * Kvec[0] + cos_bt * Kvec[1]
    return q1, q2

q1_exact, q2_exact = exact_orbit(t_plot.flatten())

# ─────────── plots ─────────────────────────────────────────────────────
plt.figure(figsize=(7, 3.5))
plt.plot(q1_exact, q2_exact, label='exact',  c='tab:blue')
plt.plot(q1_plot,  q2_plot, '--', label='PINN', c='tab:orange')
plt.xlabel('$q_1$'); plt.ylabel('$q_2$')
plt.title('Uniform-B orbit: analytic vs PINN')
plt.legend(); plt.axis('equal'); plt.grid(True); plt.tight_layout()

plt.figure(figsize=(7, 3.5))
plt.semilogy(loss_history, c='tab:green')
plt.xlabel('epoch'); plt.ylabel('physics loss')
plt.title('Training loss'); plt.tight_layout()
plt.show()
```

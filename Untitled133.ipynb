# **Attention on a line segment**

The trajectory is a line segment $\{y=1; 0<1<6\}$ and the observer is located at $(0, 4)$. The graph of the entropy of all points along the segment has a peak at $x=3.746$.


```python
import numpy as np
import matplotlib.pyplot as plt

# ----------------------------
# Settings
# ----------------------------
x_min, x_max = 0.0, 6.0
N = 600                    # number of samples on the segment
tau = 0.5                  # softmax temperature (smaller => sharper, lower H)
anchor = np.array([4.0, 0.0])  # observer/anchor (ax, ay)

# ----------------------------
# Sample the segment y=1, 0<x<6
# ----------------------------
eps = 1e-6  # avoid exact endpoints for numerical stability
x = np.linspace(x_min + eps, x_max - eps, N)
y = np.ones_like(x)
X = np.stack([x, y], axis=1)  # shape (N, 2)

# ----------------------------
# Cosine-similarity attention & entropy at each point
# ----------------------------
# Vectors from anchor to all points
V = X - anchor[None, :]
Vn = np.linalg.norm(V, axis=1)
V_unit = V / (Vn[:, None] + 1e-12)

H = np.empty(N)
for i in range(N):
    vi = V_unit[i]                  # unit vector for the reference point i
    sims = V_unit @ vi              # cosine similarities to all j
    logits = sims / tau             # temperature scaling
    m = logits.max()                # stabilize softmax
    w = np.exp(logits - m)
    p = w / w.sum()
    H[i] = -(p * (np.log(p + 1e-12))).sum()  # entropy in nats

# Find maximum-entropy location (for reference)
imax = int(np.argmax(H))
x_argmax, H_max = x[imax], H[imax]
print(f"Max entropy at x ≈ {x_argmax:.6f}, H_max ≈ {H_max:.6f} nats")

# ----------------------------
# Plot 1: Entropy vs x
# ----------------------------
plt.figure(figsize=(8.2, 3.6))
plt.plot(x, H, lw=2)
plt.axvline(x_argmax, color='r', ls='--', alpha=0.6,
            label=f"max at x≈{x_argmax:.3f}")
plt.xlabel('x (on the segment y=1, 0 < x < 6)')
plt.ylabel('Entropy H(x) [nats]')
plt.title(f'Entropy along y=1 with anchor at ({anchor[0]}, {anchor[1]}) and τ={tau}')
plt.legend()
plt.tight_layout()
plt.show()

# ----------------------------
# Plot 2: Segment colored by -Entropy (hotter = lower entropy)
# ----------------------------
plt.figure(figsize=(7.2, 3.0))
sc = plt.scatter(x, y, c=-H, s=12, cmap='viridis')
plt.plot([anchor[0]], [anchor[1]], 'x', ms=9, label='Anchor')
plt.colorbar(sc, label='-Entropy (hotter = lower entropy)')
plt.ylim(0.7, 1.3)
plt.xlabel('x'); plt.ylabel('y')
plt.title('Segment y=1 colored by -Entropy')
plt.legend()
plt.tight_layout()
plt.show()
```

# **Attention on circle**

The next code produces the entropy along the unit circle with anchor at
$(−0.5,0)$.

```python
import numpy as np
import matplotlib.pyplot as plt

# ----------------------------
# Settings
# ----------------------------
N = 1000                 # number of points on the circle
tau = 0.5                # softmax temperature (smaller => sharper, lower H)
anchor = np.array([-0.5, 0.0])   # observer/anchor (ax, ay)

# ----------------------------
# Sample the unit circle
# ----------------------------
phi = np.linspace(0, 2*np.pi, N, endpoint=False)
X = np.stack([np.cos(phi), np.sin(phi)], axis=1)  # (N, 2)

# ----------------------------
# Cosine-similarity attention & entropy at each point
# ----------------------------
V = X - anchor[None, :]                   # vectors from anchor
Vn = np.linalg.norm(V, axis=1)
V_unit = V / (Vn[:, None] + 1e-12)        # unit vectors

H = np.empty(N)
for i in range(N):
    vi = V_unit[i]                        # unit vector for reference i
    sims = V_unit @ vi                    # cosine similarities to all j
    logits = sims / tau                   # temperature scaling
    m = logits.max()                      # stabilize softmax
    w = np.exp(logits - m)
    p = w / w.sum()
    H[i] = -(p * (np.log(p + 1e-12))).sum()  # entropy in nats

# ----------------------------
# Figure 1: Entropy H(φ) vs φ
# ----------------------------
plt.figure(figsize=(8.4, 3.6))
plt.plot(phi, H, lw=2)
plt.xlabel(r'$\phi$ (radians)')
plt.ylabel('Entropy H(φ) [nats]')
plt.title('Entropy on Unit Circle with Anchor at (-0.5, 0)')
plt.tight_layout()
plt.show()

# ----------------------------
# Figure 2: Circle colored by -Entropy (hotter = lower entropy)
# ----------------------------
plt.figure(figsize=(5.2, 5.2))
sc = plt.scatter(X[:, 0], X[:, 1], c=-H, s=10, cmap='viridis')
plt.plot([anchor[0]], [anchor[1]], 'x', ms=9, label='Anchor (-0.5, 0)')
plt.axis('equal')
plt.xlabel('x'); plt.ylabel('y')
plt.title('Unit Circle colored by -Entropy (hotter = lower entropy)')
plt.colorbar(sc, label='-Entropy')
plt.legend()
plt.tight_layout()
plt.show()
```

# **The Ellipse**

```python
import numpy as np
import matplotlib.pyplot as plt

# ----------------------------
# Ellipse & attention settings
# ----------------------------
a, b = 2.0, 1.0          # ellipse semiaxes: x^2/a^2 + y^2/b^2 = 1
N = 1600                 # number of samples on the ellipse
tau = 0.5                # softmax temperature (smaller => sharper, lower H)
anchor = np.array([0.0, 0.0])  # observer at origin

# ----------------------------
# Sample the ellipse
# ----------------------------
theta = np.linspace(0, 2*np.pi, N, endpoint=False)
X = np.stack([a*np.cos(theta), b*np.sin(theta)], axis=1)   # points on ellipse

# ----------------------------
# Cosine-similarity attention & entropy at each point
# ----------------------------
V = X - anchor[None, :]                     # vectors from anchor
Vn = np.linalg.norm(V, axis=1)
V_unit = V / (Vn[:, None] + 1e-12)          # unit vectors

H = np.empty(N)
for i in range(N):
    vi = V_unit[i]                          # unit vector for reference point i
    sims = V_unit @ vi                      # cosine similarities to all j
    logits = sims / tau                     # temperature scaling
    m = logits.max()                        # stabilize softmax
    w = np.exp(logits - m)
    p = w / w.sum()
    H[i] = -(p * (np.log(p + 1e-12))).sum()  # entropy in nats

# ----------------------------
# Figure 1: Entropy H(θ) vs θ
# ----------------------------
plt.figure(figsize=(8.4, 3.6))
plt.plot(theta, H, lw=2)
plt.xlabel(r'$\theta$ (radians)')
plt.ylabel('Entropy H(θ) [nats]')
plt.title('Entropy along Ellipse (a=2, b=1) with Anchor at Origin')
plt.tight_layout()
plt.show()

# ----------------------------
# Figure 2: Ellipse colored by -Entropy (hotter = lower entropy)
# ----------------------------
plt.figure(figsize=(6.2, 5.2))
sc = plt.scatter(X[:, 0], X[:, 1], c=-H, s=10, cmap='viridis')
plt.plot([anchor[0]], [anchor[1]], 'x', ms=9, label='Anchor (0,0)')
plt.axis('equal')
plt.xlabel('x'); plt.ylabel('y')
plt.title('Ellipse colored by -Entropy (anchor at origin)')
plt.colorbar(sc, label='-Entropy (hotter = lower entropy)')
plt.legend()
plt.tight_layout()
plt.show()
```

# **The parabola**

# **Bernoulli Lemniscate**

```python
import numpy as np
import matplotlib.pyplot as plt

# ----------------------------
# Settings
# ----------------------------
a = 1.0           # lemniscate scale
N = 2000          # samples along the curve
tau = 0.5         # softmax temperature (smaller => sharper, lower entropy)
anchor = np.array([0.0, 0.0])  # observer at the origin

# ----------------------------
# Bernoulli lemniscate parameterization
# (x^2 + y^2)^2 = 2 a^2 (x^2 - y^2)
# ----------------------------
t = np.linspace(0, 2*np.pi, N, endpoint=False)
sqrt2 = np.sqrt(2.0)
x = (sqrt2 * a * np.cos(t)) / (1.0 + np.sin(t)**2)
y = (sqrt2 * a * np.sin(t) * np.cos(t)) / (1.0 + np.sin(t)**2)
X = np.stack([x, y], axis=1)  # shape (N, 2)

# ----------------------------
# Vectors from anchor; drop points extremely close to the crossing (origin)
# ----------------------------
V = X - anchor[None, :]
Vn = np.linalg.norm(V, axis=1)
mask = Vn > 1e-6           # avoid zero-length vectors at the intersection
Xv = X[mask]
tv = t[mask]
V = V[mask]
Vn = Vn[mask]
V_unit = V / (Vn[:, None] + 1e-12)  # unit directions

# ----------------------------
# Entropy along the curve (cosine-similarity softmax)
# ----------------------------
H = np.empty(V_unit.shape[0])
for i in range(V_unit.shape[0]):
    vi = V_unit[i]
    sims = V_unit @ vi                 # cosine similarities to all j
    logits = sims / tau
    m = logits.max()                   # stabilize softmax
    w = np.exp(logits - m)
    p = w / w.sum()
    H[i] = -(p * (np.log(p + 1e-12))).sum()  # entropy in nats

# ----------------------------
# Figure 1: Entropy profile H(t)
# ----------------------------
plt.figure(figsize=(8.4, 3.6))
plt.plot(tv, H, lw=2)
plt.xlabel('t (radians)')
plt.ylabel('Entropy H(t) [nats]')
plt.title('Entropy along Bernoulli Lemniscate (anchor at origin)')
plt.tight_layout()
plt.show()

# ----------------------------
# Figure 2: Lemniscate colored by -Entropy (hotter = lower entropy)
# ----------------------------
plt.figure(figsize=(6.6, 5.6))
sc = plt.scatter(Xv[:, 0], Xv[:, 1], c=-H, s=10, cmap='viridis')
plt.plot([anchor[0]], [anchor[1]], 'x', ms=10, label='Anchor (origin)')
plt.axis('equal')
plt.xlabel('x'); plt.ylabel('y')
plt.title('Lemniscate colored by -Entropy (hotter = lower entropy)')
plt.colorbar(sc, label='-Entropy')
plt.legend()
plt.tight_layout()
plt.show()
```

Here is a  script that computes average entropy on a Bernoulli lemniscate across 12 observers placed on a circle of radius 2 (centered at the origin), then plots (1) the heat map on the curve colored by $H$

 and (2) the average entropy profile vs. the curve parameter
$t$.

```python
import numpy as np
import matplotlib.pyplot as plt

# ----------------------------
# Settings
# ----------------------------
a = 1.0            # lemniscate scale
N = 1600           # number of samples along the curve
M = 12             # number of observers
obs_radius = 2.0   # observer ring radius
tau = 0.5          # softmax temperature (smaller => sharper, lower entropy)

# ----------------------------
# Bernoulli lemniscate parameterization
# (x^2 + y^2)^2 = 2 a^2 (x^2 - y^2)
# ----------------------------
t = np.linspace(0, 2*np.pi, N, endpoint=False)
sqrt2 = np.sqrt(2.0)
X = np.stack([
    (sqrt2 * a * np.cos(t)) / (1.0 + np.sin(t)**2),
    (sqrt2 * a * np.sin(t) * np.cos(t)) / (1.0 + np.sin(t)**2)
], axis=1)  # shape (N, 2)

# ----------------------------
# Observer (anchor) positions on circle of radius 2
# ----------------------------
phi = np.linspace(0, 2*np.pi, M, endpoint=False)
anchors = np.stack([obs_radius * np.cos(phi), obs_radius * np.sin(phi)], axis=1)  # (M, 2)

def entropy_for_anchor(anchor, X, tau=0.5):
    """Entropy H(i) at each curve point i for a single anchor (cosine-softmax)."""
    V = X - anchor[None, :]                 # vectors from anchor to curve points
    Vn = np.linalg.norm(V, axis=1)
    V_unit = V / (Vn[:, None] + 1e-12)      # unit directions
    H = np.empty(X.shape[0])
    for i in range(X.shape[0]):
        vi = V_unit[i]
        sims = V_unit @ vi                  # cosine similarities to all j
        logits = sims / tau
        m = logits.max()
        w = np.exp(logits - m)
        p = w / w.sum()
        H[i] = -(p * (np.log(p + 1e-12))).sum()  # nats
    return H

# ----------------------------
# Compute per-anchor entropies and average
# ----------------------------
H_all = np.stack([entropy_for_anchor(a0, X, tau=tau) for a0 in anchors], axis=0)  # (M, N)
H_avg = H_all.mean(axis=0)        # average entropy across observers
negH_avg = -H_avg                 # for coloring (hotter = lower average entropy)

# ----------------------------
# Figure 1: Lemniscate colored by -Average Entropy, with observer crosses
# ----------------------------
plt.figure(figsize=(7.2, 6.0))
sc = plt.scatter(X[:, 0], X[:, 1], c=negH_avg, s=10, cmap='viridis')
plt.scatter(anchors[:, 0], anchors[:, 1], marker='x', s=120, linewidths=2.0,
            c='k', label='Observers')
plt.axis('equal')
plt.xlabel('x'); plt.ylabel('y')
plt.title('Lemniscate colored by -Average Entropy (12 observers on radius-2 circle)')
plt.colorbar(sc, label='- Average Entropy (hotter = lower)')
plt.legend(loc='upper right', frameon=True)
plt.tight_layout()
plt.show()

# ----------------------------
# Figure 2: Average entropy profile vs parameter t
# ----------------------------
plt.figure(figsize=(8.6, 3.6))
plt.plot(t, H_avg, lw=2)
plt.xlabel('t (radians)')
plt.ylabel('Average entropy $\\overline{H}(t)$ [nats]')
plt.title('Average entropy along the lemniscate (12 observers)')
plt.tight_layout()
plt.show()
```

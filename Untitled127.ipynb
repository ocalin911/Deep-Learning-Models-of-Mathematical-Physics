# **The wave equation in 1D**

Consider the Cauchy problem
\begin{align*}
\partial_t^2 u(t,x) - \partial_x^2 u(t,x) &= t\,x, && (t,x)\in (0,\infty)\times R,\\
u(0,x) &= \ln\!\bigl(1+x^2\bigr), && x\in R,\\
\partial_t u(0,x) &= 2e^x, && x\in R.
\end{align*}
By d'Alembert’s formula, the classical solution is


$$
u_{\rm exact}(t,x)
= \tfrac12\!\left[\ln\!\bigl(1+(x+t)^2\bigr)+\ln\!\bigl(1+(x-t)^2\bigr)\right]
+ 2e^x\sinh t \;+\; \tfrac16\,t^3 x.
$$


Using the single-network approach, we approximate $u$ with


$$
u_{\theta}(t,x)
= \ln\!\bigl(1+x^2\bigr) \;+\; 2t\,e^x \;+\; t^2\,\mathrm{nn}(\theta; t,x),
$$


and train by minimizing the PDE residual loss.

```python
# TensorFlow 2.x PINN for 1D wave equation with source: u_tt - u_xx = t*x
# IC: u(0,x)=ln(1+x^2), u_t(0,x)=2 e^x
# Plots: (1) loss vs epochs (log); (2) all profiles for t={0.1,0.5,1,1.5,2} on one graph (exact vs NN)

import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, Model
import matplotlib.pyplot as plt

# ----------------------------
# Problem setup
# ----------------------------
# Training domain
T_MAX = 2.0
X_MAX = 4.0
NT, NX = 128, 256  # collocation grid resolution

t_grid = np.linspace(0.0, T_MAX, NT, dtype=np.float32)
x_grid = np.linspace(-X_MAX, X_MAX, NX, dtype=np.float32)
TT, XX = np.meshgrid(t_grid, x_grid, indexing='ij')
t_train = TT.reshape(-1, 1)
x_train = XX.reshape(-1, 1)

# Right-hand side f(t,x) = t*x
def f_np(t, x):
    return t * x

# Exact solution (for validation / plots)
def u_exact_np(t, x):
    # t, x arrays broadcastable to same shape
    return (0.5*(np.log(1.0+(x+t)**2) + np.log(1.0+(x-t)**2))
            + 2.0*np.exp(x)*np.sinh(t) + (1.0/6.0)*t**3 * x)

# ----------------------------
# Neural network core (nn_\theta)
# ----------------------------
def build_mlp(widths=(64,64,64), activation='tanh'):
    inp = layers.Input(shape=(2,))
    z = inp
    for w in widths:
        z = layers.Dense(w, activation=activation)(z)
    out = layers.Dense(1, activation=None)(z)
    return Model(inp, out, name="nn_core")

nn = build_mlp(widths=(64,64,64), activation='tanh')

# Ansatz: u_theta(t,x) = ln(1+x^2) + 2 t e^x + t^2 * nn(t,x)
def u_theta(t, x):
    # t, x are tf tensors of shape (N,1)
    nn_in = tf.concat([t, x], axis=1)
    nn_out = nn(nn_in)  # (N,1)
    return tf.math.log(1.0 + x**2) + 2.0*t*tf.exp(x) + (t**2)*nn_out

# PDE residual r = u_tt - u_xx - t*x
def residual(t, x):
    with tf.GradientTape(persistent=True) as tape2:
        tape2.watch([t, x])
        with tf.GradientTape(persistent=True) as tape1:
            tape1.watch([t, x])
            u = u_theta(t, x)
        u_t = tape1.gradient(u, t)
        u_x = tape1.gradient(u, x)
        del tape1
    u_tt = tape2.gradient(u_t, t)
    u_xx = tape2.gradient(u_x, x)
    del tape2
    return u_tt - u_xx - t*x

# Loss = MSE of residual over collocation points
def loss_fn(t, x):
    r = residual(t, x)
    return tf.reduce_mean(tf.square(r))

optimizer = tf.keras.optimizers.Adam(learning_rate=1e-3)

# ----------------------------
# Training loop
# ----------------------------
EPOCHS = 2000
loss_history = []

t_tf = tf.constant(t_train, dtype=tf.float32)
x_tf = tf.constant(x_train, dtype=tf.float32)

@tf.function
def train_step():
    with tf.GradientTape() as tape:
        loss = loss_fn(t_tf, x_tf)
    grads = tape.gradient(loss, nn.trainable_variables)
    optimizer.apply_gradients(zip(grads, nn.trainable_variables))
    return loss

for epoch in range(1, EPOCHS+1):
    loss_val = float(train_step().numpy())
    loss_history.append(loss_val)
    if epoch % 200 == 0:
        print(f"Epoch {epoch:4d} | PDE loss: {loss_val:.3e}")

# ----------------------------
# Evaluation / plotting
# ----------------------------
def u_theta_np(t_array, x_array):
    # returns u_theta evaluated on (t,x) grid with t fixed or vectorized
    T, X = np.meshgrid(np.atleast_1d(t_array).astype(np.float32),
                       np.atleast_1d(x_array).astype(np.float32),
                       indexing='ij')
    t_flat = T.reshape(-1, 1)
    x_flat = X.reshape(-1, 1)
    u_flat = u_theta(tf.constant(t_flat), tf.constant(x_flat)).numpy().reshape(T.shape)
    return T, X, u_flat

# 1) Loss curve
plt.figure(figsize=(6,4))
plt.plot(loss_history)
plt.yscale('log')
plt.xlabel('Epoch')
plt.ylabel('PDE loss')
plt.title('Training Loss (log scale)')
plt.tight_layout()
plt.show()

# 2) All profiles on one graph: t = 0.1, 0.5, 1.0, 1.5, 2.0
ts_to_plot = [0.1, 0.5, 1.0, 1.5, 2.0]
x_plot = np.linspace(-X_MAX, X_MAX, 800, dtype=np.float32)

plt.figure(figsize=(9,6))
for t0 in ts_to_plot:
    # exact
    u_ex = u_exact_np(t0*np.ones_like(x_plot), x_plot)
    # nn
    _, _, U_nn = u_theta_np(np.array([t0], dtype=np.float32), x_plot)
    u_nn = U_nn[0]

    plt.plot(x_plot, u_ex, label=f'Exact t={t0}')
    plt.plot(x_plot, u_nn, '--', label=f'NN t={t0}')

plt.xlabel('x')
plt.ylabel('u(t,x)')
plt.title('Exact vs. PINN Solution Profiles (all times on one graph)')
plt.legend(ncol=2)
plt.tight_layout()
plt.show()
```

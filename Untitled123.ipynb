# **Radial Reduction of the Poisson Equation**

Consider the Poisson equation with homogeneous boundary conditions
$$
\Delta v(x) = 12|x| + 6, \qquad x \in B(0,1) \subset \mathbb{R}^2, $$
$$
v(x) = 0, \qquad  x \in \partial B(0,1).
$$

```python
# radial_poisson_3d_nn.py
# Requirements: pip install tensorflow matplotlib numpy

import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

# -------------------
# Settings
# -------------------
SEED = 123
np.random.seed(SEED)
tf.random.set_seed(SEED)

EPOCHS = 4000
BATCH = 1024
LR = 1e-4
R_MIN = 1e-2    # avoid r=0 in the 2/r term (we'll also add a smoothness penalty at r=0)
LAMBDA_ORIGIN = 1e-4  # small penalty for psi'(0) ≈ 0

# Exact solution and RHS
def rhs(r):        # 12 r + 6
    return 12.0 * r + 6.0

def psi_exact(r):  # r^3 + r^2 - 2
    return r**3 + r**2 - 2.0

# -------------------
# Model: phi_theta(r), then psi = (1 - r) * phi_theta(r)
# -------------------
def make_model(width=64, depth=3):
    inp = keras.Input(shape=(1,), dtype=tf.float32)  # r in [0,1]
    z = inp
    for _ in range(depth):
        z = layers.Dense(width, activation="swish")(z)
    out = layers.Dense(1, activation=None)(z)  # phi_theta(r)
    return keras.Model(inp, out)

phi_model = make_model(width=64, depth=3)
optimizer = keras.optimizers.Adam(LR)

# Helper: compute psi, psi', psi'' via autodiff
@tf.function
def psi_and_derivs(r):
    # r: (N,1)
    with tf.GradientTape(persistent=True) as t2:
        t2.watch(r)
        with tf.GradientTape() as t1:
            t1.watch(r)
            phi = phi_model(r, training=True)                 # (N,1)
            psi = (1.0 - r) * phi                             # hard embed psi(1)=0
        dpsi_dr = t1.gradient(psi, r)                         # (N,1)
    d2psi_dr2 = t2.gradient(dpsi_dr, r)                       # (N,1)
    del t1, t2
    return psi, dpsi_dr, d2psi_dr2

# Training step
@tf.function
def train_step():
    # sample r in (R_MIN, 1], uniform
    r = tf.random.uniform((BATCH, 1), minval=R_MIN, maxval=1.0, dtype=tf.float32)

    psi, psi_r, psi_rr = psi_and_derivs(r)
    pde_res = psi_rr + (2.0 / r) * psi_r - rhs(r)             # residual

    # small smoothness penalty at origin: use a tiny batch exactly at r=0
    r0 = tf.zeros((64, 1), dtype=tf.float32)
    with tf.GradientTape() as t:
        t.watch(r0)
        phi0 = phi_model(r0, training=True)
        psi0 = (1.0 - r0) * phi0
    psi_r0 = t.gradient(psi0, r0)                             # should be ≈ 0 for smooth radial soln

    loss_pde = tf.reduce_mean(tf.square(pde_res))
    loss_origin = tf.reduce_mean(tf.square(psi_r0))
    loss = loss_pde + LAMBDA_ORIGIN * loss_origin

    grads = tf.GradientTape()._tape  # (workaround not needed; use standard tape below)

    # Standard gradient path:
    with tf.GradientTape() as tape:
        # recompute for gradient (TF needs fresh forward pass)
        psi2, psi_r2, psi_rr2 = psi_and_derivs(r)
        pde_res2 = psi_rr2 + (2.0 / r) * psi_r2 - rhs(r)
        # origin again
        r0b = tf.zeros((64, 1), dtype=tf.float32)
        with tf.GradientTape() as t2:
            t2.watch(r0b)
            phi0b = phi_model(r0b, training=True)
            psi0b = (1.0 - r0b) * phi0b
        psi_r0b = t2.gradient(psi0b, r0b)

        loss_pde_b = tf.reduce_mean(tf.square(pde_res2))
        loss_origin_b = tf.reduce_mean(tf.square(psi_r0b))
        loss_b = loss_pde_b + LAMBDA_ORIGIN * loss_origin_b

    grads = tape.gradient(loss_b, phi_model.trainable_variables)
    optimizer.apply_gradients(zip(grads, phi_model.trainable_variables))
    return loss_b, loss_pde_b, loss_origin_b

# -------------------
# Train
# -------------------
hist_total, hist_pde, hist_origin = [], [], []
for ep in range(1, EPOCHS + 1):
    total, lpde, lorg = train_step()
    hist_total.append(float(total.numpy()))
    hist_pde.append(float(lpde.numpy()))
    hist_origin.append(float(lorg.numpy()))
    if ep % 100 == 0 or ep == 1:
        print(f"Epoch {ep:4d} | total: {hist_total[-1]:.3e} | PDE: {hist_pde[-1]:.3e} | origin: {hist_origin[-1]:.3e}")

# -------------------
# Evaluate & Plot
# -------------------
def psi_pred(r1d):
    r_tf = tf.constant(r1d.reshape(-1,1), dtype=tf.float32)
    phi = phi_model(r_tf, training=False).numpy()
    return ((1.0 - r1d.reshape(-1,1)) * phi).ravel()

rs = np.linspace(0.0, 1.0, 400, dtype=np.float32)
psi_nn = psi_pred(rs)
psi_ex = psi_exact(rs)

plt.figure(figsize=(7,5))
plt.plot(rs, psi_ex, 'k--', label='Exact  $r^3 + r^2 - 2$')
plt.plot(rs, psi_nn, label='NN (radial, hard BC)')
plt.xlabel('radius  $r$')
plt.ylabel(r'$\psi(r)$')
plt.title('Radial Poisson in $\mathbb{R}^3$: NN vs Exact')
plt.grid(alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()

plt.figure(figsize=(7,5))
plt.semilogy(hist_total, label='total')
plt.semilogy(hist_pde,   label='PDE')
plt.semilogy(hist_origin,label='origin penalty')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.title('Training losses')
plt.grid(True, which='both', alpha=0.5)
plt.legend()
plt.tight_layout()
plt.show()
```

# **Hamilton-Jacobi Equation**

**Free 1D particle trajectory found using the action function**

In this example the Lagrangian is $L=\frac{1}{2} \dot q^2$ and the momentum is given by $p= \dot q$.
We look for the solution which satisfies the boundary conditions $q(0)=1$ and $q(1) =3$, by means of the action $S$ by solving
$
\dot q = \frac{\partial S}{\partial q}.
$

The function $S(q,t)$ satisfies the Hamilton--Jacobi equation
$$\partial_t S + \frac{1}{2} (\partial_q S)^2 =0, $$
with initial condition $S(q_0, t) =0$, for all $t$.

```python
# ============================================================
# PINN for Hamilton–Jacobi (free particle, q0=1) + trajectory
# Part 1: learn S(q,t) solving S_t + 1/2 (S_q)^2 = 0
#         with S(q,1) = 0.5*(q-1)^2 and S(1,t) = 0
# Part 2: learn q(t) via dq/dt = S_q(q(t), t), with q(0)=1, q(1)=3
# Exact solutions: S(q,t) = (q-1)^2 / (2t),  q(t) = 1 + 2t
# ============================================================

import numpy as np
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import matplotlib.pyplot as plt

# -------------------------
# Global precision & seeds
# -------------------------
tf.keras.backend.set_floatx("float64")
np.random.seed(0)
tf.random.set_seed(0)

q0_const = 1.0  # anchor point

# -------------------------
# Small helper: MLP builder
# -------------------------
def build_mlp(input_dim, width=64, depth=4, act="tanh", out_dim=1):
    inputs = keras.Input(shape=(input_dim,))
    x = inputs
    for _ in range(depth):
        x = layers.Dense(width, activation=act)(x)
    outputs = layers.Dense(out_dim, activation=None)(x)
    return keras.Model(inputs, outputs)

# ============================================================
# Part 1: Train S(q,t) for free particle HJ PDE
# ============================================================
S_model = build_mlp(input_dim=2, width=64, depth=4, act="tanh", out_dim=1)
opt_S = keras.optimizers.Adam(1e-3)

# Domain & training hyperparameters
q_min, q_max = 0.0, 4.0     # cover around q0=1 and target q=3
t_min, t_max = 0.01, 1.0     # avoid t=0 singularity

EPOCHS_S = 1000
N_pde     = 512              # collocation points for PDE residual
N_anchor  = 256              # anchor line q=1

lambda_anchor = 1.0

@tf.function
def train_step_S():
    # Collocation points for PDE
    q_pde = tf.random.uniform((N_pde, 1), minval=q_min, maxval=q_max, dtype=tf.float64)
    t_pde = tf.random.uniform((N_pde, 1), minval=t_min, maxval=t_max, dtype=tf.float64)


    # Anchor at q=q0: S(q0,t) = 0
    t_anchor = tf.random.uniform((N_anchor, 1), minval=t_min, maxval=t_max, dtype=tf.float64)
    q_anchor = tf.ones_like(t_anchor, dtype=tf.float64) * q0_const

    with tf.GradientTape() as tape:
        # PDE residual: S_t + 1/2 (S_q)^2 = 0
        with tf.GradientTape(persistent=True) as pdetape:
            pdetape.watch([q_pde, t_pde])
            S_vals = S_model(tf.concat([q_pde, t_pde], axis=1))
        S_q = pdetape.gradient(S_vals, q_pde)
        S_t = pdetape.gradient(S_vals, t_pde)
        del pdetape

        res = S_t + 0.5 * tf.square(S_q)
        loss_pde = tf.reduce_mean(tf.square(res))


        # Anchor at q=q0
        S_anch = S_model(tf.concat([q_anchor, t_anchor], axis=1))
        loss_anchor = tf.reduce_mean(tf.square(S_anch - 0.0))

        loss_total = loss_pde + lambda_anchor * loss_anchor

    grads = tape.gradient(loss_total, S_model.trainable_variables)
    opt_S.apply_gradients(zip(grads, S_model.trainable_variables))
    return loss_pde, loss_anchor, loss_total

print("==> Training S(q,t) with q0 = 1")
for epoch in range(1, EPOCHS_S + 1):
    lp, lanc, ltot = train_step_S()
    if epoch % 500 == 0:
        print(f"[S] epoch {epoch:4d} | PDE {lp.numpy():.3e} | anchor {lanc.numpy():.3e} | total {ltot.numpy():.3e}")

# ============================================================
# Part 2: Train q(t) with hard BCs (q(0)=1, q(1)=3)
# ============================================================
# Embed BCs exactly: q(t) = 1 + 2 t + t(1-t) * n(t)
base_q = build_mlp(input_dim=1, width=64, depth=3, act="tanh", out_dim=1)

class QModel(keras.Model):
    def __init__(self, base):
        super().__init__()
        self.base = base
    def call(self, t):
        return (1.0 + 2.0 * t) + t * (1.0 - t) * self.base(t)

q_model = QModel(base_q)
opt_q = keras.optimizers.Adam(1e-3)

EPOCHS_Q = 500
N_t = 256
tmin_train = 0.01   # avoid the t=0 singular behavior of S_q = (q-1)/t
tmax_train = 1.0

@tf.function
def train_step_q():
    # Collocation in time (avoid t=0)
    t_batch = tf.random.uniform((N_t, 1), minval=tmin_train, maxval=tmax_train, dtype=tf.float64)

    with tf.GradientTape(persistent=True) as tape:
        tape.watch(t_batch)
        # q(t)
        q_pred = q_model(t_batch)
        # dq/dt
        dq_dt = tape.gradient(q_pred, t_batch)

        # Compute S(q,t) and then S_q using the SAME tape so we can backprop through S_q
        S_vals = S_model(tf.concat([q_pred, t_batch], axis=1))
        S_q = tape.gradient(S_vals, q_pred)  # dS/dq at (q(t), t)

        # Dynamics residual
        loss_dyn = tf.reduce_mean(tf.square(dq_dt - S_q))

    grads = tape.gradient(loss_dyn, q_model.trainable_variables)
    opt_q.apply_gradients(zip(grads, q_model.trainable_variables))
    del tape
    return loss_dyn

print("\n==> Training q(t)")
for epoch in range(1, EPOCHS_Q + 1):
    ld = train_step_q()
    if epoch % 500 == 0:
        print(f"[q] epoch {epoch:4d} | dyn {ld.numpy():.3e}")

# ============================================================
# Plots
# ============================================================


#  Plot learned q(t) vs exact q_exact(t) = 1 + 2 t
t_grid = tf.cast(tf.linspace(0.0, 1.0, 201), tf.float64)[:, None]
q_pred_grid = q_model(t_grid).numpy()
q_exact_grid = (1.0 + 2.0 * t_grid.numpy())

mse_q = np.mean((q_pred_grid - q_exact_grid)**2)
print(f"\nMSE for q(t) vs q_exact(t)=1+2t on 201-point grid: {mse_q:.3e}")

plt.figure(figsize=(6,4))
plt.plot(t_grid, q_exact_grid, label=r"$q_{\mathrm{exact}}(t)=1+2t$")
plt.plot(t_grid, q_pred_grid, label=r"$q_{\mathrm{NN}}(t)$", linestyle="--")
plt.xlabel(r"$t$")
plt.ylabel(r"$q(t)$")
plt.title("Trajectory: exact vs neural network")
plt.legend()
plt.tight_layout()
plt.show()
```

# **Eiconal Equation**

**The Hyperbolic Oscillator**

The next code plots the solution for the hyperbolic oscillator with boundary conditions $q_0=1$, $q=q(3)=6$ using its closed form formula:

$$q(s) = q_0 \cosh(as)
+ \frac{q - q_0 \cosh(at)}{\sinh(at)} \sinh(as),
\qquad 0 \leq s \leq t.$$

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
q0 = 1
q = 6
t = 3
a = 1  # you can change 'a' if needed

# Define the function q(s)
def q_func(s, q0, q, t, a):
    numerator = q - q0 * np.cosh(a * t)
    denominator = np.sinh(a * t)
    return q0 * np.cosh(a * s) + (numerator / denominator) * np.sinh(a * s)

# Generate values of s between 0 and t
s_vals = np.linspace(0, t, 2000)
q_vals = q_func(s_vals, q0, q, t, a)

# Plot
plt.figure(figsize=(8, 5))
plt.plot(s_vals, q_vals, label=fr"$q(s)$ with $q_0={q0}, q={q}, t={t}, a={a}$", linewidth=2)
plt.xlabel(r"$s$", fontsize=14)
plt.ylabel(r"$q(s)$", fontsize=14)
plt.title("Plot of $q(s)$", fontsize=16)
plt.grid(True, alpha=0.3)
plt.legend(fontsize=12)
plt.show()
```

The next code finds the trajectory solution with boundary conditions $q_0=1$, $q=q(3)=6$ using the eiconal equation $(\partial_q \Phi)^2 = 2C$, with the initial condition $\Phi(q_0) =0$.

Then it compares it with the exact solution. We note some little gap for small values of $s$. The training is done for 800 epochs.

```python
import numpy as np
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import matplotlib.pyplot as plt

tf.keras.backend.set_floatx("float64")
np.random.seed(0); tf.random.set_seed(0)

# Problem
a = 1.0; q0 = 1.0; qT = 6.0; tT = 3.0

# Exact trajectory for comparison
def q_exact(s):
    num = (qT - q0 * np.cosh(a * tT))
    den = np.sinh(a * tT)
    return q0 * np.cosh(a * s) + (num / den) * np.sinh(a * s)

# Small MLPs
def mlp(in_dim, width=64, depth=3, act="swish", out_dim=1):
    x = inp = keras.Input(shape=(in_dim,))
    for _ in range(depth):
        x = layers.Dense(width, activation=act)(x)
    out = layers.Dense(out_dim, activation=None)(x)
    return keras.Model(inp, out)

# Phase network Phi(q)
phi_net = mlp(in_dim=1, width=164, depth=3, act="swish", out_dim=1)

# Trajectory base network n(s) used inside a hard-BC embedding for q(s)
n_net = mlp(in_dim=1, width=164, depth=3, act="swish", out_dim=1)

# --- Hard-BC embedding: q(0)=q0, q(tT)=qT exactly ---
class QModel(keras.Model):
    def __init__(self, base):
        super().__init__(); self.base = base
    def call(self, s):
        # linear baseline + interior deformation
        return q0 + (qT - q0) * (s / tT) + s * (tT - s) * self.base(s)

q_model = QModel(n_net)

# --- Global scalar energy C (trainable variable), not a function of s ---
C_var = tf.Variable(0.1, dtype=tf.float64, trainable=True, name="C")

opt = keras.optimizers.Adam(1e-3)

# Weights (make dynamics strong)
w_dyn = 100.0
w_phi0 = 1e-2

EPOCHS = 800
N = 656
eps = 1e-3  # avoid s=0 in collocation for gradients

@tf.function
def train_step():
    s = tf.random.uniform((N,1), minval=eps, maxval=tT, dtype=tf.float64)

    with tf.GradientTape(persistent=True) as tape:
        tape.watch(s)

        # q(s) and dq/ds (same tape so grads flow through everything)
        q_pred = q_model(s)
        dq_ds  = tape.gradient(q_pred, s)

        # Phi(q) and Phi_q
        Phi    = phi_net(q_pred)
        Phi_q  = tape.gradient(Phi, q_pred)

        # Hamiltonian H = 1/2 Phi_q^2 - 1/2 a^2 q^2
        H_val = 0.5 * tf.square(Phi_q) - 0.5 * (a**2) * tf.square(q_pred)

        # Loss terms
        # 1) generalized eiconal with GLOBAL constant energy C
        L_eic = tf.reduce_mean(tf.square(H_val - C_var))

        # 2) characteristic relation: dq/ds = Phi_q
        L_dyn = tf.reduce_mean(tf.square(dq_ds - Phi_q))

        # 3) anchor the phase at q0 to fix additive constant
        Phi_q0 = phi_net(tf.constant([[q0]], dtype=tf.float64))
        L_phi0 = tf.reduce_mean(tf.square(Phi_q0 - 0.0))

        L_tot = L_eic + w_dyn * L_dyn + w_phi0 * L_phi0

    vars_all = q_model.trainable_variables + phi_net.trainable_variables + [C_var]
    grads = tape.gradient(L_tot, vars_all)
    opt.apply_gradients(zip(grads, vars_all))
    del tape
    return L_eic, L_dyn, L_phi0, L_tot

for ep in range(1, EPOCHS+1):
    Le, Ld, Lp0, Lt = train_step()
    if ep % 500 == 0:
        print(f"ep {ep:5d} | eic {Le.numpy():.3e} | dyn {Ld.numpy():.3e} | phi0 {Lp0.numpy():.3e} | tot {Lt.numpy():.3e} | C={C_var.numpy():.4f}")

# Evaluate
s_grid = np.linspace(0, tT, 400)[:,None]
q_nn = q_model(tf.constant(s_grid, dtype=tf.float64)).numpy()
q_ex = q_exact(s_grid)
mse = np.mean((q_nn - q_ex)**2)
print(f"\nMSE(q_nn, q_exact) on 400-pt grid: {mse:.3e}")

plt.figure(figsize=(7,4.5))
plt.plot(s_grid, q_ex, label=r"$q_{\rm exact}(s)$", linewidth=2)
plt.plot(s_grid, q_nn, "--", label=r"$q_\theta(s)$ (NN)", linewidth=2)
plt.xlabel(r"$s$"); plt.ylabel(r"$q(s)$")
plt.title(rf"Trajectory on [0,{tT}] with $a={a}$, $q_0={q0}$, $q={qT}$  (C global)")
plt.grid(alpha=0.3); plt.legend(); plt.tight_layout(); plt.show()
```

In order to close that gap, in th enext code we increase the training to 8,000 epochs and we also plot the losses to see whether they get saturated.

We also introduce a biased choice of collocation points, replacing th euniform collocation points from the last code.

```python
# ============================================================
# PINN for eiconal + dynamics with three-head NN and loss plot
# Problem: L = 1/2 dot{q}^2 + (a^2/2) q^2  =>  H = 1/2 p^2 - (a^2/2) q^2
# Learn:  q(s) on [0,tT] with q(0)=q0, q(tT)=qT; also Phi(q) and a GLOBAL C
# Includes: hard BCs for q, initial-slope penalty, importance sampling near s=0
# Outputs: trajectory plot and loss-vs-epochs plot
# ============================================================

import numpy as np
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import matplotlib.pyplot as plt

# -------------------------
# Global precision & seeds
# -------------------------
tf.keras.backend.set_floatx("float64")
np.random.seed(0); tf.random.set_seed(0)

# -------------------------
# Problem parameters
# -------------------------
a  = 1.0
q0 = 1.0
qT = 6.0
tT = 3.0

# Exact trajectory for comparison
def q_exact(s):
    num = (qT - q0 * np.cosh(a * tT))
    den = np.sinh(a * tT)
    return q0 * np.cosh(a * s) + (num / den) * np.sinh(a * s)

# Exact initial slope compatible with endpoints
v0 = a * (qT - q0 * np.cosh(a * tT)) / np.sinh(a * tT)

# -------------------------
# Small MLP builder
# -------------------------
def mlp(in_dim, width=64, depth=3, act="tanh", out_dim=1):
    x = inp = keras.Input(shape=(in_dim,))
    for _ in range(depth):
        x = layers.Dense(width, activation=act)(x)
    out = layers.Dense(out_dim, activation=None)(x)
    return keras.Model(inp, out)

# Phase network Phi(q)
phi_net = mlp(in_dim=1, width=64, depth=3, act="tanh", out_dim=1)

# Base network n(s) used inside a hard-BC embedding for q(s)
n_net = mlp(in_dim=1, width=64, depth=3, act="tanh", out_dim=1)

# --- Hard-BC embedding: q(0)=q0, q(tT)=qT exactly ---
class QModel(keras.Model):
    def __init__(self, base):
        super().__init__(); self.base = base
    def call(self, s):
        # linear baseline + interior deformation that vanishes at the endpoints
        return q0 + (qT - q0) * (s / tT) + s * (tT - s) * self.base(s)

q_model = QModel(n_net)

# --- Global scalar energy C (trainable), not a function of s ---
C_var = tf.Variable(0.1, dtype=tf.float64, trainable=True, name="C")

# Optimizer
opt = keras.optimizers.Adam(1e-3)

# -------------------------
# Training hyperparameters
# -------------------------
EPOCHS = 8000
N      = 256        # collocation per epoch
eps    = 1e-3       # avoid s=0 exactly for gradients
k_bias = 2.0        # importance-sampling exponent; >1 biases to small s

# Loss weights
w_dyn  = 100.0      # strengthen dq/ds = Phi_q
w_phi0 = 1e-3       # anchor Phi at q0
w_s0   = 10.0       # initial-slope penalty

# For plotting loss vs epochs
hist_epochs  = []
hist_eic     = []
hist_dyn     = []
hist_phi0    = []
hist_s0      = []
hist_total   = []
hist_C       = []

@tf.function
def train_step():
    # Importance sampling near s=0: s = tT * u^k
    u = tf.random.uniform((N,1), 0.0, 1.0, dtype=tf.float64)
    s = tT * tf.pow(u, k_bias) + eps

    with tf.GradientTape(persistent=True) as tape:
        tape.watch(s)

        # q(s) and dq/ds (same tape so grads flow everywhere)
        q_pred = q_model(s)
        dq_ds  = tape.gradient(q_pred, s)

        # Phi(q) and Phi_q
        Phi   = phi_net(q_pred)
        Phi_q = tape.gradient(Phi, q_pred)

        # Hamiltonian H = 1/2 Phi_q^2 - 1/2 a^2 q^2
        H_val = 0.5 * tf.square(Phi_q) - 0.5 * (a**2) * tf.square(q_pred)

        # Loss terms
        # 1) generalized eiconal with GLOBAL C
        L_eic = tf.reduce_mean(tf.square(H_val - C_var))

        # 2) dynamics: dq/ds = Phi_q
        L_dyn = tf.reduce_mean(tf.square(dq_ds - Phi_q))

        # 3) anchor the phase at q0 to fix additive constant
        Phi_at_q0 = phi_net(tf.constant([[q0]], dtype=tf.float64))
        L_phi0 = tf.reduce_mean(tf.square(Phi_at_q0 - 0.0))

        # 4) initial slope penalty at s ~ 0
        s0 = tf.constant([[eps]], dtype=tf.float64)
        with tf.GradientTape() as tape0:
            tape0.watch(s0)
            q0_from_model = q_model(s0)
        dq_ds_at0 = tape0.gradient(q0_from_model, s0)
        L_s0 = tf.reduce_mean(tf.square(dq_ds_at0 - v0))

        L_tot = L_eic + w_dyn * L_dyn + w_phi0 * L_phi0 + w_s0 * L_s0

    vars_all = q_model.trainable_variables + phi_net.trainable_variables + [C_var]
    grads = tape.gradient(L_tot, vars_all)
    opt.apply_gradients(zip(grads, vars_all))
    del tape
    return L_eic, L_dyn, L_phi0, L_s0, L_tot

# -------------------------
# Train
# -------------------------
for ep in range(1, EPOCHS+1):
    Le, Ld, Lp0, Ls0, Lt = train_step()

    # record history
    hist_epochs.append(ep)
    hist_eic.append(Le.numpy())
    hist_dyn.append(Ld.numpy())
    hist_phi0.append(Lp0.numpy())
    hist_s0.append(Ls0.numpy())
    hist_total.append(Lt.numpy())
    hist_C.append(C_var.numpy())

    if ep % 500 == 0:
        print(f"ep {ep:5d} | eic {Le.numpy():.3e} | dyn {Ld.numpy():.3e} | "
              f"phi0 {Lp0.numpy():.3e} | s0 {Ls0.numpy():.3e} | "
              f"tot {Lt.numpy():.3e} | C={C_var.numpy():.4f}")

# -------------------------
# Evaluate & Plot (trajectory)
# -------------------------
s_grid = np.linspace(0, tT, 400)[:,None]
q_nn = q_model(tf.constant(s_grid, dtype=tf.float64)).numpy()
q_ex = q_exact(s_grid)
mse = np.mean((q_nn - q_ex)**2)
print(f"\nMSE(q_nn, q_exact) on 400-pt grid: {mse:.3e}")
print(f"Learned global C: {hist_C[-1]:.6f}")

plt.figure(figsize=(7,4.5))
plt.plot(s_grid, q_ex, label=r"$q_{\rm exact}(s)$", linewidth=2)
plt.plot(s_grid, q_nn, "--", label=r"$q_\theta(s)$ (NN)", linewidth=2)
plt.xlabel(r"$s$")
plt.ylabel(r"$q(s)$")
plt.title(rf"Trajectory on [0,{tT}] with $a={a}$, $q_0={q0}$, $q={qT}$  (C global)")
plt.grid(alpha=0.3); plt.legend(); plt.tight_layout(); plt.show()

# -------------------------
# Plot Loss vs Epochs (log scale)
# -------------------------
plt.figure(figsize=(7,4.5))
plt.plot(hist_epochs, hist_total, label="Total Loss", linewidth=2)
plt.plot(hist_epochs, hist_eic,   "--", label="Eiconal Loss")
plt.plot(hist_epochs, hist_dyn,   "--", label="Dynamics Loss")
plt.plot(hist_epochs, hist_phi0,  "--", label="Phase Anchor Loss")
plt.plot(hist_epochs, hist_s0,    "--", label="Init Slope Loss")
plt.yscale("log")
plt.xlabel("Epochs"); plt.ylabel("Loss")
plt.title("Loss vs. Epochs")
plt.grid(alpha=0.3); plt.legend(); plt.tight_layout(); plt.show()

# (Optional) Save figures for LaTeX:
# plt.savefig("trajectory.pdf", bbox_inches="tight")
# plt.savefig("loss_vs_epochs.pdf", bbox_inches="tight")
```

Since the gap did not disapeared, we increase the training to 10,000 epochs.

```python
# ============================================================
# PINN for eiconal + dynamics with three-head NN and loss plot
# Problem: L = 1/2 dot{q}^2 + (a^2/2) q^2  =>  H = 1/2 p^2 - (a^2/2) q^2
# Learn:  q(s) on [0,tT] with q(0)=q0, q(tT)=qT; also Phi(q) and a GLOBAL C
# Includes: hard BCs for q, importance sampling near s=0
# Outputs: trajectory plot and loss-vs-epochs plot
# NOTE: No initial-velocity (s=0 slope) loss here
# ============================================================

import numpy as np
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import matplotlib.pyplot as plt

# -------------------------
# Global precision & seeds
# -------------------------
tf.keras.backend.set_floatx("float64")
np.random.seed(0); tf.random.set_seed(0)

# -------------------------
# Problem parameters
# -------------------------
a  = 1.0
q0 = 1.0
qT = 6.0
tT = 3.0

# Exact trajectory for comparison
def q_exact(s):
    num = (qT - q0 * np.cosh(a * tT))
    den = np.sinh(a * tT)
    return q0 * np.cosh(a * s) + (num / den) * np.sinh(a * s)

# -------------------------
# Small MLP builder
# -------------------------
def mlp(in_dim, width=64, depth=3, act="tanh", out_dim=1):
    x = inp = keras.Input(shape=(in_dim,))
    for _ in range(depth):
        x = layers.Dense(width, activation=act)(x)
    out = layers.Dense(out_dim, activation=None)(x)
    return keras.Model(inp, out)

# Phase network Phi(q)
phi_net = mlp(in_dim=1, width=64, depth=3, act="tanh", out_dim=1)

# Base network n(s) used inside a hard-BC embedding for q(s)
n_net = mlp(in_dim=1, width=64, depth=3, act="tanh", out_dim=1)

# --- Hard-BC embedding: q(0)=q0, q(tT)=qT exactly ---
class QModel(keras.Model):
    def __init__(self, base):
        super().__init__(); self.base = base
    def call(self, s):
        # linear baseline + interior deformation that vanishes at the endpoints
        return q0 + (qT - q0) * (s / tT) + s * (tT - s) * self.base(s)

q_model = QModel(n_net)

# --- Global scalar energy C (trainable), not a function of s ---
C_var = tf.Variable(0.1, dtype=tf.float64, trainable=True, name="C")

# Optimizer
opt = keras.optimizers.Adam(5e-4)

# -------------------------
# Training hyperparameters
# -------------------------
EPOCHS = 10000
N      = 356        # collocation per epoch
eps    = 1e-3       # avoid s=0 exactly for gradients
k_bias = 2.9        # importance-sampling exponent; >1 biases to small s

# Loss weights
w_dyn  = 100.0      # strengthen dq/ds = Phi_q
w_phi0 = 1e-3       # anchor Phi at q0

# For plotting loss vs epochs
hist_epochs  = []
hist_eic     = []
hist_dyn     = []
hist_phi0    = []
hist_total   = []
hist_C       = []

@tf.function
def train_step():
    # Importance sampling near s=0: s = tT * u^k
    u = tf.random.uniform((N,1), 0.0, 1.0, dtype=tf.float64)
    s = tT * tf.pow(u, k_bias) + eps

    with tf.GradientTape(persistent=True) as tape:
        tape.watch(s)

        # q(s) and dq/ds (same tape so grads flow everywhere)
        q_pred = q_model(s)
        dq_ds  = tape.gradient(q_pred, s)

        # Phi(q) and Phi_q
        Phi   = phi_net(q_pred)
        Phi_q = tape.gradient(Phi, q_pred)

        # Hamiltonian H = 1/2 Phi_q^2 - 1/2 a^2 q^2
        H_val = 0.5 * tf.square(Phi_q) - 0.5 * (a**2) * tf.square(q_pred)

        # Loss terms
        # 1) generalized eiconal with GLOBAL C
        L_eic = tf.reduce_mean(tf.square(H_val - C_var))

        # 2) dynamics: dq/ds = Phi_q
        L_dyn = tf.reduce_mean(tf.square(dq_ds - Phi_q))

        # 3) anchor the phase at q0 to fix additive constant
        Phi_at_q0 = phi_net(tf.constant([[q0]], dtype=tf.float64))
        L_phi0 = tf.reduce_mean(tf.square(Phi_at_q0 - 0.0))

        L_tot = L_eic + w_dyn * L_dyn + w_phi0 * L_phi0

    vars_all = q_model.trainable_variables + phi_net.trainable_variables + [C_var]
    grads = tape.gradient(L_tot, vars_all)
    opt.apply_gradients(zip(grads, vars_all))
    del tape
    return L_eic, L_dyn, L_phi0, L_tot

# -------------------------
# Train
# -------------------------
for ep in range(1, EPOCHS+1):
    Le, Ld, Lp0, Lt = train_step()

    # record history
    hist_epochs.append(ep)
    hist_eic.append(Le.numpy())
    hist_dyn.append(Ld.numpy())
    hist_phi0.append(Lp0.numpy())
    hist_total.append(Lt.numpy())
    hist_C.append(C_var.numpy())

    if ep % 500 == 0:
        print(f"ep {ep:5d} | eic {Le.numpy():.3e} | dyn {Ld.numpy():.3e} | "
              f"phi0 {Lp0.numpy():.3e} | tot {Lt.numpy():.3e} | C={C_var.numpy():.4f}")

# -------------------------
# Evaluate & Plot (trajectory)
# -------------------------
s_grid = np.linspace(0, tT, 400)[:,None]
q_nn = q_model(tf.constant(s_grid, dtype=tf.float64)).numpy()
q_ex = q_exact(s_grid)
mse = np.mean((q_nn - q_ex)**2)
print(f"\nMSE(q_nn, q_exact) on 400-pt grid: {mse:.3e}")
print(f"Learned global C: {hist_C[-1]:.6f}")

plt.figure(figsize=(7,4.5))
plt.plot(s_grid, q_ex, label=r"$q_{\rm exact}(s)$", linewidth=2)
plt.plot(s_grid, q_nn, "--", label=r"$q_\theta(s)$ (NN)", linewidth=2)
plt.xlabel(r"$s$")
plt.ylabel(r"$q(s)$")
plt.title(rf"Trajectory on [0,{tT}] with $a={a}$, $q_0={q0}$, $q={qT}$  (C global)")
plt.grid(alpha=0.3); plt.legend(); plt.tight_layout(); plt.show()

# -------------------------
# Plot Loss vs Epochs (log scale)
# -------------------------
plt.figure(figsize=(7,4.5))
plt.plot(hist_epochs, hist_total, label="Total Loss", linewidth=2)
plt.plot(hist_epochs, hist_eic,   "--", label="Eiconal Loss")
plt.plot(hist_epochs, hist_dyn,   "--", label="Dynamics Loss")
plt.plot(hist_epochs, hist_phi0,  "--", label="Phase Anchor Loss")
plt.yscale("log")
plt.xlabel("Epochs"); plt.ylabel("Loss")
plt.title("Loss vs. Epochs (no initial-slope loss)")
plt.grid(alpha=0.3); plt.legend(); plt.tight_layout(); plt.show()

# (Optional) Save figures for LaTeX:
# plt.savefig("trajectory.pdf", bbox_inches="tight")
# plt.savefig("loss_vs_epochs.pdf", bbox_inches="tight")
```

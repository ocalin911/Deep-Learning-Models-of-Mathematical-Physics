# **HNN for harmonic Oscillator**

Many physical systems, such as a spring--mass system, can be modeled by a quadratic Hamiltonian function of the form
$$
H(q, p) = \frac{1}{2} p^2 + \frac{1}{2} q^2.
$$


We consider the corresponding Hamiltonian system

$$
q' = p, \quad p' = -q,
$$

and solve it for 25 pairs of initial conditions $(q_0, p_0)$ using the analytical solutions

$$
q(t) = q_0 \cos t + p_0 \sin t, \qquad
p(t) = p_0 \cos t - q_0 \sin t.
$$

The energy along each trajectory is given by $H_0 = \tfrac{1}{2}(q_0^2 + p_0^2)$, with $H_0 \in [0.2, 1]$.

For each trajectory, we uniformly sample 30 points $(q(t), p(t))$ and compute the corresponding phase-space derivatives $(q'(t), p'(t))$.
To simulate measurement uncertainty, Gaussian noise with variance $\sigma^2 = 0.1$ is added to the data.

These noisy pairs constitute the training dataset.


```python
#@title Hamiltonian NN (Natural) vs Baseline — Train, Validate vs Exact, Energies (Aligned) + Phase Portrait

import numpy as np
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import matplotlib.pyplot as plt

print("TensorFlow:", tf.__version__)

# -----------------------
# Configuration
# -----------------------
SEED = 42
np.random.seed(SEED)
tf.random.set_seed(SEED)

# Data generation
N_TRAJ = 25            #@param {type:"integer"}
SAMPLES_PER_TRAJ = 30  #@param {type:"integer"}
T_MIN, T_MAX = 0.0, 2.0 * np.pi
ENERGY_MIN, ENERGY_MAX = 0.2, 1.0

# Noise: sigma^2 = 0.1 => sigma = sqrt(0.1)
NOISE_STD = float(np.sqrt(0.1))

# Training
BASELINE_LR = 1e-3     #@param {type:"number"}
HNN_LR = 1e-3          #@param {type:"number"}
EPOCHS_BASELINE = 2000 #@param {type:"integer"}
EPOCHS_HNN = 2000      #@param {type:"integer"}
BATCH_SIZE = 128       #@param {type:"integer"}

# Validation rollout config
q0_val = 0.8     #@param {type:"number"}
p0_val = -0.3    #@param {type:"number"}
T_roll = 12.56   #@param {type:"number"}  # ~4 periods
dt = 0.02        #@param {type:"number"}

# -----------------------
# Ground-truth SHO system
# H = 1/2 p^2 + 1/2 q^2 => q' = p, p' = -q
# -----------------------
def true_qprime(q, p): return p
def true_pprime(q, p): return -q

def solve_trajectory(q0, p0, t):
    q = q0 * np.cos(t) + p0 * np.sin(t)
    p = p0 * np.cos(t) - q0 * np.sin(t)
    return q, p

def sample_initial_conditions(n_traj, energy_min, energy_max, rng):
    H0 = rng.uniform(energy_min, energy_max, size=(n_traj,))
    r = np.sqrt(2.0 * H0)  # radius in (q,p)
    phi = rng.uniform(0, 2.0 * np.pi, size=(n_traj,))
    q0 = r * np.cos(phi)
    p0 = r * np.sin(phi)
    return q0, p0, H0

def generate_dataset(n_traj=25, samples_per_traj=30, noise_std=np.sqrt(0.1), seed=42):
    rng = np.random.default_rng(seed)
    q0_all, p0_all, H0_all = sample_initial_conditions(n_traj, ENERGY_MIN, ENERGY_MAX, rng)
    X_list, Y_list = [], []
    X_hnn_list, Y_hnn_list = [], []  # HNN: input q, label p'

    for q0, p0 in zip(q0_all, p0_all):
        t = rng.uniform(T_MIN, T_MAX, size=(samples_per_traj,))
        t.sort()
        q, p = solve_trajectory(q0, p0, t)
        qprime = true_qprime(q, p)
        pprime = true_pprime(q, p)

        # Add Gaussian noise
        q_noisy = q + rng.normal(0.0, noise_std, size=q.shape)
        p_noisy = p + rng.normal(0.0, noise_std, size=p.shape)
        qprime_noisy = qprime + rng.normal(0.0, noise_std, size=qprime.shape)
        pprime_noisy = pprime + rng.normal(0.0, noise_std, size=pprime.shape)

        # Baseline: X=(q,p), Y=(q',p')
        X_list.append(np.stack([q_noisy, p_noisy], axis=1))
        Y_list.append(np.stack([qprime_noisy, pprime_noisy], axis=1))

        # HNN reduced: input q, target p' (noisy); loss enforces U'(q)+p'~0
        X_hnn_list.append(q_noisy.reshape(-1, 1))
        Y_hnn_list.append(pprime_noisy.reshape(-1, 1))

    X = np.concatenate(X_list, axis=0).astype(np.float32)
    Y = np.concatenate(Y_list, axis=0).astype(np.float32)
    X_hnn = np.concatenate(X_hnn_list, axis=0).astype(np.float32)
    Y_hnn = np.concatenate(Y_hnn_list, axis=0).astype(np.float32)
    return (X, Y), (X_hnn, Y_hnn), H0_all

# Generate data
(X, Y), (X_hnn, Y_hnn), H0_all = generate_dataset(
    n_traj=N_TRAJ, samples_per_traj=SAMPLES_PER_TRAJ, noise_std=NOISE_STD, seed=SEED
)
print(f"Data -> Baseline X: {X.shape}, Y: {Y.shape}; HNN X: {X_hnn.shape}, Y: {Y_hnn.shape}")

# -----------------------
# Baseline model: (q,p)->(q',p')
# -----------------------
def build_baseline():
    model = keras.Sequential([
        layers.Input(shape=(2,)),
        layers.Dense(200, activation='tanh'),
        layers.Dense(200, activation='tanh'),
        layers.Dense(2, activation=None),
    ])
    model.compile(optimizer=keras.optimizers.Adam(learning_rate=BASELINE_LR), loss='mse')
    return model

baseline = build_baseline()
print("Training baseline model...")
baseline.fit(X, Y, epochs=EPOCHS_BASELINE, batch_size=BATCH_SIZE, verbose=0)
baseline_loss = baseline.evaluate(X, Y, verbose=0)
print(f"Baseline training MSE: {baseline_loss:.6f}")

# -----------------------
# HNN (natural): learn U(q) with L_red = ||U'(q) + p'||^2
# -----------------------
def build_potential_net():
    q_in = keras.Input(shape=(1,))
    x = layers.Dense(200, activation='tanh')(q_in)
    x = layers.Dense(200, activation='tanh')(x)
    U = layers.Dense(1, activation=None)(x)  # scalar potential
    return keras.Model(q_in, U, name="U_theta")

class HNNReduced(keras.Model):
    def __init__(self, potential_net, learning_rate=1e-3):
        super().__init__()
        self.potential_net = potential_net
        self.loss_tracker = keras.metrics.Mean(name="loss")
        self.optimizer = keras.optimizers.Adam(learning_rate=learning_rate)

    def call(self, q, training=False):
        return self.potential_net(q, training=training)

    def train_step(self, data):
        q, pprime_obs = data  # shapes: (B,1)
        # Outer tape: d(loss)/d(theta); inner tape: dU/dq
        with tf.GradientTape() as tape:
            with tf.GradientTape() as tape_q:
                tape_q.watch(q)
                U = self.potential_net(q, training=True)
            dU_dq = tape_q.gradient(U, q)   # U'(q)
            residual = dU_dq + pprime_obs   # enforce U'(q) + p' ~ 0
            loss = tf.reduce_mean(tf.square(residual))
        grads = tape.gradient(loss, self.potential_net.trainable_variables)
        self.optimizer.apply_gradients(zip(grads, self.potential_net.trainable_variables))
        self.loss_tracker.update_state(loss)
        return {"loss": self.loss_tracker.result()}

    def test_step(self, data):
        q, pprime_obs = data
        with tf.GradientTape() as tape_q:
            tape_q.watch(q)
            U = self.potential_net(q, training=False)
        dU_dq = tape_q.gradient(U, q)
        residual = dU_dq + pprime_obs
        loss = tf.reduce_mean(tf.square(residual))
        self.loss_tracker.update_state(loss)
        return {"loss": self.loss_tracker.result()}

    @property
    def metrics(self): return [self.loss_tracker]

potential_net = build_potential_net()
hnn_model = HNNReduced(potential_net, learning_rate=HNN_LR)
hnn_model.compile()

print("Training HNN-reduced model...")
hnn_model.fit(X_hnn, Y_hnn, epochs=EPOCHS_HNN, batch_size=BATCH_SIZE, verbose=0)
hnn_metrics = hnn_model.evaluate(X_hnn, Y_hnn, verbose=0, return_dict=True)
print(f"HNN reduced training loss (MSE on U'(q) + p'): {hnn_metrics['loss']:.6f}")

# -----------------------
# Utilities: RK4 rollout & energies
# -----------------------
def rk4_step(f, y, dt):
    # y = [q, p]; f returns [qdot, pdot]
    k1 = np.array(f(y), dtype=np.float32)
    k2 = np.array(f(y + 0.5*dt*k1), dtype=np.float32)
    k3 = np.array(f(y + 0.5*dt*k2), dtype=np.float32)
    k4 = np.array(f(y + dt*k3), dtype=np.float32)
    return y + (dt/6.0)*(k1 + 2*k2 + 2*k3 + k4)

def rollout(f, y0, T, dt):
    n = int(np.ceil(T/dt))
    traj = np.zeros((n+1, 2), dtype=np.float32)
    traj[0] = y0
    for i in range(n):
        traj[i+1] = rk4_step(f, traj[i].copy(), dt)
    t = np.linspace(0.0, n*dt, n+1)
    return t, traj

def baseline_dyn(baseline_model):
    def f(y):
        q, p = y
        qp = np.array([[q, p]], dtype=np.float32)
        qdot, pdot = baseline_model(qp, training=False).numpy().reshape(-1)
        return np.array([qdot, pdot], dtype=np.float32)
    return f

def hnn_force_dU(q_scalar):
    # returns dU/dq(q) using TF gradient
    q_tf = tf.convert_to_tensor([[q_scalar]], dtype=tf.float32)
    with tf.GradientTape() as tape_q:
        tape_q.watch(q_tf)
        U = potential_net(q_tf, training=False)
    dU_dq = tape_q.gradient(U, q_tf).numpy()[0,0]
    return float(dU_dq)

def hnn_dyn(potential_net):
    def f(y):
        q, p = y
        qdot = p
        pdot = -hnn_force_dU(float(q))
        return np.array([qdot, pdot], dtype=np.float32)
    return f

def energy_true(q, p):
    return 0.5*(p**2 + q**2)

def energy_hnn_raw(q, p):
    q_tf = tf.convert_to_tensor([[q]], dtype=tf.float32)
    U = potential_net(q_tf, training=False).numpy()[0,0]
    return 0.5*(p**2) + float(U)

# -----------------------
# Validation: (q0_val, p0_val), compare trajectories
# -----------------------
y0 = np.array([q0_val, p0_val], dtype=np.float32)

# Exact trajectory (analytic)
t_exact = np.arange(0.0, T_roll + 1e-12, dt)
q_exact, p_exact = solve_trajectory(y0[0], y0[1], t_exact)
traj_exact = np.stack([q_exact, p_exact], axis=1).astype(np.float32)

# Baseline rollout
t_b, traj_b = rollout(baseline_dyn(baseline), y0, T_roll, dt)

# HNN rollout
t_h, traj_h = rollout(hnn_dyn(potential_net), y0, T_roll, dt)

# -----------------------
# Energy alignment for HNN (Fix A): match at t=0
# -----------------------
H_true0 = energy_true(y0[0], y0[1])
H_hnn0  = energy_hnn_raw(y0[0], y0[1])
C_align = H_true0 - H_hnn0
def energy_hnn_aligned(q, p): return energy_hnn_raw(q, p) + C_align
print(f"Energy alignment constant C_align = {C_align:.6f}")

# -----------------------
# Plots: q(t), p(t)
# -----------------------
plt.figure(figsize=(10,4))
plt.subplot(1,2,1)
plt.plot(t_exact, traj_exact[:,0], label="Exact q(t)")
plt.plot(t_b, traj_b[:,0], '--', label="Baseline q(t)")
plt.plot(t_h, traj_h[:,0], ':', label="HNN q(t)")
plt.xlabel("time"); plt.ylabel("q(t)"); plt.legend(); plt.title("Position vs time")

plt.subplot(1,2,2)
plt.plot(t_exact, traj_exact[:,1], label="Exact p(t)")
plt.plot(t_b, traj_b[:,1], '--', label="Baseline p(t)")
plt.plot(t_h, traj_h[:,1], ':', label="HNN p(t)")
plt.xlabel("time"); plt.ylabel("p(t)"); plt.legend(); plt.title("Momentum vs time")
plt.tight_layout(); plt.show()

# -----------------------
# Plot: energy vs time (HNN energy aligned)
# -----------------------
E_exact = energy_true(traj_exact[:,0], traj_exact[:,1])
E_base_trueH = energy_true(traj_b[:,0], traj_b[:,1])
E_hnn_modelH = np.array([energy_hnn_aligned(q, p) for q,p in traj_h], dtype=np.float32)

plt.figure(figsize=(7,4))
plt.plot(t_exact, E_exact, label="Exact (true H)", linewidth=2)
plt.plot(t_b, E_base_trueH, '--', label="Baseline rollout (true H)")
plt.plot(t_h, E_hnn_modelH, ':', label="HNN rollout (learned Hθ, aligned)")
plt.xlabel("time"); plt.ylabel("Energy"); plt.title("Energy vs time (validation trajectory)")
plt.legend(); plt.tight_layout(); plt.show()

# -----------------------
# NEW: Phase portrait (q, p)
# -----------------------
plt.figure(figsize=(5.8,5))
plt.plot(traj_exact[:,0], traj_exact[:,1], label="Exact", linewidth=2)
plt.plot(traj_b[:,0],    traj_b[:,1],    '--', label="Baseline")
plt.plot(traj_h[:,0],    traj_h[:,1],    ':',  label="HNN")
plt.xlabel("q"); plt.ylabel("p"); plt.title("Phase portrait: (q, p)")
plt.legend(); plt.axis('equal'); plt.tight_layout(); plt.show()
```
